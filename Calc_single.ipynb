## 🔹 0.1 What Is Calculus?

**Calculus** is the math of:

- **Change** → Derivatives  
- **Accumulation** → Integrals

---

### 🔹 Calculus Answers:

- “How fast is something changing?” → **Derivative**  
- “How much total has accumulated?” → **Integral**

We begin with **derivatives**, but to understand them, we must master:

# 🚪 THE FOUNDATION: LIMITS

---

## 🔹 1. What Is a Limit?

### ✏️ Definition:

A **limit** describes what a function is *approaching* as the input value gets *closer and closer* to a point.

### ✅ Plain English:

> If I walk toward a wall but never touch it, the **limit** is where I’m heading.

### 📏 Formal Notation:

$$
\lim_{x \to a} f(x) = L
$$

Means: as $x$ gets close to $a$, $f(x)$ gets close to $L$.

---

### 🧠 Examples:

#### ✅ Example 1:
Let $f(x) = 3x + 1$

Then:
$$
\lim_{x \to 2} f(x) = 3(2) + 1 = 7
$$

#### ❗ Example 2 (Weird):

Let:
$$
f(x) = \begin{cases}
x^2 & \text{if } x \ne 3 \\
5 & \text{if } x = 3
\end{cases}
$$

Then:
$$
\lim_{x \to 3} f(x) = 9 \quad \text{(not 5!)}
$$

Because the **limit cares about approach**, not the value at $x = 3$.

---

## 🔹 2. Graphical Understanding

A **limit** is what the **graph is approaching near a point** — even if:

- It jumps  
- Has a hole  
- Isn’t defined at that point  

### 🎯 Rule:

A limit **exists** only if:

$$
\lim_{x \to a^-} f(x) = \lim_{x \to a^+} f(x)
$$

(left-hand = right-hand)

---

## 🔢 3. Python Code: Visualize a Limit

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a piecewise function with a hole
def f(x):
    return np.where(x == 3, 5, x**2)

x_vals = np.linspace(2, 4, 1000)
y_vals = f(x_vals)

plt.plot(x_vals, y_vals, label='f(x)', color='blue')
plt.scatter(3, 5, color='red', label='f(3) = 5', zorder=5)
plt.scatter(3, 9, color='black', marker='o', facecolors='none',
            label='hole at (3, 9)', zorder=5)
plt.axvline(3, color='gray', linestyle='--', linewidth=1)
plt.title("Limit of f(x) as x → 3")
plt.legend()
plt.grid(True)
plt.show()
```

---

### 🧠 Mnemonic:

> **"Limits look, not touch."**  
They care about what happens *near* a point, not the value *at* the point.

````python
import numpy as np

# Define the function f(x) = x^2
def f(x):
    return x**2

# Create arrays of x values approaching 3 from the left and right
x_left = np.array([2.9, 2.99, 2.999, 2.9999])
x_right = np.array([3.1, 3.01, 3.001, 3.0001])

# Evaluate f(x) for each side
f_left = f(x_left)
f_right = f(x_right)

# Print the values to show the function approaching a limit from both sides
print("Approaching x = 3 from the left:")
for x_val, f_val in zip(x_left, f_left):
    print(f"f({x_val}) = {f_val}")

print("\nApproaching x = 3 from the right:")
for x_val, f_val in zip(x_right, f_right):
    print(f"f({x_val}) = {f_val}")

# Final estimate of the limit (both sides should agree)
limit_estimate = f(3)
print(f"\nEstimated limit as x → 3 is: {limit_estimate}")
````

## 🔹 1. Concept: The Derivative as Slope of a Tangent

We want to measure **how fast a curve is changing** at a specific point.

- For straight lines → slope is constant  
- For curves → slope varies  
- We use a **tangent line** to estimate this slope — the **derivative**

---

## 🔹 2. Graphing a Curve and Its Tangent Line in Python

We’ll:

- Plot $f(x) = x^2$  
- Choose a point (e.g., $x = 2$)  
- Draw the **tangent line** at that point

```python
import numpy as np
import matplotlib.pyplot as plt  # For graphing

# Define the function we're analyzing
def f(x):
    return x**2  # This is the curve y = x²

# Choose the x-value we're interested in (the point of tangency)
x0 = 2
y0 = f(x0)  # y-value at that point

# List of step sizes (h) to try — smaller h means closer to the tangent line
h_values = [1, 0.5, 0.1, 0.01]

# Set up x-values for plotting the curve and lines
x_vals = np.linspace(0, 4, 200)  # 200 points from x=0 to x=4
y_vals = f(x_vals)  # Calculate the corresponding y-values for the curve

# Create the plot
plt.figure(figsize=(10, 6))

# Plot the original curve
plt.plot(x_vals, y_vals, label='f(x) = x²', color='blue')

# Mark the point we're analyzing
plt.scatter([x0], [y0], color='black', zorder=5)

# Loop through each h value to plot different secant lines
for h in h_values:
    x1 = x0 + h            # Second point x = x0 + h
    y1 = f(x1)             # Corresponding y-value at x1
    slope = (y1 - y0) / h  # This is the slope of the secant line

    # Create the full secant line using point-slope form: y = m(x - x0) + y0
    secant_line = slope * (x_vals - x0) + y0

    # Plot the secant line (dashed for clarity)
    plt.plot(x_vals, secant_line, linestyle='--',
             label=f'h = {h}, slope ≈ {round(slope, 3)}')

# Title and axis labels
plt.title('Secant Lines Approaching the Tangent Line at x=2')
plt.xlabel('x')
plt.ylabel('f(x)')

# Add a grid and light axis lines
plt.grid(True)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)

# Show the legend to explain each line
plt.legend()

# Show the full plot
plt.show()
```

---

## 🔎 What This Shows You

- The **blue curve** is $f(x) = x^2$  
- The **red dashed line** is the **tangent line** at $x = 2$  
- That red line’s slope is the derivative:  
  $f'(2) = 2 \cdot 2 = 4$

---

Change `x0 = 2` to something like `x0 = 1` or `x0 = 3` and re-run the code.  
You’ll see how the **tangent line tilts**, since slope = derivative **changes at each x**.
