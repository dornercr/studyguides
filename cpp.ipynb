# 📘 SECTION 1: The Anatomy of a C++ Program

---

## 1.1 💡 How C++ Works Internally

C++ is a **compiled language**, meaning:

- You write **source code** → `.cpp`  
- The **compiler** (e.g., `g++`) turns it into **machine code**  
- The final **executable** runs natively on your CPU — no interpreter or VM needed

### 🧱 Compilation Stages:

1. **Preprocessing** — Handles `#include`, `#define`, etc.  
2. **Compilation** — Syntax/type checking → assembly code  
3. **Assembly** — Converts to CPU-specific machine code  
4. **Linking** — Combines `.o` object files and libraries → final binary

---

## 1.2 🔍 Dissecting the Minimal C++ Program

```cpp
#include <iostream>  // Preprocessor directive

int main() {         // Entry point
    std::cout << "Hello, world!\n";  // Output to console
    return 0;        // Exit status
}
```

### Line-by-Line Breakdown:

- `#include <iostream>`  
  → Imports input/output declarations (`std::cout`, `std::cin`)

- `int main()`  
  → Mandatory entry point. Returns an integer to OS.

- `std::cout << "Hello"`  
  → Outputs string to the terminal

#### 🆚 `\n` vs `std::endl`

| Syntax        | Behavior                         |
|---------------|----------------------------------|
| `\n`          | Newline only                     |
| `std::endl`   | Newline **+ buffer flush**       |

- `std::endl` is more expensive due to flushing.

- `return 0;`  
  → Signals success (by convention)

---

## 1.3 ⚙️ Memory Model and Execution

When a C++ program runs:

```
+---------------------+
|     Code Segment    | ← compiled instructions
+---------------------+
|  Global/Static Data | ← global vars and constants
+---------------------+
|         Heap        | ← dynamic memory (grows up)
+---------------------+
|         Stack       | ← function calls, locals (grows down)
+---------------------+
```

| Region         | Purpose                          |
|----------------|----------------------------------|
| Stack          | Function calls, local variables  |
| Heap           | `new`, `malloc` — manually managed |
| Static/Global  | Global variables, static vars    |
| Code Segment   | Actual compiled instructions     |

---

## 1.4 🧠 Memory-Safe vs. Unsafe

C++ gives you **raw control over memory** — powerful but risky.

```cpp
int* ptr = nullptr;  // ✅ Safe null pointer
*ptr = 10;           // 💣 SEGFAULT if dereferenced
```

✅ You must **initialize**, **allocate**, and **free** memory properly.  
We’ll cover `new`, `delete`, and smart pointers later.

````python
%%writefile main.cpp
#include <iostream>

// Global/Static Memory Segment
int global_var = 42;

int main() {
    // ----------------------------
    // 1. Minimal Program Output
    // ----------------------------
    std::cout << "Hello, world!\\n";             // newline without flush
    std::cout << "Hello with flush!" << std::endl;  // newline + flush

    // ----------------------------
    // 2. Memory Model Demo
    // ----------------------------
    int local_var = 10;                         // Stack
    int* heap_var = new int(99);                // Heap

    std::cout << "Global var: " << global_var << std::endl;
    std::cout << "Local var: " << local_var << std::endl;
    std::cout << "Heap var: " << *heap_var << std::endl;

    // Show memory addresses (optional)
    std::cout << "Address of global_var: " << &global_var << std::endl;
    std::cout << "Address of local_var: " << &local_var << std::endl;
    std::cout << "Address of heap_var: " << heap_var << std::endl;

    // ----------------------------
    // 3. Unsafe Pointer Example
    // ----------------------------
    int* unsafe_ptr = nullptr;
    // *unsafe_ptr = 5;  // 💣 Uncommenting this causes SEGFAULT

    // ----------------------------
    // 4. Return Status
    // ----------------------------
    delete heap_var;  // Free manually allocated memory
    return 0;         // Success
}
````
