# 📘 1.2 Learning Resources – Study Guide

## 🔧 Main Tool: Git
**Git** is a **version control system**.

It helps you:
- Track code changes
- Collaborate with teammates
- Revert to earlier versions if something breaks

---

## 🧠 Why Git Is Important
You can:
- **Track every change** made to the code
- **Work together** without overwriting each other's changes
- **Fix mistakes** by reverting to earlier versions
- **Test new ideas** using branches before merging into the main project

---

## 🔗 Branching and Merging
- **Branch**: A copy of the code where you can test something new safely
- **Merge**: Combine the changes from a branch into the main project

---

## 🧰 Tools Used
- You’ll use **GitLab** (not GitHub) in this course
- You’ll watch **Git Essential Training (01:23:00)** on LinkedIn Learning to learn how Git works

---

## 🧠 Quick Facts to Memorize

| Term             | Meaning                                      |
|------------------|----------------------------------------------|
| Git              | Tracks code changes over time                |
| Version Control  | Keeps a history of all code edits            |
| Branch           | A safe place to try new code ideas           |
| Merge            | Bring branch code into the main project      |
| GitLab           | The platform you’ll use in this course       |

---
