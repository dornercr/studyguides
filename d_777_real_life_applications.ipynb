# 🧠 2.2 Algorithms Overview – Explained Simply

## 💡 What Is an Algorithm?
An **algorithm** is a **step-by-step set of instructions** a computer follows to solve a problem or perform a task.

> Think of it like a recipe: Just like a recipe tells you what to do in what order to make a dish, an algorithm tells the computer what to do and when.

---

## 📋 Characteristics of a Good Algorithm

| Characteristic        | What It Means (Simple Definition)                                |
|------------------------|------------------------------------------------------------------|
| **Unambiguous**        | Every step must be clear and have only one meaning               |
| **Finite**             | The algorithm must end eventually (not go on forever)            |
| **Well-defined input** | You know exactly what data you’re starting with                 |
| **Well-defined output**| You know exactly what result you’re expecting at the end         |
| **Effective**          | Each step must be doable with simple instructions                |
| **Feasible**           | It should be possible to actually run the steps on a computer    |
| **Language-independent** | The steps work no matter what programming language is used    |

---

## 📊 What Are Data Flow Diagrams?

A **Data Flow Diagram (DFD)** is a drawing that shows how **data moves through a system**.

- It uses **symbols** to show:
  - Where the data **starts**
  - How the data **moves**
  - Where the data **ends**

> Businesses use DFDs to visualize their processes and find ways to make them better.

---

## 🧪 Example

**Pizza Shop DFD**:
- Customer → Cashier → Kitchen → Delivery

This shows the **flow of data (orders)** through the pizza shop system.


# 🧠 2.3 Introduction to Algorithms – Study Guide

## 💡 What Is an Algorithm?
An algorithm is a **sequence of steps** a computer follows to solve a computational problem.

Algorithms can be written in:
- English  
- Pseudocode  
- Programming languages  
- Flowcharts or diagrams

---

## 🧪 Computational Problem

A computational problem always includes:

| Component | Example                                |
|----------|-----------------------------------------|
| Input    | Array: `[13, 7, 45, -3, 92, 17]`        |
| Question | "What is the largest number?"          |
| Output   | `92`                                   |

📌 Algorithms are used to **answer questions about input data**.

---

## 🧰 Example Algorithm: FindMax()

FindMax(inputArray):
    max = inputArray[0]
    for i from 1 to inputArray.size:
        if inputArray[i] > max:
            max = inputArray[i]
    return max

✅ This finds the **maximum value** in an array.

---

## 🗂️ Real-World Applications and Matching Algorithms

| Domain         | Problem (Input + Question)                      | Algorithm Used             |
|----------------|--------------------------------------------------|----------------------------|
| DNA Analysis   | Longest matching sequence in 2 DNA strings       | Longest Common Substring   |
| Search Engines | Product ID lookup in sorted inventory            | Binary Search              |
| Navigation     | Shortest walking route from point A to B         | Dijkstra’s Shortest Path   |

🧠 Tip: Real-world problems are usually combinations of **smaller subproblems**, solved with known efficient algorithms.

---

## ❗ NP-Complete Problems

Some problems have no known efficient algorithm. These are **NP-complete** problems:

- No known **polynomial-time** solution  
- Not proven impossible either  
- If **one** NP-complete problem is solved efficiently, **all** can be

---

## 🧪 Example: Clique Problem (NP-Complete)

| Component | Example                                                                 |
|-----------|-------------------------------------------------------------------------|
| Input     | Graph of people (nodes = people, edges = friendships) + K (group size) |
| Question  | Do K people all know each other?                                       |
| Output    | `True` or `False` (Boolean)

🔍 K = 3 → Yes  
🔍 K = 4 → No

This is the **Clique Decision Problem** — it’s **NP-complete**.
