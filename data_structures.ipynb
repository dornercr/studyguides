# 📚 1.1 Data Structures – Study Guide

## 🔑 What is a Data Structure?
A **data structure** organizes, stores, and allows operations on data such as:
- Accessing
- Updating
- Searching
- Inserting
- Removing

---

## 📦 Basic Data Structures

| Data Structure | Description | Example |
|----------------|-------------|---------|
| **Record**     | A group of **fields** with names (like a row in a table). | Employee: `{firstName: "Maria", lastName: "Lu", title: "Developer", salary: 95000}` |
| **Array**      | An **ordered list** of items, accessed by index. | `[35, 20, 7, 18, 10]` |
| **Linked List**| A list made of **nodes**, each pointing to the next. | `10 → 5 → 17 → null` |
| **Binary Tree**| A tree where each node has up to **2 children**. | `20 → (12, 23), 12 → (11, 18)` |
| **Hash Table** | Uses a **hash function** to map keys to "buckets". | `bucket[3] → [53, 363]` |
| **Heap**       | A **tree** where each node is ≥ (max-heap) or ≤ (min-heap) its children. | `57 → (42, 19), 42 → (13, 6)` |
| **Graph**      | A set of **vertices** connected by **edges**. | `A — B — C — D — E` |

---

## ✅ True/False Quick Review

1. A linked list stores items in an **unordered** way.  
   ❌ False – It's **ordered** by pointers.

2. A binary tree node can have 0, 1, or 2 children.  
   ✅ True

3. Linked list nodes can store **records**.  
   ✅ True

4. Arrays allow access using **indexes**.  
   ✅ True

---

## 📌 Choosing the Right Data Structure

Pick a structure based on:
- Data type
- Required operations

**Example**:  
Need fast insertions? ➝ Use a **linked list** instead of an array.

---

## 🆚 Array vs. Linked List (Insert B)

### Array Insertion (middle of list):
1. Shift items to make space.
2. Insert at correct index.  
   ➝ Takes **time** for large arrays.

### Linked List Insertion:
1. Create a new node.
2. Point it to the next item.
3. Update previous node's pointer.  
   ➝ **No shifting** needed.

---

## 🧠 Memory Check (Insertion)

| Scenario | Items Shifted |
|----------|---------------|
| Insert at **end** of 999-item array | 0 |
| Insert at **end** of linked list    | 0 |
| Insert at **start** of 999-item array | 999 |
| Insert at **start** of linked list    | 0 |

---

# 🔁 Summary

- **Arrays** = Easy access, slow inserts/deletes.
- **Linked Lists** = Flexible inserts/deletes.
- **Trees/Heaps** = Good for sorting/searching.
- **Hash Tables** = Fast lookups.
- **Graphs** = Show relationships.

## 📌 What is an Algorithm?

An **algorithm** is a step-by-step procedure to solve a computational problem or perform a calculation. Algorithms can be expressed in:
- English
- Pseudocode
- Programming languages (e.g., Python, Java)
- Hardware-level descriptions

---

## 💡 Key Concepts

- A **computational problem** includes:
  - **Input**: Data provided to the algorithm
  - **Question**: What needs to be determined
  - **Output**: The result/solution

---

## 🔍 Example: Find Maximum in Array

### 🔢 Input:
Array of integers: `13, 7, 45, -3, 92, 17`

### ❓ Question:
What is the **maximum** value in the array?

### ✅ Output:
`92`

### 🧠 Algorithm (Pseudocode):
```plaintext
FindMax(inputArray) {
   max = inputArray[0]
   for (i = 1; i < inputArray.size; ++i) {
      if (inputArray[i] > max) {
         max = inputArray[i]
      }
   }
   return max
}
```

## 🔐 NP-Complete Problems

- **NP-complete problems**:
  - No known efficient algorithm (polynomial time)
  - Not proven impossible to solve efficiently
  - Efficient solution to one implies solutions to all

### Example: Clique Problem

> **Input**: Graph of friendships + integer K  
> **Question**: Do K people all know each other?  
> **Output**: Boolean (`True` or `False`)

If K=3 → **Yes** (Xiao, Sean, Tanya)  
If K=4 → **No**

````python
#find the maximum value in an array
def find_max(input_array):
  #assume the first elemenet is the maximum initially
  max_val = input_array[0]

  #loop through the array starting at index 1
  for i in range(1, len(input_array)):
    #if the current element is greater than the current max, update max
    if input_array[i] > max_val:
      max_val = input_array[i]

  #return the largest value found
  return max_val

#example usage
arr = [13, 7, 45, -3, 92, 17]
print("Maximum value:", find_max(arr)) #output 92
````

````python
#count frequency of a specified word in a list
def count_frequency(word_list, target_word):
  #initialize counter to 0
  count = 0

  #iterate through each word in a list
  for word in word_list:
    #if the word matches the target, increment the counter
    if word == target_word:
      count += 1

  #return the total count of the target word
  return count

#example usage:
words = ["apple", "banana", "apple", "orange", "banana", "apple"]
target = "apple"

#print how many times 'apple' appears in the list
print(f"Frequency of '{target}': ", count_frequency(words, target)) #output = 3
````

````python
# 3. Binary Search (for sorted array)
def binary_search(sorted_array, target):
    # Set the initial search range to cover the whole array
    left, right = 0, len(sorted_array) - 1

    # Continue searching as long as the range is valid
    while left <= right:
        # Find the middle index
        mid = (left + right) // 2

        # If target is found at the middle, return the index
        if sorted_array[mid] == target:
            return mid  # return value instead if preferred

        # If the middle value is less than the target, search the right half
        elif sorted_array[mid] < target:
            left = mid + 1

        # If the middle value is greater than the target, search the left half
        else:
            right = mid - 1

    # If target is not found, return -1
    return -1

# Example usage:
employees = ["Alice", "Bob", "Charlie", "Derek", "Eva"]
target_name = "Charlie"
index = binary_search(employees, target_name)

# Display the result of the search
if index != -1:
    print(f"{target_name} found at index {index}")
else:
    print(f"{target_name} not found")

# ----------------------------------------------------------
````

````python
#4. Detect a clique of size k (brute-force method for small graphs)
import itertools  # For generating combinations

def is_clique(graph, k):
    # Iterate through all combinations of k nodes in the graph
    for subset in itertools.combinations(graph, k):
        # Check if every pair in the subset is mutually connected
        if all(b in graph[a] for a, b in itertools.combinations(subset, 2)):
            return True  # Found a clique
    return False  # No clique of size k found

# Example social graph represented as an adjacency list
social_graph = {
    "Xiao": {"Sean", "Tanya"},
    "Sean": {"Xiao", "Tanya", "Eve"},
    "Eve": {"Sean", "Tanya", "Sima"},
    "Tanya": {"Xiao", "Sean", "Eve", "Mario"},
    "Mario": {"Tanya", "Sima"},
    "Sima": {"Eve", "Mario"}
}

# Check for cliques of size 3 and 4
print("Clique of size 3 exists:", is_clique(social_graph, 3))  # Output: True
print("Clique of size 4 exists:", is_clique(social_graph, 4))  # Output: False
````
