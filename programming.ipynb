# LESSON 5: Loops (for, while, repeat until)
🔑 20% That Powers 80% of Automation:
Loops repeat code until a condition is false.

Use when:

You want to run code N times (for)

You don’t know how many times — just "until" it's done (while)

Loop Type	When to Use
for	Counted repetition (definite)
while	Repeat until condition is false
✅ Print 1 to 5 in All Four Languages
🔹 Python — for and while
# for loop
for i in range(1, 6):
    print(i)

# while loop
i = 1
while i <= 5:
    print(i)
    i += 1
🔹 C — for and while
#include <stdio.h>

int main() {
    // for loop
    for (int i = 1; i <= 5; i++) {
        printf("%d\n", i);
    }

    // while loop
    int j = 1;
    while (j <= 5) {
        printf("%d\n", j);
        j++;
    }

    return 0;
}
🔹 C++ — identical to C
#include <iostream>
using namespace std;

int main() {
    for (int i = 1; i <= 5; i++) {
        cout << i << endl;
    }

    int j = 1;
    while (j <= 5) {
        cout << j << endl;
        j++;
    }

    return 0;
}
🔹 Rust — loop, while, for
fn main() {
    // for loop (Rust's cleanest)
    for i in 1..=5 {
        println!("{}", i);
    }

    // while loop
    let mut j = 1;
    while j <= 5 {
        println!("{}", j);
        j += 1;
    }

    // loop (infinite unless manually broken)
    let mut k = 1;
    loop {
        if k > 5 { break; }
        println!("{}", k);
        k += 1;
    }
}
🔍 Loop Structures Compared
Language	for Loop	while Loop	Special
Python	for x in range()	while cond:	No do-while
C	for(init;cond;upd)	while(cond)	do { ... } while(cond);
C++	Same as C	Same as C	Adds break / continue
Rust	for x in start..end	while	loop {} infinite loop

# LESSON 7: Arrays & Lists — Store Many Values
🔑 The 20% That Powers 80% of Data Work:
An array (or list) stores multiple values under one name.

Use it to loop through data, search, sort, etc.

Python has flexible lists.
C/C++/Rust use arrays or vectors (for dynamic sizes).

✅ Storing: 5 Scores → 95, 88, 76, 100, 67
🔹 Python (list)
scores = [95, 88, 76, 100, 67]

print(scores[0])      # Access
scores.append(80)     # Add new
print(len(scores))    # Length
🔹 C (fixed-size array)
#include <stdio.h>

int main() {
    int scores[5] = {95, 88, 76, 100, 67};
    printf("%d\n", scores[0]);     // Access
    return 0;
}
🛑 You can't change size at runtime — use malloc() or linked list for dynamic size.

🔹 C++ (array or vector)
#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> scores = {95, 88, 76, 100, 67};

    cout << scores[0] << endl;     // Access
    scores.push_back(80);          // Add
    cout << scores.size() << endl; // Length
    return 0;
}
🔹 Rust (array or vector)
fn main() {
    let scores = [95, 88, 76, 100, 67];        // fixed-size array
    println!("{}", scores[0]);                // Access

    let mut scores_vec = vec![95, 88, 76, 100, 67];
    scores_vec.push(80);                      // Add
    println!("{}", scores_vec.len());         // Length
}
🧠 Static vs Dynamic

C++	✅ (int arr[5])	✅ (vector<int>)	Use std::vector often
Rust	✅ ([T; N])	✅ (Vec<T>)	Use Vec for real projects
