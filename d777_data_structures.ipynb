# 2.1 Study Guide: Data Structures

---

## 🔹 What is a Data Structure?

A method for organizing and managing data.  
**Key operations:**
- Access
- Update
- Search
- Insert
- Delete

---

## 🔹 Core Types

| Type         | Description                                                             |
|--------------|-------------------------------------------------------------------------|
| Record       | A group of named fields (example: name, age, score).                    |
| Array        | An ordered list of items, accessed by position (index).                 |
| Linked List  | A sequence of nodes; each node points to the next.                      |
| Binary Tree  | A structure where each node has up to two children (left and right).    |
| Hash Table   | Uses a key to map items quickly; no specific order.                     |
| Heap         | A tree where each parent is larger (max-heap) or smaller (min-heap).    |
| Graph        | A collection of connected nodes (items) and edges (links).              |

---

## 🔹 Quick Facts

- A linked list is unordered.  
  ❌ **False** — The order comes from the node links.

- Binary tree nodes can have 0, 1, or 2 children.  
  ✅ **True**

- A list node can store a record with multiple fields.  
  ✅ **True**

- Items in an array are accessed by index.  
  ✅ **True**

---

## 🔹 Choosing the Right Structure

Choose based on:
- The kind of data
- The operations you need to perform

**Examples:**
- Need fast inserting? → Use a **linked list**
- Need quick lookups? → Use a **hash table**

---

## 🔹 Inserting: Array vs Linked List

**Array:** Insert at index 1
- Shift items right to make space
- Add new item

**Linked List:** Insert between A and C
- Create new node
- Update links: A → new → C  
(No shifting needed)

---

## 🔹 Practice Questions

**Q1:** How many items shift when adding to the end of a 999-item array?  
**A:** 0

**Q2:** How many shift when adding to the end of a 999-item linked list?  
**A:** 0

**Q3:** How many shift when adding to the start of a 999-item array?  
**A:** 999

**Q4:** How many shift when adding to the start of a 999-item linked list?  
**A:** 0

# 2.2 Study Guide: Algorithms Overview

---

## 🔹 What Is an Algorithm?

An **algorithm** is a set of commands that must be followed for a computer to perform calculations or solve a problem.

It provides step-by-step instructions for completing a task or achieving a result.

To dive deeper, refer to:
**"What Is An Algorithm? Definition, Types, Characteristics"**

---

## 🔹 Characteristics of an Algorithm

A good algorithm should have these core properties:

- **Unambiguous**: Every step is clearly defined  
- **Finiteness**: It completes after a finite number of steps  
- **Well-defined Inputs**: Accepts valid inputs clearly  
- **Well-defined Outputs**: Produces expected output  
- **Effectiveness**: Each step can be carried out exactly  
- **Feasibility**: Realistically implementable on hardware  
- **Language Independence**: Logic should apply across languages

For full explanation, see:  
**"Characteristics of Algorithms"**

---

## 🔹 Data Flow Diagrams (DFDs)

Businesses run on structured systems and workflows.  
A **Data Flow Diagram (DFD)** is a visual tool used to model:

- How data moves through a system  
- How processes transform data  
- Where data is stored  
- How external entities interact with the system

DFDs help uncover inefficiencies and improve processes.

To learn about DFDs, symbols, levels (0, 1, 2...), and use cases, refer to:  
**"Data Flow Diagram Examples, Symbols, Types, and Tips"**

---

## ✅ Summary

- Algorithms define **how to solve problems**.
- Good algorithms are **clear, finite, effective**, and **language-neutral**.
- Data Flow Diagrams visualize **how information flows** through systems.
