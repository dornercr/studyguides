# 2.1 Study Guide: Data Structures

---

## 🔹 What is a Data Structure?

A method for organizing and managing data.  
**Key operations:**
- Access
- Update
- Search
- Insert
- Delete

---

## 🔹 Core Types

| Type         | Description                                                             |
|--------------|-------------------------------------------------------------------------|
| Record       | A group of named fields (example: name, age, score).                    |
| Array        | An ordered list of items, accessed by position (index).                 |
| Linked List  | A sequence of nodes; each node points to the next.                      |
| Binary Tree  | A structure where each node has up to two children (left and right).    |
| Hash Table   | Uses a key to map items quickly; no specific order.                     |
| Heap         | A tree where each parent is larger (max-heap) or smaller (min-heap).    |
| Graph        | A collection of connected nodes (items) and edges (links).              |

---

## 🔹 Quick Facts

- A linked list is unordered.  
  ❌ **False** — The order comes from the node links.

- Binary tree nodes can have 0, 1, or 2 children.  
  ✅ **True**

- A list node can store a record with multiple fields.  
  ✅ **True**

- Items in an array are accessed by index.  
  ✅ **True**

---

## 🔹 Choosing the Right Structure

Choose based on:
- The kind of data
- The operations you need to perform

**Examples:**
- Need fast inserting? → Use a **linked list**
- Need quick lookups? → Use a **hash table**

---

## 🔹 Inserting: Array vs Linked List

**Array:** Insert at index 1
- Shift items right to make space
- Add new item

**Linked List:** Insert between A and C
- Create new node
- Update links: A → new → C  
(No shifting needed)

---

## 🔹 Practice Questions

**Q1:** How many items shift when adding to the end of a 999-item array?  
**A:** 0

**Q2:** How many shift when adding to the end of a 999-item linked list?  
**A:** 0

**Q3:** How many shift when adding to the start of a 999-item array?  
**A:** 999

**Q4:** How many shift when adding to the start of a 999-item linked list?  
**A:** 0

```python
# Cell 1 – Setup
from IPython.display import display, Markdown, clear_output
import ipywidgets as widgets

# Flashcards: (front, back)
flashcards = [
    # ✅ Definitions & Core Types
    ("What is a data structure?", "A method for organizing and managing data to support operations like access, insert, update, delete, and search."),
    ("What is a record?", "A group of named fields (like a row with columns: name, age, score)."),
    ("What is an array?", "An ordered list of items accessed by position (index)."),
    ("What is a linked list?", "A sequence of nodes; each node points to the next."),
    ("What is a binary tree?", "A tree structure where each node has up to two children: left and right."),
    ("What is a hash table?", "A structure that maps keys to values for fast access, with no defined order."),
    ("What is a heap?", "A tree where each parent is greater than (max-heap) or less than (min-heap) its children."),
    ("What is a graph?", "A collection of connected nodes and edges representing relationships."),

    # 🔍 Quick Facts (True/False)
    ("Is a linked list unordered?", "False — The order comes from node-to-node links."),
    ("Can binary tree nodes have 0, 1, or 2 children?", "True"),
    ("Can a list node store a record with multiple fields?", "True"),
    ("Are array items accessed by index?", "True"),

    # 🧠 Choosing a Structure
    ("How should you choose a data structure?", "Base it on the data and the required operations: access, insert, search, etc."),
    ("What structure is best for fast insertion?", "Linked list"),
    ("What structure is best for fast lookups?", "Hash table"),

    # ⚙️ Insert Operation Comparison
    ("What happens when inserting into an array at index 1?", "You shift later items to the right, then insert the new one."),
    ("What happens when inserting into a linked list between A and C?", "You create a new node and update links: A → new → C (no shifting needed)."),

    # ❓ Practice Questions
    ("How many items shift when adding to the end of a 999-item array?", "0"),
    ("How many items shift when adding to the end of a 999-item linked list?", "0"),
    ("How many items shift when adding to the start of a 999-item array?", "999"),
    ("How many items shift when adding to the start of a 999-item linked list?", "0"),
]

# Cell 2 – Flashcard Widget
card_index = 0
showing_front = True

output = widgets.Output()

# Buttons
flip_btn = widgets.Button(description="Flip", button_style='primary')
next_btn = widgets.Button(description="Next", button_style='success')
prev_btn = widgets.Button(description="Previous", button_style='info')
reset_btn = widgets.Button(description="Reset", button_style='warning')

def show_card():
    output.clear_output()
    with output:
        side = flashcards[card_index][0] if showing_front else flashcards[card_index][1]
        label = "**Q:**" if showing_front else "**A:**"
        display(Markdown(f"{label} {side}"))

def on_flip_clicked(_):
    global showing_front
    showing_front = not showing_front
    show_card()

def on_next_clicked(_):
    global card_index, showing_front
    card_index = (card_index + 1) % len(flashcards)
    showing_front = True
    show_card()

def on_prev_clicked(_):
    global card_index, showing_front
    card_index = (card_index - 1) % len(flashcards)
    showing_front = True
    show_card()

def on_reset_clicked(_):
    global card_index, showing_front
    card_index = 0
    showing_front = True
    show_card()

# Hook up events
flip_btn.on_click(on_flip_clicked)
next_btn.on_click(on_next_clicked)
prev_btn.on_click(on_prev_clicked)
reset_btn.on_click(on_reset_clicked)

# Display
btns = widgets.HBox([prev_btn, flip_btn, next_btn, reset_btn])
show_card()
display(output, btns)
```
