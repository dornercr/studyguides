## Vector Fundamentals and Representation

### 1. Core Definition

A **Vector** is a mathematical object characterized by both **magnitude** (the length or size) and **direction** (the orientation in space). In machine learning, it is formally defined as an element of a **vector space** (a set of objects that can be scaled and added together following specific algebraic rules).

### 2. Why It Matters

Vectors allow computers to represent multi-dimensional data numerically. Exams frequently test your ability to distinguish between **scalars** (single numbers) and vectors, and your understanding of how vectors serve as the foundational input for algorithms.

### 3. Core Concepts

* **Magnitude:** The total length of the vector arrow.
* **Direction:** Where the arrow points in n-dimensional space.
* **Dimensions:** The number of components (elements) within the vector.

### 4. Common Exam Traps

* **Confusing Vectors with Scalars:** A vector must have direction; a scalar is just a magnitude.
* **Dimension Misalignment:** Assuming a vector is just a "list" without realizing it exists in a specific geometric space.

### 5. Memory Hooks

**V.M.D.**: **V**ectors have **M**agnitude and **D**irection.

### 6. Active Recall Questions

1. What is the primary difference between a scalar and a vector?
2. In machine learning, what do we call the collection of rules that vectors must follow?
3. True or False: A vector can exist without a direction.

**Answers: 1. A scalar has only magnitude; a vector has magnitude and direction. 2. Vector Space. 3. False.**

## Vector Representation

### 1. Core Definition

**Vector Representation** is the process of expressing a geometric vector as an ordered list of numbers. This is typically done as a **Row Vector** (a horizontal array, e.g., $[1, 2, 3]$) or a **Column Vector** (a vertical array).

### 2. Why It Matters

Computers cannot "see" arrows; they require numerical arrays to perform calculations. Exams test your ability to convert visual or conceptual data into these formats. Standard notation in linear algebra usually defaults to **Column Vectors**.

### 3. Core Concepts

* **Array:** A data structure used to store the vector's components.
* **Component:** An individual number within the vector representing a specific dimension.
* **Orientation:** Whether the vector is laid out horizontally (row) or vertically (column).

### 4. Step-by-Step Process: Representing a 2D Vector

1. Identify the horizontal displacement ($x$).
2. Identify the vertical displacement ($y$).
3. Place them in brackets: $[x, y]$ for a row or $\begin{bmatrix} x \\ y \end{bmatrix}$ for a column.

### 5. Common Exam Traps

* **Transposition Errors:** Forgetting that switching a row to a column (and vice versa) is a specific operation called **transposition** ($V^T$).
* **Index Confusion:** Mixing up the order of components; $[3, 2]$ is a different vector than $[2, 3]$.

### 6. Worked Example

**Task:** Represent a movement of 5 units East and 3 units North as a column vector.
**Solution:**

$$\begin{bmatrix} 5 \\ 3 \end{bmatrix}$$

### 7. Memory Hooks

**R-H / C-V**: **R**ow is **H**orizontal / **C**olumn is **V**ertical.

### 8. Active Recall Questions

1. What is the most common default orientation for vectors in linear algebra textbooks?
2. What is the mathematical term for flipping a row vector into a column vector?
3. Represent the vector $[7, 8]$ as a column vector.

**Answers: 1. Column Vector. 2. Transposition. 3. $\begin{bmatrix} 7 \\ 8 \end{bmatrix}$**

## Vector Addition and Subtraction

### 1. Core Definition

**Vector Addition/Subtraction** is the operation of combining two vectors of the same **dimension** (the number of elements) by adding or subtracting their corresponding components.

### 2. Why It Matters

In machine learning, this is used to move "weights" or update models. Exams test if you realize that you **cannot** add vectors of different sizes.

### 3. Formulas

**Addition:**

$$A + B = [a_1 + b_1, a_2 + b_2, \dots, a_n + b_n]$$


**Read aloud:** "Vector A plus Vector B equals a new vector where each component is the sum of the components from A and B."

### 4. Step-by-Step Process

1. Check that both vectors have the same number of elements.
2. Add/Subtract the first elements.
3. Add/Subtract the second elements.
4. Repeat for all $n$ elements to form the resulting vector.

### 5. Common Exam Traps

* **Dimension Mismatch:** Trying to add a 2D vector to a 3D vector. This is impossible.
* **Sign Errors:** Forgetting to distribute a negative sign during subtraction (e.g., $2 - (-3) = 5$).

### 6. Worked Example

**Solve:** $[5, 2] - [3, -1]$
**Step 1:** $5 - 3 = 2$
**Step 2:** $2 - (-1) = 3$
**Result:** $[2, 3]$

### 7. Active Recall Questions

1. Can you add $V = [1, 2]$ and $W = [1, 2, 3]$?
2. What is the result of $[10, 5] + [-2, 3]$?
3. Geometrically, what happens when you add two vectors?

**Answers: 1. No, dimensions do not match. 2. $[8, 8]$. 3. You place the tail of the second vector at the head of the first.**

## Dot Product

### 1. Core Definition

The **Dot Product** is an algebraic operation that takes two equal-length sequences of numbers (vectors) and returns a single **scalar** (a single number).

### 2. Why It Matters

It measures the "overlap" or similarity between two vectors. It is the most high-yield formula for exams involving similarity scores or neural network layers.

### 3. Formulas

**Algebraic Definition:**


$$V_1 \cdot V_2 = \sum_{i=1}^{n} X_i Y_i$$


**Read aloud:** "The dot product of V-one and V-two equals the sum from i equals one to n of X-sub-i times Y-sub-i."

**Variables:**

* $X_i, Y_i$: Corresponding components.
* $n$: Total dimensions.

### 4. Step-by-Step Process

1. Multiply the first components together.
2. Multiply the second components together.
3. Continue for all components.
4. **Sum** all the resulting products into one number.

### 5. Common Exam Traps

* **Returning a Vector:** The dot product **always** results in a single number, never a vector.
* **Dimension Mismatch:** Like addition, dimensions must match.

### 6. Worked Example

**Vectors:** $A = [1, 2, 3]$, $B = [4, 5, 6]$
**Calculation:** $(1 \times 4) + (2 \times 5) + (3 \times 6) = 4 + 10 + 18 = 32$.

### 7. Memory Hooks

**Multiply, then Sum**: The two steps to the dot product.

### 8. Active Recall Questions

1. What is the dot product of $[1, 0]$ and $[0, 1]$?
2. If the dot product of two vectors is a very large positive number, what does that suggest about their directions?
3. True or False: The dot product of two 3D vectors is a 3D vector.

**Answers: 1. 0. 2. They point in a similar direction. 3. False (it is a scalar).**

## Vector Magnitude (Length)

### 1. Core Definition

**Vector Magnitude** (often called the **Euclidean Norm** or $L_2$ norm) is the geometric length of a vector from its tail to its head.

### 2. Why It Matters

Used to calculate distances between data points. Exams test your ability to apply the Pythagorean theorem in higher dimensions.

### 3. Formulas

**Formula:**


$$\|V\| = \sqrt{X_1^2 + X_2^2 + \dots + X_n^2}$$


**Read aloud:** "The magnitude of V equals the square root of the sum of the squares of its components."

### 4. Step-by-Step Process

1. Square every number in the vector (this makes all numbers positive).
2. Add all those squared values together.
3. Take the square root of the total.

### 5. Common Exam Traps

* **Forgetting the Square Root:** Students often stop after adding the squares.
* **Negative Components:** Squaring a negative number (e.g., $-3^2$) results in a positive number ($9$). Magnitude is never negative.

### 6. Worked Example

**Vector:** $V = [3, -4]$
**Step 1:** $3^2 = 9$ and $(-4)^2 = 16$
**Step 2:** $9 + 16 = 25$
**Step 3:** $\sqrt{25} = 5$
**Magnitude:** $5$

### 7. Memory Hooks

**S.S.R.**: **S**quare, **S**um, **R**oot.

### 8. Active Recall Questions

1. What is the magnitude of the "Zero Vector" $[0, 0, 0]$?
2. Calculate the magnitude of $V = [1, 1]$.
3. Why do we square the components before adding them?

**Answers: 1. 0. 2. $\sqrt{2} \approx 1.41$. 3. To ensure the result is positive and to follow the Pythagorean theorem for distance.**

A **Linear Combination** is an expression constructed from a set of vectors by multiplying each vector by a constant (scalar) and adding the results together.

### 2. Why It Matters

This is how neural networks create new information from input data. Exams test if you can "reach" a target vector using a combination of base vectors.

### 3. Formulas

**Form:**

* $V$: Original vectors.

### 4. Common Exam Traps

* **Misunderstanding "Span":** The "Span" is simply the set of all possible linear combinations you can make with a set of vectors.

### 5. Active Recall Questions

## Linear Independence

### 1. Core Definition

A set of vectors is **Linearly Independent (LI)** if no vector in the set can be written as a linear combination of the others. If one *can* be written as a combination of others, the set is **Linearly Dependent (LD)**.

### 2. Why It Matters

LI vectors provide unique, non-redundant information. In machine learning, redundant (LD) features can lead to model errors (like multicollinearity).

### 3. Core Concepts

* **Redundancy:** LD vectors are "redundant" because they don't add a new direction.
* **The Zero Test:** A set is LI if the only way to get the zero vector using a linear combination is to set all scalars to zero.

### 4. Common Exam Traps

* **Over-counting Dimensions:** If you have 4 vectors in a 3D space, they **must** be Linearly Dependent. You cannot have more LI vectors than you have dimensions.
* **The Zero Vector:** Any set containing the zero vector $[0, 0, \dots, 0]$ is automatically Linearly Dependent.

### 5. Worked Example

**Set:** $V_1 = [1, 0]$, $V_2 = [2, 0]$
**Analysis:** $V_2$ is just $2 \times V_1$. Since $V_2$ is a multiple of $V_1$, they are **Linearly Dependent**.

### 6. Active Recall Questions

1. If you have 2 vectors pointing in the exact same direction, are they LI or LD?
2. Is the set $\{[1, 0, 0], [0, 1, 0], [0, 0, 1]\}$ Linearly Independent?
3. Can a set of 3 vectors in 2D space be Linearly Independent?

**Answers: 1. LD. 2. Yes (they are the standard basis vectors). 3. No (too many vectors for the space).**

## Orthogonal and Orthonormal Vectors

### 1. Core Definition

**Orthogonal Vectors** are vectors that are perpendicular (at a 90-degree angle) to each other. Their **dot product is exactly zero**.
**Orthonormal Vectors** are orthogonal vectors that also have a **magnitude (length) of exactly one**.

### 2. Why It Matters

These vectors are the "gold standard" for data coordinate systems because they don't overlap. Exams test your ability to prove orthogonality using the dot product.

### 3. Formulas / Rules

* **Orthogonal Rule:** $V \cdot W = 0$
* **Orthonormal Rule:** $V \cdot W = 0$ AND $\|V\| = 1, \|W\| = 1$

### 4. Step-by-Step Process: Checking for Orthogonality

1. Calculate the dot product of the two vectors.
2. If the result is 0, they are **orthogonal**.
3. If the result is not 0, they are not orthogonal.

### 5. Common Exam Traps

* **Confusing the terms:** All orthonormal vectors are orthogonal, but not all orthogonal vectors are orthonormal.
* **The Zero Vector:** The zero vector is technically orthogonal to every vector (because the dot product will be zero), but it cannot be orthonormal (because its length is 0, not 1).

### 6. Memory Hooks

**Ortho = 90 / Normal = 1**: **Ortho**gonal means 90 degrees; **Normal**ized means length of 1.

### 7. Active Recall Questions

1. If $A \cdot B = 0$, what is the angle between the vectors?
2. What two conditions must be met for vectors to be "orthonormal"?
3. Are $[1, 0]$ and $[0, 5]$ orthogonal? Are they orthonormal?

**Answers: 1. 90 degrees. 2. They must be orthogonal AND have a magnitude of 1. 3. They are orthogonal (dot product is 0), but not orthonormal (length of the second vector is 5, not 1).**

## NumPy Array Fundamentals (The `ndarray`)

### 1. Core Definition

An **ndarray** (N-dimensional array) is the central data structure of the NumPy library. Unlike Python lists, it is a **homogeneous** container (all elements must be of the same data type) stored in **contiguous memory** (elements are placed side-by-side in the computer's RAM).

### 2. Why It Matters

Exams test your understanding of why we use NumPy over standard Python lists. The primary reasons are **memory efficiency** and **computational speed** due to its low-level C implementation.

### 3. Core Concepts

* **dtype:** The data type of the elements (e.g., `float64`, `int32`).
* **shape:** A tuple representing the size of each dimension (e.g., `(3,)` for a 3D vector).
* **rank:** The number of dimensions (a vector has a rank of 1).

### 4. Step-by-Step Process (Creation)

1. **Import NumPy:** `import numpy as np`
2. **From List:** `v = np.array([1, 2, 3])`
3. **Placeholders:** `np.zeros(3)` or `np.ones(3)`
4. **Range-based:** `np.arange(0, 10, 2)` (start, stop, step)

### 5. Common Exam Traps

* **List vs Array Syntax:** In a list, `[1, 2] + [3, 4]` results in `[1, 2, 3, 4]` (concatenation). In NumPy, it results in `[4, 6]` (addition).
* **Shape vs Size:** `shape` returns the dimensions `(3,)`, while `size` returns the total number of elements `3`.

### 6. Worked Example

**Task:** Create a 5-dimensional vector of zeros and change the third element to 10.

```python
v = np.zeros(5)      # [0., 0., 0., 0., 0.]
v[2] = 10            # [0., 0., 10., 0., 0.]
print(v.dtype)       # float64 (default)

```

### 7. Active Recall Questions

1. What happens if you try to put a string into a NumPy array of integers?
2. What attribute tells you the dimensions of a vector?
3. Why is a NumPy array faster than a Python list for math?

**Answers: 1. NumPy will attempt to "upcast" all elements to strings or throw an error (homogeneity rule). 2. `.shape`. 3. It uses contiguous memory and pre-compiled C code.**

## Dot Product Implementation

### 1. Core Definition

The implementation of the algebraic dot product $V_1 \cdot V_2 = \sum X_i Y_i$ in code.

2. Execute `result = a @ b`.

## Broadcasting

### 1. Core Definition

**Broadcasting** is a set of rules that allows NumPy to perform arithmetic operations on arrays of different shapes.

### 2. Why It Matters

This is a **High-Yield Exam Topic**. It allows you to perform operations (like adding a bias term or scaling features) without manually looping or copying data.

### 3. The Rules

1. **Right-Alignment:** Compare dimensions starting from the right (trailing dimension).
2. **Compatibility:** Two dimensions are compatible if:
* They are equal, OR
* One of them is 1.



### 4. Step-by-Step Process (Logic)

1. Look at Shape A and Shape B.
2. If one has fewer dimensions, prepends 1s to the left.
3. Compare each dimension. If any are unequal and neither is 1, broadcasting fails.

### 5. Common Exam Traps

* **Failed Broadcast:** A vector of shape `(3,)` cannot be broadcast with a vector of shape `(2,)`.
* **Implicit Stretching:** NumPy doesn't actually copy the data to make the shapes match; it uses memory "strides" to act as if it did.

### 6. Worked Example

**Scenario:** Scaling a 3D vector by a scalar.

```python
v = np.array([1, 2, 3]) # Shape (3,)
s = 10                  # Scalar (effectively shape (1,))
res = v * s             # [10, 20, 30] - Scalar "stretched" to (3,)

```

### 7. Active Recall Questions

1. Can an array of shape `(4, 3)` broadcast with an array of shape `(3,)`?
2. Can an array of shape `(4, 3)` broadcast with an array of shape `(4,)`?
3. What is the final shape if you add `(3, 1)` and `(1, 5)`?

**Answers: 1. Yes (3 matches 3). 2. No (trailing dimensions 3 and 4 do not match). 3. (3, 5).**

## Vectorization

### 1. Core Definition

**Vectorization** is the practice of replacing explicit `for` loops in Python with array expressions.

### 2. Why It Matters

In ML, datasets have millions of rows. Loops in Python are slow; vectorized operations in NumPy are executed in optimized C, often 10x-100x faster.

### 3. Core Concepts

* **SIMD:** Single Instruction, Multiple Data (the CPU hardware capability NumPy exploits).
* **Ufuncs:** Universal functions (like `np.add`, `np.sin`, `np.exp`) that automatically apply element-wise.

### 4. Memory Hooks

**Loops are for Lists, Vectors are for Velocity.**

### 5. Active Recall Questions

1. What is the technical term for a function that operates element-wise on an entire array?
2. True or False: Vectorization reduces the time complexity ($O$ notation) of an algorithm.
3. Why does vectorization use less power/time than a standard loop?

**Answers: 1. Universal Function (ufunc). 2. False (it reduces the constant factor/overhead, but adding $N$ numbers is still $O(N)$). 3. It avoids the overhead of the Python interpreter for every single element.**

## Matrices and Dimensionality

### 1. Core Definition

A **Matrix** is a rectangular array of numbers arranged in **rows** and **columns**. It is a 2D mathematical object. If a vector is a point in space, a matrix represents a collection of points or a **linear transformation** (a function that moves or stretches space).

### 2. Why It Matters

In ML, a single row in a matrix usually represents one **observation** (e.g., one person), and each column represents a **feature** (e.g., age, weight). Exams test your ability to identify the "Shape" of data and perform transformations.

### 3. Core Concepts

* **Dimension (Shape):** Written as $R \times C$ (Rows by Columns).
* **Element ($A_{i,j}$):** The number located at the $i$-th row and $j$-th column.
* **Transpose ($A^T$):** An operation that flips a matrix over its diagonal, switching rows and columns.

### 4. Step-by-Step Process: Transposing a Matrix

1. Take the first row of the original matrix.
2. Write it as the first column of the new matrix.
3. Repeat for all rows.
4. If the original shape was $2 \times 3$, the new shape will be $3 \times 2$.

### 5. Common Exam Traps

## Matrix-Vector Multiplication (The Dot Product Evolution)

### 1. Core Definition

**Matrix-Vector Multiplication** is the process of multiplying a matrix by a vector to produce a new vector. This is the fundamental "workhorse" of neural networks.

### 2. Why It Matters

This operation represents a model making a prediction. The matrix contains the "weights" (patterns), the vector is the "input," and the result is the "output." Exams heavily focus on whether multiplication is **defined** (possible) based on shapes.

### 3. Formulas / Rules

For the product $A \mathbf{v}$ to exist:
**The number of columns in $A$ must equal the number of elements in $\mathbf{v}$.**

**Formula:**


$$y_i = \sum_{j=1}^{n} A_{ij} v_j$$


**Read aloud:** "The $i$-th element of the output vector is the dot product of the $i$-th row of the matrix and the input vector."

### 4. Step-by-Step Process

1. Check the shape: $(R \times C) \times (C \times 1)$. The inner numbers ($C$) must match.
2. The resulting vector will have the shape $(R \times 1)$.
3. Take the first row of the matrix and the vector. Calculate their **dot product**. This is your first result element.
4. Move to the second row and repeat.

### 5. Common Exam Traps

* **Order Matters:** $A \mathbf{v}$ is not the same as $\mathbf{v} A$. In fact, $\mathbf{v} A$ is usually impossible due to shape rules.
* **Shape Calculation:** If you multiply a $10 \times 3$ matrix by a $3 \times 1$ vector, the result is $10 \times 1$.

### 6. Worked Example

**Matrix $A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$, Vector $\mathbf{v} = \begin{bmatrix} 5 \\ 6 \end{bmatrix}$**

1. Row 1 Dot Product: $(1 \times 5) + (2 \times 6) = 5 + 12 = 17$
2. Row 2 Dot Product: $(3 \times 5) + (4 \times 6) = 15 + 24 = 39$
**Result:** $\begin{bmatrix} 17 \\ 39 \end{bmatrix}$

### 7. Active Recall Questions

1. Can you multiply a $4 \times 5$ matrix by a vector with 4 elements?
2. What is the resulting shape of $(100 \times 50) \times (50 \times 1)$?
3. If a matrix represents a transformation, what does the vector represent?

**Answers: 1. No (Columns 5 $\neq$ Elements 4). 2. $100 \times 1$. 3. The input data being transformed.**

## Derivatives and the Rate of Change

### 1. Core Definition

A **Derivative** is a mathematical tool that measures how a function changes as its input changes. Geometrically, it represents the **slope** of the line tangent to a curve at a specific point.

### 2. Why It Matters

In machine learning, we use derivatives to determine how changing a model's **weights** (parameters) will change the **error** (loss). If the derivative is negative, increasing the weight reduces the error.

### 3. Core Concepts

* **Slope:** If a derivative is 5, then for every 1 unit you move right ($x$), the output ($y$) moves up 5.
* **Optimization:** Finding where the derivative is **zero** usually identifies the "peak" (maximum) or "valley" (minimum) of a function.

### 4. Formulas / Rules

For a function $f(x) = x^n$, the power rule derivative is:


$$f'(x) = n \cdot x^{n-1}$$


**Read aloud:** "The derivative of x to the n equals n times x to the power of n minus one."

### 5. Common Exam Traps

* **Constants:** The derivative of a constant (like 5 or 100) is always **0** because constants don't change.
* **Linear Functions:** The derivative of $f(x) = 3x$ is just **3**. The $x$ disappears.

### 6. Worked Example

**Task:** Find the derivative of the error function $E(w) = w^2$.
**Step 1:** Apply the power rule ($n=2$).
**Step 2:** Result is $2 \cdot w^{2-1} = 2w$.
**Meaning:** If the weight ($w$) is 10, the slope is 20. If we want to reach the minimum (where $w=0$), we should move in the opposite direction of this positive slope.

### 7. Memory Hooks

**Bring it down, subtract one**: Bring the exponent down to multiply, then subtract one from the power.

## Partial Derivatives and the Gradient

### 1. Core Definition

A **Partial Derivative** is the derivative of a function with multiple inputs (like $x$ and $y$), where we treat all other variables as constants. The **Gradient** ($\nabla$) is simply a vector containing all the partial derivatives of a function.

### 2. Why It Matters

Machine learning models have thousands of weights. The gradient tells the computer the direction of "steepest ascent"—which way to move every single weight at once to change the error the most.

### 3. Core Concepts

* **$\frac{\partial f}{\partial x}$:** Notation for the partial derivative of $f$ with respect to $x$.
* **Gradient Vector ($\nabla f$):** A vector $[\frac{\partial f}{\partial w_1}, \frac{\partial f}{\partial w_2}, \dots]$ that points "uphill."

### 4. Step-by-Step Process: Finding a Gradient

1. Identify all variables in the function (e.g., $w_1$ and $w_2$).
2. Differentiate the function with respect to $w_1$ (treat $w_2$ as a zero).
3. Differentiate the function with respect to $w_2$ (treat $w_1$ as a zero).
4. Combine these results into a vector.

### 5. Common Exam Traps

* **Treating Variables as Variables:** When taking a partial derivative of $x^2 + y^2$ with respect to $x$, students often forget that $y^2$ becomes **0**, not $2y$.
* **Direction:** The gradient points **up**. To minimize error, models move in the direction of the **negative gradient** ($-\nabla f$).

### 6. Worked Example

**Function:** $f(x, y) = x^2 + 3y$
**Partial with respect to $x$:** $2x$
**Partial with respect to $y$:** $3$
**Gradient Vector:** $\nabla f = [2x, 3]$

### 7. Active Recall Questions

1. If the gradient of a loss function is $[0, 0]$, what does that indicate about the current weights?
2. What is the partial derivative of $f(w, b) = wb$ with respect to $w$?
3. In which direction does a model move to *decrease* error: the gradient or the negative gradient?

**Answers: 1. You have reached a local minimum or maximum (an "optimum"). 2. $b$ (since $w$ is the variable and $b$ is treated as a constant). 3. The negative gradient.**

## Gradient Descent (The "Learning" Algorithm)

### 1. Core Definition

**Gradient Descent** is an iterative optimization algorithm used to find the minimum of a function. It updates parameters by taking small steps in the direction of the negative gradient.

### 2. Why It Matters

This is how almost all modern AI "learns." Exams often ask about the "Learning Rate" and how it affects this process.

### 3. Formulas / Rules

**The Update Rule:**


$$w_{new} = w_{old} - \eta \cdot \nabla f(w_{old})$$

* $w$: Weights
* $\eta$ (Eta): **Learning Rate** (a small scalar like 0.01)
* $\nabla f$: The Gradient

### 4. Common Exam Traps

* **Learning Rate too high:** The model "overshoots" the minimum and fails to converge.
* **Learning Rate too low:** The model takes forever to learn (inefficient).

### 5. Active Recall Questions

1. What is the purpose of the "Learning Rate" in Gradient Descent?
2. If the gradient is positive, will the weight increase or decrease during the next step of Gradient Descent?
3. What is the "Stop Condition" for Gradient Descent?

**Answers: 1. To control the size of the steps taken toward the minimum. 2. Decrease (because we subtract the gradient). 3. When the gradient is zero or changes are smaller than a set threshold.**

## Symbolic vs. Numerical Differentiation

### 1. Core Definition

* **Symbolic Differentiation:** Using code to find the exact algebraic formula of a derivative (e.g., turning $x^2$ into $2x$).
* **Numerical/Automatic Differentiation:** Calculating the slope at a specific point without necessarily finding the general formula.

### 2. Why It Matters

Exams often test your ability to recognize the **SymPy** library for symbolic math and understand how optimization loops are structured in Python.

### 3. Core Concepts

* **SymPy:** The standard library for symbolic mathematics in Python.
* **Symbols:** Variables that are treated as mathematical objects rather than numeric values.
* **diff():** The function used to find derivatives.

### 4. Step-by-Step Process (Symbolic)

1. **Import:** `from sympy import symbols, diff`
2. **Define Symbols:** `x, y = symbols('x y')`
3. **Define Function:** `f = x**2 + 3*x`
4. **Differentiate:** `derivative = diff(f, x)`

### 5. Common Exam Traps

* **Symbol Declaration:** Forgetting to declare a variable as a symbol before using it will cause a `NameError`.
* **Evaluation:** `diff()` returns a formula. To get a number, you must use `.subs()` to substitute a value into the result.

### 6. Worked Example

```python
from sympy import symbols, diff

x = symbols('x')
f = x**3 + 2*x

# Find derivative
df = diff(f, x)
print(df) # Output: 3*x**2 + 2

# Evaluate at x = 2
val = df.subs(x, 2)
print(val) # Output: 14

```

### 7. Active Recall Questions

1. Which Python library is used for symbolic differentiation?
2. What does the `.subs()` method do in SymPy?
3. True or False: `diff(x**2, x)` returns the number 4 if $x=2$.

**Answers: 1. SymPy. 2. It substitutes a variable with a specific value. 3. False (it returns the symbolic expression `2*x`).**

## Implementing Gradient Descent in Python

### 1. Core Definition

Creating a loop that updates a weight variable based on a gradient and a learning rate.

### 2. Why It Matters

This is the "how it works under the hood" question. You may be asked to identify the logic of a training loop.

### 3. Step-by-Step Process (The Loop)

1. **Initialize:** Set a starting weight (`w`) and a learning rate (`lr`).
2. **Calculate Gradient:** Determine the slope at the current `w`.
3. **Update:** Subtract `lr * gradient` from `w`.
4. **Repeat:** Loop until the gradient is near zero.

### 4. Worked Example (Manual Gradient Descent)

```python
# Function: f(w) = w^2 | Gradient: 2*w
w = 10              # Starting weight
lr = 0.1            # Learning rate
iterations = 20

for i in range(iterations):
    gradient = 2 * w
    w = w - (lr * gradient)
    print(f"Iteration {i}: w = {w:.4f}")

```

### 5. Common Exam Traps

* **Direction:** Forgetting to **subtract** the gradient. If you add it, the weight will go "uphill" to infinity (Gradient Ascent).
* **Convergence:** If the weights stop changing, the model has converged. If they get larger and larger, the learning rate is likely too high.

## Probability Basics and Conditional Probability

### 1. Core Definition

**Probability** is a numerical measure (between 0 and 1) of the likelihood that an event will occur. **Conditional Probability** is the probability of an event occurring *given* that another event has already occurred.

### 2. Why It Matters

Exams test your ability to calculate probabilities in multi-step scenarios, specifically where one event influences another (dependency).

### 3. Core Concepts

* **Sample Space ($\Omega$):** The set of all possible outcomes.
* **Independence:** Two events are independent if the occurrence of one does not change the probability of the other ($P(A|B) = P(A)$).
* **Mutually Exclusive:** Events that cannot happen at the same time ($P(A \cap B) = 0$).

### 4. Formulas / Rules

**Conditional Probability:**


$$P(A|B) = \frac{P(A \cap B)}{P(B)}$$


**Read aloud:** "The probability of A given B equals the probability of A and B divided by the probability of B."

### 5. Common Exam Traps

* **Denominator Confusion:** In $P(A|B)$, the denominator is always the "given" condition ($B$).
* **Independence Assumption:** Never assume two events are independent unless the problem explicitly states it or $P(A \cap B) = P(A)P(B)$.

### 6. Active Recall Questions

1. If $P(A) = 0.5$ and $P(B) = 0.2$, and they are independent, what is $P(A \cap B)$?
2. What does $P(A|B) = 1$ mean in plain English?

**Answers: 1. $0.5 \times 0.2 = 0.1$. 2. It means if event B happens, event A is guaranteed to happen.**

## Bayes' Theorem (EXTREMELY HIGH YIELD)

### 1. Core Definition

**Bayes' Theorem** is a formula used to update the probability of a hypothesis (A) as more evidence (B) becomes available.

### 2. Why It Matters

This is the heart of ML classification (e.g., Spam Filters). Exams often use "Medical Test" or "Spam" word problems to test this.

### 3. Formulas / Rules

**Formula:**


$$P(A|B) = \frac{P(B|A)P(A)}{P(B)}$$


**Symbol Definitions:**

* **$P(A|B)$ (Posterior):** Probability of hypothesis after seeing evidence.
* **$P(B|A)$ (Likelihood):** Probability of evidence given the hypothesis.
* **$P(A)$ (Prior):** Initial belief before seeing evidence.
* **$P(B)$ (Evidence):** Total probability of the evidence across all hypotheses.

### 4. Step-by-Step Process (Word Problems)

1. Identify the Prior $P(A)$ (the base rate of the condition).
2. Identify the Likelihood $P(B|A)$ (how often the test is correct).
3. Calculate the total probability $P(B)$ (often using the Law of Total Probability).
4. Plug into the formula.

### 5. Common Exam Traps

* **Base Rate Fallacy:** Ignoring the Prior. Even if a test is 99% accurate, if the disease is 1 in a million, a positive result still likely means you don't have it.

### 6. Worked Example

**Scenario:** 1% of the population has a disease ($P(D)=0.01$). A test is 90% accurate for those with it ($P(+|D)=0.9$) and has a 5% false positive rate for those without it ($P(+|\text{no } D)=0.05$).
**Calculation:** 1. $P(D \cap +) = 0.01 \times 0.9 = 0.009$
2. $P(\text{no } D \cap +) = 0.99 \times 0.05 = 0.0495$
3. $P(+) = 0.009 + 0.0495 = 0.0585$
4. **$P(D|+)$:** $0.009 / 0.0585 \approx 0.15$ (Only a 15% chance you actually have the disease).

### 7. Memory Hooks

**P.L.E.P.** (Posterior = Likelihood $\times$ Prior / Evidence).

## MLE vs. MAP Estimation

### 1. Core Definition

* **Maximum Likelihood Estimation (MLE):** Choosing parameters that make the observed data most probable. (Uses only data).
* **Maximum A Posteriori (MAP):** Choosing parameters that are most probable given the data **and** a prior belief. (Uses data + prior).

### 2. Why It Matters

This explains the "Frequentist" vs. "Bayesian" divide in ML. Exams test your understanding of when to use which.

### 3. Common Exam Traps

* **Uniform Prior:** If the prior in MAP is uniform (equal weight everywhere), **MAP = MLE**.
* **Data Volume:** As you get more data, the influence of the Prior in MAP fades away, and it converges to the MLE.

### 4. Memory Hooks

* **MLE = Data Only.**
* **MAP = Data + My Opinion (Prior).**

### 4. Worked Example

**Task:** Find the probability that a student scores above 90 on an exam where the mean is 75 and the standard deviation is 7.

```python
from scipy import stats

# Define the distribution
exam_dist = stats.norm(loc=75, scale=7)

# Probability of scoring 90 or LESS
prob_less_than_90 = exam_dist.cdf(90)

# Probability of scoring MORE than 90
prob_more_than_90 = 1 - prob_less_than_90
print(f"{prob_more_than_90:.4f}") # Output: 0.0161 (~1.6%)

```

### 5. Common Exam Traps

* **PDF vs. CDF:** The PDF at a single point in a continuous distribution isn't the probability of that point (which is technically 0); it's the density. Always use **CDF** for probability ranges.
* **Scale vs. Variance:** In `stats.norm`, the `scale` parameter is the **Standard Deviation**, not the Variance. If a problem gives you Variance = 25, you must use `scale=5`.

## Descriptive Statistics (NumPy/SciPy)

### 1. Core Definition

Functions that summarize the characteristics of a dataset (mean, median, mode, variance, etc.).

### 2. Why It Matters

Essential for "Exploratory Data Analysis" (EDA). You must know the syntax for calculating these across specific axes of a matrix.

### 3. Core Commands

* **`np.mean(data, axis=0)`**: Average. `axis=0` calculates the mean of each **column** (feature).
* **`np.median(data)`**: The middle value (robust to outliers).
* **`np.var(data)`**: Average of squared deviations from the mean.
* **`np.std(data)`**: Square root of variance.
