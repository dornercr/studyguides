````python
# @title
programming_basics = [

# ---------- 1. VARIABLES ----------
("Variables — Concept",
"A variable (named storage) holds a value in memory. \
Python is dynamically typed (type decided at runtime). \
C, Rust, C++ are statically typed (type fixed at compile time).",
"V M T → Variable, Memory, Type.\nKindergarten: 'Very Many Toys.'"),

("Variables — Examples",
"Python: x=10\nC: int x=10;\nRust: let x:i32=10;\nC++: int x=10;",
"X I → Example, Initialize.\nKindergarten: 'Xylophones Inspire.'"),

# ---------- 2. TYPES ----------
("Types — Concept",
"A type (classification of data) defines allowed values and operations. \
Common: int, float, string/char, bool.",
"T I F B → Types: Int, Float, Bool.\nKindergarten: 'Tigers In Forests Bite.'"),

("Types — Examples",
"Python: x=3.14\nC: float x=3.14;\nRust: let y:f64=3.14;\nC++: double x=3.14;",
"N D → Number, Decimal.\nKindergarten: 'Nice Duck.'"),

# ---------- 3. OPERATORS ----------
("Operators — Arithmetic",
"Arithmetic operators = + - * / % (modulus).\n\n\
Python: 5%2\nC: 5%2\nRust: 5%2\nC++: 5%2",
"A A M → Add, Arithmetic, Mod.\nKindergarten: 'All Ants March.'"),

("Operators — Comparison",
"Comparison operators = == != < > <= >=.\nReturn true/false.",
"C C T → Compare, Check, True.\nKindergarten: 'Cats Chase Tails.'"),

("Operators — Logic",
"Logical operators = and/&&, or/||, not/!.\n\
Python: if a and b: ...\nC/C++/Rust: if(a && b) { }",
"L A O → Logic, And, Or.\nKindergarten: 'Lions And Owls.'"),

# ---------- 4. CONTROL FLOW ----------
("Conditionals — If/Else",
"A conditional runs code if a condition is true.\n\n\
Python: if x>0: print(x)\nC: if(x>0){printf(\"%d\",x);}\
\nRust: if x>0 { println!(\"{}\",x); }\nC++: if(x>0){std::cout<<x;}",
"I E → If, Else.\nKindergarten: 'Ice Everywhere.'"),

("Loops — For",
"A for loop repeats for a known range.\n\n\
Python: for i in range(3): print(i)\nC: for(int i=0;i<3;i++){}\nRust: for i in 0..3{}\nC++: for(int i=0;i<3;i++){}",
"F R → For, Repeat.\nKindergarten: 'Fish Run.'"),

("Loops — While",
"A while loop repeats until a condition is false.\n\n\
Python: while x>0: x-=1\nC: while(x>0){x--;}\nRust: while x>0 {x-=1;}\nC++: while(x>0){x--;}",
"W C → While, Condition.\nKindergarten: 'Whales Cry.'"),

# ---------- 5. FUNCTIONS ----------
("Functions — Concept",
"A function (reusable code block) takes input (parameters) and returns output. \
Encourages modularity (breaking programs into parts).",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

("Functions — Examples",
"Python: def add(a,b): return a+b\nC: int add(int a,int b){return a+b;}\
\nRust: fn add(a:i32,b:i32)->i32{a+b}\nC++: int add(int a,int b){return a+b;}",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),

# ---------- 6. DATA STRUCTURES ----------
("Arrays — Concept",
"Array = fixed-size, contiguous memory block. \
Efficient but size cannot change.",
"A F → Array, Fixed.\nKindergarten: 'Apples Fall.'"),

("Arrays — Examples",
"C: int nums[3]={1,2,3};\nRust: let arr=[1,2,3];\nC++: int nums[3]={1,2,3};",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

("Lists/Vectors — Dynamic",
"List/Vector = resizable array.\n\n\
Python: nums=[1,2,3]\nRust: let v=vec![1,2,3];\nC++: std::vector<int> v={1,2,3};",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

("Maps/Dicts — Key-Value",
"Dict/Map = store key→value pairs.\n\n\
Python: ages={\"bob\":25}\nRust: HashMap::new()\nC++: std::map<string,int> ages;",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Sets — Unique",
"Set = collection of unique items.\n\n\
Python: {1,2,3}\nRust: use std::collections::HashSet;\nC++: std::set<int> s;",
"S U → Set, Unique.\nKindergarten: 'Snakes Unite.'"),

# ---------- 7. INPUT/OUTPUT ----------
("Input — User",
"Get input from user.\n\n\
Python: input()\nC: scanf()\nRust: stdin().read_line()\nC++: std::cin",
"I U → Input, User.\nKindergarten: 'Ice Underwear.'"),

("Output — Print",
"Show output to console.\n\n\
Python: print()\nC: printf()\nRust: println!()\nC++: std::cout",
"O P → Output, Print.\nKindergarten: 'Owls Play.'"),

# ---------- 8. ERROR HANDLING ----------
("Error Handling — Concept",
"Error handling = controlled way to manage unexpected events (e.g., divide by zero).\n\
Python: exceptions, C: return codes, Rust: Option/Result, C++: exceptions.",
"E C H → Error, Control, Handle.\nKindergarten: 'Elephants Catch Hats.'"),

("Error Handling — Examples",
"Python: try/except\nC: if(den==0){}\nRust: Option/Result\nC++: try/catch",
"T R → Try, Result.\nKindergarten: 'Turtles Run.'"),

# ---------- 9. COMPILATION ----------
("Compilation vs Interpretation",
"Compiler (C, C++, Rust): translates code into machine code before execution (fast, safe). \
Interpreter (Python): executes line by line at runtime (flexible, easy debugging).",
"C I → Compile, Interpret.\nKindergarten: 'Cats Ice-skate.'"),
]

data_structures = [

# ---------- 1. VARIABLES ----------
("Variables — Concept",
"""A variable (named storage) holds a value in memory.
Typing:
- Python is dynamically typed (type decided at runtime).
- C, Rust, C++ are statically typed (type fixed at compile time).""",
"V M T → Variable, Memory, Type.\nKindergarten: 'Very Many Toys.'"),

("Variables — Examples",
"""Python:
    x = 10
    name = "Alice"

C:
    int x = 10;
    char name[] = "Alice";

Rust:
    let x: i32 = 10;
    let mut name: &str = "Alice";

C++:
    int x = 10;
    std::string name = "Alice";""",
"X I → Example, Initialize.\nKindergarten: 'Xylophones Inspire.'"),

# ---------- 2. TYPES ----------
("Types — Concept",
"""A type (classification of data) defines allowed values and operations.
Common: int, float/double, string/char, bool.""",
"T I F B → Types: Int, Float, Bool.\nKindergarten: 'Tigers In Forests Bite.'"),

("Types — Examples",
"""Python:
    n = 42            # int
    x = 3.14          # float
    ok = True         # bool
    s = "hi"          # str

C:
    int n = 42;
    float x = 3.14f;
    _Bool ok = 1;     // or bool with <stdbool.h>
    char s[] = "hi";

Rust:
    let n: i32 = 42;
    let x: f64 = 3.14;
    let ok: bool = true;
    let s: &str = "hi";

C++:
    int n = 42;
    double x = 3.14;
    bool ok = true;
    std::string s = "hi";""",
"N D → Number, Decimal.\nKindergarten: 'Nice Duck.'"),

# ---------- 3. OPERATORS ----------
("Operators — Arithmetic",
"""Arithmetic: +  -  *  /  % (modulus)

Python:
    print(7 % 3)

C:
    int r = 7 % 3;

Rust:
    let r = 7 % 3;

C++:
    int r = 7 % 3;""",
"A A M → Add, Arithmetic, Mod.\nKindergarten: 'All Ants March.'"),

("Operators — Comparison",
"""Comparison: ==  !=  <  >  <=  >=  (returns boolean)

Python:
    print(5 > 2)

C:
    int b = (5 > 2);

Rust:
    let b = 5 > 2;

C++:
    bool b = (5 > 2);""",
"C C T → Compare, Check, True.\nKindergarten: 'Cats Chase Tails.'"),

("Operators — Logic",
"""Logic: and/&&   or/||   not/!

Python:
    if ok and ready:
        print("go")

C:
    if (ok && ready) { printf("go"); }

Rust:
    if ok && ready { println!("go"); }

C++:
    if (ok && ready) { std::cout << "go"; }""",
"L A O → Logic, And, Or.\nKindergarten: 'Lions And Owls.'"),

# ---------- 4. CONTROL FLOW ----------
("Conditionals — If/Else",
"""A conditional runs code when a condition is true.

Python:
    if x > 0:
        print(x)
    else:
        print("neg or zero")

C:
    if (x > 0) {
        printf("%d", x);
    } else {
        printf("neg or zero");
    }

Rust:
    if x > 0 {
        println!("{}", x);
    } else {
        println!("neg or zero");
    }

C++:
    if (x > 0) {
        std::cout << x;
    } else {
        std::cout << "neg or zero";
    }""",
"I E → If, Else.\nKindergarten: 'Ice Everywhere.'"),

("Loops — For",
"""A for loop repeats for a known range.

Python:
    for i in range(3):
        print(i)

C:
    for (int i = 0; i < 3; i++) {
        printf("%d\\n", i);
    }

Rust:
    for i in 0..3 {
        println!("{}", i);
    }

C++:
    for (int i = 0; i < 3; i++) {
        std::cout << i << "\\n";
    }""",
"F R → For, Repeat.\nKindergarten: 'Fish Run.'"),

("Loops — While",
"""A while loop repeats until the condition becomes false.

Python:
    while x > 0:
        x -= 1

C:
    while (x > 0) {
        x--;
    }

Rust:
    while x > 0 {
        x -= 1;
    }

C++:
    while (x > 0) {
        x--;
    }""",
"W C → While, Condition.\nKindergarten: 'Whales Cry.'"),

# ---------- 5. FUNCTIONS ----------
("Functions — Concept",
"""A function (reusable code block) takes inputs (parameters) and may return output.
Functions promote modularity (split programs into parts).""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

("Functions — Examples",
"""Python:
    def add(a, b):
        return a + b

C:
    int add(int a, int b) {
        return a + b;
    }

Rust:
    fn add(a: i32, b: i32) -> i32 {
        a + b
    }

C++:
    int add(int a, int b) {
        return a + b;
    }""",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),



# ---------- 6. DATA STRUCTURES (quick tour, detailed deck below) ----------
("Arrays — Concept",
"""Array = fixed size, contiguous memory block.
Pros: O(1) access. Cons: fixed size, costly insert/delete.""",
"A F → Array, Fixed.\nKindergarten: 'Apples Fall.'"),

("Arrays — Examples",
"""Python:
    nums = [1, 2, 3]      # behaves like dynamic array

C:
    int nums[3] = {1, 2, 3};

Rust:
    let arr = [1, 2, 3];

C++:
    int nums[3] = {1, 2, 3};""",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

("Lists/Vectors — Dynamic",
"""Dynamic array that can grow and shrink.

Python:
    nums = [1, 2, 3]
    nums.append(4)

C (manual via heap):
    int *arr = malloc(3 * sizeof(int));
    arr[0]=1; arr[1]=2; arr[2]=3;
    arr = realloc(arr, 5 * sizeof(int));
    free(arr);

Rust:
    let mut v = vec![1, 2, 3];
    v.push(4);

C++:
    std::vector<int> v = {1, 2, 3};
    v.push_back(4);""",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

("Maps/Dicts — Key Value",
"""Key to value mapping. Hash maps average O(1) lookup.

Python:
    ages = {"bob": 25}
    print(ages["bob"])

C (struct array demo):
    struct Pair { char key[20]; int value; };
    struct Pair dict[2] = { {"bob",25}, {"alice",30} };

Rust:
    use std::collections::HashMap;
    let mut m = HashMap::new();
    m.insert("bob", 25);

C++:
    std::map<std::string, int> m;     // ordered tree map
    m["bob"] = 25;""",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Sets — Unique",
"""Set holds unique items only.

Python:
    s = {1, 2, 3}
    s.add(4)

C (manual check):
    int arr[3] = {1,2,3};
    int val = 2, exists = 0;
    for (int i=0;i<3;i++) if (arr[i]==val) { exists = 1; break; }

Rust:
    use std::collections::HashSet;
    let mut s = HashSet::new();
    s.insert(1);

C++:
    std::set<int> s;
    s.insert(1);""",
"S U → Set, Unique.\nKindergarten: 'Snakes Unite.'"),

# ---------- 7. I/O ----------
("Input — User",
"""Python:
    name = input("Enter name: ")

C:
    char name[64];
    scanf("%63s", name);

Rust:
    use std::io::{self, Read};
    let mut s = String::new();
    io::stdin().read_line(&mut s).unwrap();

C++:
    std::string name;
    std::cin >> name;""",
"I U → Input, User.\nKindergarten: 'Ice Umbrella.'"),

("Output — Print",
"""Python:
    print("hello")

C:
    printf("hello\\n");

Rust:
    println!("hello");

C++:
    std::cout << "hello\\n";""",
"O P → Output, Print.\nKindergarten: 'Owls Play.'"),

# ---------- 8. ERROR HANDLING ----------
("Error Handling — Concept",
"""Handle unexpected events in a controlled way.
Python and C++ use exceptions. C uses return codes. Rust uses Option/Result.""",
"E C H → Error, Control, Handle.\nKindergarten: 'Elephants Catch Hats.'"),

("Error Handling — Examples",
"""Python:
    try:
        x = 1 / 0
    except ZeroDivisionError:
        print("err")

C:
    if (den == 0) {
        printf("divide by zero");
    }

Rust:
    fn divide(a: i32, b: i32) -> Option<i32> {
        if b == 0 { None } else { Some(a / b) }
    }

C++:
    try {
        int x = 1/0;   // example only
    } catch (...) {
        std::cout << "err";
    }""",
"T R → Try, Result.\nKindergarten: 'Turtles Run.'"),

# ---------- 9. RUNTIME MODEL ----------
("Compilation vs Interpretation",
"""Compiler (C, C++, Rust) turns code into machine code before running.
Interpreter (Python) executes line by line at runtime.""",
"C I → Compile, Interpret.\nKindergarten: 'Cats Ice-skate.'"),

# ---------- ARRAYS ----------
("Array — Concept",
"""Array = fixed-size, contiguous memory collection.
Pros: O(1) random access.
Cons: fixed size, O(n) insert/delete.""",
"A F C → Array, Fixed, Contiguous.\nKindergarten: 'All Frogs Clap.'"),

("Array — Examples",
"""Python:
    nums = [1, 2, 3]
    print(nums[0])

C:
    int nums[3] = {1,2,3};
    printf("%d\\n", nums[0]);

Rust:
    let arr = [1, 2, 3];
    println!("{}", arr[0]);

C++:
    int nums[3] = {1, 2, 3};
    std::cout << nums[0] << "\\n";""",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

# ---------- LISTS / VECTORS ----------
("List/Vector — Concept",
"""Dynamic array. Resizable. Fast access and append.
Most used general-purpose structure.""",
"L D R → List, Dynamic, Resizable.\nKindergarten: 'Little Dogs Run.'"),

("List/Vector — Examples",
"""Python:
    nums = [1, 2, 3]
    nums.append(4)

C (manual dynamic array):
    int *arr = malloc(3 * sizeof(int));
    arr[0]=1; arr[1]=2; arr[2]=3;
    arr = realloc(arr, 5 * sizeof(int));
    free(arr);

Rust:
    let mut v = vec![1, 2, 3];
    v.push(4);

C++:
    std::vector<int> v = {1, 2, 3};
    v.push_back(4);""",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

# ---------- STACK ----------
("Stack — Concept",
"""Stack uses LIFO: last in, first out.
Use cases: recursion, undo, parsing.""",
"S L → Stack, LIFO.\nKindergarten: 'Snakes Leap.'"),

("Stack — Examples",
"""Python:
    stack = []
    stack.append(1)
    x = stack.pop()

C:
    int stack[100];
    int top = -1;
    stack[++top] = 1;       // push
    int x = stack[top--];   // pop

Rust:
    let mut s: Vec<i32> = Vec::new();
    s.push(1);
    let x = s.pop();

C++:
    std::stack<int> s;
    s.push(1);
    int x = s.top();
    s.pop();""",
"P P → Push, Pop.\nKindergarten: 'Puppies Play.'"),

# ---------- QUEUE ----------
("Queue — Concept",
"""Queue uses FIFO: first in, first out.
Use cases: scheduling, buffering, BFS.""",
"Q F → Queue, FIFO.\nKindergarten: 'Queen Flies.'"),

("Queue — Examples",
"""Python:
    from collections import deque
    q = deque()
    q.append(1)      # enqueue
    x = q.popleft()  # dequeue

C (circular buffer sketch):
    int q[100], front = 0, rear = 0, cap = 100;
    q[rear] = 1;            rear = (rear + 1) % cap;   // enqueue
    int x = q[front];       front = (front + 1) % cap; // dequeue

Rust:
    use std::collections::VecDeque;
    let mut q: VecDeque<i32> = VecDeque::new();
    q.push_back(1);
    let x = q.pop_front();

C++:
    std::queue<int> q;
    q.push(1);
    int x = q.front();
    q.pop();""",
"E D → Enqueue, Dequeue.\nKindergarten: 'Elephants Dance.'"),

# ---------- SET ----------
("Set — Concept",
"""Set stores unique items only.
Great for membership tests and deduplication.""",
"S U → Set, Unique.\nKindergarten: 'Silly Unicorn.'"),

("Set — Examples",
"""Python:
    s = {1, 2, 3}
    s.add(4)

C (simulate with search):
    int arr[3] = {1,2,3};
    int val = 2, exists = 0;
    for (int i=0;i<3;i++) if (arr[i]==val) { exists = 1; break; }

Rust:
    use std::collections::HashSet;
    let mut s: HashSet<i32> = HashSet::new();
    s.insert(1);

C++:
    std::set<int> s;
    s.insert(1);""",
"I U → Insert, Unique.\nKindergarten: 'Ice Unicorn.'"),

# ---------- MAP / DICT ----------
("Map/Dict — Concept",
"""Key to value store.
Hash map: average O(1) lookup. Tree map: O(log n) + ordering.""",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Map/Dict — Examples",
"""Python:
    ages = {"bob": 25}
    print(ages["bob"])

C (struct array demo):
    struct Pair { char key[20]; int value; };
    struct Pair dict[2] = { {"bob",25}, {"alice",30} };
    printf("%s is %d\\n", dict[0].key, dict[0].value);

Rust:
    use std::collections::HashMap;
    let mut m: HashMap<&str, i32> = HashMap::new();
    m.insert("bob", 25);
    println!("{}", m["bob"]);

C++:
    std::map<std::string, int> m;   // tree map (ordered)
    m["bob"] = 25;
    std::cout << m["bob"] << "\\n";""",
"K V → Key, Value.\nKindergarten: 'Kangaroos Visit.'"),
]

variables = [

# ---------- CONCEPT ----------
("Variables — Concept",
"""Definition: Named storage for a value (lives in memory).
Key ideas: declaration, initialization, type, scope, lifetime.""",
"V D T S L → Variable, Decl, Type, Scope, Lifetime.\nKindergarten: 'Very Daring Tigers Smile Loudly.'"),

# ---------- DECLARATION vs INITIALIZATION ----------
("Variables — Declare vs Initialize",
"""Declare = introduce name and type. Initialize = give first value.

Python:
    x = 0                 # declare+init (dynamic)

C:
    int x; x = 0;         # separate
    int y = 1;            # combined

Rust:
    let x: i32; x = 0;
    let y: i32 = 1;

C++:
    int x; x = 0;
    int y = 1;""",
"D I → Declare, Initialize.\nKindergarten: 'Dogs Ice-skate.'"),

# ---------- TYPE SYSTEM ----------
("Variables — Types",
"""Type = classification of data (int, float, bool, string).
Dynamic typing (Python) vs static typing (C, Rust, C++).

Python:
    a = 42          # int
    b = 3.14        # float
    ok = True       # bool

C:
    int a = 42;
    float b = 3.14f;
    _Bool ok = 1;   // or bool with <stdbool.h>

Rust:
    let a: i32 = 42;
    let b: f64 = 3.14;
    let ok: bool = true;

C++:
    int a = 42;
    double b = 3.14;
    bool ok = true;""",
"T D S → Type, Dynamic, Static.\nKindergarten: 'Turtles Dance Softly.'"),

# ---------- TYPE INFERENCE ----------
("Variables — Type Inference",
"""Inference = compiler deduces type from value.

Python:
    x = 10                 # runtime typing

Rust:
    let x = 10;            # infers i32
    let y = 3.14;          # infers f64

C++:
    auto x = 10;           # int
    auto y = 3.14;         # double""",
"I V → Inference, Value.\nKindergarten: 'Iguanas Vacation.'"),

# ---------- MUTABILITY ----------
("Variables — Mutability",
"""Mutable = can change after init. Immutable = cannot change.

Python:
    x = 1
    x = 2                   # rebinding allowed

C:
    int x = 1; x = 2;

Rust:
    let x = 1;              // immutable
    let mut y = 1; y = 2;   // mutable

C++:
    int x = 1; x = 2;""",
"M I → Mutable, Immutable.\nKindergarten: 'Mice Imagine.'"),

# ---------- CONSTANTS ----------
("Variables — Constants",
"""Constant = value that never changes.

Python:
    PI = 3.14159            # by convention uppercase

C:
    const int N = 10;

Rust:
    const N: i32 = 10;
    static MAX: i32 = 100;  // global static

C++:
    const int N = 10;""",
"C F S → Const, Fixed, Static.\nKindergarten: 'Cats Find Sun.'"),

# ---------- REFERENCES and POINTERS (preview) ----------
("Variables — Refs and Pointers (Preview)",
"""Reference = alias to another variable. Pointer = holds an address.

C:
    int v = 10;
    int *p = &v;     // pointer to v

Rust:
    let v = 10;
    let r = &v;      // reference (borrow)

C++:
    int v = 10;
    int &r = v;      // reference""",
"R P → Reference, Pointer.\nKindergarten: 'Rabbits Play.'"),

# ---------- SCOPE ----------
("Variables — Scope",
"""Scope = where a name is visible.

Python:
    x = 1
    def f():
        y = 2           # local

C:
    int x = 1;
    void f(){ int y = 2; }

Rust:
    let x = 1;
    { let y = 2; }      // y ends here

C++:
    int x = 1;
    void f(){ int y = 2; }""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- LIFETIME ----------
("Variables — Lifetime",
"""Lifetime = how long a value exists.

Python:
    def f():
        a = 1  # destroyed after f returns

C:
    void f(){ int a = 1; }  // stack lifetime

Rust:
    // compiler checks lifetimes via borrowing rules

C++:
    { int a = 1; }  // a destroyed at block end""",
"L B → Lifetime, Block.\nKindergarten: 'Lambs Bounce.'"),

# ---------- SHADOWING ----------
("Variables — Shadowing",
"""Shadowing = new declaration hides an older one in inner scope.

Python:
    x = 1
    def f():
        x = 2  # hides outer x inside f

Rust:
    let x = 1;
    let x = x + 1;  // new x shadows old

C++:
    int x = 1;
    { int x = 2; }  // inner x hides outer""",
"S H → Shadow, Hide.\nKindergarten: 'Shy Hedgehog.'"),

# ---------- INITIALIZATION PITFALLS ----------
("Variables — Init Pitfalls",
"""Uninitialized use can cause bugs.

C:
    int x; printf("%d", x);      // undefined

C++:
    int x; std::cout << x;       // undefined

Rust:
    let x: i32; // must assign before use -> compile error

Python:
    # NameError if you read a variable not yet defined""",
"U I → Uninitialized, Invalid.\nKindergarten: 'Umbrellas Itch.'"),

# ---------- NAMING ----------
("Variables — Naming",
"""Use clear, descriptive names. Prefer lower_snake_case in Python/Rust, camelCase or snake_case in C/C++ projects.

Python:
    max_count = 10

Rust:
    let max_count = 10;

C/C++:
    int max_count = 10;""",
"N C → Name, Clarity.\nKindergarten: 'New Car.'"),
]


functions = [

# ---------- CONCEPT ----------
("Functions — Concept",
"""A function is a reusable block of code.
- Takes input (parameters).
- May return output.
- Promotes modularity (splitting programs into parts).""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

# ---------- SIMPLE FUNCTION ----------
("Functions — Simple Example",
"""Python:
    def hello():
        print("Hello!")

C:
    void hello() {
        printf("Hello!\\n");
    }

Rust:
    fn hello() {
        println!("Hello!");
    }

C++:
    void hello() {
        std::cout << "Hello!\\n";
    }""",
"H → Hello.\nKindergarten: 'Happy Hippo.'"),

# ---------- PARAMETERS ----------
("Functions — Parameters & Return",
"""Python:
    def add(a, b):
        return a + b

C:
    int add(int a, int b) {
        return a + b;
    }

Rust:
    fn add(a: i32, b: i32) -> i32 {
        a + b
    }

C++:
    int add(int a, int b) {
        return a + b;
    }""",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),

# ---------- VOID / NO RETURN ----------
("Functions — No Return (Void)",
"""Python:
    def log(msg):
        print(msg)   # no explicit return

C:
    void log(const char* msg) {
        printf("%s", msg);
    }

Rust:
    fn log(msg: &str) {
        println!("{}", msg);
    }

C++:
    void log(const std::string& msg) {
        std::cout << msg;
    }""",
"V N → Void, No-return.\nKindergarten: 'Very Nice.'"),

# ---------- SCOPE ----------
("Functions — Scope",
"""Local variables exist only inside a function.
Global variables exist everywhere.

Python:
    x = 10
    def f():
        y = 20  # local

C:
    int x = 10;
    void f() {
        int y = 20;  // local
    }

Rust:
    let x: i32 = 10;
    fn f() {
        let y: i32 = 20;
    }

C++:
    int x = 10;
    void f() {
        int y = 20;
    }""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- CONCEPT ----------
("Functions — Concept",
"""Definition: Reusable block of code.
- Takes inputs (parameters).
- May return outputs.
- Improves modularity.""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

# ---------- CALL SEMANTICS ----------
("Functions — Call Semantics",
"""Value = copy of argument.
Reference = direct access to original.

Python:
    def inc(lst): lst.append(1)

C:
    void inc(int *x){ (*x)++; }

Rust:
    fn inc(x: &mut i32){ *x += 1; }

C++:
    void inc(int &x){ x++; }""",
"V R → Value, Ref.\nKindergarten: 'Very Round.'"),

# ---------- PURE vs SIDE EFFECT ----------
("Functions — Purity vs Side Effect",
"""Pure = same input → same output, no external change.
Side effect = modifies outside state (I/O, globals).""",
"P S → Pure, Side.\nKindergarten: 'Pink Snake.'"),

# ---------- SCOPE ----------
("Functions — Scope",
"""Local = variable inside a function.
Global = variable accessible everywhere.""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- RECURSION ----------
("Functions — Recursion",
"""Definition: Function calls itself.
Requires base case to stop recursion.

Python:
    def fact(n): return 1 if n<=1 else n*fact(n-1)""",
"R B → Recursion, Base.\nKindergarten: 'Red Balloon.'"),

# ---------- HIGHER ORDER ----------
("Functions — Higher Order",
"""Definition: Functions that take or return other functions.
Closure = captures surrounding variables.

Python:
    def make_adder(k): return lambda x:x+k""",
"H C → Higher, Closure.\nKindergarten: 'Happy Cat.'"),

# ---------- OVERLOAD / GENERIC ----------
("Functions — Overload & Generics",
"""Overload = same name, different parameters.
Generic = type-flexible functions.

C++:
    int add(int,int);
    double add(double,double);
    template<typename T>T maxT(T a,T b);""",
"O G → Overload, Generic.\nKindergarten: 'Orange Giraffe.'"),

# ---------- VARIADIC ----------
("Functions — Variadic",
"""Definition: Accept variable number of arguments.

Python:
    def total(*xs): return sum(xs)

C:
    int total(int n,...); // stdarg.h

C++:
    template<typename...T> int total(T...xs){ return (0+...+xs);}""",
"V A → Variadic Args.\nKindergarten: 'Very Antsy.'"),

# ---------- ERROR ----------
("Functions — Error Handling",
"""C = return codes.
C++ = exceptions.
Rust = Result/Option.
Python = exceptions.""",
"E R → Error, Result.\nKindergarten: 'Elephant Runs.'"),

# ---------- INLINE / LINK ----------
("Functions — Inline & Linkage",
"""Inline = hint to expand function at call site.
Linkage = scope across files (internal vs external).""",
"I L → Inline, Link.\nKindergarten: 'Ice Lion.'"),

# ---------- ASYNC ----------
("Functions — Asynchronous",
"""Sync = blocks until finished.
Async = returns immediately, completes later.

Python: async/await
Rust: async/await
C++20: coroutines
C: threads/event loops""",
"A S → Async, Sync.\nKindergarten: 'Ant Sings.'"),
]
logic = [

# ---------- CONCEPT ----------
("Logic — Concept",
"""Definition: Study of reasoning with true/false statements.
Purpose:
- Prove theorems
- Build formulas
- Show truth

Example:
Judy likes round things (T).
Donuts are round (T).
→ Judy loves donuts.""",
"L R T → Logic, Reasoning, Truth.\nKindergarten: 'Little Red Turtle.'"),

# ---------- PROPOSITIONS ----------
("Logic — Propositions",
"""Definition: Statement that is true or false.

Examples:
- All squares are yellow.
- x = 2
- AB bisects CD

Key: Words, symbols, or both — must have truth value.""",
"P T F → Proposition, True, False.\nKindergarten: 'Purple Tall Frog.'"),

# ---------- TRUTH VALUE ----------
("Logic — True/False",
"""Accept truth values as given, even if unrealistic.

Example: If '2+2=5' is true, accept it.

Tip: Logic = rules of the problem, not always reality.""",
"T F C → True, False, Context.\nKindergarten: 'Tiny Funny Cat.'"),

# ---------- CRITICAL THINKING ----------
("Logic — Critical Thinking",
"""Definition: Making new links from known truths.
Method: if–then.

Example:
x=5, y=1
- If z=x+y → z=6
- If z=x×y → z=5""",
"C I T → Critical, If-then, Truths.\nKindergarten: 'Cool Iguana Talks.'"),

# ---------- SUMMARY ----------
("Logic — Summary",
"""- Logic = reasoning with true/false.
- Propositions = must be T/F.
- Accept given truth values.
- Use if–then to connect truths.""",
"L P C → Logic, Propositions, Connections.\nKindergarten: 'Little Piggy Climbs.'"),

# ---------- CONCEPT ----------
("Fallacies — Concept",
"""Definition: Errors in reasoning, sound logical but lack proof.
Key: Flawed reasoning ≠ always false conclusion.

Example:
'All rabbits love carrots because one does.'
→ Hasty Generalization.""",
"L F R → Logical, Fallacy, Reasoning.\nKindergarten: 'Little Funny Rabbit.'"),

# ---------- HASTY GENERALIZATION ----------
("Fallacies — Hasty Generalization",
"""Definition: Conclusion from too little evidence.
Example: Judy’s baby should talk at 8 months since Jill’s did.
Flaw: One case ≠ all.""",
"H G E → Hasty, Generalization, Evidence.\nKindergarten: 'Happy Green Elephant.'"),

# ---------- CIRCULAR REASONING ----------
("Fallacies — Circular Reasoning",
"""Definition: Claim proves itself; no outside proof.
Examples:
- 'Because I said so.'
- 'Stars are painted because they’re painted.'
Flaw: Claim = proof.""",
"C R S → Circular, Reasoning, Same.\nKindergarten: 'Curly Red Snake.'"),

# ---------- FALSE CAUSE ----------
("Fallacies — False Cause",
"""Definition: Mistaking correlation for causation.
Example: Swing arms → doors open.
Flaw: Timing ≠ cause; doors are automatic.""",
"F C C → False, Cause, Correlation.\nKindergarten: 'Funny Clever Cat.'"),

# ---------- LIMITED CHOICE ----------
("Fallacies — Limited Choice",
"""Definition: Only some options given, others ignored.
Example: Survey lists only blue, red, yellow.
Flaw: Excludes real options.""",
"L C O → Limited, Choice, Options.\nKindergarten: 'Little Cute Octopus.'"),

# ---------- SUMMARY ----------
("Fallacies — Summary",
"""- Fallacies = flawed reasoning.
Types:
- Hasty Gen → not enough evidence
- Circular → claim proves itself
- False Cause → correlation ≠ causation
- Limited Choice → unfair options""",
"F L T → Fallacies, Logic, Types.\nKindergarten: 'Funny Little Turtle.'"),

]

ml_week1 = [

# ---------- CONCEPT ----------
("ML — What is Machine Learning?",
"""Mitchell (1997): A program learns from experience E on task T with measure M if performance at T improves with E.
Plain English: Algorithms improve with practice by learning from data, not rules.""",
"M L E T M → Machine, Learning, Experience, Task, Measure.\nKindergarten: 'Many Little Elephants Teach Math.'"),

# ---------- FORMAL SETUP ----------
("ML — Formal Setup",
"""Input X = features (e.g., age, glucose).
Output Y = labels (e.g., diabetes yes/no).
Hypothesis H = possible models.
Loss L = penalty for mistakes.
Risk R(f) = expected loss under D.
Empirical Risk R̂(f) = average loss on data.

👉 Solve:
1. Choose model f(x).
2. Compute predictions f(x_i).
3. Apply loss L(f(x_i), y_i).
4. Average across samples = R̂(f).""",
"X Y H L R → Input, Output, Hypothesis, Loss, Risk.\nKindergarten: 'Xylophones Yell Happy Little Rabbits.'"),

# ---------- EMPIRICAL RISK EXAMPLE ----------
("ML — Empirical Risk Example",
"""Model: f(x)=2x
Data: (1,2),(2,4),(3,7)
Predictions → (2,4,6)
Errors → (0,0,-1)
Squares → (0,0,1)
Avg = (0+0+1)/3 = 0.33

👉 Solve:
Prediction − Truth → Error → Square → Average.""",
"E R → Empirical Risk.\nKindergarten: 'Eager Rabbit.'"),

# ---------- INDUCTIVE BIAS ----------
("ML — Inductive Bias",
"""Models assume patterns:
- Linear → straight lines
- Trees → axis splits
- Nets → layered features
No bias → no generalization.

👉 Solve:
Identify what assumptions a model makes about patterns.""",
"I B → Inductive, Bias.\nKindergarten: 'Ice Bear.'"),

# ---------- GENERALIZATION ----------
("ML — Generalization",
"""Training accuracy ≠ real-world accuracy.
Gap = Train error − Test error.
Goal: minimize gap, not just training loss.

👉 Solve:
1. Compute training error.
2. Compute test error.
3. Gap = Train − Test.""",
"G G → Generalization, Gap.\nKindergarten: 'Green Giraffe.'"),

# ---------- HISTORY ----------
("ML — History & Milestones",
"""1950s: Turing Test
1957: Perceptron
1980s: Backpropagation
1990s: SVMs
2012: AlexNet
2016: AlphaGo
2018+: Transformers
2020+: Foundation models

👉 Takeaway: Shift from hand-engineered features → learned representations.""",
"H T P S A → History, Turing, Perceptron, SVMs, AlexNet.\nKindergarten: 'Happy Tiger Plays Soccer Always.'"),

# ---------- SUPERVISED ----------
("ML — Supervised Learning",
"""Data has inputs (x) and labels (y).
Tasks:
- Regression → numbers (loss=MSE).
- Classification → categories (loss=Cross-Entropy).

👉 Solve Linear Regression by Hand:
1. Compute means (x̄,ȳ).
2. Compute slope a = Sxy/Sxx.
3. Compute intercept b = ȳ − a·x̄.
4. Line = ax+b.""",
"S R C → Supervised, Regression, Classification.\nKindergarten: 'Silly Red Cat.'"),

# ---------- UNSUPERVISED ----------
("ML — Unsupervised Learning",
"""Data has only x (no labels).
Tasks: clustering, dimensionality reduction, anomaly detection.

👉 Solve K-means:
1. Pick k centers.
2. Assign points to nearest center.
3. Recompute centers.
4. Repeat until stable.""",
"U C → Unsupervised, Clustering.\nKindergarten: 'Ugly Cat.'"),

# ---------- REINFORCEMENT ----------
("ML — Reinforcement Learning",
"""Modeled as MDP: states, actions, rewards.
Goal: maximize discounted return E[Σ γ^t r_t].

👉 Solve Q-Learning Update:
Q(s,a) ← Q(s,a) + α[r + γ max Q(s′,a′) − Q(s,a)]
Ex: Old Q=5, r=2, γ=0.9, max=10, α=0.1
Target = 2+0.9*10=11
Error = 11−5=6
Update = 5+0.1*6=5.6.""",
"R Q → Reinforcement, Q-learning.\nKindergarten: 'Rabbit Quest.'"),

# ---------- DEEP LEARNING ----------
("ML — Deep Learning",
"""Learns layered features.
Universal Approximation: nets with enough units can approximate any continuous function.
Architectures: CNNs (images), RNNs (sequences), Transformers (attention).

👉 Solve:
Pick architecture suited to data type (image, sequence, text).""",
"D L N → Deep, Layers, Nets.\nKindergarten: 'Dancing Little Narwhal.'"),

# ---------- METRICS ----------
("ML — Evaluation Metrics",
"""Why not accuracy? Imbalance hides error.
Metrics:
- Accuracy = (TP+TN)/All
- Precision = TP/(TP+FP)
- Recall = TP/(TP+FN)
- F1 = 2PR/(P+R)

👉 Solve Example:
TP=40, FP=10, FN=5, TN=45
Accuracy=0.85
Precision=0.80
Recall=0.89
F1=0.84""",
"A P R F → Accuracy, Precision, Recall, F1.\nKindergarten: 'Apple Pie Really Fun.'"),

# ---------- CASE STUDY ----------
("ML — Healthcare Case Study",
"""Task: predict diabetes.
Inputs: glucose, BMI, age.
Output: probability of diabetes.
Metric: ROC-AUC.

Risks:
- False negatives → missed patients.
- Bias → unfair predictions.
- Privacy → HIPAA compliance.""",
"H C D → Healthcare, Case, Diabetes.\nKindergarten: 'Happy Cute Doctor.'"),

# ---------- LIMITS ----------
("ML — Limits & Misconceptions",
"""Limits:
- No Free Lunch: no best algorithm for all tasks.
- Bias–Variance tradeoff: simple = underfit, complex = overfit.
- Spurious correlations: models may cheat.

Misconceptions:
- More data ≠ always better.
- 100% accuracy = usually leakage.
- Deep learning ≠ always best (tabular → XGBoost better).""",
"L N B → Limits, NFL, Bias-variance.\nKindergarten: 'Little Naughty Bear.'"),

# ---------- DATA SYMBOLS ----------
("Definitions — Data Symbols",
"""x (feature): the input used to predict (e.g., house size).
y (label/target): the true answer (e.g., house price).
(x_i, y_i): the i-th training example.
m: number of training examples.
Dataset: the full collection of m examples.

Solve:
Given (1,2),(2,4),(3,5) ⇒ m=3; x_2=2, y_2=4.""",
"X Y M → x, y, m.\nKindergarten: 'Xylophone Yodels Merrily.'"),

# ---------- MODEL SYMBOLS ----------
("Definitions — Model Symbols",
"""θ (theta): parameter vector learned from data.
θ0 (intercept/bias): prediction when x=0.
θ1 (slope/weight): change in prediction per +1 in x.
Parameters: all θ values the algorithm tunes.

Solve:
If θ0=0.5, θ1=1.2 ⇒ predict rises by 1.2 when x increases by 1.""",
"T P W → Theta, Parameters, Weight.\nKindergarten: 'Tiny Penguin Waves.'"),

# ---------- HYPOTHESIS FUNCTION ----------
("Definition — Hypothesis hθ(x)",
"""hθ(x): prediction rule (the model).
For linear regression: hθ(x) = θ0 + θ1 x.
Meaning: plug x and θ into the line to get predicted y.

Example:
θ0=0.4, θ1=0.8, x=2 ⇒ hθ(2)=0.4+0.8·2=2.0.""",
"H L P → Hypothesis, Line, Prediction.\nKindergarten: 'Happy Little Parrot.'"),

# ---------- PREDICTION & ERROR ----------
("Definitions — Prediction & Error",
"""Prediction (ŷ): model’s guess = hθ(x).
Error/residual: ŷ − y (signed difference).
Absolute error: |ŷ − y|.
Squared error: (ŷ − y)^2.

Example:
ŷ=2.0, y=4 ⇒ error = -2, squared error = 4.""",
"P E S → Prediction, Error, Squared.\nKindergarten: 'Puppy Eats Snacks.'"),

# ---------- COST FUNCTION ----------
("Definition — Cost J(θ) (MSE)",
"""J(θ): wrongness score across the dataset.
Mean Squared Error (with 1/(2m) scaling):
J(θ) = (1/(2m)) Σ_i (hθ(x_i) − y_i)^2
Σ: summation over all training examples.

Example (1 point x=2,y=4, θ0=0,θ1=0):
ŷ=0 ⇒ J = (1/(2·1))·(0−4)^2 = 8.""",
"J M Σ → J, MSE, Summation.\nKindergarten: 'Jelly Makes Sandwiches.'"),

# ---------- GRADIENT / PARTIAL DERIVATIVE ----------
("Definition — Gradient ∂J/∂θ_j",
"""∂J/∂θ_j: slope of cost with respect to parameter θ_j.
Meaning: direction and magnitude to change θ_j to lower J.
For linear regression with MSE (single feature):
∂J/∂θ0 = (1/m) Σ (hθ(x_i) − y_i)
∂J/∂θ1 = (1/m) Σ (hθ(x_i) − y_i) x_i

Example (x=2,y=4, θ0=0,θ1=0):
∂J/∂θ0 = -4, ∂J/∂θ1 = -8.""",
"G S D → Gradient, Slope, Direction.\nKindergarten: 'Giraffe Slides Down.'"),

# ---------- LEARNING RATE ----------
("Definition — Learning Rate α",
"""α (alpha): step size in gradient descent.
Large α: faster but can diverge.
Small α: stable but slow.

Tip: try values like 0.001, 0.01, 0.1; watch J(θ) decrease.""",
"A S C → Alpha, Step, Control.\nKindergarten: 'Ant Steps Carefully.'"),

# ---------- GRADIENT DESCENT ----------
("Definition — Gradient Descent (Update Rule)",
"""Algorithm to minimize J(θ).
Update:
θ_j := θ_j − α · (∂J/∂θ_j)

Meaning: move θ opposite the gradient to roll downhill on the cost surface.

Example (x=2,y=4; θ0=0,θ1=0; α=0.1):
Gradients: (-4, -8).
θ0=0.4, θ1=0.8 ⇒ h(2)=2 (closer to 4).""",
"G U M → Gradient, Update, Minimize.\nKindergarten: 'Goose Uses Map.'"),

# ---------- MULTI-POINT EXAMPLE ----------
("Definitions — Multi-point Example (Compute J & Gradients)",
"""Data: (1,2),(2,4),(3,5), m=3; start θ0=0,θ1=0
Preds: [0,0,0]  Errors: [-2,-4,-5]
Cost:
J = (1/(2·3))[(−2)^2+(−4)^2+(−5)^2] = 7.5
Gradients:
∂J/∂θ0 = -11/3 ≈ -3.67
∂J/∂θ1 = -25/3 ≈ -8.33
α=0.1 ⇒ θ0=0.367, θ1=0.833 (errors shrink).""",
"M C G → Multi, Cost, Gradients.\nKindergarten: 'Mouse Counts Grapes.'"),

# ---------- “WHAT EACH THING IS” QUICK CARD ----------
("Definitions — What Each Symbol Is",
"""x: input feature (e.g., size) • y: target label (e.g., price)
(x_i, y_i): i-th training point • m: number of points
θ0: intercept • θ1: slope • θ=(θ0,θ1): all parameters
hθ(x): prediction rule (model) • ŷ=hθ(x): predicted y
J(θ): cost (MSE with 1/(2m) scaling) • Σ: sum over data
∂J/∂θ_j: gradient component • α: learning rate
Update rule: θ_j ← θ_j − α·∂J/∂θ_j""",
"S D P → Symbols, Definitions, Pointers.\nKindergarten: 'Seal Draws Pictures.'"),

# ---------- THE LEARNING STORY ----------
("ML — The Learning Loop (Concept)",
"""Steps of supervised learning:
1. Predict with current line (hθ(x)).
2. Compare to real y (error).
3. Measure overall wrongness (J).
4. Adjust θ to reduce J.
5. Repeat until J is small.

👉 Key: ML is not about one guess, it’s about looping to improve.""",
"L P C A → Loop, Predict, Compare, Adjust.\nKindergarten: 'Little Puppy Chases Apples.'"),

# ---------- WHY COST EXISTS ----------
("ML — Why We Need a Cost Function",
"""One guess can be good or bad.
We need a single number to tell us:
'How wrong is this line overall?'

👉 Cost = average wrongness across all points.
This lets us compare lines and pick the best.""",
"C N W → Cost, Number, Wrongness.\nKindergarten: 'Cat Needs Water.'"),

# ---------- WHY SQUARED ----------
("ML — Why Square Errors?",
"""- If we just sum errors, positives + negatives cancel.
- Absolute values work, but are harder to optimize.
- Squared makes errors positive AND smooths math.
👉 Squared error curve = bowl shape → one global minimum.""",
"S P B → Squared, Positive, Bowl.\nKindergarten: 'Seal Plays Ball.'"),

# ---------- GRADIENT DESCENT INTUITION ----------
("ML — Gradient Descent Intuition",
"""Imagine standing on a hill (the cost function).
- Gradient = slope of the hill.
- Step downhill = reduce cost.
- Learning rate α = step size.
👉 Repeat: small downhill steps until you reach the bottom.""",
"H S B → Hill, Slope, Bottom.\nKindergarten: 'Horse Slides Backwards.'"),

# ---------- LEARNING RATE INTUITION ----------
("ML — Learning Rate α Intuition",
"""- Too big → overshoot, bounce around.
- Too small → super slow.
- Just right → steady downhill walk.

👉 α is like turning your walking speed knob.""",
"S F J → Small, Fast, Just-right.\nKindergarten: 'Snail Flies Jolly.'"),

# ---------- SYMBOLS IN PLAIN ENGLISH ----------
("ML — Symbols as Knobs",
"""- θ0: starting point of the line (intercept).
- θ1: tilt of the line (slope).
- hθ(x): the line’s guess (prediction).
- J(θ): score of how wrong the line is.
- ∂J/∂θ: tells us how to twist θ to reduce wrongness.

👉 Think: knobs you twist until guesses are good.""",
"K G W → Knobs, Guess, Wrongness.\nKindergarten: 'Koala Grabs Watermelon.'"),

# ---------- THE BIG PICTURE ----------
("ML — Big Picture of Week 1",
"""1. Data: (x,y) pairs.
2. Hypothesis: line hθ(x)=θ0+θ1x.
3. Cost: J(θ) = wrongness score.
4. Gradient: slope of J.
5. Descent: update θ to lower J.
👉 Repeat until the line fits data well.""",
"D H C G → Data, Hypothesis, Cost, Gradient.\nKindergarten: 'Duck Hugs Cute Goat.'"),

# ---------- PYTHON DEMO ----------
("ML — Linear Regression in Python",
"""import numpy as np
from sklearn.linear_model import LinearRegression

# Example data: house sizes vs prices
X = np.array([[1],[2],[3],[4],[5]])   # sizes (1000 sq ft)
y = np.array([200,400,600,800,1000])  # prices ($100k)

model = LinearRegression()
model.fit(X, y)

print("Slope θ1:", model.coef_[0])
print("Intercept θ0:", model.intercept_)
print("Prediction for x=6:", model.predict([[6]])[0])

👉 Demo: Fit line, check slope & intercept, make prediction.""",
"P C S → Python, Code, Sklearn.\nKindergarten: 'Pink Cat Sleeps.'"),

# ---------- SUPERVISED LR BY HAND ----------
("ML — Linear Regression by Hand (Supervised)",
"""Data: (1,2), (2,3), (3,5)
Step 1: Compute means → x̄=2, ȳ=3.33
Step 2: Compute slope → a=Sxy/Sxx=3/2=1.5
Step 3: Intercept → b=ȳ − a·x̄=3.33 −1.5·2=0.33
Line → ŷ=1.5x+0.33

👉 Each 1-unit increase in x raises y by ~1.5.""",
"L S I → Line, Slope, Intercept.\nKindergarten: 'Little Snail Igloo.'"),

# ---------- KEY TAKEAWAYS ----------
("Week 1 — Key Takeaways",
"""- ML = programming by example (E,T,M).
- Supervised = labeled data → predictions.
- Linear regression = predict with a line.
- Cost = MSE measures wrongness.
- Gradient descent = adjust θ to minimize cost.
- Symbols: x,y,m,θ,hθ,J,α all have clear roles.
👉 Core idea: Predict → Compare → Adjust → Repeat.""",
"K L C G → Key, Linear, Cost, Gradient.\nKindergarten: 'Kind Lion Catches Grapes.'"),

]

mit_algorithms_101 = [

# =========================================================
# 2) PROBLEMS vs. ALGORITHMS
# =========================================================

("Computational Problem — Responsibility",
"Inputs (given data) and outputs (answers) connected by a relation (rule).\n"
"Often expressed with a predicate (True/False test).\n"
"Example: 'Does index i hold the number 5?'"),

("Computational Problem — Mnemonic",
"I O R P → Inputs, Outputs, Relation, Predicate.\nKindergarten: 'Inchworms Open Red Presents.'"),

("Algorithm — Responsibility",
"A step-by-step method (function) that returns one correct output for each input.\nMust handle arbitrarily large inputs.\nAlways correct (valid output)."),

("Algorithm — Mnemonic",
"F C S → Function, Correct, Scalable.\nKindergarten: 'Frogs Climb Stairs.'"),


# =========================================================
# 3) BIRTHDAY MATCHING EXAMPLE
# =========================================================

("Birthday Matching — Problem",
"Decide if two students share the same birthday."),

("Birthday Matching — Algorithm",
"Keep a record (list or set).\nFor each student:\n- If birthday in record → return match.\n- Else add it.\nIf no repeats → return none.\n\nEfficiency:\n- Record = list → O(n^2).\n- Record = hash set → O(n)."),

("Birthday Matching — Mnemonic",
"L H → List = Hard, Hash = Handy.\nKindergarten: 'Lions Hug.'"),


# =========================================================
# 4) PROVING CORRECTNESS
# =========================================================

("Correctness Proof — Responsibility",
"Use induction:\n- Base case: trivial start (0 students → no match).\n- Step: assume true for k, prove for k+1.\nMaintain invariant: after k steps, rule still holds."),

("Correctness Proof — Mnemonic",
"B S I → Base, Step, Invariant.\nKindergarten: 'Bears Sing Inside.'"),


# =========================================================
# 5) MEASURING EFFICIENCY
# =========================================================

("Efficiency — Responsibility",
"Count operations, not clock time.\nUse asymptotic analysis.\nBig O = worst, Ω = best, Θ = tight (both)."),

("Efficiency — Growth Rates",
"O(1) constant, O(log n) logarithmic, O(n) linear, O(n^2) quadratic, O(n^c) polynomial, O(c^n) exponential."),

("Efficiency — Mnemonic",
"C L N Q P E → Constant, Log, Linear, Quadratic, Poly, Exponential.\nKindergarten: 'Cats Love Nice Quiet Purple Eggs.'"),


# =========================================================
# 6) RAM MODEL OF COMPUTATION
# =========================================================

("RAM Model — Responsibility",
"Assume any memory cell accessed in constant time.\nWord = fixed chunk (e.g. 64 bits).\nAllowed ops: read/write word, arithmetic, logic.\nReading n items always costs Ω(n)."),

("RAM Model — Mnemonic",
"C W O → Constant access, Word size, Ops.\nKindergarten: 'Cats Wear Overalls.'"),


# =========================================================
# 7) DATA STRUCTURES
# =========================================================

("Data Structures — Responsibility",
"Static array (fixed box).\nList (ordered collection).\nDictionary/Hash (fast key-value lookup).\nTree (hierarchical)."),

("Data Structures — Mnemonic",
"A L D T → Array, List, Dict, Tree.\nKindergarten: 'Apples Like Delicious Treats.'"),


# =========================================================
# 8) STRATEGIES TO SOLVE PROBLEMS
# =========================================================

("Problem-Solving Strategies — Responsibility",
"1) Reduce to known problem.\n2) Design new algorithm (recursive, iterative).\n3) Paradigms: divide-and-conquer, greedy, dynamic programming, graph."),

("Problem-Solving — Mnemonic",
"R N P → Reduce, New, Paradigms.\nKindergarten: 'Rabbits Need Pie.'"),


# =========================================================
# 9) COURSE ROADMAP
# =========================================================

("Course Roadmap — Responsibility",
"Quiz 1: Data structures + sorting\nQuiz 2: Graphs + shortest paths\nFinal: Dynamic programming"),


# =========================================================
# 10) FLASHCARDS (Q/A style)
# =========================================================

("Q: What is a computational problem?",
"A: A relation between inputs (given data) and outputs (answers)."),

("Q: What is an algorithm?",
"A: A step-by-step function that gives one correct output per input."),

("Q: What is induction?",
"A: A proof method with a base case (start) and inductive step (extend to k+1)."),

("Q: What are Big O, Ω, Θ?",
"A: O = upper bound (worst), Ω = lower bound (best), Θ = tight bound (both)."),

("Q: What does the RAM model assume?",
"A: Any memory cell can be accessed in constant time."),


# =========================================================
# 11) PRACTICE PROBLEMS
# =========================================================

("Practice — Define Problem",
"Task: Define if an array has a negative number.\nInputs: array A.\nOutput: Yes if ∃ i with A[i]<0, else No.\nPredicate true if element <0."),

("Practice — Max Element Algorithm",
"Set max ← first element.\nFor each element: if > max, update.\nReturn max.\nRunning time O(n)."),

("Practice — Max Correctness Proof",
"Invariant: after k elements, max holds maximum so far.\nBase: first element correct.\nStep: compare next; update if larger.\nAt end: max = true maximum."),

("Practice — Duplicate Search Comparison",
"O(n^2) check every pair vs. O(n) with hash set.\nBoth correct, hash set asymptotically faster."),

("Practice — RAM Model Word Size",
"64-bit CPU → 2^64 ≈ 1.8×10^19 bytes (≈ 20 exabytes)."),

 # =========================================================
# 1) INTERFACES vs DATA STRUCTURES
# =========================================================

("Interface vs Data Structure — Responsibility",
"Interface = what you can do (menu of actions + meaning).\n"
"Data structure = how it’s built (storage + algorithms to run ops).\n"
"Key idea: keep WHAT separate from HOW (clean design; easy swaps)."),

("Interface vs Data Structure — Mnemonic",
"W H S → What, How, Swap.\nKindergarten: 'Worms Hug Sandwiches.'"),

# =========================================================
# 2) TWO INTERFACES (Sequence & Set)
# =========================================================

("Interfaces We Care About — Responsibility",
"Sequence (ordered items: first, second, …).\n"
"Set (unique items; cares about membership; may order by key).\n"
"(Lecture 2 focuses on Sequence)."),

("Sequence & Set — Mnemonic",
"O U → Ordered, Unique.\nKindergarten: 'Otters Unite.'"),

# =========================================================
# 3) SEQUENCE — STATIC INTERFACE (fixed size)
# =========================================================

("Sequence (Static) — Responsibility",
"Items x0..x_{n-1}. Ops: build(X), length(), iter(), get_at(i), set_at(i).\n"
"Natural solution: static array (contiguous memory)."),

("Static Array — Performance",
"get_at/set_at: O(1) (constant)\n"
"length: O(1)\n"
"build: Θ(n) (allocate + initialize)\n"
"Word-size need: w ≥ log n (word must hold any index up to n)."),

("Static Array — Python sketch",
"```python\n# Conceptual: Python list acts like an array for O(1) reads/writes\nA = [10, 20, 30]\nA[1] = 99           # set_at(1, 99)\nx = A[1]            # get_at(1)\nlenA = len(A)       # length()\nfor v in A: ...     # iter()\n```"),

# =========================================================
# 4) MEMORY ALLOCATION MODEL
# =========================================================

("Allocation Model — Responsibility",
"Assume we can allocate an array of size n in Θ(n) time.\n"
"(Keeps space/time realistic; no 'instant infinite arrays')."),

# =========================================================
# 5) SEQUENCE — DYNAMIC INTERFACE (grow/shrink)
# =========================================================

("Sequence (Dynamic) — Responsibility",
"Add ops: insert_at(i, x), delete_at(i).\n"
"Common fast targets (special cases): insert_first/last, delete_first/last,\n"
"get_first/last, set_first/last (ends can be cheaper on some structures)."),

# =========================================================
# 6) LINKED LISTS (pointer-based)
# =========================================================

("Linked List — Responsibility",
"Nodes store (value + next pointer). Keep head (and often tail).\n"
"Pointers fit in one word → following a pointer is O(1)."),

("Linked List — Performance",
"insert_first / delete_first: O(1)\n"
"get_at(i) / set_at(i): Θ(i) (worst Θ(n)) — must walk i steps\n"
"get_last: O(1) if you store tail (augmentation)"),

("Linked List — Python sketch",
"```python\n# Python tip: use collections.deque for O(1) appends/pops at both ends\nfrom collections import deque\nq = deque([1,2,3])\nq.appendleft(0)   # insert_first in O(1)\nq.append(4)       # insert_last in O(1)\nq.popleft()       # delete_first in O(1)\nq.pop()           # delete_last in O(1)\n```"),

# =========================================================
# 7) STATIC ARRAYS FOR DYNAMIC OPS (why it hurts)
# =========================================================

("Static Array for Dynamic Ops — Responsibility",
"insert_at/delete_at: Θ(n) (shift many items to keep indices).\n"
"insert_last/delete_last: Θ(n) if resizing/copying is needed (no room)."),

# =========================================================
# 8) DYNAMIC ARRAYS (how Python lists grow)
# =========================================================

("Dynamic Array — Responsibility",
"Keep length n (used) and capacity size (reserved). Slots [0..n-1]=data; [n..size-1]=empty.\n"
"insert_last(x): if n<size → write at a[n] (O(1)); else resize then write."),

("Dynamic Array — Amortized Analysis",
"Grow by a factor (e.g., double). Resizes at 1,2,4,8,…\n"
"Total copied over n appends is Θ(n) → average O(1) per append (amortized)."),

("Dynamic Array — Python sketch",
"```python\nA = []\nfor x in range(1_000):  # appends are amortized O(1)\n    A.append(x)\nA.pop()                  # delete_last (amortized O(1) with a shrink policy)\nA[123] = 42              # get/set at index are O(1)\n```"),

# =========================================================
# 9) QUICK COMPARISON TABLE (big-O)
# =========================================================

("Static vs Linked vs Dynamic Array — Performance",
"get_at/set_at:  Array O(1) | List Θ(n) | DynArray O(1)\n"
"insert_first/delete_first: Array Θ(n) | List O(1) | DynArray Θ(n)\n"
"insert_last: Array Θ(n) | List Θ(n) or O(1 w/ tail) | DynArray O(1) amortized\n"
"delete_last: Array Θ(n) | List Θ(n) or O(1 w/ doubly+tail) | DynArray O(1) amortized\n"
"insert_at/delete_at (middle): Θ(n) for all (shift or walk)."),

("Choosing Structures — Mnemonic",
"R E A → Random, Ends, Average appends.\n"
"Array wins Random, List wins Ends, DynArray wins Average appends.\n"
"Kindergarten: 'Rabbits Eat Apples.'"),

# =========================================================
# 10) KEY TERMS (parentheses)
# =========================================================

("Key Terms — Responsibility",
"Interface (menu/spec): what ops mean.\n"
"Data structure (storage): how ops are built.\n"
"Word RAM (simple computer model): constant-time word access.\n"
"Word (w bits): fixed-size 'number box' (often 64 bits).\n"
"Address (memory location): where an item lives.\n"
"Pointer (stored address): number that points to a node.\n"
"Contiguous (back-to-back in memory): no gaps.\n"
"Allocate (reserve memory): ask OS for space.\n"
"Initialize (set starting values): e.g., fill zeros.\n"
"Amortized (averaged across many steps): occasional big cost, small on average.\n"
"Augmentation (extra stored info): e.g., keep tail pointer."),

# =========================================================
# 11) FLASHCARDS (Q/A)
# =========================================================

("Q: Interface vs data structure?",
"A: Interface = what ops mean; data structure = how they’re implemented."),

("Q: Why array get_at is O(1)?",
"A: Contiguous memory + offset math: address = base + i."),

("Q: Why must w ≥ log n?",
"A: The word must hold any index up to n (to address elements in constant time)."),

("Q: Why doubling → amortized O(1) append?",
"A: Resizes at powers of two; total copies over n appends Θ(n), so average per append is O(1)."),

("Q: When are linked lists better than arrays?",
"A: Frequent insert/delete at the front (ends), little random access."),

# =========================================================
# 12) PRACTICE (with solutions)
# =========================================================

("P1 — Pick a structure (solution included)",
"Need: (a) fast random get/set, (b) frequent append, (c) rare deletes.\n"
"Pick dynamic array (Python list). get/set O(1); append amortized O(1); deletes acceptable.\n"
"```python\nA = []\nfor x in data: A.append(x)\nA[i] = 7\n```"),

("P2 — Index math (solution included)",
"Array base address = 10, one word per element. Address of element i = 10 + i (base + offset)."),

("P3 — Linked-list tail (solution included)",
"Goal: O(1) get_last(). Keep a tail pointer and maintain it on edits.\n"
"(Augmentation = store extra info to speed ops.)"),

("P4 — insert_first cost (solution included)",
"Static array insert_first is Θ(n): must shift n items right to keep indices aligned."),

("P5 — Amortized append proof (solution included)",
"Doubling capacity → copies: 1+2+4+...+2^{⌊log n⌋} = Θ(n). So total Θ(n) for n appends → O(1) amortized."),

("P6 — When list beats dyn array? (solution included)",
"Heavy insert_first/delete_first with little random access (queue/deque ends). List gives O(1) at ends; dyn array pays Θ(n) at front.\n"
"```python\nfrom collections import deque\nq = deque(); q.appendleft(1); q.popleft()\n```"),

("P7 — Word size intuition (solution included)",
"To name any of n positions, you need numbers up to n. That needs about log2(n) bits → w must grow with n."),

("P8 — insert_at(i, x) in dyn array (solution included)",
"Θ(n): even with extra capacity, shifting elements i..n-1 right costs linear time.\n"
"```python\nA.insert(i, x)   # conceptual: Python does the shifts under the hood\n```"),

("P9 — Doubly-linked list advantage (solution included)",
"delete_last becomes O(1) with prev pointers + tail. In singly list you must walk to find the predecessor (Θ(n))."),

("P10 — Build cost (solution included)",
"Static array build is Θ(n): allocate + initialize + copy items.")

]

ml_lifecycle_week1 = [

# ---------- LIFECYCLE ----------
("ML Lifecycle — Concept",
"""Machine learning follows a structured lifecycle with 5 iterative phases:
1. Define the project
2. Acquire & explore data
3. Modeling
4. Evaluation
5. Maintenance""",
"D A M E M → Dogs Always Make Everyone Merry",
"""**Meaning**
The machine learning lifecycle is a structured, repeatable process used to design, build, test, and maintain ML systems. It ensures disciplined, scientific progress instead of ad-hoc trials.

1. **Define the Project** – Clarify goal, metric, and baseline.
2. **Acquire & Explore Data** – Gather, clean, and understand data.
3. **Modeling** – Train and tune algorithms.
4. **Evaluation** – Measure accuracy and interpret results.
5. **Maintenance** – Deploy, monitor, and retrain as data changes.

**Code Example**
```python
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2)
model = RandomForestClassifier()
model.fit(X_train, y_train)
preds = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, preds))
```"""),

("ML Lifecycle — Define the Project",
"""Questions:
- What problem are you solving?
- What metric measures success?
- What baseline must you beat?""",
"P M B → Problem, Metric, Baseline.\nKindergarten: 'Pandas Make Bread.'",
"""**Meaning**
Defining the project sets the foundation for the entire lifecycle. You must clearly describe **what problem** the model solves, **how success** is measured, and **what baseline** defines minimum acceptable performance.

1. **Problem** – The specific question to answer (e.g., “Can we predict home prices?”).
2. **Metric** – The quantitative measure of success (accuracy, precision, MSE, etc.).
3. **Baseline** – The simplest comparison model (like always predicting the mean).

This ensures all future steps align with measurable objectives.

**Code Example**
```python
y_true = [200000, 250000, 300000]
y_pred_baseline = [250000, 250000, 250000]  # simple average predictor
mse = mean_squared_error(y_true, y_pred_baseline)
print("Baseline MSE:", mse)
```"""),

("ML Lifecycle — Acquire & Explore Data",
"""Tasks:
- Collect the right data
- Clean missing values
- Perform EDA
- Engineer features""",
"C C E F → Collect, Clean, Explore, Features.\nKindergarten: 'Cats Cook Easy Food.'",
"""**Meaning**
The data stage ensures that your model has a solid foundation. You **collect**, **clean**, and **analyze** data to prepare it for learning.
Exploratory Data Analysis (EDA) helps you visualize relationships, detect errors, and design new **features** that improve performance.

1. **Collect Data** – Choose relevant, trustworthy sources.
2. **Clean Data** – Handle missing or corrupted values.
3. **Explore Data** – Use visualizations and statistics to find patterns.
4. **Feature Engineering** – Create variables that make learning easier.

**Code Example**
```python
df = sns.load_dataset("titanic").dropna(subset=["age", "fare"])
sns.scatterplot(data=df, x="age", y="fare", hue="class")

df["fare_per_age"] = df["fare"] / df["age"]
print(df[["age", "fare", "fare_per_age"]].head())
```"""),

("ML Lifecycle — Modeling",
"""Steps:
- Train candidate models
- Select variables
- Tune hyperparameters""",
"T S T → Train, Select, Tune.\nKindergarten: 'Tigers Sing Tunes.'",
"""**Meaning**
Modeling is where algorithms learn from data. You train different models, adjust settings (hyperparameters), and evaluate which approach generalizes best.
Good modeling focuses on balancing **accuracy**, **simplicity**, and **interpretability**, ensuring the model performs well on unseen data.

1. **Train** – Fit multiple candidate models.
2. **Select** – Compare their performance on validation data.
3. **Tune** – Adjust hyperparameters to optimize results.

**Code Example**
```python
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target)
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
print("Coefficients:", model.coef_)
```"""),

("ML Lifecycle — Evaluation",
"""Ask:
- Does the model beat the baseline?
- Are results meaningful?""",
"B M → Baseline, Meaningful.\nKindergarten: 'Bears Moo.'",
"""**Meaning**
Evaluation verifies that your model actually solves the problem it was designed for. You test it on unseen data, compare it to the baseline, and analyze its behavior with metrics and visualizations.
The goal is to confirm that performance improvements are **real**, **interpretable**, and **generalizable**.

1. **Compare to Baseline** – Ensure measurable improvement.
2. **Use Proper Metrics** – Choose accuracy, F1-score, or MSE based on task.
3. **Interpret Results** – Use confusion matrices or ROC curves for insight.

**Code Example**
```python
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
print("Confusion Matrix:\\n", confusion_matrix(y_test, y_pred))
```"""),

("ML Lifecycle — Maintenance",
"""Plan for:
- Monitoring after deployment
- Retraining as data changes""",
"M R → Monitor, Retrain.\nKindergarten: 'Monkeys Run.'",
"""**Meaning**
Maintenance ensures long-term model reliability. Once deployed, a model’s performance can degrade as the world changes—a phenomenon called **model drift**.
Regular monitoring, retraining, and documentation keep the system effective, transparent, and fair.

1. **Monitor Performance** – Track accuracy and drift in production.
2. **Retrain Periodically** – Update with new data and label corrections.
3. **Document Changes** – Maintain reproducibility and trust.

**Code Example**
```python
joblib.dump(model, "model_v1.pkl")

# Later retraining
model = joblib.load("model_v1.pkl")
model.fit(new_X, new_y)
joblib.dump(model, "model_v2.pkl")
```"""),

 ("""1️⃣ D A M E M → Dogs Always Make Everyone Merry 🐶🎉\n
2️⃣ Define the Project → Pandas Make Bread 🐼🍞\n
3️⃣ Acquire & Explore Data → Cats Cook Easy Food 🐱🍲\n
4️⃣ Modeling → Tigers Sing Tunes 🐯🎵\n
5️⃣ Evaluation → Bears Moo 🐻🐮\n
6️⃣ Maintenance → Rabbits Make Umbrellas 🐰☂️""",
"""1. Define the Project: Identify the problem, metric, and baseline for success. (P M B → Problem, Metric, Baseline.)\n
2. Acquire & Explore Data: Collect, clean, explore, and engineer features. (C C E F → Collect, Clean, Explore, Features.)\n
3. Modeling: Train models, select variables, and tune hyperparameters. (T S T → Train, Select, Tune.)\n
4. Evaluation: Check if the model beats the baseline and is meaningful. (B M → Baseline, Meaningful.)\n
5. Maintenance: Retrain with new data, monitor drift, and update documentation. (R M U → Retrain, Monitor, Update.)\n
6. D A M E M (Concept): Machine learning follows a structured lifecycle with five iterative phases — Define, Acquire, Model, Evaluate, Maintain.""",
"ML Lifecycle — Checkpoint Card"),

# ---------- WHY IT MATTERS ----------
("ML Lifecycle — Why It Matters",
"""If one stage is poor, the project fails.""",
"M D M → Metric, Data, Maintenance.\nKindergarten: 'Mice Dig Mounds.'",
"""**Meaning**
Every stage in the lifecycle directly affects final model performance. A single weak phase can cause failure:
• Unclear **metrics** → wrong objectives
• Low-quality **data** → biased models
• Missing **maintenance** → drift and degradation

Following the lifecycle enforces a scientific, reproducible approach to building trustworthy models.

**Code Example**
```python
for step in ["Metric clarity","Data quality","Maintenance plan"]:
    print("✔️", step)
```"""),

("Scikit-learn — Design Philosophy",
"""Consistent interfaces → systematic experimentation.
Roles:
- Estimator: .fit()
- Transformer: .transform()
- Predictor: .predict()
- Inspection: params vs learned attrs""",
"E T P I → Estimator, Transformer, Predictor, Inspection.\nKindergarten: 'Elephants Talk Pretty Instantly.'",
"""**Meaning**
Perfect — let’s unpack this **step by step** so you fully understand **what is being done**, **how it works**, and **why Scikit-learn was designed this way**.

Scikit-learn’s **unified API** provides a consistent structure across all models and preprocessing tools.
Every algorithm — whether it’s Logistic Regression, Random Forest, or K-Nearest Neighbors — uses the same methods: `.fit()`, `.transform()`, `.predict()`, and `.score()`.

This makes it easy to **swap models**, **chain data transformations**, and **compare results** systematically — which is why Scikit-learn dominates both research and production pipelines.

---

### 💡 Core Design Roles

| Role            | Purpose                                                     | Key Methods                             | Example                       |
| --------------- | ----------------------------------------------------------- | --------------------------------------- | ----------------------------- |
| **Estimator**   | Learns from data                                            | `.fit(X, y)`                            | `model.fit(X_train, y_train)` |
| **Transformer** | Changes or prepares data                                    | `.transform(X)` or `.fit_transform(X)`  | `scaler.transform(X_train)`   |
| **Predictor**   | Makes predictions                                           | `.predict(X)` or `.score(X, y)`         | `model.predict(X_test)`       |
| **Inspection**  | Access parameters (inputs) and learned attributes (outputs) | attributes like `.coef_`, `.intercept_` | `model.coef_`                 |

---

**Step-by-Step Explanation**

1. `scaler = StandardScaler().fit(X_train)`
   - Creates a **Transformer** that computes mean and standard deviation per feature.
   - These values are stored in `scaler.mean_` and `scaler.scale_`.

2. `X_scaled = scaler.transform(X_train)`
   - Applies standardization (subtract mean, divide by std).
   - Produces normalized `X_scaled`, making features comparable in scale.
   - Shortcut: `StandardScaler().fit_transform(X_train)` combines both steps.

3. `model = LogisticRegression()`
   - Creates an **Estimator + Predictor**.
   - Initially untrained, it learns once `.fit()` is called.

4. `model.fit(X_scaled, y_train)`
   - Trains the model using scaled data.
   - Finds optimal weights that minimize classification error via **log-loss optimization**.
   - Stores learned parameters:
     - `model.coef_` (feature weights)
     - `model.intercept_` (bias term)

5. `print("Accuracy:", model.score(X_scaled, y_train))`
   - Evaluates model performance (default metric = **accuracy**).
   - `.score()` compares predictions with true labels.

6. `print("Learned Coefficients:", model.coef_)`
   - Displays learned feature weights.
   - Positive coefficients increase the likelihood of class 1; negative reduce it.

---

**Why It Matters**

Because **all Scikit-learn components share the same API**, you can:
- Replace `LogisticRegression()` with `RandomForestClassifier()` instantly.
- Combine transformations + estimators using Pipelines (`Pipeline([('scaler', StandardScaler()), ('model', LogisticRegression())])`).
- Automate and reproduce experiments easily.

This unified design supports **reproducibility**, **scalability**, and **modular ML workflows**.

---

**Summary**
- `.fit()` → learns
- `.transform()` → preprocesses
- `.predict()` / `.score()` → makes or evaluates predictions
- Attributes ending with `_` → store learned results

This consistent structure makes machine learning workflows **clean, testable, and easily extensible**.

---

**Code Example**
```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

scaler = StandardScaler().fit(X_train)
X_scaled = scaler.transform(X_train)

model = LogisticRegression()
model.fit(X_scaled, y_train)

print("Accuracy:", model.score(X_scaled, y_train))
print("Learned Coefficients:", model.coef_)
```"""),

("Scikit-learn — Advantage",
"""Defaults allow quick baselines, then refinement.""",
"D B T → Defaults, Baseline, Tuning.\nKindergarten: 'Dogs Bark Together.'",
"""**Meaning**
Scikit-learn simplifies rapid prototyping and iterative improvement.
Start with sensible **defaults** to build a **baseline**, then refine models systematically using **grid search**, **cross-validation**, and **pipelines**.
This workflow supports the full lifecycle from idea to production with transparent, reproducible results.

**Code Example**
```python
grid = GridSearchCV(LogisticRegression(), {"C": [0.1, 1, 10]}, cv=5)
grid.fit(X_train, y_train)
print("Best parameters:", grid.best_params_)
```"""),

# ---------- CATEGORIES ----------
("ML Categories — Overview",
"""Four learning types:
1. Supervised
2. Unsupervised
3. Semi-supervised
4. Reinforcement""",
"S U S R → Supervised, Unsupervised, Semi, Reinforcement.\nKindergarten: 'Silly Unicorns Swim Rapidly.'",
"""**Meaning**
All machine-learning tasks fall into one of four learning categories, depending on whether the data includes labels and how feedback is provided.

1. **Supervised Learning** – Models learn from labeled examples (x → y).
2. **Unsupervised Learning** – Models find structure or clusters without labels.
3. **Semi-Supervised Learning** – A small labeled subset guides learning from a large unlabeled pool.
4. **Reinforcement Learning** – An agent interacts with an environment and learns from rewards over time.

Identifying the correct category determines how to prepare data, choose algorithms, and design evaluation metrics.

**Code Example**
```python
model = DecisionTreeClassifier().fit(X_train, y_train)
print("Prediction:", model.predict(X_test[:1]))
```"""),

("Supervised Learning — Concept",
"""Data has inputs (x) and known outputs (y).""",
"I O G → Input, Output, Generalize.\nKindergarten: 'Igloos Open Gates.'",
"""**Meaning**
Supervised learning teaches a model to map inputs to known outputs using labeled data.
The goal is to approximate a function f that generalizes from training examples to unseen cases.

1. **Classification** – Predict categorical labels (spam / not spam, disease / no disease).
2. **Regression** – Predict continuous values (prices, temperatures).

The model learns relationships between x and y so it can make accurate predictions on future data.

**Code Example**
```python
rf = RandomForestClassifier().fit(X_train, y_train)
print("Accuracy:", rf.score(X_test, y_test))
```"""),

("Supervised Learning — Algorithms",
"""Examples: kNN, Naive Bayes, Trees, Regression, SVM, Neural.""",
"N D L S N → Nearest, Decision, Linear, SVM, Neural.\nKindergarten: 'Nice Dogs Like Soft Noses.'",
"""**Meaning**
Supervised learning spans a wide family of algorithms, each with strengths for different data shapes and sizes.

• **k-Nearest Neighbors** – Classifies by proximity in feature space.
• **Naïve Bayes** – Uses conditional probability based on Bayes’ theorem.
• **Decision Trees / Random Forests** – Split data using learned rules.
• **Linear & Logistic Regression** – Fit simple mathematical relationships.
• **Support Vector Machines (SVMs)** – Find optimal separating boundaries.
• **Neural Networks** – Model complex nonlinear mappings through layers.

**Code Example**
```python
models = [KNeighborsClassifier(), DecisionTreeClassifier(), SVC()]
for m in models:
    m.fit(X_train, y_train)
    print(m.__class__.__name__, "→", m.score(X_test, y_test))
```"""),

("Unsupervised Learning — Concept",
"""No labels; find structure/patterns.""",
"C A D A → Clustering, Association, Dim reduction, Anomaly.\nKindergarten: 'Cats Are Dancing Around.'",
"""**Meaning**
Unsupervised learning discovers hidden structure in unlabeled data.
The model groups similar samples, compresses information, or identifies anomalies without explicit targets.

1. **Clustering** – Group items by similarity (e.g., k-means).
2. **Association Rules** – Find co-occurring patterns (bread → butter).
3. **Dimensionality Reduction** – Compress features (PCA, t-SNE).
4. **Anomaly Detection** – Spot rare or unusual events.

These methods reveal patterns humans might overlook, guiding exploratory analysis or downstream modeling.

**Code Example**
```python
kmeans = KMeans(n_clusters=3).fit(X)
print("Cluster labels:", kmeans.labels_[:10])
```"""),

("Semi-Supervised Learning — Concept",
"""Few labeled, many unlabeled examples.""",
"L U P → Labeled, Unlabeled, Precision.\nKindergarten: 'Lions Use Paint.'",
"""**Meaning**
Semi-supervised learning bridges the gap between supervised and unsupervised methods.
It uses a small set of labeled examples to guide learning from a much larger pool of unlabeled data, improving performance when labeling is costly.

Typical pipeline:
1. Train an initial model on labeled data.
2. Use it to label or weight the unlabeled set.
3. Retrain using both to refine accuracy.

**Code Example**
```python
label_prop = LabelPropagation().fit(X, y_partial)
print("Predicted labels:", label_prop.transduction_[:10])
```"""),

("Reinforcement Learning — Concept",
"""Agent acts → gets reward → learns policy.""",
"S A R → State, Action, Reward.\nKindergarten: 'Snakes Always Run.'",
"""**Meaning**
Reinforcement learning trains an **agent** to make sequential decisions through interaction with an **environment**.
At each step, the agent observes the **state**, takes an **action**, and receives a **reward** signal.
Over time, it learns a **policy** that maximizes long-term cumulative reward.

Applications include robotics, game playing, and recommendation systems.
The balance between exploration (trying new actions) and exploitation (using known rewards) defines learning efficiency.

**Code Example**
```python
for episode in range(3):
    state = env.reset()
    done = False
    while not done:
        action = policy(state)
        state, reward, done, _ = env.step(action)
```"""),

# ---------- LEARNING STRATEGIES ----------
("Learning — Instance-Based",
"""Memorize examples; compare new ones (kNN).""",
"M S D → Memorize, Simple, Distance.\nKindergarten: 'Mice See Dogs.'",
"""**Meaning**
Instance-based learning memorizes examples from the training data and predicts outcomes for new inputs by comparing them to stored cases.
Instead of forming an explicit model, it relies on **similarity metrics** (like Euclidean distance) to decide which training samples are most relevant.

This approach is simple and intuitive but can become slow as datasets grow, and its accuracy depends heavily on feature scaling and noise control.

**Code Example**
```python
knn = KNeighborsClassifier(n_neighbors=3).fit(X_train, y_train)
print("Prediction for first sample:", knn.predict(X_test[:1]))
```"""),

("Learning — Model-Based",
"""Fit model that summarizes patterns.""",
"F S M → Fit, Summarize, Model.\nKindergarten: 'Frogs Swim Madly.'",
"""**Meaning**
Model-based learning builds an abstract mathematical representation of data relationships.
Rather than storing all examples, the model learns parameters that summarize how inputs relate to outputs.
This allows for **faster predictions**, **less storage**, and **better generalization** when designed properly.

Examples include linear regression, logistic regression, and neural networks.
The trade-off is that incorrect assumptions about model form can limit accuracy.

**Code Example**
```python
reg = LinearRegression().fit(X_train, y_train)
print("Coefficients:", reg.coef_)
```"""),

("ML Challenges",
"""Issues: Quantity, Quality, Sampling, Features, Overfitting.""",
"Q Q S F O → Quantity, Quality, Sampling, Features, Overfit.\nKindergarten: 'Quick Queens Sing Funny Opera.'",
"""**Meaning**
Machine learning success depends on the quality and representativeness of data, as well as sound model design.
Common pitfalls can distort predictions, waste computation, and lead to biased results.

1. **Data Quantity** – Too few samples cause unreliable learning.
2. **Data Quality** – Incomplete or biased inputs degrade accuracy.
3. **Sampling Bias** – Non-representative data misleads the model.
4. **Feature Engineering** – Poor features hide important relationships.
5. **Overfitting vs. Underfitting** – Balancing memorization and generalization is critical.

Recognizing and addressing these challenges early prevents wasted effort and ensures trustworthy outcomes.

**Code Example**
```python
train_acc = model.score(X_train, y_train)
test_acc = model.score(X_test, y_test)
print("Train vs Test:", train_acc, test_acc)
```"""),

# ---------- WRAP-UP ----------
("Week 1 — Wrap-up",
"""We:
- Defined ML (P,T,E framework)
- Compared ML vs programming
- Saw iconic examples (digits, faces, Netflix, AlphaGo)
- Introduced lifecycle + Scikit-learn
- Classified learning types
- Discussed challenges (data, features, fitting)
Takeaway: ML = process, data, critical thinking — not just algorithms.""",
"P E L C C → Process, Examples, Lifecycle, Categories, Challenges.\nKindergarten: 'Puppies Eat Lemon Cake Cold.'"),
]


data_cleaning = [

# ---------- 1) DATA INSPECTION ----------
("Data Inspection — df.head()",
"""Shows the first N rows (default 5). Quick peek at shape and columns.
Example:
    df.head(10)
**Use case:** **Open a fresh CSV and quickly verify the columns and sample values look sane before doing anything else.**""",
"Head shows the head. 3rd grade: 'Heads first!'"),

("Data Inspection — df.info()",
"""Prints column names, non-null counts, and dtypes. Great for sanity checks.
Example:
    df.info()
**Use case:** **Check which columns are strings vs numbers and where nulls appear so you can plan cleaning.**""",
"Info gives info. 'Ask info first.'"),

("Data Inspection — df.describe()",
"""Summary stats for numeric columns. Use include='all' to see more types.
Example:
    df.describe()
**Use case:** **Spot outliers quickly (min/max) and scale ranges before modeling.**""",
"Describe describes. 'Stats snapshot.'"),

# ---------- 2) MISSING DATA HANDLING ----------
("Missing Data — df.isnull().sum()",
"""Counts missing values per column so you know what to fix.
Example:
    df.isnull().sum()
**Use case:** **Quantify missingness to choose between drop vs fill (e.g., 2% vs 40% missing).**""",
"Is null? Sum it. 'Count the holes.'"),

("Missing Data — df.dropna()",
"""Drops rows (or columns) with any NA by default. Use subset/how/axis to control.
Example:
    df.dropna(subset=['age'])
**Use case:** **Remove rows with missing target y before training a supervised model.**""",
"Drop NA drops empties. 'Bye-bye blanks.'"),

("Missing Data — df.fillna(value)",
"""Fills missing entries with a value or strategy like mean or median.
Example:
    df['age'] = df['age'].fillna(df['age'].median())
**Use case:** **Fill numeric gaps with median and categorical gaps with mode to keep rows for modeling.**""",
"Fill NA fills holes. 'Patch the gaps.'"),

# ---------- 3) CLEANING AND TRANSFORMATION ----------
("Cleaning — df.drop_duplicates()",
"""Removes duplicate rows. Use subset and keep to control behavior.
Example:
    df = df.drop_duplicates(subset=['id'])
**Use case:** **Deduplicate customer table on 'id' so each customer appears once.**""",
"Drop dupes, keep uniques. 'One of each.'"),

("Cleaning — df.rename(columns={'old':'new'})",
"""Renames columns with a dict. Set inplace=False to return a copy.
Example:
    df = df.rename(columns={'cust_id':'customer_id'})
**Use case:** **Standardize mixed headers like 'Cust ID' → 'customer_id' for clean code and joins.**""",
"Rename to new name. 'New labels.'"),

("Cleaning — df.astype({'col':'type'})",
"""Converts column dtypes, like to int, float, string, category, datetime.
Example:
    df = df.astype({'age':'int64'})
**Use case:** **Fix numbers stored as text ('12,345') or parse date strings to datetime for time ops.**""",
"Astype changes type. 'Make it the right kind.'"),

("Cleaning — df.replace({'col': {'old':'new'}})",
"""Replaces matching values. Works with lists, dicts, regex if needed.
Example:
    df = df.replace({'city': {'NYC':'New York'}})
**Use case:** **Unify messy labels ('NYC','New York City') into a single category before grouping.**""",
"Replace swaps values. 'Old to new.'"),

("Cleaning — df.reset_index(drop=True)",
"""Resets the index to 0..n-1. Use drop=True to discard old index.
Example:
    df = df.reset_index(drop=True)
**Use case:** **After filtering or concatenating, reset row numbering for clean exports and plots.**""",
"Reset to fresh numbers. 'Start counting again.'"),

("Cleaning — df.drop(columns=['col'])",
"""Removes columns. You can also use axis=1.
Example:
    df = df.drop(columns=['temp_col'])
**Use case:** **Remove leak-prone fields (like 'price_after_discount') before training.**""",
"Drop column to remove. 'Take it out.'"),

# ---------- 4) SELECTION AND FILTERING ----------
("Selection — df.loc[row_labels, col_labels]",
"""Label-based selection and assignment with conditions.
Example:
    df.loc[df['score']>90, 'grade'] = 'A'
**Use case:** **Create or edit a column only for rows meeting a business rule (e.g., high-score = 'A').**""",
"Loc uses labels. 'Names, not numbers.'"),

("Selection — df.iloc[row_idx, col_idx]",
"""Integer-position selection, like NumPy slicing.
Example:
    df.iloc[0:5, 0:3]
**Use case:** **Grab the first 5 rows and first 3 columns for a quick debug print.**""",
"ILOC uses integers. 'Numbers, not names.'"),

("Filtering — df[df['col'] > value]",
"""Boolean filter that keeps rows matching a condition.
Example:
    high = df[df['price'] > 100]
**Use case:** **Build a 'premium_products' subset for a focused report.**""",
"True stays, False goes. 'Keep the matches.'"),

# ---------- 5) AGGREGATION AND ANALYSIS ----------
("Aggregation — df.groupby('col').agg('mean')",
"""Groups rows by key and computes aggregates.
Example:
    df.groupby('team')['points'].agg('mean')
**Use case:** **Compute average points per team for league standings.**""",
"Group then compute. 'Buckets then math.'"),

("Sorting — df.sort_values('col', ascending=False)",
"""Orders rows by column values. Can sort by multiple columns.
Example:
    df.sort_values(['team','points'], ascending=[True, False])
**Use case:** **Rank top 10 customers by total spend.**""",
"Sort to order. 'Big to small or small to big.'"),

("Counts — df['col'].value_counts()",
"""Counts each unique value in a Series. Good for category checks.
Example:
    df['color'].value_counts()
**Use case:** **See the distribution of product colors to find rare categories.**""",
"Value counts count values. 'How many of each.'"),

("Apply — df.apply(func, axis=1 or 0)",
"""Runs a function across rows or columns. Use sparingly for speed.
Example:
    df['bmi'] = df.apply(lambda r: r['wt']/((r['ht']/100)**2), axis=1)
**Use case:** **Compute a custom metric (e.g., BMI or profit_margin) that needs multiple columns.**""",
"Apply applies a function. 'Do a custom rule.'"),

("Pivot Table — df.pivot_table(values, index, columns)",
"""Summarizes data like Excel Pivot. Choose aggfunc.
Example:
    df.pivot_table(values='sales', index='region', columns='quarter', aggfunc='sum')
**Use case:** **Create a region-by-quarter sales matrix for management.**""",
"Pivot to summarize. 'Flip and add up.'"),

# ---------- 6) COMBINING AND MERGING ----------
("Combine — pd.concat([df1, df2])",
"""Stacks DataFrames by rows (axis=0) or columns (axis=1).
Example:
    big = pd.concat([df1, df2], axis=0, ignore_index=True)
**Use case:** **Stack monthly CSVs into one year-to-date table.**""",
"Concat connects frames. 'Stick them together.'"),

("Merge — pd.merge(df1, df2, on='key')",
"""SQL-style join on columns. Control how: inner, left, right, outer.
Example:
    out = pd.merge(a, b, on='id', how='left')
**Use case:** **Join orders with customers on 'id' to add demographics.**""",
"Merge mixes by key. 'Match and mix.'"),

("Join — df1.join(df2)",
"""Join on index by default. Good after setting index.
Example:
    out = df1.join(df2, how='inner')
**Use case:** **Attach feature table aligned by index after setting both to the same index.**""",
"Join joins by index. 'Line up by labels.'"),

("Append — df1.append(df2) (deprecated)",
"""Old way to add rows. Use pd.concat instead in modern pandas.
Example:
    pd.concat([df1, df2], ignore_index=True)
**Use case:** **Replace legacy append() in old notebooks with concat for speed and future safety.**""",
"Do not use append. 'Concat instead.'"),

# ---------- CATEGORY OVERVIEWS ----------
("Overview — Data Inspection",
"""Head, Info, Describe answer: What do I have? Are types and stats sensible?
Use these first to avoid surprises later.
**Use case:** **Kickoff of any project: run head/info/describe to build a quick data profile.**""",
"See it before fix it. 'Look, then work.'"),

("Overview — Missing Data",
"""Find, drop, or fill missing values. Choose a strategy that fits the task.
Document what you remove or impute.
**Use case:** **Create a small 'missingness report' before modeling so stakeholders agree on the plan.**""",
"Find holes, then patch. 'Spot and fix.'"),

("Overview — Cleaning & Transform",
"""Rename, type-cast, replace, remove dupes, drop junk columns, reset index.
Keep transformations in a script or pipeline.
**Use case:** **Build a reproducible preprocessing pipeline that you can reuse in training and inference.**""",
"Clean first, model later. 'Make it tidy.'"),

("Overview — Select & Filter",
"""Select by labels with loc, by positions with iloc, filter with booleans.
Chain these for clear, readable queries.
**Use case:** **Create a 'study cohort' by filtering to the target population for analysis.**""",
"Pick and keep. 'Choose what matters.'"),

("Overview — Aggregate & Analyze",
"""Group, sort, count, apply, pivot to summarize patterns for decisions.
Start simple, then refine.
**Use case:** **Produce an executive summary table showing totals and averages by region.**""",
"Group then learn. 'Buckets tell stories.'"),

("Overview — Combine & Merge",
"""Concat to stack, merge to join by key, join to align by index.
Check row counts before and after to catch mistakes.
**Use case:** **Assemble a single 'master table' by merging fact tables with lookup tables.**""",
"Combine with care. 'Match, then merge.'"),
]

Ml_concepts = [

# ---------- LINEAR MODELS ----------

("Linear Regression — Concept & Equation",
"""Linear Regression models a **continuous relationship** between inputs and output.

**Equation:**
    ŷ = θ₀ + θ₁x₁ + θ₂x₂ + ... + θ_d x_d

**Plain English:**
Predicted value = intercept + (each feature × its weight).

**Applications:**
- Predicting housing prices
- Forecasting sales
- Predicting customer lifetime value

**Advantages:**
- Easy to interpret
- Fast and explainable

**Disadvantages:**
- Assumes linear relationships
- Sensitive to outliers
- Poor with high-dimensional or correlated data""",
"Line fits points → Straight line predictions"),

("Logistic Regression — Concept & Equation",
"""Used for **binary classification** (0 or 1 outcome).

**Equation:**
    p(y=1|x) = 1 / (1 + e^(−θᵗx))

**Plain English:**
Predicts probability between 0–1 using a sigmoid curve.

**Applications:**
- Spam vs. not spam
- Disease prediction
- Customer churn detection

**Advantages:**
- Interpretable & fast
- Works well for linearly separable data

**Disadvantages:**
- Assumes linear boundary
- Limited for complex/nonlinear patterns""",
"Logistic → Log-odds → 0 or 1 output"),

("Ridge Regression — Concept",
"""A linear model with **L2 regularization**, which adds a penalty for large weights.

**Equation:**
    J(θ) = Σ (yᵢ - ŷᵢ)² + λ Σ θⱼ²

**Plain English:**
Keeps coefficients small to reduce overfitting.

**Applications:**
- Predictive maintenance
- Regularized forecasting
- Medical regression tasks

**Advantages:**
- Less prone to overfitting
- Handles multicollinearity

**Disadvantages:**
- Keeps all predictors (no feature selection)
- λ requires tuning""",
"Ridge = Regularization reduces risk"),

("Lasso Regression — Concept",
"""Linear model with **L1 regularization**, which forces some coefficients to zero.

**Equation:**
    J(θ) = Σ (yᵢ - ŷᵢ)² + λ Σ |θⱼ|

**Plain English:**
Removes unimportant features automatically.

**Applications:**
- Feature selection in predictive models
- Sparse linear regression

**Advantages:**
- Performs feature selection
- Reduces overfitting

**Disadvantages:**
- Can underperform with correlated variables""",
"Lasso = Least Absolute Shrinkage → Some weights = 0"),

# ---------- TREE-BASED MODELS ----------

("Decision Tree — Concept",
"""Splits data recursively using decision rules (e.g., “if income < 50k”).

**Plain English:**
A tree of yes/no questions to make predictions.

**Applications:**
- Customer churn prediction
- Credit scoring
- Loan approval

**Advantages:**
- Explainable & visual
- Handles mixed data types

**Disadvantages:**
- Overfits easily
- Sensitive to noisy data/outliers""",
"Tree = Rules that split data"),

("Random Forest — Concept",
"""An **ensemble** of decision trees using bootstrap samples and averaging predictions.

**Equation (conceptually):**
    ŷ = (1/N) Σ fᵢ(x)

**Plain English:**
Many trees vote → reduces noise and overfitting.

**Applications:**
- Credit risk modeling
- Disease prediction
- Sales forecasting

**Advantages:**
- Robust, reduces overfitting
- Handles missing data well

**Disadvantages:**
- Slower training
- Less interpretable than a single tree""",
"Forest = Many trees vote together"),

("Gradient Boosting — Concept",
"""Builds trees sequentially, each correcting previous errors.

**Equation (conceptual):**
    Fₘ(x) = Fₘ₋₁(x) + α hₘ(x)

**Plain English:**
Each new model learns the mistakes of the last one.

**Applications:**
- Predicting insurance claims
- Emissions or energy modeling

**Advantages:**
- High accuracy
- Handles complex nonlinear relationships

**Disadvantages:**
- Sensitive to outliers
- Requires careful tuning""",
"Boosting = Learn from mistakes"),

("XGBoost — Concept",
"""Extreme Gradient Boosting = efficient, parallelized gradient boosting framework.

**Plain English:**
A super-fast, regularized boosting algorithm.

**Applications:**
- Fraud detection
- Loan default prediction
- Kaggle competitions

**Advantages:**
- State-of-the-art performance
- Handles missing data

**Disadvantages:**
- Complex tuning
- Poor on sparse categorical data""",
"XGBoost = Extreme speed + performance"),

("LightGBM — Concept",
"""Light Gradient Boosting Machine uses **leaf-wise tree growth**.

**Plain English:**
Builds trees from the most promising leaves for speed and accuracy.

**Applications:**
- Credit scoring
- Flight delay prediction
- Health analytics

**Advantages:**
- Extremely fast
- Handles large datasets well

**Disadvantages:**
- Can overfit small data
- Requires large memory""",
"LightGBM = Lightning-fast boosting"),

# ---------- UNSUPERVISED LEARNING: CLUSTERING ----------

("K-Means Clustering — Concept",
"""Partitions data into K clusters minimizing intra-cluster variance.

**Equation:**
    minimize Σ Σ ||xᵢ - μₖ||²

**Plain English:**
Group points so each cluster center (μₖ) is close to its members.

**Applications:**
- Customer segmentation
- Image compression
- Recommendation grouping

**Advantages:**
- Simple, scalable
- Works well on spherical clusters

**Disadvantages:**
- Must predefine K
- Sensitive to initialization and scale""",
"K-Means = Keep means close"),

("Hierarchical Clustering — Concept",
"""Builds a tree (dendrogram) of clusters.

**Plain English:**
Groups data step-by-step based on similarity distance.

**Applications:**
- Gene expression analysis
- Document similarity
- Fraud pattern grouping

**Advantages:**
- Easy to visualize
- No need for fixed K

**Disadvantages:**
- Computationally heavy
- Struggles with large datasets""",
"Hierarchy = Nested clusters"),

("Gaussian Mixture Model (GMM) — Concept",
"""Models data as a mixture of Gaussian distributions.

**Equation:**
    p(x) = Σ πₖ N(x | μₖ, Σₖ)

**Plain English:**
Each cluster is shaped like an ellipse; GMM gives probability of belonging to each.

**Applications:**
- Customer segmentation
- Anomaly detection
- Image segmentation

**Advantages:**
- Soft clustering (probabilistic)
- Captures complex cluster shapes

**Disadvantages:**
- Requires tuning of K
- Sensitive to initialization""",
"GMM = Gaussian-shaped groups"),

# ---------- ASSOCIATION RULES ----------

("Apriori Algorithm — Concept",
"""Finds **frequent itemsets** and generates association rules.

**Plain English:**
Discovers “if X → then Y” relationships in data.

**Applications:**
- Market basket analysis (“if buys bread → buys butter”)
- Recommendation engines
- Promotion optimization

**Advantages:**
- Interpretable & intuitive
- Solid baseline for retail analytics

**Disadvantages:**
- Computationally expensive
- Generates many redundant rules""",
"Apriori = Association patterns appear repeatedly"),

]

linear_algebra = [

# ---------- SCALARS: CONCEPT ----------
("What does a scalar represent?",
"A scalar represents magnitude only—it tells “how much,” not “which way.”",
"S = Size only (no direction).",
"""**Meaning**
A scalar is the simplest mathematical quantity—it expresses **magnitude only**, without direction.
Scalars tell *how much* of something exists, like 5 kilograms or 10 seconds, but not *where* or *which way*.
They form the foundation for all higher-dimensional quantities like vectors and matrices.

**Code Example**
```python
import numpy as np
s = np.array(7.5)
print("Scalar value:", s)
print("Rank:", np.ndim(s))  # 0D means scalar
```"""
),

# ---------- SCALARS: EXAMPLES ----------
("Give three everyday examples of scalars.",
"Temperature (25 °C), mass (70 kg), time (5 s).",
"TMT = Temperature, Mass, Time.",
"""**Meaning**
Common physical measurements such as **temperature**, **mass**, and **time** are scalars because they have a numerical value but no direction.
No matter which way you face, 25 °C remains 25 °C—it doesn’t “point” anywhere.

**Code Example**
```python
temperature = 25  # °C
mass = 70         # kg
time = 5          # s
print(f"Temperature={temperature}°C, Mass={mass}kg, Time={time}s")
```"""
),

# ---------- SCALARS: MATHEMATICAL SET ----------
("What mathematical set do scalars belong to?",
"The real numbers (ℝ).",
"R = Real numbers represent scalar values.",
"""**Meaning**
Scalars belong to the **set of real numbers (ℝ)**.
Each scalar corresponds to one real value—positive, negative, or zero—and follows standard arithmetic operations.

**Code Example**
```python
import numpy as np
s = np.array(-3.14)
print("Scalar:", s, "is real:", isinstance(float(s), float))
```"""
),

# ---------- SCALARS: ATOMS OF MATH ----------
("Why are scalars called the atoms of mathematics?",
"Because they are the simplest one-dimensional quantities from which higher-dimensional objects build.",
"Atom = base of everything.",
"""**Meaning**
Scalars are called the “atoms” of mathematics because complex structures like vectors, matrices, and tensors
are built by stacking or combining scalar elements. Every data structure ultimately reduces to scalars.

**Code Example**
```python
import numpy as np
vector = np.array([2, 4, 6])
print("Each element is scalar:", all(isinstance(x, (int, float)) for x in vector))
```"""
),

# ---------- SCALARS: SPEED VS VELOCITY ----------
("What is the physical difference between speed and velocity?",
"Speed is a scalar (magnitude only); velocity is a vector (magnitude + direction).",
"Speed = scalar; Velocity = vector.",
"""**Meaning**
Speed describes *how fast* something moves, while velocity describes *how fast and in what direction*.
Because direction matters for velocity but not speed, speed is classified as a scalar quantity.

**Code Example**
```python
speed = 60  # km/h
velocity = [60, 0]  # km/h east
print("Speed:", speed, "| Velocity:", velocity)
```"""
),

# ---------- SCALARS: PYTHON DEFINITION ----------
("How do you define and inspect a scalar in Python?",
"Prints 7.5 and rank 0 → it’s a 0D NumPy array.",
"ndim = 0 ⇒ scalar.",
"""**Meaning**
In NumPy, a scalar is represented as an array with **zero dimensions (0D)**.
It behaves like a plain number but retains NumPy’s data type and broadcasting features.

**Code Example**
```python
import numpy as np
s = np.array(7.5)
print("Scalar:", s)
print("Number of dimensions:", np.ndim(s))
```"""
),

# ---------- SCALARS: ARITHMETIC ----------
("Perform scalar arithmetic.",
"Output = 686.7 → scalar × scalar = scalar (no direction).",
"Scalar × Scalar = Scalar.",
"""**Meaning**
Scalars obey standard arithmetic rules. Multiplying or dividing scalars produces another scalar,
since direction is not introduced through numeric operations.

**Code Example**
```python
mass = 70   # kg
g = 9.81    # m/s²
weight = mass * g
print("Weight (N):", weight)
```"""
),

 # ---------- VECTOR CONCEPT ----------
("How does a vector extend the idea of a scalar?",
"It adds direction to magnitude—answering both “how much” and “which way.”",
"Vector = Value + Direction.",
"""**Meaning**
A **vector** extends a scalar by including **direction** as well as **magnitude**.
While a scalar tells *how much*, a vector tells both *how much* and *which way*.
In mathematics, a vector is an **ordered tuple of numbers** that represents a point or direction in space.
Geometrically, it’s drawn as an arrow with **length** (magnitude) and **orientation** (direction).
Vectors are the building blocks of physics and computer graphics because they encode motion, force, and position changes.

**Code Example**
import numpy as np
v = np.array([3, 4])
print("Vector:", v)
print("Magnitude:", np.linalg.norm(v))
Output → Magnitude = 5 (√(3² + 4²))
"""),

# ---------- VECTOR NOTATION ----------
("Write the 3-D form of a vector.",
"(\\vec{v} = [v_1, v_2, v_3]^T)",
"3D = (v₁, v₂, v₃)ᵀ.",
"""**Meaning**
A 3D vector can be written as a **column** or **row** form, representing its components along the x-, y-, and z-axes.
Mathematically:
v = [v1, v2, v3]^T
Each component represents projection on one axis.
In higher dimensions, this idea generalizes to **n-dimensional vectors** used in machine learning and linear algebra.

**Code Example**
import numpy as np
v = np.array([1, 2, 3])
print("Vector:", v)
print("Shape:", v.shape)
Output → (3,) → a 1D array with three components.
"""),

# ---------- VECTOR EXAMPLES ----------
("Give three examples of vector quantities.",
"Velocity, force, displacement.",
"VFD = Velocity, Force, Displacement.",
"""**Meaning**
In physics, many quantities require both magnitude and direction:
- **Velocity** – describes speed and direction of motion.
- **Force** – describes push/pull intensity and direction.
- **Displacement** – describes change in position.
These all depend on direction, which distinguishes them from scalars like mass or time.

**Code Example**
import numpy as np
velocity = np.array([5, 0])      # 5 m/s east
force = np.array([0, -9.8])      # downward
displacement = np.array([3, 4])  # 3 east, 4 north
print("Velocity:", velocity, "Force:", force, "Displacement:", displacement)
"""),

# ---------- DOT PRODUCT ----------
("What operation measures alignment between two vectors?",
"The dot product.",
"Dot = Degree of Togetherness.",
"""**Meaning**
The **dot product** measures how aligned two vectors are.
Mathematically, a·b = |a||b|cos(θ).
If vectors point in the same direction, the cosine term is near 1 → large positive value.
If perpendicular, the result is 0 → no alignment.
If opposite, negative → opposite orientation.
It’s the basis for measuring similarity in physics and machine learning (e.g., cosine similarity).

**Code Example**
import numpy as np
a = np.array([3, 4])
b = np.array([4, 3])
dot = np.dot(a, b)
theta = np.arccos(dot / (np.linalg.norm(a)*np.linalg.norm(b)))
print("Dot:", dot, "Angle (radians):", theta)
"""),

# ---------- CROSS PRODUCT ----------
("What does the cross product produce?",
"A vector perpendicular to both inputs → measures rotation or area.",
"Cross = Creates Right-angle Vector.",
"""**Meaning**
The **cross product** of two 3D vectors produces another **vector** that is perpendicular to both inputs.
Its magnitude equals the **area of the parallelogram** formed by the two vectors: |a × b| = |a||b|sin(θ).
It encodes **rotational direction** and **orientation** following the right-hand rule.
This operation is essential in physics (torque, angular momentum) and 3D graphics (surface normals).

**Code Example**
import numpy as np
a = np.array([1, 0, 0])
b = np.array([0, 1, 0])
cross = np.cross(a, b)
print("Cross product:", cross)
Output → [0 0 1] (perpendicular to both)
"""),

# ---------- PYTHON: MAGNITUDE ----------
("Create a 2D vector and find its magnitude.",
"Magnitude = 5 → √(3² + 4²).",
"3-4-5 Triangle Rule.",
"""**Meaning**
The **magnitude (length)** of a vector measures its size regardless of direction.
For vector v = [x, y], |v| = √(x² + y²).
It’s used to normalize vectors for unit direction or to compare distances in geometric space.

**Code Example**
import numpy as np
v = np.array([3, 4])
mag = np.linalg.norm(v)
print("Vector:", v, "Magnitude:", mag)
"""),

# ---------- PYTHON: DOT + ANGLE ----------
("Compute dot product and angle between two vectors.",
"Dot = 11; angle ≈ 0.17985 rad → shows alignment.",
"Dot = alignment; Angle = orientation.",
"""**Meaning**
The **dot product** quantifies how similar two directions are.
The angle between them is found using the arccosine of their cosine similarity:
θ = cos⁻¹((a·b)/|a||b|).
A small angle means high alignment; a 90° angle means orthogonality.

**Code Example**
import numpy as np
v = np.array([3, 4])
w = np.array([1, 2])
dot = np.dot(v, w)
theta = np.arccos(dot / (np.linalg.norm(v)*np.linalg.norm(w)))
print("Dot:", dot, "Angle (radians):", theta)
"""),

 # ---------- MATRICES: CONCEPT ----------
("What is a matrix?",
"A 2D grid of numbers representing a linear transformation or stacked vectors.",
"Matrix = Multi-vector map.",
"""**Meaning**
A **matrix** is a rectangular array of numbers arranged in rows and columns.
It represents a **linear transformation** that can scale, rotate, or shear vectors.
Each column of a matrix corresponds to the image of a basis vector under this transformation.
Matrices generalize vectors to two dimensions and form the foundation of linear algebra, computer graphics, and machine learning.

**Code Example**
import numpy as np
A = np.array([[1, 2],
              [3, 4]])
print("Matrix A:\n", A)
print("Shape:", A.shape)
Output → (2, 2) meaning 2 rows and 2 columns.
"""),

# ---------- MATRICES: ACTION ----------
("What happens when a matrix multiplies a vector?",
"It transforms the vector—scaling, rotating, or shearing it.",
"Matrix × Vector = Transformation.",
"""**Meaning**
When a **matrix multiplies a vector**, the result is another vector that has been **transformed** by the operation.
Each entry in the result represents a **weighted combination** of the original vector’s components.
This process describes geometric effects such as rotation, scaling, projection, or shearing.

**Code Example**
import numpy as np
A = np.array([[1, 2],
              [3, 4]])
x = np.array([2, 1])
result = A @ x
print("A·x =", result)
Output → [4, 10] showing how the vector is transformed.
"""),

# ---------- MATRICES: OPERATORS ----------
("Why are matrices called operators?",
"Because they act on vectors to change direction or magnitude.",
"Operator = Action on vector.",
"""**Meaning**
A **matrix** is often called an **operator** because it performs an **operation** on vectors.
When applied, it changes their direction and/or magnitude, producing a new vector.
This concept is fundamental to physics, computer graphics, and data science, where matrices represent coordinate transformations or systems of equations.

**Code Example**
import numpy as np
A = np.array([[0, -1],
              [1,  0]])
v = np.array([1, 0])
print("Rotated vector:", A @ v)
Output → [0, 1] meaning the vector was rotated 90°.
"""),

# ---------- MATRICES: COLUMNS ----------
("What do the columns of a matrix represent?",
"How the basis vectors are stretched or rotated in the new space.",
"Columns = New basis directions.",
"""**Meaning**
Each **column** of a matrix represents how one **basis vector** is transformed in the new coordinate system.
For example, the first column shows where the x-axis maps, and the second column shows where the y-axis maps.
This provides a geometric interpretation of the matrix as a transformation of space.

**Code Example**
import numpy as np
A = np.array([[2, 0],
              [0, 3]])
print("Column 1 (x-axis transform):", A[:, 0])
print("Column 2 (y-axis transform):", A[:, 1])
Output → Columns show scaling factors for each basis direction.
"""),

# ---------- PYTHON: MATRIX MULTIPLICATION ----------
("Define a 2×2 matrix and multiply it by a vector.",
"A·x = [4, 10] → shows transformation effect.",
"Matrix times Vector = New Vector.",
"""**Meaning**
Matrix multiplication applies a linear transformation to a vector.
The rows of the matrix combine the vector’s components to produce new values in transformed space.
This operation is the core of most computations in computer graphics, robotics, and deep learning.

**Code Example**
import numpy as np
A = np.array([[1, 2],
              [3, 4]])
x = np.array([2, 1])
print("A·x =", A @ x)
Output → [4, 10]
"""),

# ---------- PYTHON: DETERMINANT & INVERSE ----------
("Compute determinant and inverse.",
"detA = -2.0 → invertible; inverse rotates/scales back.",
"Det = Scale factor; Inv = Undo transform.",
"""**Meaning**
The **determinant** measures how a matrix scales space — its “area” effect for 2D transformations.
If det(A) ≠ 0, the matrix is **invertible**, meaning there exists another matrix that reverses its transformation.
The **inverse matrix** undoes the effect of the original matrix by restoring any transformed vector.

**Code Example**
import numpy as np
A = np.array([[1, 2],
              [3, 4]])
detA = np.linalg.det(A)
invA = np.linalg.inv(A)
print("Determinant:", detA)
print("Inverse:\n", invA)
Output → det(A) = -2.0 (invertible), inverse reverses the effect.
"""),
]

trigonometry_mastery = [

# ---------- TRIANGLE SIDES ----------
("Triangle Sides — Concept",
"A right triangle has three sides: opposite, adjacent, and hypotenuse.",
"OAH → Opposite, Adjacent, Hypotenuse.",
"**Meaning**\nEvery right triangle includes one **right angle (90°)**, and the sides are named **relative to the reference angle θ**:\n- **Opposite side** – lies across from θ (forms vertical in diagram).\n- **Adjacent side** – lies next to θ and touches it (but not the hypotenuse).\n- **Hypotenuse** – longest side, always opposite the right angle.\n\nThese three sides define the trigonometric ratios:\n**sin θ = Opposite / Hypotenuse**  \n**cos θ = Adjacent / Hypotenuse**  \n**tan θ = Opposite / Adjacent**\n\n**Code Example**\nimport math\nθ = math.radians(30)\nhyp = 10\nopp = hyp * math.sin(θ)\nadj = hyp * math.cos(θ)\nprint(f'Opposite={opp:.2f}, Adjacent={adj:.2f}, Hypotenuse={hyp}')"
,
"https://i.sstatic.net/pLloD.png"),

("Triangle Sides — Relation (Pythagoras)",
"a² + b² = c².",
"Pythagoras: Perpendicular + Base = Hypotenuse².",
"**Meaning**\nIn any right triangle, the square of the **hypotenuse (c)** equals the sum of the squares of the other two sides:\n**a² + b² = c².**\nThis defines how the three sides relate geometrically and numerically.\nIt’s the foundation of Euclidean geometry, distance formulas, and vector norms.\n\n**Code Example**\nimport math\na, b = 3, 4\nc = math.sqrt(a**2 + b**2)\nprint('Hypotenuse =', c)"
,"https://www.chilimath.com/wp-content/uploads/2023/08/the-pythagorean-theorem-formula-illustrated.png"),

("Triangle Sides — Trig Ratios Summary",
"sin θ = O/H, cos θ = A/H, tan θ = O/A.",
"SOH-CAH-TOA = sine, cosine, tangent.",
"**Meaning**\nThe trigonometric ratios link sides of a right triangle to an angle θ:\n- **sin θ** (sine) = Opposite / Hypotenuse  \n- **cos θ** (cosine) = Adjacent / Hypotenuse  \n- **tan θ** (tangent) = Opposite / Adjacent  \n\nThese ratios remain constant for all similar triangles (same angles, different sizes).\n\n**Code Example**\nimport math\nθ = math.radians(45)\nprint('sin:', round(math.sin(θ),3), 'cos:', round(math.cos(θ),3), 'tan:', round(math.tan(θ),3))"
,
 "https://www.onlinemathlearning.com/image-files/trig-ratios.png"),

("Triangle Sides — Identifying Relative to Angle",
"Label sides relative to θ: Opposite, Adjacent, Hypotenuse.",
"Angle decides the names.",
"**Meaning**\nThe names “opposite” and “adjacent” **depend on which angle θ** you’re using as reference.\n- The side across from θ → Opposite.\n- The side touching θ (besides hypotenuse) → Adjacent.\n- The longest side opposite 90° → Hypotenuse.\n\nSwitching the reference angle changes which side is opposite vs. adjacent.\n\n**Code Example**\n# Example for a right triangle with sides (3,4,5)\nangle_ref = 'smaller_angle'\nprint('Opposite = 3, Adjacent = 4, Hypotenuse = 5 (for smaller angle)')"
,"https://www.onlinemathlearning.com/image-files/trigonometry-hypotenuse-opposite-adjacent.png"),

("Triangle Sides — Distance Formula Link",
"d = √((x₂−x₁)² + (y₂−y₁)²).",
"Distance = diagonal = hypotenuse.",
"**Meaning**\nThe **distance formula** is a direct application of the **Pythagorean theorem** in the coordinate plane.  \nThe difference between x and y coordinates form the triangle’s **adjacent** and **opposite** sides,  \nand the line connecting the two points is the **hypotenuse**.\n\n**Code Example**\nimport math\nx1, y1 = 0, 0\nx2, y2 = 3, 4\nd = math.dist((x1, y1), (x2, y2))\nprint('Distance between points:', d)"
,"https://www.gstatic.com/education/formulas2/553212783/en/distance_formula.svg"),

("Triangle Sides — Area Formula",
"Area = ½ × base × height.",
"Half base times height.",
"**Meaning**\nThe area of a right triangle is half the product of its **base** and **height** (which correspond to adjacent and opposite sides for the right angle).  \nThis formula derives from the area of a rectangle divided by 2.\n\n**Code Example**\nbase, height = 3, 4\narea = 0.5 * base * height\nprint('Triangle area:', area)"
,"https://www.inchcalculator.com/wp-content/uploads/2022/12/triangle-area-formula.png"),

("Triangle Sides — Connection to the Unit Circle",
"Unit circle triangles have hypotenuse = 1 → coordinates = (cos θ, sin θ).",
"Unit circle = triangle on a circle.",
"**Meaning**\nWhen you draw a right triangle inside the **unit circle**, its hypotenuse equals 1.  \nThis makes the sides numerically equal to **cos θ** (adjacent) and **sin θ** (opposite).  \nThus, trigonometric functions are just normalized triangle ratios on a circle.\n\n**Code Example**\nimport numpy as np\nθ = np.pi/6\nadj = np.cos(θ)\nopp = np.sin(θ)\nprint(f'On unit circle: cos(θ)={adj:.3f}, sin(θ)={opp:.3f}'",
"https://telemeter.wndsn.com/img/unitcircle-sohcahtoa-6fnc.jpg"),

# ---------- DAY 1 ----------
("Day 1 – Geometry of Angles and Triangles",
"A triangle’s angles measure rotation; total = 180°. Angles are measured in degrees (360° per circle) or radians (2π per circle).",
"Radian = radius-unit turn.",
"""**Meaning**
Angles express rotation.  One radian is the angle that cuts an arc equal to the radius.
Because a circle’s circumference is 2πr, a complete revolution equals 2π radians or 360°.
Radians feel natural because they link arc length directly to radius.

**Code Example**
import numpy as np, matplotlib.pyplot as plt
theta = np.linspace(0, 2*np.pi, 100)
plt.plot(np.cos(theta), np.sin(theta))
plt.gca().set_aspect('equal')
plt.title("One full rotation = 2π radians = 360°")
plt.show(),
""",
 "https://mammothmemory.net/images/user/base/Maths/Geometry/Internal%20Angles%20of%20Polygons/example-1-of-a-triangles-internal-angles.ccf9363.jpg"),

# ---------- DAY 2 ----------
("Day 2 – Defining sin, cos, tan",
"sin θ = O/H,  cos θ = A/H,  tan θ = O/A.",
"SOH-CAH-TOA = sine, cosine, tangent.",
"""**Meaning**
Sine, cosine, and tangent compare sides of a right triangle.
The ratios depend only on the angle, not triangle size.
They extend beyond 0°–90° using the unit circle.

**Code Example**
import math
θ = math.radians(30)
print(math.sin(θ), math.cos(θ), math.tan(θ))
""",
 "https://mathsathome.com/wp-content/uploads/2023/07/sohcahtoa-1024x581.png"),

# ---------- DAY 3 ----------
("Day 3 – Tangent as Slope",
"tan θ equals slope = rise / run.",
"Tan tells the tilt.",
"""**Meaning**
Tangent measures steepness.
A line through the origin with slope m makes an angle θ such that tan θ = m.
In calculus, the tangent line shares this slope locally with a curve.

**Code Example**
import numpy as np, matplotlib.pyplot as plt
x = np.linspace(0, 1, 50)
plt.plot(x, np.tan(np.radians(45))*x)
plt.title("tan(45°) = 1 → slope = 1")
plt.show()
""",
 "https://mathbooks.unl.edu/PreCalculus/images/imagesChap13/TangentSlope.png"),

# ---------- DAY 4 ----------
("Day 4 – Inverse Functions",
"Inverse trig functions return the angle from a ratio.",
"'arc' means angle recovery.",
"""**Meaning**
arcsin, arccos, and arctan undo the basic trig functions.
They return principal-value angles so that the mapping stays one-to-one.

**Code Example**
import math
print(math.degrees(math.asin(0.5)))   # 30°
"""),

# ---------- DAY 5 ----------
("Day 5 – Degrees ↔ Radians",
"1° = π/180 rad; 180° = π rad; 360° = 2π rad.",
"Half circle, half π.",
"""**Meaning**
A radian measures arc length over radius.
Since a circle’s circumference is 2πr, converting degrees to radians multiplies by π/180.

**Code Example**
import math
deg = 60
rad = deg * math.pi / 180
print(rad)
"""),

# ---------- DAY 6 ----------
("Day 6 – Sine, Cosine, Tangent Graphs",
"sin θ and cos θ oscillate between −1 and 1 with period 2π.",
"Spin becomes wave.",
"""**Meaning**
As the radius of the unit circle rotates, its x- and y-coordinates trace cosine and sine waves.
Tangent repeats every π and diverges when cos θ = 0.

**Code Example**
import numpy as np, matplotlib.pyplot as plt
θ = np.linspace(0, 2*np.pi, 200)
plt.plot(θ, np.sin(θ), label='sin θ')
plt.plot(θ, np.cos(θ), label='cos θ')
plt.legend(); plt.show()
"""),

# ---------- DAY 7 ----------
("Day 7 – The Unit Circle",
"Each point on the unit circle = (cos θ, sin θ).",
"Circle of 1 — sin up, cos across.",
"""**Meaning**
The unit circle connects angles to coordinates.
For a radius of 1, x = cos θ and y = sin θ.
Quadrants determine the sign of each function.

**Code Example**
import numpy as np, matplotlib.pyplot as plt
θ = np.linspace(0, 2*np.pi, 400)
plt.plot(np.cos(θ), np.sin(θ))
plt.gca().set_aspect('equal')
plt.title("Unit Circle – (cos θ, sin θ)")
plt.show()
"""),

# ---------- DAY 8 ----------
("Day 8 – Pythagorean Identity",
"sin²θ + cos²θ = 1.",
"Circle truth: sin² + cos² = 1.",
"""**Meaning**
From the unit-circle equation x² + y² = 1, substituting x = cos θ and y = sin θ gives
sin²θ + cos²θ = 1 — the fundamental relationship of trigonometry.

**Code Example**
import numpy as np
θ = np.linspace(0, 2*np.pi, 5)
print(np.sin(θ)**2 + np.cos(θ)**2)
"""),

# ---------- DAY 9 ----------
("Day 9 – The Dot Product",
"a·b = |a||b|cos θ.",
"Dot = degree of alignment.",
"""**Meaning**
The dot product measures how parallel two vectors are.
It equals the product of their magnitudes times cos θ, linking algebraic multiplication with angle.

**Code Example**
import numpy as np
a = np.array([1,0]); b = np.array([np.sqrt(2)/2, np.sqrt(2)/2])
θ = np.degrees(np.arccos(np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))))
print(θ)
"""),

# ---------- DAY 10 ----------
("Day 10 – The Cross Product",
"|a×b| = |a||b|sin θ → gives perpendicular vector and area.",
"Cross creates area.",
"""**Meaning**
The cross product produces a vector perpendicular to both inputs.
Its magnitude equals the parallelogram area spanned by a and b.

**Code Example**
import numpy as np
a = np.array([3,0,0]); b = np.array([0,4,0])
cross = np.cross(a,b)
print("Cross:", cross, "Magnitude:", np.linalg.norm(cross))
"""),

# ---------- DAY 11 ----------
("Day 11 – Arctan and Angle Finding",
"θ = arctan(y/x); use arctan2(y,x) for correct quadrant.",
"atan2 → angle true.",
"""**Meaning**
arctan finds the angle from slope y/x but cannot tell the correct quadrant.
arctan2(y,x) fixes this by considering the sign of both components.

**Code Example**
import numpy as np
θ = np.degrees(np.arctan2(1,1))
print(θ)   # 45°
"""),

# ---------- DAY 12 ----------
("Day 12 – Trig in Rotations",
"R(θ) = [[cos θ, −sin θ], [sin θ, cos θ]].",
"Cos keeps, sin swaps.",
"""**Meaning**
The rotation matrix turns vectors in 2-D space by angle θ around the origin.
It preserves length and orthogonality, implementing pure rotation.

**Code Example**
import numpy as np
θ = np.pi/4
R = np.array([[np.cos(θ), -np.sin(θ)], [np.sin(θ), np.cos(θ)]])
v = np.array([1,0])
print(R @ v)
"""),

# ---------- DAY 13 ----------
("Day 13 – Waves & Harmonic Motion",
"x(t) = A cos(ωt + φ).",
"Circle’s shadow = wave.",
"""**Meaning**
Simple harmonic motion is projection of uniform circular motion.
A controls amplitude, ω controls frequency, and φ controls phase shift.

**Code Example**
import numpy as np, matplotlib.pyplot as plt
t = np.linspace(0, 2*np.pi, 200)
plt.plot(t, np.cos(t))
plt.title("x(t) = cos(t)")
plt.show()
"""),

# ---------- DAY 14 ----------
("Day 14 – Real-World Applications",
"Trig appears in physics, graphics, AI, and engineering.",
"Angles connect worlds.",
"""**Meaning**
Trigonometry underlies analysis of rotation, projection, and similarity.
Used in computer vision, robotics, acoustics, and machine learning (cosine similarity).

**Code Example**
import numpy as np
a = np.array([1,0]); b = np.array([np.sqrt(2)/2, np.sqrt(2)/2])
cos_sim = np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
print("Cosine similarity:", cos_sim)
"""),

# ---------- DAY 15 ----------
("Day 15 – Review & Synthesis",
"All trigonometry is geometry written in algebraic form.",
"sin = height, cos = width, tan = slope.",
"""**Meaning**
Every trigonometric relationship comes from geometry.
Sine measures vertical projection, cosine horizontal, tangent slope, and radians link arc to angle.

**Code Example**
import numpy as np
θ = np.linspace(0, 2*np.pi, 8)
for t in θ:
    print("θ:", round(t,2), "sin:", round(np.sin(t),2), "cos:", round(np.cos(t),2))
"""),

(
"Day 16 – The Unit Circle (Deep Dive)",
"The unit circle maps every angle θ to coordinates (cos θ, sin θ) — extending trig to all angles.",
"3-4-6 → √3/2, √2/2, 1/2 — mirror across x and y axes.",
"""**Meaning**
The **unit circle** is a circle centered at (0, 0) with radius 1.
Every point P on the circle corresponds to an angle θ measured from the positive x-axis: P = (cos θ, sin θ).
This extends right-triangle ratios to all angles, making sine and cosine periodic with period 2π.
Signs of sin θ and cos θ depend on the quadrant:

| Quadrant | Angle Range | cos θ | sin θ |
|:-|:-|:-|:-|
| I | 0 → π/2 | + | + |
| II | π/2 → π | – | + |
| III | π → 3π/2 | – | – |
| IV | 3π/2 → 2π | + | – |

Key angles to know (degrees → radians → cos θ / sin θ):

0° (0) = 1, 0  • 30° (π/6) = √3/2, 1/2  • 45° (π/4) = √2/2, √2/2  • 60° (π/3) = 1/2, √3/2  • 90° (π/2) = 0, 1
120° (2π/3) = –1/2, √3/2  • 135° (3π/4) = –√2/2, √2/2  • 150° (5π/6) = –√3/2, 1/2  • 180° (π) = –1, 0
270° (3π/2) = 0, –1  • 360° (2π) = 1, 0

Geometrically, drawing a radius r = 1 to a point (cos θ, sin θ) forms a right triangle whose adjacent = cos θ and opposite = sin θ.
Hence sin θ and cos θ are simply the sides of that triangle.

**Code Example**
import numpy as np, matplotlib.pyplot as plt
theta = np.linspace(0, 2*np.pi, 360)
x = np.cos(theta)
y = np.sin(theta)
plt.plot(x, y)
plt.axis('equal')
plt.xlabel('cos θ')
plt.ylabel('sin θ')
plt.title('Unit Circle')
plt.show()

Applications:
• Convert between angle ↔ coordinates.
• Compute rotations in 2D: [x′ y′] = [[cos θ  −sin θ],[sin θ  cos θ]] [x y].
• Model periodic motion (sine waves).
• Define all other trig functions (tan θ = sin θ / cos θ).

Key Insight — The unit circle is the geometric universe of trigonometry: every sin θ and cos θ value is simply a coordinate on that circle.
"""),

]

ml_chapter1 = [

# ---------- SECTION 1.1 ----------
("ML Basics — What It Is (and Isn’t)",
"""Machine Learning teaches computers to make predictions or decisions from examples rather than rules.
It transforms data into learned patterns that generalize to new information.""",
"R D K → Rules, Data, Knowledge. Kindergarten: 'Robots Drink Kool-Aid.'",
"""**Meaning**
Machine Learning is about *learning from data*, not programming logic.
Instead of defining step-by-step instructions, we let algorithms infer patterns and relationships.

1. **Traditional Programming** = Rules + Data → Answers
2. **Machine Learning** = Data + Answers → Rules
3. Learning = Adjusting parameters to minimize error.
"""),

("ML Basics — Simple Linear Model (Code)",
"""Train a model to predict exam scores from hours studied using linear regression.""",
"I M C T A → Import, Make, Create, Train, Ask. Kindergarten: 'I Make Clever Teachers Awesome.'",
"""**Meaning**
This code demonstrates the workflow of a simple ML model: import, make data, create features, train, and ask for predictions.

**Code Example**
```python
# I M C T A → Import, Make, Create, Train, Ask. (Kindergarten: "I Make Clever Teachers Awesome.")

import pandas as pd
from sklearn.linear_model import LinearRegression

# Make data
data = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5],
    'Score': [50, 55, 65, 70, 80]
})

# Create X (features) and y (target)
X = data[['Hours']]
y = data['Score']

# Train model
model = LinearRegression()
model.fit(X, y)

# Ask it to predict a new value
print(model.predict([[6]]))
```"""),

# ---------- SECTION 1.2 ----------
("ML Impact — Why It Matters",
"""Machine Learning matters because it uncovers complex patterns that humans can’t see.
It powers recommendations, fraud detection, medical diagnosis, and automation across industries.""",
"A S F → Apps, Scans, Fraud. Kindergarten: 'Ants See Food.'",
"""**Meaning**
ML scales human decision-making, automating repetitive analysis and prediction.

1. **Automation** – models work tirelessly and consistently.
2. **Scalability** – thousands of predictions per second.
3. **Usefulness** – works for simple or global-scale data projects.
"""),

("ML Impact — Predicting from Patterns (Code)",
"""Demonstrates a perfect mathematical pattern where ML learns y = 2x automatically.""",
"I C T M → Import, Create, Train, Make (prediction). Kindergarten: 'I Count Tiny Mice.'",
"""**Meaning**
This code shows a model recognizing a simple relationship between x and y by training itself.

**Code Example**
```python
# I C T M → Import, Create, Train, Make (prediction). (Kindergarten: "I Count Tiny Mice.")

import numpy as np
from sklearn.linear_model import LinearRegression

# Create data
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2,4,6,8,10])

# Train model
model = LinearRegression()
model.fit(X, y)

# Make prediction
print(model.predict([[6]]))
```"""),

# ---------- SECTION 1.3 ----------
("ML Process — How It Works",
"""All ML systems follow a similar learning loop: input, training, testing, and deployment.
Training uses data feedback to minimize error through optimization.""",
"I P S T T → Import, Prepare, Split, Train, Test. Kindergarten: 'I Play Smart Train Tracks.'",
"""**Meaning**
Learning is an iterative process that improves model accuracy with each feedback cycle.

1. **Input** – feed in features.
2. **Prepare** – structure and clean data.
3. **Split** – separate training and testing sets.
4. **Train** – fit the model.
5. **Test** – measure accuracy.
"""),

("ML Process — The Learning Loop (Code)",
"""Implements the ML workflow: split data, train, and test accuracy.""",
"I P S T T → Import, Prepare, Split, Train, Test. Kindergarten: 'I Play Smart Train Tracks.'",
"""**Meaning**
This example shows the data lifecycle: prepare, split, train, and test performance.

**Code Example**
```python
# I P S T T → Import, Prepare, Split, Train, Test. (Kindergarten: "I Play Smart Train Tracks.")

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import pandas as pd

# Prepare data
data = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5, 6],
    'Score': [50, 55, 60, 65, 70, 75]
})

# Split data
X = data[['Hours']]
y = data['Score']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0)

# Train
model = LinearRegression()
model.fit(X_train, y_train)

# Test accuracy
print(model.score(X_test, y_test))
```"""),

# ---------- SECTION 1.4 ----------
("ML Ethics — The Human Side",
"""Humans remain central in ML — they choose the data, interpret results, and enforce ethics.
Bad data leads to bad outcomes, no matter how advanced the algorithm.""",
"H E T → Humans, Ethics, Trust. Kindergarten: 'Horses Eat Tacos.'",
"""**Meaning**
Responsible ML balances technical power with human oversight.
People define the problem, interpret predictions, and ensure fairness and transparency.

1. **Human Oversight** – set goals and constraints.
2. **Ethical Data** – remove bias, protect privacy.
3. **Trust** – explain results and ensure accountability.
"""),

("ML Ethics — Human Oversight (Code)",
"""Shows how an outlier affects model predictions — human review is essential.""",
"I P T P → Import, Prepare, Train, Predict. Kindergarten: 'I Paint Tiny Penguins.'",
"""**Meaning**
A single bad data point can distort model learning.
This demonstrates why humans must clean and validate data before training.

**Code Example**
```python
# I P T P → Import, Prepare, Train, Predict. (Kindergarten: "I Paint Tiny Penguins.")

from sklearn.linear_model import LinearRegression
import pandas as pd

# Prepare data with an outlier
data = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 10],
    'Score': [50, 55, 60, 65, 200]
})

# Train model
model = LinearRegression()
model.fit(data[['Hours']], data['Score'])

# Predict
print(model.predict([[6]]))
```""")
]
ml_chapter2 = [

# ---------- SECTION 2.1 ----------
("Data Importance — Why Data Matters",
"""Machine Learning depends entirely on data quality.
Without clean, balanced, and relevant data, even the best algorithms fail.
Data is the fuel that powers learning — every number, word, or image tells part of a larger story.""",
"D B C → Data, Bias, Cleanliness. Kindergarten: 'Dogs Bring Cookies.'",
"""**Meaning**
Data gives models something to learn from.
Good data reveals patterns; bad data hides them.
Bias, incompleteness, and noise cause poor predictions.

1. **Incomplete Data** – missing values reduce accuracy.
2. **Biased Data** – unequal representation leads to unfair results.
3. **Noisy Data** – random errors confuse the model.
"""),

("Data Importance — Showing the Power of Data (Code)",
"""Shows how a single outlier can distort the learning process and model predictions.""",
"I C T P → Import, Create, Train, Plot. Kindergarten: 'I Catch Tall Penguins.'",
"""**Meaning**
An unrealistic data point drastically alters the fitted model line, proving that bad data misleads the learner.

**Code Example**
```python
# I C T P → Import, Create, Train, Plot. (Kindergarten: "I Catch Tall Penguins.")

import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Create data with an outlier
data = pd.DataFrame({
    'Hours_Studied': [1, 2, 3, 4, 5, 100],
    'Exam_Score': [40, 50, 60, 70, 80, 85]
})

# Train model
model = LinearRegression()
model.fit(data[['Hours_Studied']], data['Exam_Score'])

# Plot
plt.scatter(data['Hours_Studied'], data['Exam_Score'])
plt.plot(data['Hours_Studied'], model.predict(data[['Hours_Studied']]), color='red')
plt.title("Outlier Changes the Line")
plt.show()
```"""),

# ---------- SECTION 2.2 ----------
("Data Types — The Kinds of Data You’ll Meet",
"""Different data forms require different handling:
Structured (tables), Unstructured (text, video), and Semi-Structured (JSON/XML).
ML models prefer numeric formats, so everything must become numbers.""",
"S U S → Structured, Unstructured, Semi-Structured. Kindergarten: 'Snakes Use Sunglasses.'",
"""**Meaning**
Recognizing data type determines preprocessing strategy.
Structured data fits neatly into tables.
Unstructured data must be transformed into numeric features.

1. **Structured** – rows/columns, e.g., databases.
2. **Unstructured** – free-form media or text.
3. **Semi-Structured** – flexible formats like JSON.
"""),

("Data Types — Seeing Different Data Shapes (Code)",
"""Demonstrates the three major data types: structured, unstructured, and semi-structured.""",
"I S U S → Import, Structured, Unstructured, Semi-Structured. Kindergarten: 'I See Unicorn Stars.'",
"""**Meaning**
This example contrasts three data representations that ML engineers frequently encounter.

**Code Example**
```python
# I S U S → Import, Structured, Unstructured, Semi-Structured. (Kindergarten: "I See Unicorn Stars.")

import pandas as pd
import json

# Structured data
structured = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Cathy'],
    'Age': [25, 30, 27],
    'City': ['New York', 'Boston', 'Chicago']
})

# Unstructured data
unstructured = ["I love learning!", "ML is amazing!", "Data rocks!"]

# Semi-structured data (JSON)
semi = json.dumps({
    "employees": [
        {"name": "Alice", "role": "Data Scientist"},
        {"name": "Bob", "role": "Engineer"}
    ]
}, indent=2)

print(structured)
print(unstructured)
print(semi)
```"""),

# ---------- SECTION 2.3 ----------
("Data Workflow — The Data Science Process",
"""The Data Science Workflow is a repeating cycle:
Collect, Explore, Clean, Visualize, Model, Evaluate, and Communicate.
Each loop improves data understanding and model accuracy.""",
"C E C V M → Collect, Explore, Clean, Visualize, Model. Kindergarten: 'Cats Eat Crunchy Veggie Muffins.'",
"""**Meaning**
The workflow guides every ML project.
It ensures disciplined handling from raw data to actionable insights.

1. **Collect** – gather data.
2. **Explore** – describe and understand.
3. **Clean** – fix missing or inconsistent entries.
4. **Visualize** – detect trends and outliers.
5. **Model** – train algorithms.
"""),

("Data Workflow — Mini Data Workflow (Code)",
"""Implements a miniature end-to-end data workflow from collection to modeling.""",
"I C C V M → Import, Collect, Clean, Visualize, Model. Kindergarten: 'I Cook Cold Veggie Muffins.'",
"""**Meaning**
A short demonstration of the full data pipeline: from gathering to cleaning to visualizing to modeling.

**Code Example**
```python
# I C C V M → Import, Collect, Clean, Visualize, Model. (Kindergarten: "I Cook Cold Veggie Muffins.")

import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Collect & explore
data = pd.DataFrame({'Hours': [1, 2, 3, 4, 5], 'Score': [40, 50, 60, 70, 80]})
print(data.describe())

# Clean
data = data.fillna(method='ffill')

# Visualize
plt.scatter(data['Hours'], data['Score'])
plt.title("Study Hours vs Exam Score")
plt.show()

# Model
model = LinearRegression()
model.fit(data[['Hours']], data['Score'])
print("Model trained successfully.")
```"""),

# ---------- SECTION 2.4 ----------
("Data Quality — Data Quality and Bias",
"""Good data must be representative and accurate.
Bias occurs when training data favors one group or outcome.
Incomplete, inconsistent, or duplicated data also degrade model reliability.""",
"B I Q → Bias, Imbalance, Quality. Kindergarten: 'Bears Ignore Quokkas.'",
"""**Meaning**
Models mirror their data.
If the data is biased, predictions will be too.
Inspect, balance, and test datasets before training.

1. **Bias** – unfair representation of classes.
2. **Imbalance** – uneven group counts.
3. **Quality** – completeness and correctness.
"""),

("Data Quality — Spotting Bias and Outliers (Code)",
"""Visualizes biased group data and score imbalance using bar and scatter plots.""",
"I C V V → Import, Create, Visualize, Visualize Again. Kindergarten: 'I Count Very Vividly.'",
"""**Meaning**
Graphs expose dataset imbalance and outliers, helping prevent biased learning.

**Code Example**
```python
# I C V V → Import, Create, Visualize, Visualize Again. (Kindergarten: "I Count Very Vividly.")

import pandas as pd
import matplotlib.pyplot as plt

# Create biased data
data = pd.DataFrame({
    'Group': ['A']*8 + ['B']*2,
    'Score': [80, 82, 85, 83, 81, 79, 88, 90, 55, 58]
})

# Visualize imbalance
data['Group'].value_counts().plot(kind='bar', title='Group Counts')
plt.show()

# Visualize scores
plt.scatter(data['Group'], data['Score'])
plt.title('Score Distribution by Group')
plt.show()
```""")
]

ml_chapter3 = [

# ---------- SECTION 3.1 ----------
("Data Prep — Why Data Preparation Matters",
"""Machine Learning models understand numbers, not words.
Before training, all information must be cleaned and converted into a numerical format.
Bad data leads to bad models; clean data leads to reliable learning.""",
"I C L → Import, Create, Look. Kindergarten: 'I Clean Laundry.'",
"""**Meaning**
Data preparation is the foundation of every ML workflow.
It corrects missing values, fixes data types, removes duplicates, and standardizes formats.

1. **Missing values** – fill or remove.
2. **Wrong types** – convert to numeric or categorical.
3. **Inconsistent entries** – normalize capitalization and spacing.
"""),

("Data Prep — Seeing the Problem (Code)",
"""Shows messy data containing missing values, wrong types, and duplicates before cleaning.""",
"I C L → Import, Create, Look. Kindergarten: 'I Clean Laundry.'",
"""**Meaning**
Raw data almost always contains noise.
We preview it first to understand issues that must be fixed before modeling.

**Code Example**
```python
# I C L → Import, Create, Look. (Kindergarten: "I Clean Laundry.")

import pandas as pd

# Create messy data
data = pd.DataFrame({
    'Age': [25, None, 35, 'thirty', 40],
    'Income': [50000, 60000, None, 80000, 120000],
    'City': ['New York', 'Boston', 'Boston', None, 'New York']
})

print("Before cleaning:")
print(data)
```"""),

# ---------- SECTION 3.2 ----------
("Missing Data — Handling Missing Values",
"""Missing data disrupts learning and prediction.
Common fixes include removing incomplete rows, filling gaps with averages, or predicting values using models.""",
"I C F P → Import, Create, Fill, Print. Kindergarten: 'I Clean Fish Ponds.'",
"""**Meaning**
Most ML systems require complete data matrices.
Filling (imputation) keeps sample size stable and preserves relationships.

1. **Remove** – delete missing entries.
2. **Fill** – impute mean, median, or mode.
3. **Predict** – model missing values from other features.
"""),

("Missing Data — Filling the Gaps (Code)",
"""Uses mean imputation to fill missing numeric values in a small dataset.""",
"I C F P → Import, Create, Fill, Print. Kindergarten: 'I Clean Fish Ponds.'",
"""**Meaning**
Simple mean filling repairs incomplete columns so algorithms run without errors.

**Code Example**
```python
# I C F P → Import, Create, Fill, Print. (Kindergarten: "I Clean Fish Ponds.")

import pandas as pd
import numpy as np

# Create dataset with missing values
data = pd.DataFrame({
    'Age': [25, np.nan, 30, 35, np.nan],
    'Salary': [40000, 50000, np.nan, 70000, 80000]
})

# Fill missing values with column mean
data['Age'].fillna(data['Age'].mean(), inplace=True)
data['Salary'].fillna(data['Salary'].mean(), inplace=True)

print("After filling missing values:")
print(data)
```"""),

# ---------- SECTION 3.3 ----------
("Data Types — Fixing Wrong Formats",
"""Numbers stored as text or inconsistent capitalization confuse ML algorithms.
Convert data to correct numeric types and standardize text formats.""",
"I C F P → Import, Create, Fix, Print. Kindergarten: 'I Cut Funny Paper.'",
"""**Meaning**
Converting and normalizing ensures that equal items are treated equally.

1. **Convert** strings to numbers using coercion.
2. **Normalize** capitalization with `.str.title()` or similar.
3. **Inspect** outputs before modeling.
"""),

("Data Types — Converting and Cleaning (Code)",
"""Fixes data type errors and inconsistent capitalization using pandas.""",
"I C F P → Import, Create, Fix, Print. Kindergarten: 'I Cut Funny Paper.'",
"""**Meaning**
After conversion, all data types become machine-readable.

**Code Example**
```python
# I C F P → Import, Create, Fix, Print. (Kindergarten: "I Cut Funny Paper.")

import pandas as pd

data = pd.DataFrame({
    'Age': ['25', '30', 'thirty-five', '40'],
    'City': ['boston', 'Boston', 'BOSTON', 'new york']
})

# Convert and clean
data['Age'] = pd.to_numeric(data['Age'], errors='coerce')
data['City'] = data['City'].str.title()

print("Cleaned data:")
print(data)
```"""),

# ---------- SECTION 3.4 ----------
("Scaling — Scaling and Normalizing Data",
"""Different feature scales can distort model learning.
Scaling rescales numbers to similar ranges; normalization evens distributions.""",
"I C A P → Import, Create, Apply, Print. Kindergarten: 'I Cook Apple Pie.'",
"""**Meaning**
Feature scaling ensures fair weighting between variables.
Essential for distance-based algorithms and neural networks.

1. **Scaling** – shrink range (e.g., 0–1).
2. **Normalization** – adjust to unit norms or zero mean.
3. **Equality** – keeps one feature from overpowering another.
"""),

("Scaling — Scaling Features (Code)",
"""Uses MinMaxScaler to scale numeric data between 0 and 1.""",
"I C A P → Import, Create, Apply, Print. Kindergarten: 'I Cook Apple Pie.'",
"""**Meaning**
Scaled values maintain proportional relationships while fitting uniform ranges.

**Code Example**
```python
# I C A P → Import, Create, Apply, Print. (Kindergarten: "I Cook Apple Pie.")

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Create data
data = pd.DataFrame({
    'Age': [20, 30, 40, 50, 60],
    'Income': [20000, 40000, 60000, 80000, 100000]
})

# Apply scaling
scaler = MinMaxScaler()
scaled = scaler.fit_transform(data)

print("Scaled data:")
print(scaled)
```"""),

# ---------- SECTION 3.5 ----------
("Encoding — Categorical Data to Numbers",
"""Categorical variables must be converted into numbers for ML models.
Label encoding assigns integers; one-hot encoding creates binary columns.""",
"I C E P → Import, Create, Encode, Print. Kindergarten: 'I Catch Electric Penguins.'",
"""**Meaning**
Encoding converts words into numeric form so models can compute with them.
One-hot encoding avoids false numerical order assumptions.

1. **Label Encode** – assign IDs to categories.
2. **One-Hot Encode** – binary columns per category.
3. **Use Numeric Features** – enables mathematical learning.
"""),

("Encoding — Turning Words into Numbers (Code)",
"""Demonstrates one-hot encoding of color categories into numeric arrays.""",
"I C E P → Import, Create, Encode, Print. Kindergarten: 'I Catch Electric Penguins.'",
"""**Meaning**
Converts categorical text into interpretable binary vectors.

**Code Example**
```python
# I C E P → Import, Create, Encode, Print. (Kindergarten: "I Catch Electric Penguins.")

import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Create data
data = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})

# Encode
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(data[['Color']])

print("Encoded data:")
print(encoded)
```"""),

# ---------- SECTION 3.6 ----------
("Feature Engineering — Creating Better Inputs",
"""Feature engineering builds new variables that reveal deeper relationships.
Examples include BMI from height and weight, or month from date.""",
"I C C P → Import, Create, Compute, Print. Kindergarten: 'I Create Cool Projects.'",
"""**Meaning**
Newly engineered features increase model insight and accuracy.

1. **Combine** related variables into informative ratios.
2. **Extract** components (like month, day) from timestamps.
3. **Transform** text or geographic data into numerical features.
"""),

("Feature Engineering — Making New Features (Code)",
"""Adds a derived BMI feature to a dataset.""",
"I C C P → Import, Create, Compute, Print. Kindergarten: 'I Create Cool Projects.'",
"""**Meaning**
Shows how to compute new attributes to enhance predictive power.

**Code Example**
```python
# I C C P → Import, Create, Compute, Print. (Kindergarten: "I Create Cool Projects.")

import pandas as pd

# Create base data
data = pd.DataFrame({
    'Height_m': [1.8, 1.6, 1.9],
    'Weight_kg': [75, 60, 90]
})

# Compute new feature
data['BMI'] = data['Weight_kg'] / (data['Height_m'] ** 2)

print("Data with new feature:")
print(data)
```"""),

# ---------- SECTION 3.7 ----------
("Pipelines — Putting It All Together",
"""Pipelines automate preprocessing and modeling steps.
They guarantee identical transformations for training and testing.""",
"I B P → Import, Build, Print. Kindergarten: 'I Build Pyramids.'",
"""**Meaning**
Pipelines simplify ML workflows by chaining operations.
They ensure consistent scaling, encoding, and modeling each run.

1. **Import** preprocessing and model classes.
2. **Build** sequential pipeline steps.
3. **Print/Run** to verify the structure.
"""),

("Pipelines — Building a Simple Pipeline (Code)",
"""Creates a pipeline combining scaling and linear regression modeling.""",
"I B P → Import, Build, Print. Kindergarten: 'I Build Pyramids.'",
"""**Meaning**
Once defined, the pipeline performs preprocessing automatically.

**Code Example**
```python
# I B P → Import, Build, Print. (Kindergarten: "I Build Pyramids.")

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

# Build pipeline
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LinearRegression())
])

print("Pipeline ready to use!")
```""")
]

pandas = [

# ---------- 1. VARIABLES ----------
("Pandas — I/O (Reading & Writing Data (R E T T J P → Rabbits Eat Tiny Treats Jumping Playfully.)",
"""Definitions:
read_csv → Load a CSV text file into a DataFrame.
read_excel → Load an Excel .xls or .xlsx workbook into a DataFrame.
to_csv → Save a DataFrame as a CSV file.
to_excel → Save a DataFrame as an Excel .xlsx file.
read_json → Load JSON structured text into a DataFrame.
read_parquet → Load a Parquet columnar analytics file into a DataFrame.""",
"""1️⃣ R R T T R R → Rocket Rabbits Take Tiny Rainbow Roads 🚀🐇🛣️🌈\n
2️⃣ read_csv → Import CSV text files into DataFrames. 📄➡️📊\n
3️⃣ read_excel → Load Excel spreadsheets (.xls/.xlsx). 📘➡️📊\n
4️⃣ to_csv → Export DataFrames to CSV. 📊➡️📄\n
5️⃣ to_excel → Export DataFrames to Excel. 📊➡️📘\n
6️⃣ read_json → Load structured JSON documents. 🧱➡️📊\n
7️⃣ read_parquet → Load fast columnar Parquet files. 📦➡️📊""",
"""1. read_csv: Reads comma-separated text files into DataFrames, handling delimiters, quoting, missing values, and date parsing.
2. read_excel: Loads Excel spreadsheets using openpyxl/xlrd, supporting sheet selection and type inference.
3. to_csv: Saves a DataFrame to a CSV text file for portability and external use.
4. to_excel: Exports a DataFrame into an Excel workbook for reporting or business workflows.
5. read_json: Loads JSON in multiple orientations (records, split, index, columns) into DataFrames.
6. read_parquet: Reads optimized columnar Parquet files used in analytics and big-data pipelines.
7. Concept: The I/O category gives you the 6 essential ways to bring data in and out of pandas for nearly all real-world pipelines.""",
"I/O — Checkpoint Card"),

("Pandas — Indexing & Selecting (L I B A → Lions Inspect Big Apples.\nKindergarten: 'Lions eat apples)",
"""Definitions:
.loc → Select rows or columns by label.
.iloc → Select rows or columns by integer position.
[] basic selection → Direct bracket-based indexing or column selection.
.at / .iat → Fast scalar lookup by label or position.""",
"""1️⃣ L I B A → Little Iguanas Build Airplanes 🦎🛠️🛩️\n
2️⃣ .loc → Label-based indexing. 🔤➡️📊\n
3️⃣ .iloc → Integer-position indexing. 🔢➡️📊\n
4️⃣ [] → Basic selection shorthand. ✂️📊\n
5️⃣ .at/.iat → Fast scalar cell access. ⚡🔍""",
"""1. .loc: Select rows/columns using LABELS — names, strings, index values.
2. .iloc: Select rows/columns using integer positions — 0-based indexing.
3. []: Shorthand for selecting a column or slicing rows (df['col'], df[0:10]).
4. .at/.iat: Extremely fast single-cell lookup where speed matters.
5. Concept: Indexing controls HOW you access and slice your DataFrame — one of the most critical Pandas skills.""",
"Indexing & Selecting — Checkpoint Card"),

("Pandas — Cleaning & Preprocessing (I N D F A R R D → In November Dogs Find All Red Roses Daily)",
"""Definitions:
.isna → Identify missing values.
.notna → Identify non-missing values.
.dropna → Remove missing rows or columns.
.fillna → Replace missing values with a fill value.
.astype → Convert a column to a new data type.
.rename → Change row or column labels.
.replace → Replace specific values in a DataFrame.
.drop → Remove rows or columns by label.""",
"""1️⃣ L I B A → Little Iguanas Build Airplanes 🦎🛠️🛩️\n
2️⃣ .loc → Label-based indexing. 🔤➡️📊\n
3️⃣ .iloc → Integer-position indexing. 🔢➡️📊\n
4️⃣ [] → Basic selection shorthand. ✂️📊\n
5️⃣ .at/.iat → Fast scalar cell access. ⚡🔍""",
"""1️⃣ I N D F A R R D → Icy Ninjas Dive Fearlessly Across Rocky River Dams 🥷❄️🌊\n
2️⃣ isna → Detect missing values. 🚫➡️🔍\n
3️⃣ notna → Detect non-missing values. ✔️🔍\n
4️⃣ dropna → Remove rows with missing values. 🧹\n
5️⃣ fillna → Fill missing values. 🩹\n
6️⃣ astype → Convert column dtypes. 🔠🔢\n
7️⃣ rename → Rename columns/labels. 🏷️\n
8️⃣ replace → Replace specific values. 🔄\n
9️⃣ drop → Drop columns/rows. ✂️""",
"""1. isna: Returns True/False masks for missing entries (NaN, None).
2. notna: Opposite of isna — useful for filtering clean rows.
3. dropna: Remove rows/columns with missing data for strict pipelines.
4. fillna: Fill missing entries with constants, medians, means, etc.
5. astype: Change column data types for consistency (int, float, str).
6. rename: Rename columns or index labels to maintain clean schema.
7. replace: Swap specific values or patterns – great for categories.
8. drop: Remove unneeded columns or rows.
9. Concept: These 8 operations handle almost every cleaning workflow in ETL and ML prep.""",
"Cleaning — Checkpoint Card"),

("Pandas — Combining Data (M C J → Monkeys Carry Juice.) ",
"""Definitions:
merge → Combine DataFrames using SQL-style joins on matching keys.
concat → Stack DataFrames vertically or horizontally.
join → Combine DataFrames by aligning their index labels.""",
"""1️⃣ M C J → Moose Carry Jars 🫎🫙\n
2️⃣ merge → SQL-style joins. 🔗\n
3️⃣ concat → Stack DataFrames. 📚\n
4️⃣ join → Join on index. 🔧""",
"""1. merge: Perform SQL-style joins between two DataFrames (left, right, inner, outer).
2. concat: Stack DataFrames vertically (axis=0) or side-by-side (axis=1).
3. join: Shortcut for joining on index; cleaner than merge for index-aligned data.
4. Concept: Combining unifies multiple datasets into one — essential for analytics and ML feature building.""",
"Combining — Checkpoint Card"),

("Pandas — Aggregation & Grouping (G A A P → Giraffes Analyze All Patterns.))",
"""Definitions:
.groupby → Split data into groups based on keys.
.agg → Apply one or more aggregation functions to each group.
.apply → Run a custom function across groups.
.pivot_table → Compute summary tables with row/column grouping.""",
"""1️⃣ G A A P → Giant Ants Act Patiently 🐜⏳\n
2️⃣ groupby → Split into groups. 🍱\n
3️⃣ agg → Aggregate with multiple functions. 🧮\n
4️⃣ apply → Custom operations per group. 🛠️\n
5️⃣ pivot_table → Multi-dimensional summaries. 📊""",
"""1. groupby: Split data into groups by key(s) for grouped analysis.
2. agg: Apply multiple summary stats to columns (sum, mean, count).
3. apply: Run custom functions across grouped subsets.
4. pivot_table: Create multi-index summary tables by row/column categories.
5. Concept: Aggregation condenses raw data into meaningful summaries — core of analytics/reporting.""",
"Aggregation — Checkpoint Card"),

("Pandas — Reshaping (M P S → Mice Paint Statues.)",
"""Definitions:
melt → Convert wide DataFrames into long format.
pivot → Convert long DataFrames back into wide format using column values.
stack / unstack → Reshape MultiIndex levels into rows or columns.""",
"""1️⃣ M P S → Monkeys Paint Statues 🐒🎨🗿\n
2️⃣ melt → Long format. 📏\n
3️⃣ pivot → Wide format. 📐\n
4️⃣ stack/unstack → Rotate index levels. 🔁""",
"""1. melt: Convert wide tables into long format for ML-ready structure.
2. pivot: Reverse of melt — build wide tables from long format.
3. stack/unstack: Pivot index levels to columns and back.
4. Concept: Reshaping changes the layout of your DataFrame without altering the underlying data.""",
"Reshaping — Checkpoint Card"),

("Pandas — Datetime Tools (T D R → Turtles Dig Rivers.)",
"""Definitions:
to_datetime → Convert strings or numbers into datetime objects.
.dt accessor → Extract datetime components (year, month, day, hour).
.resample → Group time-series data by a regular frequency (daily, weekly, monthly).""",
"""1️⃣ T D R → Tigers Dance Rapidly 🐅💃⏱️\n
2️⃣ to_datetime → Convert strings to timestamps. 🕰️\n
3️⃣ .dt accessor → Extract year/month/day. 📅\n
4️⃣ resample → Resample time series. 📈""",
"""1. to_datetime: Convert raw text dates into pandas Timestamp objects.
2. .dt accessor: Extract components like year, month, weekday, hour.
3. resample: Group time series into new frequencies (daily, monthly).
4. Concept: These tools make time-based data analyzable and ML-ready.""",
"Datetime — Checkpoint Card"),

("Pandas — Descriptive Analytics (D V U N → Dragons Visit Unicorn Nations)",
"""Definitions:
.describe → Summary statistics for numeric or object columns.
.value_counts → Count frequency of unique values.
.unique → Show each unique value in a column.
.nunique → Count unique values in a column.""",
"""1️⃣ D V U N → Dragons Visit Unicorn Nations 🐉🦄🌍\n
2️⃣ describe → Summary stats. 📊\n
3️⃣ value_counts → Frequency counts. 🔢\n
4️⃣ unique → Distinct values. 🎯\n
5️⃣ nunique → Count unique values. 🔍""",
"""1. describe: Quick statistical summary of numerical columns.
2. value_counts: Shows how often each category appears.
3. unique: Shows all distinct entries in a column.
4. nunique: Number of unique values in a column.
5. Concept: These functions summarize and characterize your dataset fast.""",
"Descriptive Analytics — Checkpoint Card"),

]

numpy = [

# ==============================================================
# ---------- 1. ARRAY CREATION (Pareto: 6) ----------
# ==============================================================

("NumPy — Array Creation (Z A L F O E → Zebras Always Lift Five Orange Eggs.)",
"""Definitions:
np.zeros → Create an array filled with zeros.
np.ones → Create an array filled with ones.
np.arange → Create evenly spaced values (start, stop, step).
np.linspace → Create N evenly spaced values between two endpoints.
np.full → Create an array filled with a repeated value.
np.empty → Create an uninitialized array with allocated memory only.""",
"""1️⃣ Z A L F O E → Zebras Always Lift Five Orange Eggs 🦓🥚\n
2️⃣ zeros → Make an all-zero array. 0️⃣🧱\n
3️⃣ ones → Make an all-one array. 1️⃣🧱\n
4️⃣ arange → Range-like sequences. 🔢➡️📏\n
5️⃣ linspace → Even spacing between limits. ↔️📐\n
6️⃣ full → Fill array with a constant. 🧱🎨\n
7️⃣ empty → Allocate memory without filling. ⚠️📦""",
"""1. zeros: Allocates arrays initialized entirely to 0, useful for starting computation buffers.
2. ones: Same as zeros but filled with 1, often used for initial feature matrices.
3. arange: Creates arithmetic sequences like Python's range but returns an ndarray.
4. linspace: Perfect for generating fixed-size numeric grids for ML, plotting, simulations.
5. full: Used to create constant-valued matrices efficiently.
6. empty: Most low-level — allocates memory but doesn’t initialize (fast but unsafe).
7. Concept: These six cover 95 percent of array creation in ML, scientific computing, and transformations.""",
"Array Creation — Checkpoint Card"),


# ==============================================================
# ---------- 2. ARRAY INSPECTION & INFO (Pareto: 5) ----------
# ==============================================================

("NumPy — Array Inspection (S D S I N → Sharks Dive Slowly Into Night.)",
"""Definitions:
shape → Shows the dimensions of the array.
ndim → Number of array dimensions.
size → Total number of elements.
dtype → Data type of array elements.
itemsize → Memory size per element in bytes.""",
"""1️⃣ S D S I N → Small Ducks Sip Iced Nectar 🦆🥤\n
2️⃣ shape → Array dimensions. 📐\n
3️⃣ ndim → # of dimensions. 🌌\n
4️⃣ size → Total element count. 🔢\n
5️⃣ dtype → Data type. 🔠\n
6️⃣ itemsize → Bytes per element. 💾""",
"""1. shape: Returns (rows, columns, …) — crucial for debugging and ML pipelines.
2. ndim: Reveals how deep the array structure is (1D, 2D, 3D…).
3. size: Computes total element count, used in memory/perf modeling.
4. dtype: Shows exact type, essential for numerical stability.
5. itemsize: Helps measure memory footprint per element.
6. Concept: These five functions give you full understanding of an array's structure and memory footprint.""",
"Array Inspection — Checkpoint Card"),


# ==============================================================
# ---------- 3. INDEXING, SLICING, MASKING (Pareto: 5) ----------
# ==============================================================

("NumPy — Indexing & Slicing (S F B A M → Snakes Fly Between Angry Mountains.)",
"""Definitions:
[] slicing → Slice arrays by ranges (rows, cols).
boolean_mask → Filter elements using boolean conditions.
fancy_indexing → Index arrays with lists/arrays of indices.
np.where → Conditional selection returning indices.
np.take → Take elements by index along an axis.""",
"""1️⃣ S F B A M → Spiders Fry Bacon At Midnight 🕷️🥓🌙\n
2️⃣ [] slicing → Range slicing. ✂️📊\n
3️⃣ boolean_mask → Conditional filtering. 🎭\n
4️⃣ fancy_indexing → Index by arrays. 🌈🔍\n
5️⃣ where → Conditional index finder. ❓📍\n
6️⃣ take → Grab by index fast. 🎯""",
"""1. slicing: The backbone of array manipulation — extract subarrays with row/col ranges.
2. boolean_mask: Filters elements by logical conditions (x > 5).
3. fancy_indexing: Lets you reorder or specifically pick element positions.
4. where: Returns indices (or values) satisfying a condition — essential for ML thresholding.
5. take: Faster alternative to fancy indexing along a specific axis.
6. Concept: Indexing/slicing controls everything: selection, filtering, reordering — 80% of real NumPy work.""",
"Indexing — Checkpoint Card"),


# ==============================================================
# ---------- 4. MATH & VECTOR OPERATIONS (Pareto: 6) ----------
# ==============================================================

("NumPy — Math Operations (A M E L S D → Ants March Every Leaf South Daily.)",
"""Definitions:
np.sum → Sum values across axes.
np.mean → Average values across axes.
np.std → Standard deviation computation.
np.min / np.max → Find extrema along axes.
np.exp → Exponential elementwise.
np.log → Natural logarithm elementwise.""",
"""1️⃣ A M E L S D → Angry Monkeys Eat Lemons Slowly Daily 🙈🍋\n
2️⃣ sum → Total values. ➕\n
3️⃣ mean → Average. ⚖️\n
4️⃣ std → Spread/variation. 📉📈\n
5️⃣ min/max → Extrema. ⬇️⬆️\n
6️⃣ exp → Exponential. 🚀\n
7️⃣ log → Natural log. 🧮""",
"""1. sum: Used constantly in reduction operations, probability normalization.
2. mean: Basis of statistics, losses, model evaluation.
3. std: Measures spread — essential for normalization.
4. min/max: Often used for clipping, scaling, range finding.
5. exp: Foundation of softmax, logistic regression, and distributions.
6. log: Used for log-loss, entropy, stabilization.
7. Concept: These six functions represent the mathematical backbone of all ML and numerical methods.""",
"Math Ops — Checkpoint Card"),


# ==============================================================
# ---------- 5. LINEAR ALGEBRA (Pareto: 5) ----------
# ==============================================================

("NumPy — Linear Algebra (D M I S E → Dragons Multiply Inside Silent Emeralds.)",
"""Definitions:
np.dot → Dot product of vectors/matrices.
np.matmul → Matrix multiplication obeying broadcasting rules.
np.linalg.inv → Compute the inverse of a matrix.
np.linalg.solve → Solve linear systems Ax = b.
np.linalg.eig → Eigenvalues/eigenvectors of a square matrix.""",
"""1️⃣ D M I S E → Dolphins Make Ice Sculptures Elegantly 🐬❄️🗿\n
2️⃣ dot → Vector/matrix dot product. 🎯\n
3️⃣ matmul → Matrix multiply. 🧮\n
4️⃣ inv → Matrix inverse. 🔄\n
5️⃣ solve → Solve Ax=b. 📘\n
6️⃣ eig → Eigen decomposition. 💎""",
"""1. dot: Core operation of neural nets, geometry, projections.
2. matmul: Generalized matrix-multiply with broadcast semantics — recommended over dot.
3. inv: Compute inverse (rare in ML but useful in closed-form solutions).
4. solve: Numerically stable method to solve Ax=b without computing inverse.
5. eig: Foundation of PCA, spectral methods, decompositions.
6. Concept: NumPy linear algebra is the foundation for ML, PCA, optimization, and scientific computing.""",
"Linear Algebra — Checkpoint Card"),


# ==============================================================
# ---------- 6. RESHAPING & STRUCTURE (Pareto: 6) ----------
# ==============================================================

("NumPy — Reshaping (R T F R S C → Robots Take Flights Riding Silver Clouds.)",
"""Definitions:
reshape → Change array dimensionality without copying data.
ravel → Flatten array to 1D.
transpose → Swap axes (rows ↔ columns).
swapaxes → Swap any two axes.
expand_dims → Insert a new axis.
squeeze → Remove axes of size 1.""",
"""1️⃣ R T F R S C → Rabbits Travel Far Riding Soft Clouds 🐇☁️\n
2️⃣ reshape → New shape. 🔄\n
3️⃣ ravel → Flatten. 📏\n
4️⃣ transpose → Swap axes. 🔁\n
5️⃣ swapaxes → Exchange dims. ↔️\n
6️⃣ expand_dims → Add axis. ➕📐\n
7️⃣ squeeze → Remove 1-size axes. ➖📐""",
"""1. reshape: Changes dimensionality; crucial for ML batching.
2. ravel: Gives a flattened 1D view of the data.
3. transpose: Swap row/column orientation or rotate dims.
4. swapaxes: Lower-level control for axis reordering.
5. expand_dims: Add batch or channel dim for ML models.
6. squeeze: Clean up unnecessary dimensions (e.g., (1, 32, 32)).
7. Concept: Reshaping is how you prepare arrays for models, vectorization, and pipeline transformations.""",
"Reshaping — Checkpoint Card"),


# ==============================================================
# ---------- 7. BROADCASTING (Pareto: 3) ----------
# ==============================================================

("NumPy — Broadcasting (B R T → Bears Ride Tricycles.)",
"""Definitions:
scalar_broadcast → Automatically expand scalars across arrays.
vector_broadcast → Align vector along compatible dims.
np.broadcast_to → Explicitly broadcast an array to a new shape.""",
"""1️⃣ B R T → Bats Read Textbooks 🦇📚\n
2️⃣ scalar_broadcast → Scalar → array. 📈\n
3️⃣ vector_broadcast → Vector expansion. ↔️\n
4️⃣ broadcast_to → Manual broadcasting. 🪄""",
"""1. scalar_broadcast: Adding/multiplying scalars to arrays — automatic alignment.
2. vector_broadcast: Aligning 1D vectors to 2D/3D arrays.
3. broadcast_to: Explicit request to broadcast to a new shape.
4. Concept: Broadcasting replaces loops — this is where NumPy gets its speed.""",
"Broadcasting — Checkpoint Card"),


# ==============================================================
# ---------- 8. RANDOM GENERATION (Pareto: 5) ----------
# ==============================================================

("NumPy — Random (R U N N B → Rabbits Use Nightly Nature Buffets.)",
"""Definitions:
rand → Uniform random values in [0,1).
randn → Standard normal values (mean 0, std 1).
random.randint → Random integers in a range.
random.choice → Randomly select elements.
random.shuffle → Shuffle arrays in-place.""",
"""1️⃣ R U N N B → Raccoons Use Nuts Nightly Beforewinter 🦝🥜❄️\n
2️⃣ rand → Uniform sample. 📊\n
3️⃣ randn → Normal distribution. 📈\n
4️⃣ randint → Random ints. 🎲\n
5️⃣ choice → Random selection. 🎯\n
6️⃣ shuffle → Shuffle order. 🔀""",
"""1. rand: Core for simulations, noise, initialization.
2. randn: Gives Gaussian-distributed values — essential for ML weight init.
3. randint: Sample discrete values.
4. choice: Weighted/optional random selection from arrays.
5. shuffle: In-place permutation of data (common in ML train/test prep).
6. Concept: Randomness drives simulation, ML initialization, stochastic processes.""",
"Random — Checkpoint Card"),

]

matplotlib = [

# ==============================================================
# ---------- 1. BASIC PLOTTING (Pareto: 6) ----------
# ==============================================================

("Matplotlib — Basic Plotting (P S H B L T → Penguins Skate Happily Before Long Trips.)",
"""Definitions:
plt.plot → Draw line plots of x vs y.
plt.scatter → Draw scatter plots of individual points.
plt.bar → Draw bar charts for categorical comparisons.
plt.hist → Draw histograms of frequency distributions.
plt.boxplot → Draw box-and-whisker plots for statistical spread.
plt.title / plt.xlabel / plt.ylabel → Add titles and axis labels.""",
"""1️⃣ P S H B L T → Purple Snakes Hop Between Large Trees 🐍🌲\n
2️⃣ plot → Line chart. 📈\n
3️⃣ scatter → Scatter points. 🎯\n
4️⃣ bar → Bar chart. 📊\n
5️⃣ hist → Histogram. 📦\n
6️⃣ boxplot → Distribution spread. 📮\n
7️⃣ title/xlabel/ylabel → Annotate axes. 🏷️""",
"""1. plot: Core Matplotlib primitive — draws continuous lines for trends.
2. scatter: For showing relationships between paired variables.
3. bar: Great for category comparisons and frequency counts.
4. hist: Shows distribution shapes — bins, skew, peaks.
5. boxplot: Summarizes median, IQR, outliers.
6. title/xlabel/ylabel: Essential labeling for clarity.
7. Concept: These six commands represent 90 percent of beginner and intermediate plotting tasks.""",
"Basic Plotting — Checkpoint Card"),


# ==============================================================
# ---------- 2. FIGURE & AXES CONTROL (Pareto: 6) ----------
# ==============================================================

("Matplotlib — Figure & Axes (F A S T Q T → Frogs Always Sip Tea Quietly Together.)",
"""Definitions:
plt.figure → Create a new figure window.
plt.subplots → Create figure + one or multiple axes.
ax.set_title → Set title for a specific axes.
ax.set_xlim / set_ylim → Control axis ranges.
ax.set_xticks / set_yticks → Manually define tick locations.
fig.tight_layout → Automatically fix spacing of plots.""",
"""1️⃣ F A S T Q T → Foxes And Spiders Take Quiet Turns 🦊🕷️\n
2️⃣ figure → New canvas. 🖼️\n
3️⃣ subplots → Create grid of plots. 🧱\n
4️⃣ set_title → Per-axes title. 🏷️\n
5️⃣ set_xlim/ylim → Axis ranges. 🎚️\n
6️⃣ set_xticks/yticks → Tick control. 📏\n
7️⃣ tight_layout → Fix spacing. 🧩""",
"""1. figure: Starts a blank plotting canvas — controls size & DPI.
2. subplots: Most important function for multi-plot layouts.
3. set_title: Prefer ax-level methods over plt-level for clean OO style.
4. set_xlim/ylim: Force graph to show specific ranges.
5. set_xticks/yticks: Custom tick marks for category or numeric control.
6. tight_layout: Prevents overlap — a must for multi-charts.
7. Concept: Mastering figure/axes control = mastery of professional plotting.""",
"Figure & Axes — Checkpoint Card"),


# ==============================================================
# ---------- 3. STYLING & APPEARANCE (Pareto: 5) ----------
# ==============================================================

("Matplotlib — Styling (C M L G A → Cats Make Lovely Graph Accessories.)",
"""Definitions:
plt.style.use → Apply full style themes (ggplot, seaborn, classic).
ax.grid → Add gridlines for readability.
ax.legend → Display legends for labeled elements.
ax.set_facecolor → Change background color of plots.
ax.tick_params → Control tick appearance (size, direction, rotation).""",
"""1️⃣ C M L G A → Crabs Make Lemon Garlic Alfredo 🦀🍋🧄\n
2️⃣ style.use → Global theme. 🎨\n
3️⃣ grid → Gridlines. #️⃣\n
4️⃣ legend → Show labels. 🏷️\n
5️⃣ set_facecolor → Background color. 🌈\n
6️⃣ tick_params → Tick styling. 🔧""",
"""1. style.use: Switch between built-in styles for instantly better visuals.
2. grid: Helps visually align and interpret data.
3. legend: Activated when functions include label='...'.
4. set_facecolor: Used in dashboards to differentiate panels.
5. tick_params: Control rotation, sizes, directions — essential for categorical plots.
6. Concept: Styling elevates raw plots to publication-grade graphics.""",
"Styling — Checkpoint Card"),


# ==============================================================
# ---------- 4. ANNOTATION & TEXT (Pareto: 4) ----------
# ==============================================================

("Matplotlib — Annotation (A T A V → Ants Tap Apples Vigorously.)",
"""Definitions:
plt.text → Add text at a coordinate in the plot.
plt.annotate → Add annotation arrows or labels to highlight points.
ax.axhline / ax.axvline → Add horizontal/vertical reference lines.
ax.axhspan / ax.axvspan → Add shaded regions.""",
"""1️⃣ A T A V → Angry Tigers Attack Volcanos 🐅🌋\n
2️⃣ text → Place text. 📝\n
3️⃣ annotate → Arrow + label. 🎯➡️💬\n
4️⃣ axhline/axvline → Reference lines. 📏\n
5️⃣ axhspan/axvspan → Shaded regions. 🌫️""",
"""1. text: Used for labels, explanations, numeric highlights.
2. annotate: Best for calling out specific points in data storytelling.
3. axhline/axvline: Useful for thresholds and medians.
4. axhspan/axvspan: Highlight ranges (confidence intervals, events).
5. Concept: Annotation transforms data into a narrative — required for business and research plots.""",
"Annotation — Checkpoint Card"),


# ==============================================================
# ---------- 5. IMAGES & COLOR MAPS (Pareto: 4) ----------
# ==============================================================

("Matplotlib — Images (I I C P → Iguanas Inspect Colorful Paintings.)",
"""Definitions:
plt.imshow → Display 2D image data.
plt.colorbar → Add a color scale legend.
plt.cm → Access built-in colormaps.
plt.clim → Change color scale range.""",
"""1️⃣ I I C P → Insects In Cities Paint 🎨🐞\n
2️⃣ imshow → Show 2D image arrays. 🖼️\n
3️⃣ colorbar → Add legend scale. 📏\n
4️⃣ cm → Colormap. 🌈\n
5️⃣ clim → Adjust color scale. 🔥❄️""",
"""1. imshow: Core for heatmaps, images, activation maps, correlation matrices.
2. colorbar: Gives numerical interpretation of color intensity.
3. cm: Dozens of maps like viridis, plasma, inferno.
4. clim: Controls min/max mapping to color.
5. Concept: Essential for ML (feature maps), CV, scientific imaging.""",
"Images — Checkpoint Card"),


# ==============================================================
# ---------- 6. SUBPLOTS & LAYOUTS (Pareto: 6) ----------
# ==============================================================

("Matplotlib — Subplots (G S A T S C → Gorillas Stack Apples Then Share Cookies.)",
"""Definitions:
plt.subplots → Create a grid of subplots.
fig.add_subplot → Add subplot manually.
ax.sharex / sharey → Link axes of subplots.
fig.suptitle → Title for the entire figure.
plt.subplot2grid → Low-level arranged grids.
fig.subplots_adjust → Manually tune spacing.""",
"""1️⃣ G S A T S C → Giant Squids Attack Tiny Seahorses Constantly 🦑🐎\n
2️⃣ subplots → Grid of axes. 🧱\n
3️⃣ add_subplot → Manual subplot. 🛠️\n
4️⃣ sharex/sharey → Shared axes. 🔗\n
5️⃣ suptitle → Global title. 🏷️\n
6️⃣ subplot2grid → Complex layout. 🗺️\n
7️⃣ subplots_adjust → Manual spacing. 🔧""",
"""1. subplots: One of the most important commands — creates rows/cols of plots.
2. add_subplot: Adds more axes on the fly to a figure.
3. sharex/sharey: Ensures consistent scales across a dashboard.
4. suptitle: Title for entire figure (e.g., multi-chart reports).
5. subplot2grid: More control for asymmetric layouts.
6. subplots_adjust: Fine-tuning spacing when tight_layout isn’t enough.
7. Concept: Subplots enable dashboards, multi-view analysis, and research figures.""",
"Subplots — Checkpoint Card"),


# ==============================================================
# ---------- 7. SAVING & EXPORTING (Pareto: 4) ----------
# ==============================================================

("Matplotlib — Saving (S S P T → Snakes Sell Paintings To Turtles.)",
"""Definitions:
plt.savefig → Save the current figure to disk.
dpi → Resolution parameter for exports.
transparent=True → Export with transparent background.
bbox_inches='tight' → Trim whitespace around saved plots.""",
"""1️⃣ S S P T → Seals Sell Pretty Tulips 🦭🌷\n
2️⃣ savefig → Save figure. 💾\n
3️⃣ dpi → Resolution. 📏\n
4️⃣ transparent → No background. 🪟\n
5️⃣ bbox_inches → Trim margins. ✂️""",
"""1. savefig: Essential for exporting figures for slides, reports, papers.
2. dpi: Controls raster resolution — higher for publication.
3. transparent: Allows overlay on different backgrounds.
4. bbox_inches: Removes empty padding for a clean output.
5. Concept: Professional visualization requires proper exporting settings.""",
"Saving — Checkpoint Card"),

]

sklearn = [

# ==============================================================
# ---------- 1. TRAIN/TEST SPLITTING + MODEL SELECTION ----------
# ==============================================================

("sklearn — Splitting & Model Selection (T S K G R → Tigers Slice Kiwi Gently, Regularly.)",
"""Definitions:
train_test_split → Split arrays or DataFrames into train/test subsets.
KFold → Cross-validation splitter that partitions into K folds.
StratifiedKFold → KFold preserving class distribution.
GridSearchCV → Exhaustive hyperparameter search.
RandomizedSearchCV → Random hyperparameter sampling.""",
"""1️⃣ T S K G R → Tiny Spiders Knit Green Rugs 🕷️🧶\n
2️⃣ train_test_split → Train/test partitions. ✂️📊\n
3️⃣ KFold → K-way CV. 🔁\n
4️⃣ StratifiedKFold → Balanced CV. ⚖️\n
5️⃣ GridSearchCV → Full grid search. 🔍\n
6️⃣ RandomizedSearchCV → Random sampling search. 🎲""",
"""1. train_test_split: Core to evaluation — prevents leakage and overfitting.
2. KFold: Repeatedly splits data for more reliable evaluation.
3. StratifiedKFold: Preserves class ratio — critical for classification.
4. GridSearchCV: Tests all hyperparameters, best for small grids.
5. RandomizedSearchCV: Faster search across large spaces.
6. Concept: This category controls **how you validate models**, the backbone of reliable ML.""",
"Splitting — Checkpoint Card"),


# ==============================================================
# ---------- 2. PREPROCESSING (TRANSFORMERS) ----------
# ==============================================================

("sklearn — Preprocessing (S M R O B → Sharks Make Really Organized Bubbles.)",
"""Definitions:
StandardScaler → Scale data to mean=0, std=1.
MinMaxScaler → Scale data to a 0–1 range.
RobustScaler → Scale data using median/IQR (outlier-safe).
OneHotEncoder → Convert categorical variables to binary columns.
OrdinalEncoder → Convert categories to integer ranks.""",
"""1️⃣ S M R O B → Smart Monkeys Ride On Boats 🐒🛶\n
2️⃣ StandardScaler → Z-scores. ⚖️\n
3️⃣ MinMaxScaler → Normalize to 0–1. 🔢\n
4️⃣ RobustScaler → Outlier-safe scaling. 🛡️\n
5️⃣ OneHotEncoder → Binary category columns. 🔲🔲\n
6️⃣ OrdinalEncoder → Numeric ordinal labels. 🔢⬆️""",
"""1. StandardScaler: Most used — essential for distance-based models (SVM, KNN).
2. MinMaxScaler: For neural nets and when bounded ranges help.
3. RobustScaler: Best when heavy outliers exist.
4. OneHotEncoder: Turns categories into binary features.
5. OrdinalEncoder: Assigns integer labels (not suitable for unordered categories).
6. Concept: Preprocessing is how you **prepare raw data** for effective modeling.""",
"Preprocessing — Checkpoint Card"),


# ==============================================================
# ---------- 3. PIPELINES (THE CORE OF SKLEARN) ----------
# ==============================================================

("sklearn — Pipelines (P M F → Penguins Make Fire.)",
"""Definitions:
Pipeline → Chain preprocessing + model steps together.
ColumnTransformer → Apply different preprocessing to different columns.
FunctionTransformer → Wrap arbitrary Python functions as transformers.""",
"""1️⃣ P M F → Parrots Make Fudge 🦜🍫\n
2️⃣ Pipeline → End-to-end workflow. 🔗\n
3️⃣ ColumnTransformer → Per-column transforms. 📦\n
4️⃣ FunctionTransformer → Custom functions. 🛠️""",
"""1. Pipeline: Prevents leakage; serializes full workflows.
2. ColumnTransformer: One of the most powerful tools in sklearn; column-wise pipelines.
3. FunctionTransformer: Lets you wrap lambda/custom functions into sklearn API.
4. Concept: Pipelines convert sklearn from a loose toolbox into a **production-ready ML framework**.""",
"Pipelines — Checkpoint Card"),


# ==============================================================
# ---------- 4. FEATURE EXTRACTION / VECTORIZATION ----------
# ==============================================================

("sklearn — Vectorization (C T H → Cats Type Honestly.)",
"""Definitions:
CountVectorizer → Convert raw text into token count vectors.
TfidfVectorizer → Convert text into TF-IDF weighted vectors.
HashingVectorizer → One-way hashed feature vectors (no vocabulary).""",
"""1️⃣ C T H → Crows Teach Humans 🐦👨‍🏫\n
2️⃣ CountVectorizer → Bag-of-words counts. 🧮\n
3️⃣ TfidfVectorizer → TF-IDF weights. 🧠\n
4️⃣ HashingVectorizer → Fast hashing. ⚡""",
"""1. CountVectorizer: Foundation of classical NLP (bag-of-words).
2. TfidfVectorizer: Better weighting — kills stopword dominance.
3. HashingVectorizer: No vocabulary stored; great for massive text.
4. Concept: Vectorizers **translate text → numeric tensors** for ML.""",
"Vectorization — Checkpoint Card"),


# ==============================================================
# ---------- 5. MODELS (CLASSIFICATION) ----------
# ==============================================================

("sklearn — Classification Models (L K S N R → Lions Keep Saving New Rhinos.)",
"""Definitions:
LogisticRegression → Linear classifier with regularization.
KNeighborsClassifier → Distance-based classifier.
SVC (Support Vector Classifier) → Margin-based classifier.
NaiveBayes → Probabilistic classifier for text and count data.
RandomForestClassifier → Tree-based ensemble classifier.""",
"""1️⃣ L K S N R → Lemurs Keep Sipping Nectar Regularly 🐒🍯\n
2️⃣ LogisticRegression → Linear boundary. 📈\n
3️⃣ KNN → Neighborhood voting. 🗳️\n
4️⃣ SVC → Max-margin classifier. 📏\n
5️⃣ Naive Bayes → Probabilistic model. 🎲\n
6️⃣ RandomForest → Tree ensemble. 🌲🌲🌲""",
"""1. LogisticRegression: Go-to baseline classifier; works well with scaling.
2. KNN: Simple nonlinear boundaries via neighborhoods.
3. SVC: Powerful for high-dimensional data; needs scaling.
4. Naive Bayes: Perfect for text and count-based features.
5. RandomForest: Extremely strong for tabular data.
6. Concept: These 5 give you expert coverage of sklearn classification.""",
"Classification — Checkpoint Card"),


# ==============================================================
# ---------- 6. MODELS (REGRESSION) ----------
# ==============================================================

("sklearn — Regression Models (L R E S → Lobsters Ride Electric Scooters.)",
"""Definitions:
LinearRegression → Ordinary least squares regression.
Ridge → L2-penalized regression.
Lasso → L1-penalized regression for sparsity.
RandomForestRegressor → Tree-based ensemble for regression.""",
"""1️⃣ L R E S → Lions Race Every Sunday 🦁🏁\n
2️⃣ LinearRegression → Basic linear fit. 📉\n
3️⃣ Ridge → L2 regularized. 🔗\n
4️⃣ Lasso → L1 sparse model. ✂️\n
5️⃣ RandomForestRegressor → Tree ensemble. 🌳""",
"""1. LinearRegression: Simple, interpretable baseline.
2. Ridge: Stabilizes coefficients when features correlate.
3. Lasso: Produces sparse solutions (feature selection).
4. RandomForestRegressor: Best general-purpose model for tabular numeric data.
5. Concept: These four solve 80% of real regression tasks in ML.""",
"Regression — Checkpoint Card"),


# ==============================================================
# ---------- 7. METRICS (EVALUATION) ----------
# ==============================================================

("sklearn — Metrics (A P R F C → Ants Paint Rainbows For Children.)",
"""Definitions:
accuracy_score → Classification accuracy.
precision_score → True positive precision.
recall_score → Sensitivity / true positive rate.
f1_score → Harmonic mean of precision and recall.
confusion_matrix → 2x2 or NxN class comparison table.""",
"""1️⃣ A P R F C → Apes Play Rugby For Cash 🦍🏉\n
2️⃣ accuracy → Correct predictions. ✔️\n
3️⃣ precision → TP / predicted positive. 🎯\n
4️⃣ recall → TP / actual positive. 📢\n
5️⃣ f1 → Precision + recall harmonic. ⚖️\n
6️⃣ confusion_matrix → Class grid. 🗺️""",
"""1. Accuracy: OK for balanced datasets.
2. Precision: "How many selected were correct?"
3. Recall: "How many correct were found?"
4. F1: Best for imbalanced datasets.
5. Confusion matrix: Breaks performance into TP/FP/TN/FN.
6. Concept: Evaluation metrics determine **how good** your model truly is.""",
"Metrics — Checkpoint Card"),


# ==============================================================
# ---------- 8. DIMENSIONALITY REDUCTION ----------
# ==============================================================

("sklearn — Dimensionality Reduction (P T S → Penguins Take Subways.)",
"""Definitions:
PCA → Principal Component Analysis for variance-maximizing components.
TruncatedSVD → SVD for sparse/textual matrices.
SelectKBest → Select top-k features based on scoring function.""",
"""1️⃣ P T S → Pandas Teach Science 🐼📘\n
2️⃣ PCA → Variance maximization. 📉\n
3️⃣ TruncatedSVD → Sparse SVD. 🌌\n
4️⃣ SelectKBest → Top-k features. 🏆""",
"""1. PCA: Reduces features by projecting into components maximizing variance.
2. TruncatedSVD: PCA-like but works on sparse inputs (TF-IDF).
3. SelectKBest: Simple way to reduce dimensionality using statistical scores.
4. Concept: Dimensionality reduction improves speed, interpretability, and generalization.""",
"Dimensionality Reduction — Checkpoint Card"),

]




# ---------------------- Topic registry ----------------------
# List your deck variable names here
topic_vars = [
    "programming_basics",
    "variables",
    "functions",
    "data_structures",
    "logic",
    "ml_week1",
    "mit_algorithms_101",
    "ml_lifecycle_week1",
    "data_cleaning",
    "Ml_concepts",
    "linear_algebra",
    "trigonometry_mastery",
    "ml_chapter1",
    "ml_chapter2",
    "ml_chapter3",
    "pandas",
    "numpy",
    "matplotlib",
    "sklearn"

    # add more like: "flashcards_data_structures", "flashcards_algorithms"
]
````

````python
# @title
# === Flashcard Viewer Pro (Academic + LaTeX + Image Edition) ===
# Supports: Question / Answer / Mnemonic / Definition / Image
# Works in Google Colab & Jupyter
# Keyboard: ← → space k u d
# Each card shows a dedicated image block (even if empty)

import ipywidgets as widgets
from IPython.display import display, HTML, clear_output, Javascript
from html import escape
import random, time, json, markdown

# ------------------------------------------------------------
# Global CSS + MathJax setup
# ------------------------------------------------------------
display(HTML("""
<style>
.output_subarea pre, .output_subarea code, pre, code {
  white-space: pre-wrap !important;
  word-break: break-word !important;
  overflow-wrap: anywhere !important;
}
.output_wrapper, .output, .jp-OutputArea-output { max-width: 100% !important; }

.timer-overlay {
  position: fixed;
  top: 10px;
  right: 20px;
  background: rgba(255,255,255,0.9);
  border: 1px solid #ccc;
  border-radius: 8px;
  padding: 6px 10px;
  font-family: sans-serif;
  font-size: 14px;
  color: #333;
  z-index: 9999;
}
.progress-container {
  width: 100%;
  background: #f1f1f1;
  border-radius: 8px;
  overflow: hidden;
  margin: 6px 0 8px 0;
  height: 10px;
}
.progress-bar {
  height: 10px;
  background-color: #0072ce;
  width: 0%;
  transition: width 0.3s;
}
.card-box {
  background:#fafafa;
  padding:12px;
  border-radius:10px;
  border:1px solid #e5e5e5;
  margin-bottom:8px;
  font-family:sans-serif;
  font-size:15px;
  line-height:1.4;
}
.answer-box {
  font-family:sans-serif;
  font-size:14px;
  margin:6px 0;
  background:#0b1020;
  color:#e6edf3;
  padding:10px;
  border-radius:10px;
  border:1px solid #1f2638;
  white-space:pre-wrap;
}
.mnemonic-box {
  font-family:sans-serif;
  font-size:14px;
  margin-top:6px;
}
.card-image {
  display: inline-block;
  max-width: 90%;
  height: auto;
  border-radius: 10px;
  margin-top: 8px;
  box-shadow: 0 0 8px rgba(0,0,0,0.15);
}
</style>

<!-- MathJax -->
<script src="https://polyfill.io/v3/polyfill.min.js?features=es6"></script>
<script id="MathJax-script" async
src="https://cdn.jsdelivr.net/npm/mathjax@3/es5/tex-mml-chtml.js"></script>
"""))

# ------------------------------------------------------------
# Timer overlay setup
# ------------------------------------------------------------
display(HTML("<div id='timer-overlay' class='timer-overlay' style='display:none;'>⏱️ 00:00</div>"))

display(Javascript("""
if (!window.flashcardTimerSetup) {
  window.flashcardTimerSetup = true;
  window.timerInterval = null;

  window.restartOverlayTimer = function(){
    clearInterval(window.timerInterval);
    document.getElementById('timer-overlay').style.display = 'block';
    let startTime = Date.now();
    window.timerInterval = setInterval(()=>{
      let elapsed = Math.floor((Date.now() - startTime)/1000);
      let m = Math.floor(elapsed/60), s = elapsed % 60;
      document.getElementById('timer-overlay').textContent =
        `⏱️ ${String(m).padStart(2,'0')}:${String(s).padStart(2,'0')}`;
    },1000);
  }

  window.resetOverlayTimer = function(){
    clearInterval(window.timerInterval);
    document.getElementById('timer-overlay').textContent = '⏱️ 00:00';
    document.getElementById('timer-overlay').style.display = 'none';
  }
}
"""))

# ------------------------------------------------------------
# Example deck loading
# ------------------------------------------------------------
topic_vars = [
    "programming_basics", "variables", "functions",
    "data_structures", "logic", "ml_week1", "mit_algorithms_101",
    "ml_lifecycle_week1", "data_cleaning", "Ml_concepts",
    "linear_algebra", "trigonometry_mastery", "ml_chapter1", "ml_chapter2",
    "ml_chapter3", "pandas", "numpy", "matplotlib","sklearn"
]
flashcards_all = []
for name in topic_vars:
    deck = globals().get(name, [])
    flashcards_all.extend(deck)
topic_vars = ["flashcards_all"] + topic_vars
globals()["flashcards_all"] = flashcards_all

# ------------------------------------------------------------
# State variables
# ------------------------------------------------------------
current_deck_name = topic_vars[0]
flashcards = globals()[current_deck_name]
index = 0
showing_answer = False
showing_definition = False
known_cards, unknown_cards = set(), set()
review_mode = False
output = widgets.Output(layout=widgets.Layout(width='100%'))

# ------------------------------------------------------------
# Markdown + LaTeX renderer
# ------------------------------------------------------------
def render_markdown_latex(text):
    if not text:
        return ""
    html = markdown.markdown(
        text,
        extensions=['fenced_code','tables','nl2br','sane_lists','md_in_html'],
        output_format='html5'
    )
    return f"<div style='line-height:1.5; white-space:normal'>{html}</div>"

# ------------------------------------------------------------
# Main card renderer (now with dedicated image block)
# ------------------------------------------------------------
def render_card(q_text, a_text="", m_text="", d_text="", img_path=""):
    q_html = render_markdown_latex(q_text)

    # --- Question ---
    display(HTML(f"""
    <div class='card-box'>
      <b>Q:</b><br>{q_html}
    </div>
    """))

    # --- Image block (always visible) ---
    if img_path:
        img_src = img_path if img_path.startswith("http") else f"files/{escape(img_path)}"
        display(HTML(f"""
        <div class='card-box' style='text-align:center; background:#f9f9f9;'>
          <b>📷 Image:</b><br>
          <img src='{img_src}' class='card-image' alt='Flashcard image'>
        </div>
        """))
    else:
        display(HTML("""
        <div class='card-box' style='text-align:center; color:#777;'>
          📷 No image for this card.
        </div>
        """))

    # --- Answer / Mnemonic / Definition ---
    if showing_answer:
        a_html = render_markdown_latex(a_text)
        display(HTML(f"<div><b>A:</b><div class='answer-box'>{a_html}</div></div>"))

        if m_text:
            m_html = render_markdown_latex(m_text)
            display(HTML(f"<div class='mnemonic-box'><b>Mnemonic:</b><br>{m_html}</div>"))

    if showing_definition:
        d_html = render_markdown_latex(d_text)
        display(HTML(f"""
        <div class='answer-box' style='background:#102030;'>
          <b>Definition / Docs:</b><br>{d_html}
        </div>
        """))

    display(HTML("<script>MathJax.typesetPromise();</script>"))

# ------------------------------------------------------------
# Progress and update functions
# ------------------------------------------------------------
def update_progress():
    if flashcards:
        percent = int((index + 1) / len(flashcards) * 100)
        progress_bar_html.value = (
            f'<div class="progress-container">'
            f'<div class="progress-bar" style="width:{percent}%"></div>'
            f'</div>'
        )

def update_card():
    with output:
        clear_output(wait=True)
        if not flashcards:
            render_card("Welcome", "No cards loaded.", "")
            return
        card = flashcards[index]
        question, answer, mnemonic, definition, img_path = (card + ("",)*(5-len(card)))[:5]
        render_card(question, answer, mnemonic, definition, img_path)
        counter_label.value = f"Topic: {current_deck_name} — Card {index+1}/{len(flashcards)}"
        update_progress()

# ------------------------------------------------------------
# Button handlers
# ------------------------------------------------------------
def on_topic_change(change):
    global current_deck_name, flashcards, index, showing_answer, showing_definition
    if change["name"] == "value" and change["new"] != change["old"]:
        current_deck_name = change["new"]
        flashcards = globals()[current_deck_name]
        index = 0
        showing_answer = False
        showing_definition = False
        display(Javascript("resetOverlayTimer();"))
        update_card()

def show_toggle(_=None):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def toggle_definition(_=None):
    global showing_definition
    showing_definition = not showing_definition
    update_card()

def next_card(_=None):
    global index, showing_answer, showing_definition
    if index < len(flashcards) - 1:
        index += 1
    showing_answer = False
    showing_definition = False
    display(Javascript("restartOverlayTimer();"))
    update_card()

def prev_card(_=None):
    global index, showing_answer, showing_definition
    if index > 0:
        index -= 1
    showing_answer = False
    showing_definition = False
    display(Javascript("restartOverlayTimer();"))
    update_card()

def shuffle_deck(_=None):
    global flashcards, index, showing_answer, showing_definition
    random.shuffle(flashcards)
    index = 0
    showing_answer = False
    showing_definition = False
    display(Javascript("restartOverlayTimer();"))
    update_card()

def mark_known(_=None):
    known_cards.add(index)
    unknown_cards.discard(index)
    next_card()

def mark_unknown(_=None):
    unknown_cards.add(index)
    known_cards.discard(index)
    next_card()

def toggle_review(change):
    global flashcards, index, showing_answer, showing_definition, review_mode
    review_mode = change["new"]
    if review_mode:
        filtered = [flashcards[i] for i in range(len(flashcards)) if i in unknown_cards]
        flashcards[:] = filtered or flashcards
    else:
        flashcards[:] = globals()[current_deck_name]
    index = 0
    showing_answer = False
    showing_definition = False
    display(Javascript("restartOverlayTimer();"))
    update_card()

def jump_to_card(_=None):
    global index, showing_answer, showing_definition
    val = jump_input.value - 1
    if 0 <= val < len(flashcards):
        index = val
        showing_answer = False
        showing_definition = False
        display(Javascript("restartOverlayTimer();"))
        update_card()

# ------------------------------------------------------------
# UI widgets
# ------------------------------------------------------------
topic_dropdown = widgets.Dropdown(options=topic_vars, value=current_deck_name, description="Topic:")
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
definition_button = widgets.Button(description="Show Definition 📘")
next_button = widgets.Button(description="Next ▶️")
shuffle_button = widgets.Button(description="Shuffle 🔀")
known_button = widgets.Button(description="Mark Known ✅")
unknown_button = widgets.Button(description="Mark Unknown ❌")
review_toggle = widgets.ToggleButton(value=False, description="Review Unknown 🔁", icon="refresh")
jump_input = widgets.BoundedIntText(value=1, min=1, description='Jump:')
jump_button = widgets.Button(description="Go ⏩")
counter_label = widgets.Label()
progress_bar_html = widgets.HTML('<div class="progress-container"><div class="progress-bar" id="prog"></div></div>')

# ------------------------------------------------------------
# Keyboard shortcuts (Colab only)
# ------------------------------------------------------------
js = """
<script>
document.addEventListener('keydown', function(e){
  if (e.target.tagName === 'INPUT' || e.target.tagName === 'TEXTAREA') return;
  const buttons = {
    'ArrowLeft': 'prev', 'ArrowRight': 'next',
    ' ': 'show', 'k': 'known', 'u': 'unknown', 'd': 'definition'
  };
  const id = buttons[e.key];
  if (!id) return;
  google.colab.kernel.invokeFunction('flashcard_keypress', [id], {});
});
</script>
"""
display(HTML(js))

from google.colab import output as colab_output
def handle_keypress(key):
    mapping = {
        "prev": prev_card,
        "next": next_card,
        "show": show_toggle,
        "known": mark_known,
        "unknown": mark_unknown,
        "definition": toggle_definition
    }
    if key in mapping:
        mapping[key]()
colab_output.register_callback('flashcard_keypress', handle_keypress)

# ------------------------------------------------------------
# Wire up buttons + launch viewer
# ------------------------------------------------------------
topic_dropdown.observe(on_topic_change, names="value")
show_button.on_click(show_toggle)
definition_button.on_click(toggle_definition)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
shuffle_button.on_click(shuffle_deck)
known_button.on_click(mark_known)
unknown_button.on_click(mark_unknown)
review_toggle.observe(toggle_review, names="value")
jump_button.on_click(jump_to_card)

update_card()
display(widgets.VBox([
    widgets.HBox([topic_dropdown]),
    progress_bar_html,
    widgets.HBox([prev_button, show_button, definition_button, next_button, shuffle_button]),
    widgets.HBox([known_button, unknown_button, review_toggle]),
    widgets.HBox([counter_label]),
    widgets.HBox([jump_input, jump_button]),
    output
]))
````
