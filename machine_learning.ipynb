## Supervised Learning

### Definition
A machine learning method where a model learns from labeled examples. Each example has inputs (features, X) and a known correct output (label/target, y). The model learns a function f(X) → y that should generalize (work well on new data, not just training data).

### Core theory (what matters)
- Training signal: the labels provide the “answer key.”
- Objective: minimize an error (loss) on training data while avoiding overfitting (memorizing).
- Two main types:
  - Classification (categorical y)
  - Regression (numeric y)

### Data shapes in scikit-learn
- X must be 2D: (n_samples, n_features)
- y usually 1D: (n_samples,)


```python
# Supervised learning = (X, y)

X = [[1], [2], [3], [4]] # 4 samples, 1 feature
y = [0,0,1,1] #4 labels (binary classes)

print("X shape-like:", (len(X), len(X[0])))
print("y length:", len(y))
```

## Classification

### Definition
A type of supervised learning where the model predicts a **class (discrete label)**, such as `0 / 1` or `"spam" / "not spam"`.

### Core Theory
- Output is **categorical**.
- Many classifiers can output:
  - `predict()` → hard class label
  - `predict_proba()` → probabilities for each class (if supported)
- Decision boundaries can be:
  - **Linear** (e.g., logistic regression)
  - **Non-linear** (e.g., decision trees)



```python
from sklearn.tree import DecisionTreeClassifier

#A classifier learns to map X -> class label
clf = DecisionTreeClassifier(random_state = 0)

#Train on a simple pattern
X = [[1], [2], [3], [4]]
y = [0, 0, 1, 1]
clf.fit(X, y)

print("Pred([1]) =", clf.predict([[1]])[0])
print("Pred([4]) =", clf.predict([[4]])[0])
```
