# 🦀 Rust from Absolute Zero — Beginner Path

## 🔰 Lesson 0: What Is Rust?

Rust is:

- A systems programming language made by Mozilla  
- Known for **speed**, **memory safety**, and **no runtime crashes**  
- Used for:
  - Operating systems  
  - Embedded systems  
  - Games  
  - AI and WebAssembly  

---

## 🧪 Step 1: Your First Program

### ✅ Code:

```rust
fn main() {
    println!("Hello, world!");
}
```

### 🧠 Breakdown:

- `fn` → defines a function  
- `main()` → the entry point for every Rust program  
- `{}` → block of code  
- `println!()` → prints to screen  
- `!` → it’s a **macro**, not a function

### 🛠 Try It Online:
[play.rust-lang.org](https://play.rust-lang.org)

---

## 🔤 Step 2: Variables

Rust variables are **immutable by default**.

```rust
fn main() {
    let name = "Charles";
    let age = 30;

    println!("Name: {}, Age: {}", name, age);
}
```

- `let` → creates a variable  
- `{}` → placeholders in print string

---

## 🔁 Step 3: Mutable Variables

Use `mut` to make a variable changeable:

```rust
fn main() {
    let mut score = 0;
    score = score + 10;
    println!("Score: {}", score);
}
```

---

## 📊 Step 4: Numbers and Math

Basic arithmetic:

```rust
fn main() {
    let a = 5;
    let b = 2;
    let sum = a + b;
    println!("Sum: {}", sum);
}
```

Operators: `+`, `-`, `*`, `/`, `%`

---

## ❓ Step 5: Conditions

Use `if`, `else`, and logical comparisons:

```rust
fn main() {
    let age = 20;

    if age >= 18 {
        println!("You're an adult.");
    } else {
        println!("You're a minor.");
    }
}
```

Comparators: `==`, `!=`, `<`, `>`, `>=`, `<=`

---

## 🔄 Step 6: Loops

Use `for` with ranges:

```rust
fn main() {
    for i in 1..4 {
        println!("Count: {}", i);
    }
}
```

- `1..4` → ranges from 1 to 3 (end is exclusive)

# 🦀 Rust Beginner Guide — Part 2 (Functions, Types, Pattern Matching)

---

## 🧱 Step 7: Functions

Functions let you group code into reusable blocks.

### ✅ Code

```rust
fn greet(name: &str) {
    println!("Hello, {}!", name);
}

fn main() {
    greet("Charles");
}
```

### 🧠 Breakdown

- `fn` → defines a function  
- `greet(name: &str)` → takes a borrowed string (`&str`)  
- `"Charles"` is passed as an argument  
- `{}` → block of code (function body)

---

## 🧠 Step 8: Return Values

Functions can return values using the `->` syntax.

### ✅ Code

```rust
fn add(a: i32, b: i32) -> i32 {
    a + b  // no semicolon → return
}

fn main() {
    let result = add(5, 3);
    println!("Result: {}", result);
}
```

### 🧠 Rules

- `-> i32` → return type is 32-bit integer  
- No semicolon → returns the value implicitly  
- Rust is **statically typed** — types must match

---

## 🎯 Step 9: Data Types (Primitives)

Rust built-in types:

| Type   | Meaning            | Example               |
|--------|---------------------|------------------------|
| `i32`  | Signed integer      | `let x: i32 = 10;`     |
| `f64`  | Floating-point      | `let x = 3.14;`        |
| `bool` | Boolean true/false | `let done = true;`     |
| `char` | Single character    | `let letter = 'A';`    |
| `&str` | String slice        | `let name = "Rust";`   |

---

## 📦 Step 10: Tuples

Tuples group multiple values, even of different types.

### ✅ Code

```rust
fn main() {
    let person = ("Charles", 36, true);
    println!("Name: {}", person.0);    // index 0
    println!("Age: {}", person.1);     // index 1
    println!("Active: {}", person.2);  // index 2
}
```

### 🧠 Notes

- Tuples are indexed by `.0`, `.1`, etc.  
- Example tuple type: `(&str, i32, bool)`

---

## 🔃 Step 11: Match (Pattern Matching)

Rust’s `match` is like a powerful `switch`.

### ✅ Code

```rust
fn main() {
    let number = 2;

    match number {
        1 => println!("One"),
        2 => println!("Two"),
        3 => println!("Three"),
        _ => println!("Something else"), // wildcard (default)
    }
}
```
