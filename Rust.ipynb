````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output, HTML

# (question, answer, PEMDAS-style mnemonic)

flashcards = [

("🦀 What is Rust?",
"🧠 Concept:\n"
"Rust is a systems programming language known for safety, speed, and zero runtime crashes. Created by Mozilla, used in OS dev, embedded, games, AI, and WebAssembly."),

("💡 What does `fn main()` mean in Rust?",
"🧠 Concept:\n"
"Defines the main function — the entry point of every Rust program.\n\n"
"✅ Code:\n"
"    fn main() {\n"
"        println!(\"Hello, world!\");\n"
"    }"),

("🧠 What does `println!()` do?",
"🧠 Concept:\n"
"It prints to the screen. The `!` means it's a **macro**, not a regular function."),

("🧠 What is `let` used for?",
"🧠 Concept:\n"
"Declares a variable in Rust (immutable by default).\n\n"
"✅ Code:\n"
"    let name = \"Charles\";"),

("🧠 How do you make a variable mutable?",
"🧠 Concept:\n"
"Use `mut` to make it changeable.\n\n"
"✅ Code:\n"
"    let mut score = 0;\n"
"    score = score + 10;"),

("📊 How do you do math in Rust?",
"🧠 Concept:\n"
"Use `+`, `-`, `*`, `/`, `%` operators like other languages.\n\n"
"✅ Code:\n"
"    let sum = 5 + 2;\n"
"    println!(\"{}\", sum);"),

("❓ How does an if-statement look?",
"🧠 Concept:\n"
"Rust uses `if` / `else` without parentheses.\n\n"
"✅ Code:\n"
"    if age >= 18 {\n"
"        println!(\"Adult\");\n"
"    } else {\n"
"        println!(\"Minor\");\n"
"    }"),

("🔁 How do you write a loop with a range?",
"🧠 Concept:\n"
"Use `for i in 1..4` to loop from 1 to 3 (end exclusive).\n\n"
"✅ Code:\n"
"    for i in 1..4 {\n"
"        println!(\"{}\", i);\n"
"    }"),

("🎯 What does `let mut x = 5;` do?",
"🧠 Concept:\n"
"Declares a mutable variable named x with value 5."),

("🧠 What is the difference between `let` and `let mut`?",
"🧠 Concept:\n"
"`let` makes an immutable variable.\n`let mut` makes a mutable one you can change."),

("🔠 What placeholder do you use in println!?",
"🧠 Concept:\n"
"Use `{}` as a placeholder inside strings.\n\n"
"✅ Code:\n"
"    println!(\"Name: {}\", name);"),

("📍 What does `1..4` mean in a loop?",
"🧠 Concept:\n"
"A range from 1 up to but **not including** 4.\n\n"
"✅ Output:\n"
"    1\n    2\n    3")
]

index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'  # Hidden initially

def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]
        if len(card) == 3:
            question, answer, mnemonic = card
        else:
            question, answer = card
            mnemonic = ""

        html = f"""
        <div style='background:#f9f9f9;padding:16px;border-radius:10px;
                    border:1px solid #ddd;line-height:1.6;font-family:monospace;
                    white-space:pre-wrap;font-size:14px'>
        <b>Q:</b> {question}<br><br>
        {f"<b>A:</b><pre>{answer}</pre><br><b>🔁 Mnemonic:</b> {mnemonic}" if showing_answer else ""}
        </div>
        """
        display(HTML(html))

    # Update button states
    prev_button.disabled = index == 0
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"


def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Attach handlers
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Initialize display
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
````
