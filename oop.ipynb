# 📘 Section 1.1: What is Object-Oriented Programming?

---

## 🧠 Concept:

**Object-Oriented Programming (OOP)** is a way to organize code by bundling **related data** and **behavior** into reusable structures called **objects**.

---

## 🧩 Why Use It?

- ✅ Makes code **modular** and **reusable**  
- ✅ Mirrors real-world concepts (cars, users, bank accounts)  
- ✅ Easier to **maintain**, **test**, and **extend**

---

## 🧱 Key Concepts (You’ll Master These Step-by-Step)

| Concept       | Meaning                                 |
|---------------|------------------------------------------|
| **Class**     | A blueprint for creating objects         |
| **Object**    | A specific instance built from a class   |
| **Attribute** | A variable attached to an object         |
| **Method**    | A function attached to an object         |
| **Inheritance** | New class from existing one            |
| **Encapsulation** | Hide internal details (e.g., private data) |
| **Polymorphism**  | Same method name, different behavior  |

---

## 🧪 Let’s Code: Your First Class and Object

```python
# Define a class called Dog
class Dog:
    def __init__(self, name, breed):
        self.name = name        # Attribute
        self.breed = breed      # Attribute

    def bark(self):             # Method
        print(f"{self.name} says: Woof!")

# Create an object from the class
my_dog = Dog("Rex", "Labrador")

# Use the object's method
my_dog.bark()  # Output: Rex says: Woof!
```

---

## 🧠 Breakdown:

- `class Dog:` → Defines a **blueprint** for dog objects  
- `__init__()` → Constructor: sets up a new dog’s attributes  
- `self` → Refers to the **current object** (like “this” in other languages)  
- `my_dog` → A **real object** created from the class

````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output, HTML

# (question, answer, PEMDAS-style mnemonic)
flashcards = [

("What is Object-Oriented Programming (OOP)?",
"🧠 Concept:\n"
"OOP is a way of organizing code around objects — each object bundles data and behavior together.\n\n"
"🗣️ Plain English:\n"
"It’s like building reusable pieces that act like real things (dogs, users, bank accounts).\"\n\n"
"💡 Mnemonic:\n"
"OOP = Objects Own Properties\""),

("What is a class?",
"🧠 Concept:\n"
"A class is a blueprint for creating objects.\n\n"
"🗣️ Plain English:\n"
"It’s like a cookie cutter — the shape used to make real cookies (objects).\"\n\n"
"✅ Code:\n"
"    class Dog:\n"
"        pass"),

("What is an object?",
"🧠 Concept:\n"
"An object is an instance created from a class — it has actual data.\n\n"
"🗣️ Plain English:\n"
"The cookie that was made using the cutter (class).\"\n\n"
"✅ Code:\n"
"    my_dog = Dog()"),

("What is an attribute in OOP?",
"🧠 Concept:\n"
"A variable that belongs to an object.\n\n"
"🗣️ Plain English:\n"
"Like a name tag or breed of a Dog object.\"\n\n"
"✅ Code:\n"
"    self.name = name"),

("What is a method in a class?",
"🧠 Concept:\n"
"A function defined inside a class — used by objects.\n\n"
"🗣️ Plain English:\n"
"It’s something the object can do — like bark() or meow().\"\n\n"
"✅ Code:\n"
"    def bark(self):\n"
"        print(\"Woof!\")"),

("What does __init__ do?",
"🧠 Concept:\n"
"It’s the constructor — runs when a new object is created.\n\n"
"🗣️ Plain English:\n"
"It sets up the object’s starting values.\"\n\n"
"✅ Code:\n"
"    def __init__(self, name):\n"
"        self.name = name"),

("What does self mean?",
"🧠 Concept:\n"
"`self` refers to the current instance of the object — like `this` in other languages.\n\n"
"🗣️ Plain English:\n"
"‘Me’ — the current object talking about itself.\""),

("How do you create a Dog object and call its method?",
":\n"
"    class Dog:\n"
"        def __init__(self, name):\n"
"            self.name = name\n"
"        def bark(self):\n"
"            print(f\"{self.name} says: Woof!\")\n\n"
"    my_dog = Dog(\"Rex\")\n"
"    my_dog.bark()\n\n"
"🖨️ Output:\n"
"    Rex says: Woof!"),

("What's the difference between a class and an object?",
"🧠 Concept:\n"
"A class is a blueprint; an object is a real thing made from that blueprint.\n\n"
"🗣️ Plain English:\n"
"Class = design. Object = product.\""),

("What will this code print?\n\nclass Cat:\n    def __init__(self, name):\n        self.name = name\n    def meow(self):\n        print(f\"{self.name} says: Meow!\")\n\nkitty = Cat(\"Whiskers\")\nkitty.meow()",
"🧠 Execution:\n"
"Creates a Cat object named 'Whiskers' and calls its meow method.\n\n"
"🖨️ Output:\n"
"    Whiskers says: Meow!")
]


index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'  # Hidden initially

def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]
        if len(card) == 3:
            question, answer, mnemonic = card
        else:
            question, answer = card
            mnemonic = ""

        html = f"""
        <div style='background:#f9f9f9;padding:16px;border-radius:10px;
                    border:1px solid #ddd;line-height:1.6;font-family:monospace;
                    white-space:pre-wrap;font-size:14px'>
        <b>Q:</b> {question}<br><br>
        {f"<b>A:</b><pre>{answer}</pre><br><b>🔁 Mnemonic:</b> {mnemonic}" if showing_answer else ""}
        </div>
        """
        display(HTML(html))

    # Update button states
    prev_button.disabled = index == 0
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"


def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Attach handlers
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Initialize display
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
````
