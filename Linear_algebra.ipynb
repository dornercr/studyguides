## 🔹 1. What Is Linear Algebra?

### 🧠 Theory:

Linear algebra is the math of **vectors**, **matrices**, and how they **transform space**.

It helps us:

- Represent data (images, audio, word embeddings)  
- Simulate physics (3D motion, game engines)  
- Train machine learning models (data → weights → outputs)  

If arithmetic is math for single numbers, **linear algebra** is math for **structured collections of numbers**.

---

## 🔹 2. Scalars, Vectors, and Matrices

### 🧠 Theory:

- **Scalar** = Single number ➤ Example: temperature = 37°C  
- **Vector** = 1D array ➤ Example: $\vec{v} = [3, 4]$  
- **Matrix** = 2D array ➤ Example: brightness grid of an image

### 💻 Code (Fully Commented):

```python
import numpy as np  # NumPy lets us work with arrays

# Scalar = just one number
a = 5

# Vector = 1D array of numbers
v = np.array([2, 3])

# Matrix = 2D array (rows and columns)
M = np.array([
    [1, 2],
    [3, 4]
])

print("Scalar:", a)      # Scalar: 5
print("Vector:", v)      # Vector: [2 3]
print("Matrix:\n", M)    # Matrix: [[1 2]
                         #          [3 4]]
```

---

## 🔹 3. Vector Addition and Scalar Multiplication

### 🧠 Theory:

- **Vector Addition**: Add corresponding components  
  $[1, 2] + [3, 4] = [4, 6]$  
- **Scalar Multiplication**: Multiply each element  
  $2 \cdot [3, 4] = [6, 8]$

### 💻 Code:

```python
v1 = np.array([1, 2])
v2 = np.array([3, 4])

sum_v = v1 + v2         # [4 6]
scaled_v = 2 * v2       # [6 8]

print("Added Vectors:", sum_v)
print("Scaled Vector:", scaled_v)
```

---

## 🔹 4. Visualizing Vectors (2D)

### 🧠 Theory:

- A vector is like an **arrow** from the origin  
- You can **add vectors** by connecting tip-to-tail

### 💻 Code (with Visualization):

```python
import matplotlib.pyplot as plt

origin = [0], [0]  # All vectors start from (0, 0)

# Plot v1 in red
plt.quiver(*origin, v1[0], v1[1], color='r', scale=1, scale_units='xy', angles='xy')

# Plot v2 in blue
plt.quiver(*origin, v2[0], v2[1], color='b', scale=1, scale_units='xy', angles='xy')

# Plot sum_v in green
plt.quiver(*origin, sum_v[0], sum_v[1], color='g', scale=1, scale_units='xy', angles='xy')

# Format plot
plt.xlim(-1, 7)
plt.ylim(-1, 7)
plt.grid(True)
plt.gca().set_aspect('equal')
plt.title("Vector Addition — Red: v1, Blue: v2, Green: v1 + v2")
plt.show()
```

````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

# (question, answer, PEMDAS-style mnemonic)
flashcards = [

("What is linear algebra?",
"🧠 Theory:\nLinear algebra studies vectors, matrices, and how they transform space.\n\n🗣️ Plain English:\nIt’s the math behind 3D games, machine learning, and physics engines.",
"📎 PEMDAS: **L**inear algebra **L**inks **A**rrays and **T**ransformations"),

("What is a scalar?",
"🧠 Theory:\nA scalar is a single number.\n\n🗣️ Plain English:\nLike temperature = 98.6, or speed = 30 mph.",
"📎 PEMDAS: **S**calar = **S**olo"),

("What is a vector?",
"🧠 Theory:\nA vector is a 1D array with direction and magnitude.\n\n🗣️ Plain English:\nLike a wind arrow or a 2D point: [3, 4].",
"📎 PEMDAS: **V**ector **V**rooms with direction"),

("What is a matrix?",
"🧠 Theory:\nA matrix is a 2D array (rows and columns).\n\n🗣️ Plain English:\nLike a table of pixels or spreadsheet cells.",
"📎 PEMDAS: **M**atrix = **M**any numbers in a grid"),

("How do you create a scalar, vector, and matrix in NumPy?",
"💻 Code:\n```python\nimport numpy as np\n\na = 5\nv = np.array([2, 3])\nM = np.array([[1, 2], [3, 4]])\n```",
"📎 PEMDAS: **N**umPy makes **S**, **V**, and **M** easy"),

("What is vector addition?",
"🧠 Theory:\nAdd corresponding components.\n[1, 2] + [3, 4] = [4, 6]",
"📎 PEMDAS: **A**dd tips to tails"),

("What is scalar multiplication on a vector?",
"🧠 Theory:\nMultiply each element by a scalar.\n2 * [3, 4] = [6, 8]",
"📎 PEMDAS: **S**calar **S**tretches **V**ector"),

("How do you add vectors in code?",
"💻 Code:\n```python\nv1 = np.array([1, 2])\nv2 = np.array([3, 4])\nprint(v1 + v2)  # [4 6]\n```",
"📎 PEMDAS: **A**rrays + **A**rrays = **A**dded Vectors"),

("How do you multiply a vector by a scalar in code?",
"💻 Code:\n```python\nv = np.array([3, 4])\nscaled = 2 * v  # [6 8]\n```",
"📎 PEMDAS: **M**ultiply for **M**agnitude"),

("How do you visualize 2D vectors?",
"🧠 Theory:\nUse arrows from the origin, showing direction and length.",
"📎 PEMDAS: **V**ectors **V**isualized = **A**rrows"),

("How do you plot 2D vectors in Python?",
"💻 Code:\n```python\nimport matplotlib.pyplot as plt\n\norigin = [0], [0]\nplt.quiver(*origin, 3, 4, color='r', scale=1, scale_units='xy', angles='xy')\nplt.xlim(-1, 5)\nplt.ylim(-1, 5)\nplt.grid(True)\nplt.gca().set_aspect('equal')\nplt.axhline(0, color='gray')\nplt.axvline(0, color='gray')\nplt.show()\n```",
"📎 PEMDAS: **Q**uiver = **Q**uick arrows"),

("What happens when you add vectors visually?",
"🧠 Theory:\nTip-to-tail rule — the sum vector ends at the last tip.",
"📎 PEMDAS: **T**ip-**T**ail = **T**otal vector"),

("How would you represent a 2×2 matrix in Python?",
"💻 Code:\n```python\nM = np.array([[1, 2], [3, 4]])\n```",
"📎 PEMDAS: **2×2** = **2 lists** of 2 numbers")

]

index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'

def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]
        if len(card) == 2:
            question, answer = card
            mnemonic = ""
        else:
            question, answer, mnemonic = card

        display(Markdown(f"**Q:** {question}"))
        if showing_answer:
            display(Markdown(f"\n---\n\n**A:**\n{answer}"))
            if mnemonic.strip():
                display(Markdown(mnemonic))

    prev_button.disabled = index == 0
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"

def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Hook up events
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Display the first flashcard
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
````
