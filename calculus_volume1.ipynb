## 🔁 Definition of a Function

> A **function** is a rule that assigns to each input **exactly one output**.

### Notation:

- A function: `f: X → Y`
- `f(x)` means “**value of function f at input x**”

### Example:

- `f(x) = x²`
  - Domain: All real numbers ℝ
  - Range: All real numbers ≥ 0 (i.e., `[0, ∞)`)

---

## 🔎 Visualizing Functions

### 1. **Input-Output Machine**  
A function takes an input `x` and produces output `f(x)`.

```
x ──▶ [ FUNCTION ] ──▶ f(x)
```

### 2. **Mapping Diagram**

```
Domain         Range
{1,2,3,4} ──▶ {2,4,6}
```

- Multiple domain values may map to the same range value.
- No domain value maps to more than one range value.

### 3. **Graphical Representation**

- Domain: x-values along the horizontal axis
- Range: f(x) values along the vertical axis
- Example:  
  ```
  (1, 2), (2, 1), (3, 2)
  Domain = {1, 2, 3}, Range = {1, 2}
  ```

---

## ✍️ Function Representation Forms

- **Tables**: Discrete input-output values
- **Formulas**: Algebraic expressions like `f(x) = x² - 1`
- **Graphs**: Visual plots on a coordinate plane

---

## 🧾 Domain and Range Notation

### Set-Builder:

- `{x ∈ ℝ | condition}`  
  Example: `{x ∈ ℝ | 1 < x < 5}`

### Interval Notation:

- Open: `(1, 5)`
- Closed: `[1, 5]`
- Half-open: `[0, ∞)`
- Infinite: `(-∞, ∞)`

---

## 🧩 Piecewise-Defined Functions

Functions can use different expressions over different intervals.

Example:
```
f(x) = {
    x² + 1, if x < 0
    2x - 1, if x ≥ 0
}
```

Evaluate based on condition:

- `f(-2) = (-2)² + 1 = 5`
- `f(3) = 2(3) - 1 = 5`

---

## 🧪 Practice: Identifying Functions

- If any domain value maps to **more than one** range value → ❌ Not a function.
- If every domain value maps to **exactly one** range value → ✅ Valid function.

````python
from sympy import symbols, simplify, expand, Function

# Define symbolic variables
x, a, h = symbols('x a h')

# Define the function f(x)
f = x**2 + 3*x - 4

# 1. f(1)
f_1 = f.subs(x, 1)

# 2. f(a)
f_a = f.subs(x, a)

# 3. f(a + h)
f_ah = expand(f.subs(x, a + h))

# 4. (f(a + h) - f(a)) / h
difference_quotient = simplify((f_ah - f_a) / h)

# Store all answers in a dictionary for easy access
results = {
    'f(1)': f_1,
    'f(a)': f_a,
    'f(a + h)': f_ah,
    '(f(a + h) - f(a)) / h': difference_quotient
}

print(f"f(1):{f_1}\nf(a): {f_a}\nf(a + h): {f_ah}\n(f(a + h) - f(a)) / h: {difference_quotient}")
````

# 🧮 Algebraic Evaluation and Difference Quotient Using Symbolic Math

This guide walks through algebraic manipulations of a quadratic function using symbolic variables in SymPy, with detailed explanations and algebraic breakdowns.

---

## Step 1: Define the Function

We begin with the quadratic function:

$$
f(x) = x^2 + 3x - 4
$$

We will evaluate this function at various inputs and compute the **difference quotient**:

$$
\frac{f(a + h) - f(a)}{h}
$$

---

## Step 2: Evaluate $f(1)$

$$
f(1) = (1)^2 + 3(1) - 4 = 1 + 3 - 4 = 0
$$

---

## Step 3: Evaluate $f(a)$

Substitute $x = a$:

$$
f(a) = a^2 + 3a - 4
$$

This is a symbolic result — we don’t plug in a number yet.

---

## Step 4: Evaluate $f(a + h)$

Substitute $x = a + h$:

$$
f(a + h) = (a + h)^2 + 3(a + h) - 4
$$

Now expand:

$$
= a^2 + 2ah + h^2 + 3a + 3h - 4
$$

So,

$$
f(a + h) = a^2 + 2ah + h^2 + 3a + 3h - 4
$$

---

## Step 5: Compute the Difference Quotient

$$
\frac{f(a + h) - f(a)}{h}
$$

Subtract:

$$
f(a + h) - f(a) =
(a^2 + 2ah + h^2 + 3a + 3h - 4) - (a^2 + 3a - 4)
$$

Cancel matching terms:

$$
= 2ah + h^2 + 3h
$$

So:

$$
\frac{f(a + h) - f(a)}{h} = \frac{2ah + h^2 + 3h}{h}
$$

Factor numerator:

$$
= \frac{h(2a + h + 3)}{h}
$$

Cancel $h$:

$$
= 2a + h + 3
$$

---

## ✅ Final Simplified Result

$$
\frac{f(a + h) - f(a)}{h} = 2a + h + 3
$$

This represents the **average rate of change** of the function from $a$ to $a + h$.
