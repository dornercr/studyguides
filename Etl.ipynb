## 1. Build and optimize ETL data pipelines using Databricks and Apache Spark

**How to do it:**
1. Connect to data sources in Databricks *(cloud-based big data analytics platform)* (e.g., S3 *(Amazon Simple Storage Service)*, SQL Server *(relational database management system)*, API endpoints *(URL access points for applications to communicate)*) using `spark.read` methods *(Apache Spark command to load data)*.
2. **Extract** – Load raw data into Spark DataFrames *(distributed data collections with named columns)* (`spark.read.format("csv").load(path)`).
3. **Transform** – Use Spark SQL *(query language for Spark)* or PySpark *(Python API for Apache Spark)* to clean, join, and reformat data (`df.filter()`, `.withColumn()`).
4. **Load** – Write transformed data to target storage (`df.write.format("parquet").save()`) *(Parquet = columnar storage file format)*.
5. **Optimize** – Enable partitioning *(splitting data for faster queries)*, caching *(storing results in memory)*, and Delta Lake *(storage layer that adds reliability to data lakes)*.
6. **Schedule** – Use Databricks Jobs or Workflows *(automated execution tools)*.

**Key tech:** Databricks, Apache Spark *(distributed data processing engine)*, PySpark, Delta Lake.  
**Mnemonic (C E T L O S):** Clever Engineers Transform Logs, Optimize Schedules.

---

## 2. Analyze large datasets using SQL and Python

**How to do it:**
1. SQL *(Structured Query Language)* – Use queries to filter, group, and aggregate data (`SELECT COUNT(*), category FROM sales GROUP BY category`).
2. Python *(high-level programming language)* **(Pandas)** *(Python data analysis library)* – Load datasets (`pd.read_csv()`), filter (`df[df['col'] > 100]`), and summarize (`df.groupby()`).
3. Combine – Sometimes run SQL in Databricks notebooks alongside Pandas for smaller datasets.

**Key tech:** SQL, Python (Pandas), Databricks SQL.  
**Mnemonic (S P C):** Slice, Probe, Consolidate.

---

## 3. Use Python to clean, manipulate, and visualize data

**How to do it:**
1. **Clean** – Handle missing values (`df.fillna()`), remove duplicates (`df.drop_duplicates()`), fix types (`df.astype()`).
2. **Manipulate** – Merge datasets (`pd.merge()`), calculate new columns (`df['total'] = df['price'] * df['qty']`).
3. **Visualize** – Use Matplotlib *(Python plotting library)* or Seaborn *(statistical data visualization library)* (`sns.barplot(x='month', y='sales', data=df)`).

**Key tech:** Python, Pandas, Matplotlib, Seaborn.  
**Mnemonic (C M V):** Clean Messy Values.

## 4. Use APIs for seamless data exchange

**How to do it:**
1. Read API *(Application Programming Interface)* docs for the system you’re connecting to.
2. Send requests with Python’s `requests` library (`requests.get(url, headers=auth)`).
3. Parse JSON *(JavaScript Object Notation)* / XML *(Extensible Markup Language)* response (`response.json()`).
4. Write data into storage or Spark DataFrame.

**Key tech:** REST APIs *(web service architecture)*, Python requests, JSON.  
**Mnemonic (R S P W):** Read, Send, Parse, Write.

```python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output, HTML

# (question, answer, PEMDAS-style mnemonic)



flashcards = [

# ---------- 1 ----------
("Build and optimize ETL data pipelines – Responsibility",
"Build and optimize ETL data pipelines using Databricks and Apache Spark."),

("Build and optimize ETL data pipelines – Steps",
"1. Connect to data sources in Databricks (e.g., S3, SQL Server, APIs):\n"
"   df = spark.read.format('csv').load('s3://bucket/file.csv')\n"
"2. Extract – Load raw data into Spark DataFrames.\n"
"3. Transform – Clean/join/reformat with PySpark:\n"
"   df = df.filter(df['age'] > 18).withColumn('score', df['points'] * 2)\n"
"4. Load – Write transformed data:\n"
"   df.write.format('parquet').save('/mnt/output')\n"
"5. Optimize – df.repartition(4).cache()\n"
"6. Schedule – Configure Jobs in Databricks UI or Workflows."),

("Build and optimize ETL data pipelines – My Experience",
"As Staff Engineer for a government contractor, designed automated Python-based scraping modules to extract and standardize foreign language content. Deployed workflows on AWS EC2/S3, integrating transformation logic and optimized storage formats, similar to Delta Lake optimizations."),

("Build and optimize ETL data pipelines – Key tech",
"Databricks, Apache Spark, PySpark, Delta Lake."),

("Build and optimize ETL data pipelines – Mnemonic",
"C E T L O S: Clever Engineers Transform Logs, Optimize Schedules."),

# ---------- 2 ----------
("Analyze large datasets – Responsibility",
"Analyze large datasets using SQL and Python."),

("Analyze large datasets – Steps",
"1. SQL query example:\n"
"   SELECT category, COUNT(*) FROM sales GROUP BY category;\n"
"2. Python (Pandas):\n"
"   import pandas as pd\n"
"   df = pd.read_csv('data.csv')\n"
"   df[df['amount'] > 100].groupby('category').size()\n"
"3. Combine SQL & Pandas in Databricks notebooks."),

("Analyze large datasets – My Experience",
"Analyzed large-scale language proficiency datasets for the Department of State using Pandas and SQL to generate metrics for training programs. Produced Power BI dashboards for decision-makers."),

("Analyze large datasets – Key tech",
"SQL, Python (Pandas), Databricks SQL."),

("Analyze large datasets – Mnemonic",
"S P C: Slice, Probe, Consolidate."),

# ---------- 3 ----------
("Use Python to clean/manipulate/visualize – Responsibility",
"Use Python to clean, manipulate, and visualize data."),

("Use Python to clean/manipulate/visualize – Steps",
"1. Clean:\n"
"   df = df.fillna(0).drop_duplicates().astype({'price': float})\n"
"2. Manipulate:\n"
"   df = pd.merge(df1, df2, on='id')\n"
"   df['total'] = df['price'] * df['qty']\n"
"3. Visualize:\n"
"   import seaborn as sns\n"
"   sns.barplot(x='month', y='sales', data=df)"),

("Use Python to clean/manipulate/visualize – My Experience",
"Created Python scripts to standardize language datasets, merging content from multiple scraping modules. Generated visualizations in Power BI and Matplotlib to track system performance and usage trends."),

("Use Python to clean/manipulate/visualize – Key tech",
"Python, Pandas, Matplotlib, Seaborn."),

("Use Python to clean/manipulate/visualize – Mnemonic",
"C M V: Clean Messy Values."),

# ---------- 4 ----------
("Use APIs for seamless data exchange – Responsibility",
"Use APIs for seamless data exchange."),

("Use APIs for seamless data exchange – Steps",
"1. Read API docs.\n"
"2. Send request:\n"
"   import requests\n"
"   r = requests.get(url, headers={'Authorization': 'Bearer TOKEN'})\n"
"3. Parse JSON:\n"
"   data = r.json()\n"
"4. Load to DataFrame:\n"
"   df = spark.createDataFrame(data)"),

("Use APIs for seamless data exchange – My Experience",
"Integrated Department of State systems by calling internal REST APIs to fetch, validate, and push processed language training data. Automated API workflows in Python to reduce manual intervention."),

("Use APIs for seamless data exchange – Key tech",
"REST APIs, Python requests, JSON."),

("Use APIs for seamless data exchange – Mnemonic",
"R S P W: Read, Send, Parse, Write."),

# ---------- 5 ----------
("Manage data in Microsoft SharePoint – Responsibility",
"Manage data in Microsoft SharePoint."),

("Manage data in Microsoft SharePoint – Steps",
"1. Access SharePoint site and list.\n"
"2. Upload via UI or Microsoft Graph API.\n"
"3. Modify lists – Add/edit columns, set types.\n"
"4. Search with KQL:\n"
"   Title:Report AND Author:Smith"),

("Manage data in Microsoft SharePoint – My Experience",
"Maintained project documentation and report archives in SharePoint for DoD corrosion prevention projects, ensuring up-to-date records and controlled access for stakeholders."),

("Manage data in Microsoft SharePoint – Key tech",
"SharePoint, KQL, Microsoft Graph API."),

("Manage data in Microsoft SharePoint – Mnemonic",
"A U M U: Access, Upload, Modify, Use KQL."),

# ---------- 6 ----------
("Leverage Microsoft 365 tools – Responsibility",
"Leverage Microsoft 365 tools for documentation & collaboration."),

("Leverage Microsoft 365 tools – Steps",
"1. Store datasets in OneDrive.\n"
"2. Document pipeline steps in Word/OneNote.\n"
"3. Use Teams for coordination.\n"
"4. Track tasks in Planner or Jira."),

("Leverage Microsoft 365 tools – My Experience",
"Used Teams and OneNote to coordinate multi-site engineering teams at Elzly Technology. Shared technical documentation and corrosion system performance reports across Microsoft 365 apps."),

("Leverage Microsoft 365 tools – Key tech",
"Microsoft 365 (Word, Excel, OneNote, Teams)."),

("Leverage Microsoft 365 tools – Mnemonic",
"S D U T: Store, Document, Use Teams, Track."),

# ---------- 7 ----------
("Develop and maintain AWS S3 – Responsibility",
"Develop and maintain AWS S3 storage."),

("Develop and maintain AWS S3 – Steps",
"1. Upload via CLI:\n"
"   aws s3 cp file.csv s3://bucket-name/\n"
"2. Organize with prefixes.\n"
"3. Set IAM permissions.\n"
"4. Load in Spark:\n"
"   df = spark.read.format('csv').load('s3://bucket/file.csv')"),

("Develop and maintain AWS S3 – My Experience",
"Deployed language learning system storage on AWS S3 with secure IAM roles. Managed EC2-hosted applications that accessed S3, with CloudFront distribution for reduced latency."),

("Develop and maintain AWS S3 – Key tech",
"AWS S3, AWS CLI, IAM."),

("Develop and maintain AWS S3 – Mnemonic",
"U O S I: Upload, Organize, Secure, Integrate."),

# ---------- 8 ----------
("Build pipelines integrating diverse sources – Responsibility",
"Build pipelines integrating diverse sources via API."),

("Build pipelines integrating diverse sources – Steps",
"1. Gather API endpoints.\n"
"2. Extract via scheduled Python scripts.\n"
"3. Transform in Databricks:\n"
"   df = df.withColumn('clean_name', trim(df['name']))\n"
"4. Load to S3 or SQL:\n"
"   df.write.jdbc(url, table, properties=props)"),

("Build pipelines integrating diverse sources – My Experience",
"Spearheaded integration of scraping, evaluation, and UI modules into a unified AWS-hosted system. Automated ETL using API inputs from multiple content providers."),

("Build pipelines integrating diverse sources – Key tech",
"APIs, Databricks, Spark, Python."),

("Build pipelines integrating diverse sources – Mnemonic",
"G E T L: Gather, Extract, Transform, Load."),

# ---------- 9 ----------
("Create/manage SharePoint lists/libraries – Responsibility",
"Create and manage SharePoint lists and libraries."),

("Create/manage SharePoint lists/libraries – Steps",
"1. Create list – choose columns/types.\n"
"2. Import Excel/CSV.\n"
"3. Configure views for sorting/filtering.\n"
"4. Set permissions."),

("Create/manage SharePoint lists/libraries – My Experience",
"Configured SharePoint lists for tracking Navy ICCP maintenance tasks and inspection schedules. Controlled permissions to restrict access to sensitive operational data."),

("Create/manage SharePoint lists/libraries – Key tech",
"SharePoint, Microsoft 365."),

("Create/manage SharePoint lists/libraries – Mnemonic",
"C I C S: Create, Import, Configure, Secure."),

# ---------- 10 ----------
("Use Jira for issue/project tracking – Responsibility",
"Use Jira for issue and project tracking."),

("Use Jira for issue/project tracking – Steps",
"1. Create issue in Jira UI.\n"
"2. Assign to user.\n"
"3. Track via Kanban/Scrum board.\n"
"4. Log work and update status."),

("Use Jira for issue/project tracking – My Experience",
"Managed Agile sprints and tracked defects for the University of Maryland platform containerization project, using Jira to coordinate dev/ops handoffs and monitor backlog completion."),

("Use Jira for issue/project tracking – Key tech",
"Jira, Agile workflows."),

("Use Jira for issue/project tracking – Mnemonic",
"C A T L: Create, Assign, Track, Log."),

# ---------- 11 ----------
("Collaborate with cross-functional teams – Responsibility",
"Collaborate with cross-functional teams."),

("Collaborate with cross-functional teams – Steps",
"1. Hold stand-ups via Teams/Zoom.\n"
"2. Document in OneNote or Confluence.\n"
"3. Share updates in Jira comments or Teams channels."),

("Collaborate with cross-functional teams – My Experience",
"Partnered with Department of State program managers, Navy engineers, and academic developers to align technical designs with operational requirements and ensure secure handling of sensitive data."),

("Collaborate with cross-functional teams – Key tech",
"Teams, Jira, Microsoft 365."),

("Collaborate with cross-functional teams – Mnemonic",
"H D S: Hold, Document, Share.")

]


index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'  # Hidden initially

def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]
        if len(card) == 3:
            question, answer, mnemonic = card
        else:
            question, answer = card
            mnemonic = ""

        html = f"""
        <div style='background:#f9f9f9;padding:16px;border-radius:10px;
                    border:1px solid #ddd;line-height:1.6;font-family:monospace;
                    white-space:pre-wrap;font-size:14px'>
        <b>Q:</b> {question}<br><br>
        {f"<b>A:</b><pre>{answer}</pre><br><b>🔁 Mnemonic:</b> {mnemonic}" if showing_answer else ""}
        </div>
        """
        display(HTML(html))

    # Update button states
    prev_button.disabled = index == 0
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"


def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Attach handlers
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Initialize display
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
```
