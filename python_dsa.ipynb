# 🧠 Expert Guide: Python Lists (Dynamic Arrays)

---

## 1. 🔹 Definition and Structure

A **Python list** is a **dynamic array** (like `malloc` + pointer arithmetic in C):

- Ordered, indexable, mutable sequence of object references  
- Backed by **contiguous memory** storing **pointers**, not actual object data  
- Allows amortized `O(1)` `append()` due to **overallocation**

### Internals:

```
+---------+---------+---------+---------+------
|   *     |   *     |   *     |   *     |  ...
+---------+---------+---------+---------+------
   obj0     obj1      obj2      obj3
```

Each `*` is a pointer to an actual Python object on the heap.

---

## 2. 🔹 Core Operations and Their Complexity

| Operation       | Big-O Time | Notes                                    |
|------------------|------------|------------------------------------------|
| `append(x)`      | O(1)*       | Amortized; realloc if needed             |
| `pop()` / `pop(-1)` | O(1)    | Removes last item                        |
| `insert(i, x)`   | O(n)       | Shifts elements right                    |
| `del list[i]`    | O(n)       | Shifts elements left                     |
| `list[i] = x`    | O(1)       | Random access                            |
| `list1 + list2`  | O(n + m)   | Allocates new memory                     |
| `sort()`         | O(n log n) | Timsort; stable                          |
| `list.index(x)`  | O(n)       | Linear search                            |
| `list.remove(x)` | O(n)       | Finds and deletes first match            |

> *Python uses overallocation strategy (e.g., +4, then ×1.125) for amortized performance.

---

## 3. 🔹 Mutability & Identity

```python
a = [1, 2, 3]
b = a
b[0] = 99
print(a)  # → [99, 2, 3]
```

- Lists are **mutable**
- Variables point to the **same object**
- Use `.copy()` or `copy.deepcopy()` to avoid shared reference

---

## 4. 🔹 Slicing and Views

```python
a = [1, 2, 3, 4]
a[1:3]     # → [2, 3]
a[::-1]    # → [4, 3, 2, 1]
a[:3:2]    # → [1, 3]
```

- Slicing creates a **new list**, not a view  
- Does **not** mutate original list unless used in assignment

---

## 5. 🔹 List Comprehensions (Efficient Syntax)

```python
squares = [x*x for x in range(10)]
evens = [x for x in mylist if x % 2 == 0]
```

- More concise and faster than `map()` + `filter()`  
- Can be **nested** for flattening or multidimensional processing

---

## 6. 🔹 Shallow vs Deep Copy

```python
import copy

a = [[1, 2], [3, 4]]
b = a.copy()          # shallow copy
c = copy.deepcopy(a)  # deep copy

b[0][0] = 99
print(a)  # → [[99, 2], [3, 4]]
```

- Shallow copy shares inner elements  
- Use `deepcopy()` for true independent duplication

---

## 7. 🔹 Memory Management

```python
import sys

a = []
print(sys.getsizeof(a))  # base size

a.append(1)
print(sys.getsizeof(a))  # increases in chunks
```

- Python **overallocates** internal buffer space  
- Capacity expands **geometrically** to reduce reallocations

---

## 8. 🔹 Identity vs Equality

```python
a = [1, 2, 3]
b = [1, 2, 3]

a == b  # True — same values
a is b  # False — different objects
```

- `==` → value equality  
- `is` → object identity

---

## 9. 🔹 Common Pitfalls

| Pitfall                         | What Happens                        | Fix                                 |
|----------------------------------|-------------------------------------|--------------------------------------|
| `a = [[0]*3]*3`                  | Creates shallow copies              | Use list comprehension               |
| `list.remove(x)` if x not found | Raises `ValueError`                 | Use `if x in list:` check            |
| Mutating during iteration       | Skips/corrupts loop behavior        | Iterate over a copy                  |
| `a.sort()` returns `None`       | `sort()` mutates in-place           | Use `sorted(a)` for a new list       |

---

## 10. 🔹 Alternatives to List

| Type           | Use Case                                 |
|----------------|-------------------------------------------|
| `array.array`  | Numeric arrays (space-efficient)          |
| `collections.deque` | Fast appends/pops from both ends     |
| `set`          | Unordered, unique elements                |
| `tuple`        | Immutable version of list                 |
| `numpy.array`  | For vectorized math, fixed-type arrays    |

````python
# 🧠 Expert Guide: Python Lists (Dynamic Arrays)

# 1. 🔹 Definition and Structure
a = [10, 20, 30, 40]  # List of object references (ints here)
print("IDs:", id(a[0]), id(a[1]))  # Show memory addresses of the objects

# 2. 🔹 Core Operations and Their Complexity
a = [1, 2, 3]
a.append(4)          # O(1) amortized
a.pop()              # O(1)
a.insert(1, 99)      # O(n)
del a[2]             # O(n)
a[1] = 100           # O(1)
b = a + [5, 6, 7]    # O(n + m)
a.sort()             # O(n log n)
print("Index of 100:", a.index(100))  # O(n)
if 1 in a:
    a.remove(1)      # O(n)

# 3. 🔹 Mutability & Identity
import copy
a = [1, 2, 3]
b = a
b[0] = 99
print("a after b[0] = 99:", a)  # [99, 2, 3]
c = a.copy()
c[0] = 42
print("a vs c:", a, c)  # [99, 2, 3], [42, 2, 3]

# 4. 🔹 Slicing and Views
a = [1, 2, 3, 4]
print("a[1:3]:", a[1:3])   # [2, 3]
print("a[::-1]:", a[::-1])  # [4, 3, 2, 1]
print("a[:3:2]:", a[:3:2])  # [1, 3]
a[1:3] = [9, 9]
print("Modified a:", a)    # [1, 9, 9, 4]

# 5. 🔹 List Comprehensions
squares = [x*x for x in range(5)]              # [0, 1, 4, 9, 16]
evens = [x for x in range(10) if x % 2 == 0]    # [0, 2, 4, 6, 8]
nested = [[i*j for j in range(3)] for i in range(3)]  # 2D list
flat = [item for sublist in nested for item in sublist]  # flatten
print("squares:", squares)
print("evens:", evens)
print("flattened:", flat)

# 6. 🔹 Shallow vs Deep Copy
a = [[1, 2], [3, 4]]
b = a.copy()          # Shallow copy
c = copy.deepcopy(a)  # Deep copy
b[0][0] = 99
print("a after b[0][0] = 99:", a)  # [[99, 2], [3, 4]]
print("c remains:", c)            # [[1, 2], [3, 4]]

# 7. 🔹 Memory Management
import sys
a = []
print("Base size:", sys.getsizeof(a))  # Initial size
for i in range(30):
    a.append(i)
    if i % 10 == 0:
        print(f"Length {len(a)}, Size: {sys.getsizeof(a)}")

# 8. 🔹 Identity vs Equality
a = [1, 2, 3]
b = [1, 2, 3]
print("a == b:", a == b)  # True
print("a is b:", a is b)  # False
print("id(a), id(b):", id(a), id(b))

# 9. 🔹 Common Pitfalls

# Pitfall: Shallow row copies
a = [[0]*3]*3
a[0][0] = 1
print("Shallow grid:", a)  # [[1, 0, 0], [1, 0, 0], [1, 0, 0]]

# Fix with list comprehension
a = [[0 for _ in range(3)] for _ in range(3)]
a[0][0] = 1
print("Proper grid:", a)  # Only first row modified

# Pitfall: Mutating during iteration
nums = [1, 2, 3]
for x in nums[:]:  # Safe: iterate over a copy
    nums.remove(x)
print("After safe remove:", nums)

# Pitfall: sort returns None
a = [3, 1, 2]
sorted_a = sorted(a)  # Non-mutating
print("sorted_a:", sorted_a)

# 10. 🔹 Alternatives to List
from array import array
from collections import deque
import numpy as np

arr = array('i', [1, 2, 3])       # Compact numeric array
dq = deque([1, 2, 3])             # Double-ended queue
s = set([1, 2, 2, 3])             # Unordered unique values
t = (1, 2, 3)                     # Immutable
np_arr = np.array([1, 2, 3])      # For vectorized math

print("array.array:", arr)
print("deque:", dq)
print("set:", s)
print("tuple:", t)
print("numpy:", np_arr)
````

# 📘 Expert-Level Theory: Stacks

---

## 🔹 1. Definition

A **stack** is an abstract data type that follows the **Last In, First Out (LIFO)** principle:

- The **last element added** is the **first one removed**  
- You can **only interact with the top** of the stack

### Real-life analogies:
- Stack of plates  
- Browser back button history  
- Function call stack

---

## 🔹 2. Core Operations

| Operation     | Description                          | Complexity |
|---------------|--------------------------------------|------------|
| `push(x)`     | Add element `x` to the top           | O(1)       |
| `pop()`       | Remove and return top element        | O(1)       |
| `peek()`      | View top element without removing it | O(1)       |
| `is_empty()`  | Check if stack is empty              | O(1)       |
| `size()`      | Return number of elements            | O(1)       |

---

## 🔹 3. Python Implementation

### ✅ Using `list`:

```python
stack = []
stack.append(1)     # Push
stack.pop()         # Pop
stack[-1]           # Peek
len(stack) == 0     # is_empty
```

🔴 **Drawback**: `list` is inefficient if popping from front.

### ✅ Better: `collections.deque` (true O(1) behavior)

```python
from collections import deque

stack = deque()
stack.append(1)
stack.pop()
```

---

## 🔹 4. Internal Mechanics

Stacks can be implemented using:

- Dynamic arrays (e.g., Python `list`, C++ `vector`)
- Linked lists (singly or doubly linked)
- Fixed-size arrays (common in embedded systems)

---

## 🔹 5. Applications

| Domain         | Example                                   |
|----------------|--------------------------------------------|
| Language parsing | Matching parentheses, HTML tags         |
| OS/Compilers   | Call stack, context switching              |
| Algorithms     | DFS, postfix evaluation                   |
| Editors        | Undo/Redo stacks                          |
| Simulations    | Browser history, maze/backtracking        |

---

## 🔹 6. Time and Space Complexity

- **Time**: All core ops are O(1)  
- **Space**: O(n) total for `n` elements  
- **Amortized Cost** (for dynamic arrays):  
  - Occasional O(n) copy during reallocation

---

## 🔹 7. Variants and Advanced Stack Patterns

🧮 **a. Min Stack**  
Track minimum with a second stack for O(1) `get_min()`.

🧮 **b. Two-Stack Queue**  
Use two stacks to implement FIFO behavior.

🧮 **c. Postfix Evaluation**  
Evaluate expressions in Reverse Polish Notation.

🧮 **d. Infix to Postfix**  
Convert with **Shunting Yard Algorithm**.

🧮 **e. Recursion Simulation**  
Manually simulate recursion using explicit stack (e.g., DFS traversal).

---

## 🔹 8. Common Pitfalls

| Mistake                       | Fix                                      |
|-------------------------------|-------------------------------------------|
| `stack.pop()` on empty stack  | Always check with `if stack:`            |
| Modifying during iteration    | Iterate over a **copy** or loop safely   |
| Confusing LIFO with FIFO      | Know your structure's order guarantees   |
| Using stack for BFS           | Use `queue`, not `stack`                 |

---

## 🔹 9. Comparison with Related Structures

| Structure | Access Order | Use Case              |
|-----------|--------------|------------------------|
| Stack     | LIFO         | Undo, recursion        |
| Queue     | FIFO         | Scheduling, BFS        |
| Deque     | Double-ended | Flexible two-way ops   |
| List      | Indexed      | General-purpose        |

---

## 🔹 10. Memory and Performance Notes

- `list` in Python grows via **overallocation** (~1.125× growth)  
- `deque` is **doubly linked** and optimized for stack/queue behavior  
- **Call stacks** in low-level systems are bounded → can trigger `StackOverflowError`

| Basic Calculator                | Stack of stack frames    |
