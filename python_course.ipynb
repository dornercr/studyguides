# 📘 Python 6.2 & 6.3 – While Loops (Theory + Examples Only)

## 🔁 What Is a While Loop?

A `while` loop is a structure that keeps repeating code **as long as a condition is True**.

- **Loop expression**: The condition that's checked before each iteration.
- **Loop body**: The code that runs if the condition is `True`.
- **Iteration**: One complete run of the loop body.

👉 When the condition becomes `False`, the loop **stops**.

---

## 🔧 Basic Syntax

```python
while condition:
    # Code that runs while the condition is True
# Code that runs after the loop stops


````python
#basic syntax
condition = True
while condition == True:
    # Code that runs while the condition is True
# Code that runs after the loop stops
    print("True")
    condition = False
````

## 🧠 Key Concepts

A `while` loop is a structure that keeps repeating code **as long as a condition is True**.

- Use == to check equality, != to check inequality.

- Ensure something changes inside the loop, or you’ll get an infinite loop.

- Use a sentinel value (like 'q' or 0) to stop the loop safely.

📌 **Vocabulary**

| Term          | Meaning                                      |
|---------------|----------------------------------------------|
| Condition     | Rule that controls the loop                  |
| Iteration     | One full run of the loop                     |
| Sentinel      | A value that tells the loop to stop          |
| Infinite Loop | Loop that never ends unless stopped          |


````python
#Power Doubling Loop
curr_power = 2
user_char = 'y'

while user_char == 'y':
    print(curr_power)
    curr_power *= 2
    user_char = input()

print('Done')
````

````python
#🙂 Face Drawing with Sentinel 'q'

nose = '0'
user_value = '-'

while user_value != 'q':
    print(f' {user_value} {user_value} ')
    print(f'  {nose}  ')
    print(user_value * 5)
    print('\n')

    user_input = input("Enter a character ('q' for quit): \n")
    user_value = user_input[0]

print('Goodbye.\n')
````

````python
#📉 Digit Printer (Right to Left)

num = int(input())

while num > 0:
    print(num % 10)
    num = num // 10
    print(f"new num = {num}")
````

````python
#🧬 Ancestors Loop (Years)

user_year = int(input())
year = 2020
ancestors = 2

while year >= user_year:
    print(f"Ancestors in {year}: {ancestors}")
    year -= 20
    ancestors *= 2
````

````python
#🔢 GCD (Greatest Common Divisor)

num_a = int(input("Enter first positive integer: "))
num_b = int(input("Enter second positive integer: "))

while num_a != num_b:
    if num_a > num_b:
        num_a -= num_b
    else:
        num_b -= num_a

print("GCD is", num_a)
````

````python
#Random Convo Bot
import random

user_text = input()

while user_text != "Goodbye":
    rand = random.randint(0, 2)

    if rand == 0:
        print("That's interesting.")
    elif rand == 1:
        print("Tell me more.")
    elif rand == 2:
        print("I see.")

    user_text = input()

print("Goodbye")
````

````python
#average calculator with sentinel

user_input = int(input())
values_sum = 0
num_values = 0

while user_input != 0:
    values_sum += user_input
    num_values += 1
    user_input = int(input())

if num_values > 0:
    print("Average:", values_sum / num_values)
else:
    print("No values entered.")
````

# 📘 Python 6.4 – Counting with While Loops (Theory + Examples Only)

## 🔁 What Is Counting in a While Loop?

Counting loops are used when we want code to repeat a set number of times.
We do this by using a loop variable, usually named i, to count the number of iterations.

---

## 🧱 Basic Counting Structure

```python
i = 1
while i <= N:
    # do something
    i = i + 1
```
- Starts at 1, ends at N
- Be careful to update i, or you'll create an infinite loop

````python
N = 5
#basic counting structure
i = 1
while i <= N:
    print(i)
    i = i + 1
````

````python
#💵 Example: Savings Interest
years = int(input())
savings = 10000
interest_rate = 0.05
i = 1

while i <= years:
    print(f'Year {i}: ${savings:.2f}')
    savings = savings + (savings * interest_rate)
    i = i + 1
````

````python
#counting down

i = 5
while i >= 1:
  print(i)
  i = i -1
````

````python
#counting by steps

i = 0
while i <= 10:
  print(i)
  i = i + 2
````

````python
#example: US Presidential Election Years

i = 1792
while i <= 2028:
  print(i)
  i = i + 4
````

````python
#example: Factorial(N!)

N = int(input())
fact = 1
i = N - i

while i >= 1:
  fact *= i
  i -= 1

print(f"{N}! is {fact}")
````

⚡ **Shorthand Operators**

Python lets us write shortcuts for common math operations:

| Long Form     | Shorthand |
|---------------|-----------|
| i = i + 1     | i += 1    |
| i = i - 1     | i -= 1    |
| i = i * 2     | i *= 2    |
| i = i / 2     | i /= 2    |


📝 **Summary of Counting Forms**

| Type              | Example                        | Output              |
|-------------------|--------------------------------|----------------------|
| Count up by 1     | i = 1; while i <= 5:           | 1 2 3 4 5            |
| Count down        | i = 5; while i >= 1:           | 5 4 3 2 1            |
| By 2s             | i = 0; while i <= 10:          | 0 2 4 6 8 10         |
| Multiples of 5    | i = 0; while i <= 1000:        | 0 5 10 ...           |
| Odd numbers       | i = 1; while i <= 9:           | 1 3 5 7 9            |
| Backwards odd     | i = 211; while i >= 31:        | 211 209 ... 31       |
| Powers of 2       | i = 1; while i <= 256:         | 1 2 4 8 ... 256      |
