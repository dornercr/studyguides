**Definition**:  
A *requirement* is a need or condition a system must meet.  
It can be:

- **High-level** (abstract goals)
- **Detailed** (technical specs)

**Forms of Requirements**:
- **Informal**: sketches, user conversations
- **Formal**: mathematical models, structured documents

**Challenge**:  
Different stakeholders = different perspectives, leading to varying formats and abstraction levels.

**Example**:  
A business customer vs. a design engineer may interpret needs very differently.

## 2. Requirements vs. Features vs. Goals

| Term       | Definition                                                |
|------------|------------------------------------------------------------|
| Goal       | A broad, high-level objective (e.g., "automate home tasks") |
| Feature    | A function or capability that helps meet a goal (e.g., "water purifier") |
| Requirement| A detailed, implementable condition that fulfills a feature (e.g., "purifier must remove 99.9% of bacteria") |

**⚠️ Warning**:  
Goals are **too vague** to be used directly as requirements — they evolve and are difficult to test.

---

## 3. Requirement Classifications

Two main classification schemes:
- **Requirement Levels** (per Sommerville)
- **Requirement Types** (functional, non-functional, etc.)

---

## 4. Requirements Level Classification

**Sommerville’s 3 Levels**:

### 🔹 User Requirements
- High-level statements in natural language  
- Written for stakeholders  
- Example: *"System must process 20 bags/min"*

### 🔹 System Requirements
- Structured, detailed system behavior  
- Derived from user requirements  
- Found in the **SRS** (System Requirements Specification)  
- May use use-cases or technical annexes

### 🔹 Design Specifications
- Translate system requirements into **technical implementation plans**  
- Guide developers in building the system

---

## 5. Key Points

- Different levels = different levels of detail for different audiences  
- Requirements evolve:  
  **Goals → Requirements → Design**

- Use **user stories** and **use cases** to bridge communication gaps between users and developers

# 🧠 Requirements Engineering – Study Guide

---

## 1. Definition

**Requirements Engineering**:  
The process of discovering, analyzing, documenting, and managing the needs and constraints placed on a software system.

---

## 2. What Are Requirements?

- **Vary in formality**: From informal sketches to formal mathematical models  
- **Vary by stakeholder**:  
  - Business users need abstract goals  
  - Engineers need technical specifications

---

## 3. Types of Requirements

### 📌 Functional Requirements (FRs)
**What the system must do**

- Describe system behavior in response to inputs  
- Can be high-level (user-focused) or detailed (input/output/exception handling)

**Examples – Baggage System**:
- "System must process 20 bags/min."
- "If the conveyor fails, shut input feed within 3 seconds."

**Examples – POS System**:
- "Total button triggers sale to close out."
- "Item prices are listed per unit during checkout."

---

### 📌 Nonfunctional Requirements (NFRs)
**How the system performs** – quality attributes:

- Security  
- Reliability  
- Maintainability  
- Performance  
- Usability  
- Testability  
- Interoperability  
- Constraints (technical, financial, regulatory)

---

### 📌 Domain Requirements
**From the application domain** (e.g., legal or industry standards)

- May **add** new requirements  
- May **constrain** existing ones

**Examples**:
- Handling of coupons in POS  
- Conveyor rules in baggage handling

---

## 4. Levels of Requirements (Sommerville)

| Level               | Purpose                                      |
|---------------------|----------------------------------------------|
| User Requirements   | High-level, natural language, stakeholder-facing |
| System Requirements | Detailed specifications of system behavior   |
| Design Specifications | Technical implementation plans             |

**Flow**:  
`User → System → Design → Unit Testing → Integration → Acceptance`

---

## 5. Requirements Engineering Core Activities

### 🔹 Elicitation / Discovery  
- Understand stakeholder needs  
- Uncover hidden NFRs and domain constraints

---

### 🔹 Analysis and Agreement  
- Resolve ambiguity, contradiction, and incompleteness  
- Address interdependencies between requirements

Common Issues:
- Incomplete
- Contradictory
- Don’t make sense
- Interdependent

---

### 🔹 Representation (Modeling)  
- Use diagrams, text, or formal specs  
- Converts needs → system architecture

---

### 🔹 Validation  
"**Are we building the right product?**"  
Techniques:  
- Reviews  
- Walkthroughs  
- Models  
- Text inspections

---

### 🔹 Management  
- Track changes over time  
- Maintain traceability  
- Control **scope creep**

---

## 6. Domain Vocabulary

- Domain-specific terms **must** be clearly understood  
- Misinterpretations (e.g., “truck” vs. “package car”) can cause serious issues

---

## 7. Vignette – Requirements for a Brick

Even simple products need clear requirements.  
Example: **ASTM C62-17** brick spec

Specifies:
- Size  
- Strength  
- Durability  
- Weather resistance  

📌 Takeaway: **Documentation matters**, even for basic components.

---

## 8. Bodies of Knowledge

| Body               | Purpose                                                       |

| SWEBOK (2014)      | IEEE standard for software engineering education              |
| GSwE (2009)        | Graduate Software Engineering – systems focus                 |
| P&Ps of SE (Licensure) | Principles and practices for licensed engineers (USA)    |

# 🧠 Requirements Engineering Process – Study Guide

---

## 🔄 Key Stages in the Process

### 🔹 Requirements Elicitation
- Gather raw requirements from stakeholders  
- Identify all stakeholders early to avoid costly rework

---

### 🔹 Requirements Analysis
- Organize and refine requirements  
- Transform raw input into clear, high-quality needs  
- Hold collaborative sessions to align understanding

---

### 🔹 System Modeling
- Use UML, LML, or similar tools to visualize system behavior  
- Models help validate and communicate requirements effectively

---

### 🔹 Requirements Specification
- Create formal **System Requirements Specification (SRS)**  
- Follow **IEEE 830** standard for structure and clarity  
- Ensure all requirements are:
  - Verifiable  
  - Practical  
  - Signed-off by all key stakeholders

---

### 🔹 Requirements Validation
- Confirm with stakeholders: "Are we building the right product?"  
- Requirements must be:
  - Testable  
  - Unambiguous  
  - Concise

---

### 🔹 Requirements Management
- Ongoing activity across all stages  
- Track changes, maintain traceability, handle scope creep  
- Manage communication across diverse stakeholders

---

## ✅ Best Practices

### 🔸 Problem-Focused Requirements
- Define **what** to solve, not **how**  
- Avoid embedding solutions in requirement statements

### 🔸 Write Clear Requirements
- Use full, unambiguous sentences  
- Avoid jargon, abbreviations, and buzzwords  
- Ensure readability across technical and non-technical roles

### 🔸 Be Concise but Complete
- Keep requirements brief yet comprehensive  
- Ensure consistency and formatting uniformity

---

| Pitfall              | Description                                                                 |
|----------------------|-----------------------------------------------------------------------------|
| Scope Creep          | Gradual expansion of requirements; increases testing & documentation burden |

| Over-Engineering     | Extra features/code may disrupt existing system behavior                    |
| Lack of Feedback     | Delay in validation leads to late discovery of flaws                         |
| Ignoring Resistance  | Stakeholders may resist change—address early                                |
| Inconsistent Documentation | Leads to misinterpretation and implementation errors               |

- Follow **IEEE 830** format  
- Ensure requirements are:
  - Testable
