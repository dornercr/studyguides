## 1.1 Source File & Encoding

- Python source files are plain text `.py` files.
- Default encoding: `UTF-8`.
- Custom encoding: add a coding declaration in the **first** or **second** line:

  ~~~python
  # -*- coding: iso-8859-1 -*-
  ~~~

- If file contains non-UTF-8 bytes and no encoding header → `SyntaxError`.

**Mnemonic:** Ugly Trolls Fart 8 times  
**Video:** [Character Encoding in Python](https://www.youtube.com/watch?v=CxeP1dQi9rg)

---

## 1.2 Lexical Structure & Tokenization

- Token: the smallest unit in source code (produced by lexical analysis).
- Types of tokens:
  - **Keywords** – reserved words like `if`, `def`, `return`
  - **Identifiers** – user-defined names (variables, functions, etc.)
  - **Literals** – fixed values like `42`, `'hi'`, `3.14`, `True`
  - **Operators** – symbols like `+`, `-`, `*`, `==`, `and`
  - **Delimiters** – structural markers like `()`, `[]`, `{}`, `,`, `:`
- Tokens → compiled into Abstract Syntax Tree (AST).
- Whitespace and comments are ignored during tokenization.

**Mnemonics:**  
- Killer Words Defy Renaming  
- I Name Everything  
- Let It Take Real Actual Life  
- Octopuses Prefer Arithmetic  
- Don’t Ever Leave Internal Markers Empty Randomly  
- Trees Only Grow on Tokens

**Video:** [Lexical Structure of Python](https://www.youtube.com/watch?v=m_8X6moMDGc)

---

## 1.3 Identifiers (Names)

- Must start with a letter or `_`, followed by letters/digits/underscores.
- Cannot use keywords as identifiers.
- Case-sensitive.
- Naming conventions:
  - `_var`: weak “internal use” signal.
  - `__var`: triggers name mangling in classes.
  - `__name__`: dunder names reserved for Python internals (e.g., `__init__`).

**Mnemonics:**  
- I Name Everything  
- Underscore = Unofficial  
- Double Down to Mangle  
- Don’t Use Names Designed Especially Reserved

**Video:** [What Is An Identifier In Python](https://www.youtube.com/watch?v=Sf2EzJogbcc)

---

## 1.4 Literals (Fixed Values)

- **Numeric literals**: `int`, `float`, `complex`  
  Example: `42`, `3.14`, `1+2j`
- **Text & byte literals**:  
  `str`, `bytes`, `bytearray`  
  Example: `'hi'`, `b'hi'`, `bytearray(b'hi')`
- **Special constants**:  
  `True`, `False`, `None`, `Ellipsis (...)`, `NotImplemented`
  - `Ellipsis`: used in slicing (`x[...]`), type hints, or as a placeholder:
    ~~~python
    x = ...  # TODO placeholder
    ~~~
  - `NotImplemented`: returned by binary special methods when unsupported:
    ~~~python
    def __eq__(self, other):
        if not isinstance(other, MyClass):
            return NotImplemented
    ~~~

**Mnemonics:**  
- Number Formats Can Switch  
- String Bytes Always  
- The Fool Never Escapes Nothing  
- Every Lazy Line  
- No Operation There

**Video:** [Python Literals Explained](https://www.youtube.com/watch?v=8Pvv9W9ix5Y)

## 1.5 Operators & Special Methods

- Python operators are implemented via special methods:
  - `+` → `__add__`
  - `==` → `__eq__`
  - `+=` → `__iadd__`, or `__add__` fallback
  - Right-hand fallback → `__radd__`, etc.
- `=` is **not** an operator and cannot be overloaded.

**Mnemonics:**  
- Overloads Are Magic Methods  
- Right Always Gets A Shot  
- Increase If Able  
- Equals Isn’t Overridable

## 1.6 Indentation & Block Structure

- Python uses **indentation**, not braces `{}` to define blocks.
- PEP 8 recommends 4 spaces per level (no tabs).
- Mixing tabs and spaces raises a `TabError`.

**Mnemonics:**  
- Indentation Beats Braces  
- Proper Eight Paces

## 1.8 Comments & Docstrings

## 2.1 Built-in Data Types

**Definition:**  
Python provides fundamental object types that serve as the core of its runtime model. All values are objects, and each has a type.

**Core Categories:**

- **Numeric Types**: `int`, `float`, `complex`
- **Sequence Types**: `str`, `list`, `tuple`, `range`
- **Set Types**: `set`, `frozenset`
- **Mapping Type**: `dict`
- **Boolean Type**: `bool`
- **Binary Types**: `bytes`, `bytearray`, `memoryview`
- **NoneType**: `None`

**Rules & Classifications:**

- `int`: Arbitrary-precision integers
- `float`: IEEE-754 64-bit floating point
- `complex`: Real and imaginary parts
- `str`: Immutable sequence of Unicode characters
- `list`: Mutable ordered sequence
- `tuple`: Immutable ordered sequence
- `set`: Unordered collection of unique elements
- `dict`: Mutable key-value mapping
- `bool`: Subclass of `int` (`True == 1`, `False == 0`)
- `None`: Singleton object representing null value

**Edge Cases:**

- `int(1e100)` → valid, very large integer
- `float('inf')`, `float('nan')` → IEEE 754-compliant
- `set([1, 1])` → `{1}`
- `bool("False")` → `True` (non-empty strings are truthy)

**Examples:**
~~~python
a = 123                   # int
b = 3.14                  # float
c = 1 + 2j                # complex
d = "hello"               # str
e = [1, 2, 3]             # list
f = (1, 2)                # tuple
g = {"x": 1, "y": 2}      # dict
h = {1, 2, 2}             # set → {1, 2}
i = b"abc"                # bytes
j = None                  # NoneType
~~~

## 2.2 Type Conversion

**Definition:**  
Changing an object from one type to another using built-in conversion functions.

**Rules:**

- Use functions like `int()`, `float()`, `str()`, `bool()`, etc.
- Implicit coercion: `int + float` → `float`
- `bool(x)` is `False` only for: `0`, `0.0`, `None`, `''`, `[]`, `{}`, etc.

**Edge Cases:**

- `int('3.0')` → `ValueError`
- `int(' 42\n')` → `42` (whitespace stripped)
- `bool('False')` → `True` (non-empty)

**Examples:**
~~~python
int('10')              # 10
float(5)               # 5.0
str(3.14)              # '3.14'
bool([])               # False
bool(' ')              # True
~~~

**Mnemonic:** Convert If Safe, Fail Loud

## 2.3 Variable Assignment

**Definition:**  
A variable is a name bound to an object in memory. Python uses dynamic typing and late binding.

**Rules:**

- Assignment uses reference semantics.
- Multiple names can point to the same object.
- Types are determined at runtime.

**Edge Cases:**

- `a = b = []` → both `a` and `b` refer to same list.
- Changing one changes the other unless reassigned.

**Examples:**
~~~python
x = [1, 2]
y = x
x.append(3)
print(y)        # [1, 2, 3]

x = [1]
y = x
x = [2]
print(y)        # [1]
~~~

**Mnemonic:** Binding Is Not Copying  
**Video:** [Python Variables Explained](https://www.youtube.com/watch?v=lZbIKeTpz2A)

## 2.4 Object Identity vs Equality

**Definition:**

- `is`: tests identity (same memory address)
- `==`: tests equality (same value)

**Rules:**

- `is` is `True` if `id(x) == id(y)`
- Use `is` for `None`, `True`, `False`, small integers
- Use `==` for value comparison

**Edge Cases:**
~~~python
x = 256
y = 256
x is y       # True (interned)

x = 257
y = 257
x is y       # False (not interned)
~~~

**Examples:**
~~~python
x = [1]
y = [1]
x == y       # True
x is y       # False
~~~

**Mnemonic:** Same Spot vs Same Shape  
**Video:** [Python Identity vs Equality](https://www.youtube.com/watch?v=HGOBQPFzWKo)

## 2.5 Mutable vs Immutable Types

**Definition:**

- **Mutable**: Can be changed in-place
- **Immutable**: Cannot be changed after creation

**Immutable Types:** `int`, `float`, `str`, `tuple`, `frozenset`  
**Mutable Types:** `list`, `dict`, `set`, `bytearray`

**Rules:**

- Assigning to a mutable object affects all bindings
- Immutable reassignment creates a new object

**Edge Cases:**
~~~python
x = (1, [2])
x[1].append(3)      # Allowed – inner list is mutable
~~~

**Examples:**
~~~python
s = 'abc'
s += 'd'            # new string created

l = [1, 2]
l.append(3)         # modifies in place
~~~

**Mnemonic:** Touch Changes Mutable  
**Video:** [Mutable vs Immutable Types](https://www.youtube.com/watch?v=RSZQA4PoxnY)

## 2.6 id(), type(), isinstance()

**Definitions:**

- `id(x)`: Returns identity (memory surrogate)
- `type(x)`: Returns actual class of object
- `isinstance(x, T)`: Checks if `x` is instance or subclass of `T`

**Rules:**

- `type(x) is T` only matches exact type
- `isinstance()` supports inheritance and Union types

**Edge Cases:**
~~~python
class A: pass
class B(A): pass

isinstance(B(), A)   # True
type(B()) is A       # False
~~~

**Examples:**
~~~python
x = 42
print(type(x))              # <class 'int'>
print(isinstance(x, int))   # True
print(id(x))                # e.g. 140712466763184
~~~

**Mnemonic:** ID Tracks, Type Identifies, Instance Checks

````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

# (question, answer, PEMDAS-style mnemonic)
flashcards = [
("What are Python’s core built-in data type categories?",
 "`int`, `float`, `complex`, `str`, `list`, `tuple`, `range`, `set`, `frozenset`, `dict`, `bool`, `bytes`, `bytearray`, `memoryview`, `NoneType`",
 "**Mnemonic:** In Funny Class, Silly Little Turtles Run So Fast Dancing Bears Bounce Big, Making Noodles Last!"),

("What is an int in Python?",
 "An arbitrary-precision integer.\nExample:\n```python\nx = 10**100  # valid\n```",
 "**Mnemonic:** Integers Never Topple"),

("What is a float in Python?",
 "64-bit IEEE-754 floating-point number.\nSupports `float('inf')`, `float('nan')`.\nExample:\n```python\nx = 3.14\n```",
 "**Mnemonic:** Float Like Infinity Tides"),

("What is a complex number?",
 "Stores real and imaginary parts.\nExample:\n```python\nx = 3 + 2j\n```",
 "**Mnemonic:** Complex Junk = Just Imaginary"),

("What is a str?",
 "Immutable Unicode character sequence.\n```python\ns = 'hello'\ns[0] = 'H'  # error\n```",
 "**Mnemonic:** Strings Stay Stuck"),

("What is a list?",
 "Mutable ordered sequence.\n```python\nx = [1, 2]; x.append(3)\n```",
 "**Mnemonic:** Lists Love Change"),

("What is a tuple?",
 "Immutable ordered sequence.\n```python\nt = (1, 2, 3)\nt[0] = 99  # error\n```",
 "**Mnemonic:** Tuples Typically Freeze"),

("What is a set or frozenset?",
 "`set`: mutable, `frozenset`: immutable.\nBoth store unique, unordered elements.\n```python\nset([1, 1])  # {1}\n```",
 "**Mnemonic:** Sets Erase Twins"),

("What is a dict?",
 "Mutable key-value mapping.\n```python\nd = {'a': 1}; d['b'] = 2\n```",
 "**Mnemonic:** Dictionaries Index Clever Things"),

("What is the bool type?",
 "`bool` is a subclass of int. `True == 1`, `False == 0`.\n```python\nbool('False')  # True (non-empty string)\n```",
 "**Mnemonic:** Booleans Often Underlie Logic"),

("What is type conversion in Python?",
 "Changing types using `int()`, `float()`, `str()`, `bool()`, etc.",
 "**Mnemonic:** Convert If Safe, Fail Loud"),

("What happens with implicit type conversion?",
 "Mixed types in expressions auto-convert: `int + float → float`",
 "**Mnemonic:** Implicit Numbers Float"),

("How does `bool(x)` behave?",
 "Falsy values: `None`, `0`, `''`, `[]`, etc. All others → True",
 "**Mnemonic:** Big Objects Usually Live"),

("What’s a ValueError edge case with int()?",
 "```python\nint('3.0') → ValueError\nint(' 42\\n') → 42\n```",
 "**Mnemonic:** Strings With Dots Die"),

("Give examples of type conversion.",
 "```python\nint('10')\nstr(3.14)\nbool([])\nfloat(5)\n```",
 "**Mnemonic:** In Students, Brains Fail"),

("How does assignment work in Python?",
 "Assignment binds names to objects via references, not copies.",
 "**Mnemonic:** Binding Is Not Copying"),

("What is dynamic typing?",
 "Types are assigned at runtime, not declared.",
 "**Mnemonic:** Duck Typing Does It"),

("What’s the a = b = [] behavior?",
 "Both names point to the same list object.",
 "**Mnemonic:** Assignments Bounce Between"),

("How do you break shared reference?",
 "Reassign one variable:\n```python\nx = [1]; y = x; x = [2]\n```",
 "**Mnemonic:** Rebind To Break"),

("What’s the difference between `is` and `==`?",
 "`is` checks identity (same object), `==` checks value equality.",
 "**Mnemonic:** Same Spot vs Same Shape"),

("When does `is` return True?",
 "When both names point to the same object (same `id()` value).",
 "**Mnemonic:** Is Is Identity"),

("What’s an interning edge case?",
 "```python\nx = 256; y = 256; x is y  # True\nx = 257; y = 257; x is y  # False\n```",
 "**Mnemonic:** Interning Keeps Common Small Integers"),

("Example: equality vs identity for lists?",
 "```python\n[1] == [1]  # True\n[1] is [1]  # False\n```",
 "**Mnemonic:** Lists Look Equal But Are Not Identical"),
("What is a mutable object?",
 "It can be changed in-place (e.g., `list`, `dict`).",
 "**Mnemonic:** Touch Changes Mutable"),

("What is an immutable object?",
 "It cannot be changed once created (e.g., `str`, `tuple`).",
 "**Mnemonic:** Immutable Means Frozen"),

("What happens with shared mutables?",
 "Changing one reference changes the shared object.",
 "**Mnemonic:** Mutants Multiply Side Effects"),

("What happens with reassignment of immutables?",
 "Reassignment creates a new object (not in-place).",
 "**Mnemonic:** Frozen Things Break When Reassigned"),

("What’s a tricky edge case with tuples?",
 "Tuples are immutable, but can contain mutables:\n```python\nt = (1, [2]); t[1].append(3)\n```",
 "**Mnemonic:** Tuples May Harbor Secrets"),

("What does `id(x)` return?",
 "The identity (memory surrogate) of an object.",
 "**Mnemonic:** Identity Defines"),

("What does `type(x)` return?",
 "The direct type (class) of an object.",
 "**Mnemonic:** Type Is Precise"),

("What does `isinstance(x, T)` do?",
 "Returns True if `x` is an instance of `T` or a subclass.",
 "**Mnemonic:** Inheritance Supported Instantly"),

("What’s the difference between `type(x) is T` vs `isinstance()`?",
 "`type(x) is T` only checks exact match. `isinstance()` checks hierarchy.",
 "**Mnemonic:** Type Is Strict, Instance Is Soft")
]

index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'  # Hidden initially

def update_card():
    with output:
        clear_output(wait=True)
        question, answer, mnemonic = flashcards[index]
        display(Markdown(f"**Q:** {question}"))
        if showing_answer:
            display(Markdown(f"\n---\n\n**A:** {answer}\n\n{mnemonic}"))

    # Update button states
    prev_button.disabled = index == 0
    next_button.disabled = False  # always enabled unless at end
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"

def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Attach handlers
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Initialize display
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
````
