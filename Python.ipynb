# 🧠 Section 1: Language Fundamentals (with YouTube links)

## 1.1 Source File & Encoding

- A Python source file is a `.py` text file.
- Default encoding: UTF‑8.
- Custom encoding must be declared:

```python
# -*- coding: iso-8859-1 -*-
```

**Video:** Character Encoding in Python  
https://www.youtube.com/watch?v=CxeP1dQi9rg

---

## 1.2 Lexical Structure & Tokenization

- Token = smallest unit of code.
- Categories: keyword, identifier, literal, operator, delimiter.
- Tokens are compiled into an abstract syntax tree (AST).

**Video:** Lexical Structure of Python  
https://www.youtube.com/watch?v=m_8X6moMDGc

---

## 1.3 Identifiers (Names)

- Start with a letter or `_`, followed by letters/digits/underscores.
- Case-sensitive. Cannot use keywords.
- Special forms: `_name`, `__name`, `__name__`.

**Video:** What Is An Identifier In Python  
https://www.youtube.com/watch?v=Sf2EzJogbcc

---

## 1.4 Literals (Fixed Values)

- Numeric: int, float, complex.
- String types: str, bytes, bytearray.
- Special: `True`, `False`, `None`, `Ellipsis`, `NotImplemented`.

**Video:** Python Literals Explained  
https://www.youtube.com/watch?v=8Pvv9W9ix5Y

---

## 1.5 Operators & Special Methods

- Operators map to methods: `+` → `__add__`, `==` → `__eq__`, etc.
- Categories: arithmetic, bitwise, comparison, logical, identity, membership, assignment.

**Video:** Python Operators for Beginners  
https://www.youtube.com/watch?v=7BxUaeROVXI

---

## 1.6 Indentation & Block Structure

- Indentation (not braces) defines block scope.
- Consistency required. PEP 8 recommends 4 spaces.

**Video:** Python Indentation Tutorial  
https://www.youtube.com/watch?v=l4CdyZPDYE4

---

## 1.7 Line Structure

- Physical line = ends with newline.
- Logical line = may span multiple lines via `\\` or grouping.

**Video:** (covered within full tutorials; no standalone link)

---

## 1.8 Comments & Docstrings

- `#` starts a comment.
- Docstring = first string literal in a class/module/function.

**Video:** (covered within full tutorials; no standalone link)

---

## 1.9 Truth Value Testing

- Falsy: `None`, `False`, 0, `''`, `[]`, `{}`.
- Custom: `__bool__()` or `__len__()`.

**Video:** (covered within full tutorials; no standalone link)

---

## 1.10 Execution Context

- `__name__ == "__main__"` checks if script was executed directly.
- `__file__` and `sys.argv` are context variables.

**Video:** (covered within full tutorials; no standalone link)

---

## 1.11 Shebang

- `#!/usr/bin/env python3` for Unix interpreter line.
- Must be at line 1, column 0.

````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

# (question, answer, PEMDAS-style mnemonic)
flashcards = [
# 1.1 Source Encoding
("What is the default encoding of Python source files?",
 "`UTF-8` is the default encoding for `.py` files.",
 "**Mnemonic:** Ugly Trolls Fart 8 times"),

("How do you override the source encoding?",
 "Place a coding declaration in the first or second line:\n```python\n# -*- coding: iso-8859-1 -*-\n```",
 "**Mnemonic:** She’s First or Second"),

("What happens if encoding is wrong and undeclared?",
 "If the file has non-UTF-8 bytes and no encoding header, Python raises `SyntaxError`.",
 "**Mnemonic:** Stop Errors From Unicode"),

# 1.2 Token Types
("What is a keyword in Python?",
 "A `keyword` is a reserved word like `if`, `def`, `return` — cannot be used as identifiers.",
 "**Mnemonic:** Killer Words Defy Renaming"),

("What is an identifier?",
 "An `identifier` is the name of a variable, function, class, etc. Must follow naming rules.",
 "**Mnemonic:** I Name Everything"),

("What is a literal in Python?",
 "A `literal` is a fixed value: `42`, `'hi'`, `3.14`, `True`, etc.",
 "**Mnemonic:** Let It Take Real Actual Life"),

("What is an operator?",
 "`Operator` performs computation: `+`, `-`, `==`, `and`, etc.",
 "**Mnemonic:** Octopuses Prefer Arithmetic"),

("What is a delimiter?",
 "A `delimiter` separates structure: `()`, `[]`, `{}`, `,`, `:`",
 "**Mnemonic:** Don’t Ever Leave Internal Markers Empty Randomly"),

("What happens after tokenization?",
 "Tokens are compiled into an abstract syntax tree (AST). Whitespace and comments are ignored.",
 "**Mnemonic:** Trees Only Grow on Tokens"),

# 1.3 Identifiers
("What characters are valid in identifiers?",
 "Must start with a letter or `_`, followed by letters, digits, underscores. No keywords.",
 "**Mnemonic:** Start Letters Define Unique Code"),

("What’s special about `_var`?",
 "`_var` is a weak ‘internal use’ convention (not enforced).",
 "**Mnemonic:** Underscore = Unofficial"),

("What about `__var`?",
 "`__var` triggers name mangling inside classes.",
 "**Mnemonic:** Double Down to Mangle"),

("What are dunder names like `__init__`?",
 "Reserved names used by Python’s internals. Don’t define your own dunders unless required.",
 "**Mnemonic:** Don’t Use Names Designed Especially Reserved"),

# 1.4 Literals
("What are numeric literals?",
 "`int`, `float`, `complex` — like `3`, `2.5`, `1+2j`",
 "**Mnemonic:** Number Formats Can Switch"),

("What are text and byte literals?",
 "`str`, `bytes`, `bytearray` — like `'hi'`, `b'hi'`, `bytearray(b'hi')`",
 "**Mnemonic:** String Bytes Always"),

("What are special constants?",
 "`True`, `False`, `None`, `Ellipsis` (`...`), `NotImplemented`",
 "**Mnemonic:** The Fool Never Escapes Nothing"),

("How is `Ellipsis` used?",
 "Used in slicing, type hints, and as a placeholder.\n```python\nx = ...  # TODO\n```",
 "**Mnemonic:** Every Lazy Line"),

("What is `NotImplemented` used for?",
 "Return it from binary special methods if the operation isn’t supported by that operand.",
 "**Mnemonic:** No Operation There"),

# 1.5 Operators
("How does Python implement operators?",
 "Via special methods like `__add__`, `__eq__`, etc.",
 "**Mnemonic:** Overloads Are Magic Methods"),

("What if `__add__` fails?",
 "Python tries `__radd__` from the right-hand operand.",
 "**Mnemonic:** Right Always Gets A Shot"),

("What about `+=`?",
 "`+=` uses `__iadd__` if available; otherwise falls back to `__add__`.",
 "**Mnemonic:** Increase If Able"),

("Can you override `=`?",
 "No — assignment is a syntax form, not an operator.",
 "**Mnemonic:** Equals Isn’t Overridable"),

# 1.6 Indentation
("What defines block structure in Python?",
 "Indentation, not braces.",
 "**Mnemonic:** Indentation Beats Braces"),

("What’s the PEP 8 indent rule?",
 "Use 4 spaces per indentation level.",
 "**Mnemonic:** Proper Eight Paces"),

("What happens if tabs and spaces are mixed?",
 "Raises `TabError` at runtime.",
 "**Mnemonic:** Tabs And Braces Explode"),

# 1.7 Lines
("What’s a physical vs logical line?",
 "Physical ends with newline. Logical may span lines via `\\` or inside `()`, `[]`, `{}`.",
 "**Mnemonic:** Physically Long Code Bends"),

("How do you extend a statement across lines?",
 "Use a backslash or wrap inside parentheses, brackets, or braces.",
 "**Mnemonic:** Continue Inside Brackets"),

# 1.8 Comments & Docstrings
("How do comments work?",
 "`#` starts a comment that continues to end of line.",
 "**Mnemonic:** Sharp Ended"),

("What is a docstring?",
 "First triple-quoted string in a class, function, or module. Stored in `__doc__`.",
 "**Mnemonic:** Doc Lives In The First Line"),

# 1.9 Truthiness
("Which values are falsy?",
 "`False`, `None`, 0, 0.0, `''`, `[]`, `{}`, etc.",
 "**Mnemonic:** Fools Never Zero Their Stuff"),

("How do objects define truth?",
 "Implement `__bool__()` or `__len__()`.\nIf both, `__bool__()` is used.",
 "**Mnemonic:** Bool Over Length"),

# 1.10 Execution Context
("When is `__name__ == '__main__'` true?",
 "When the script is run directly (not imported).",
 "**Mnemonic:** Main Means Direct"),

("What is `__file__`?",
 "Path to the current file.",
 "**Mnemonic:** File Is Location Evidence"),

("What is `sys.argv`?",
 "List of command-line arguments passed to the script.",
 "**Mnemonic:** Argument Vector Gives Parameters"),

# 1.11 Shebang
("What is a shebang and where does it go?",
 "Starts with `#!` on the first line. Tells Unix which interpreter to run.\nExample:\n```python\n#!/usr/bin/env python3\n```",
 "**Mnemonic:** Shebang Hits Immediately")
]
index = 0
showing_answer = False
output = widgets.Output()

# Define buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")

def update_card():
    with output:
        clear_output(wait=True)
        question, answer, mnemonic = flashcards[index]
        display(Markdown(f"**Q:** {question}"))
        if showing_answer:
            display(Markdown(f"\n---\n\n**A:** {answer}\n\n{mnemonic}"))

    # Disable/enable navigation buttons
    prev_button.disabled = index == 0
    next_button.disabled = index >= len(flashcards) - 1
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"

def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

# Attach handlers
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)

# Initial display
update_card()
display(widgets.HBox([prev_button, show_button, next_button]), output)
````
