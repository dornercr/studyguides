# 🧠 Section 1: Language Fundamentals (with YouTube links)

## 1.1 Source File & Encoding

- A Python source file is a `.py` text file.
- Default encoding: UTF‑8.
- Custom encoding must be declared:

```python
# -*- coding: iso-8859-1 -*-
```

**Video:** Character Encoding in Python  
https://www.youtube.com/watch?v=CxeP1dQi9rg

---

## 1.2 Lexical Structure & Tokenization

- Token = smallest unit of code.
- Categories: keyword, identifier, literal, operator, delimiter.
- Tokens are compiled into an abstract syntax tree (AST).

**Video:** Lexical Structure of Python  
https://www.youtube.com/watch?v=m_8X6moMDGc

---

## 1.3 Identifiers (Names)

- Start with a letter or `_`, followed by letters/digits/underscores.
- Case-sensitive. Cannot use keywords.
- Special forms: `_name`, `__name`, `__name__`.

**Video:** What Is An Identifier In Python  
https://www.youtube.com/watch?v=Sf2EzJogbcc

---

## 1.4 Literals (Fixed Values)

- Numeric: int, float, complex.
- String types: str, bytes, bytearray.
- Special: `True`, `False`, `None`, `Ellipsis`, `NotImplemented`.

**Video:** Python Literals Explained  
https://www.youtube.com/watch?v=8Pvv9W9ix5Y

---

## 1.5 Operators & Special Methods

- Operators map to methods: `+` → `__add__`, `==` → `__eq__`, etc.
- Categories: arithmetic, bitwise, comparison, logical, identity, membership, assignment.

**Video:** Python Operators for Beginners  
https://www.youtube.com/watch?v=7BxUaeROVXI

---

## 1.6 Indentation & Block Structure

- Indentation (not braces) defines block scope.
- Consistency required. PEP 8 recommends 4 spaces.

**Video:** Python Indentation Tutorial  
https://www.youtube.com/watch?v=l4CdyZPDYE4

---

## 1.7 Line Structure

- Physical line = ends with newline.
- Logical line = may span multiple lines via `\\` or grouping.

**Video:** (covered within full tutorials; no standalone link)

---

## 1.8 Comments & Docstrings

- `#` starts a comment.
- Docstring = first string literal in a class/module/function.

**Video:** (covered within full tutorials; no standalone link)

---

## 1.9 Truth Value Testing

- Falsy: `None`, `False`, 0, `''`, `[]`, `{}`.
- Custom: `__bool__()` or `__len__()`.

**Video:** (covered within full tutorials; no standalone link)

---

## 1.10 Execution Context

- `__name__ == "__main__"` checks if script was executed directly.
- `__file__` and `sys.argv` are context variables.

**Video:** (covered within full tutorials; no standalone link)

---

## 1.11 Shebang

- `#!/usr/bin/env python3` for Unix interpreter line.
- Must be at line 1, column 0.
