# Section 1: Linear Algebra Basics – Study Guide

---

## 🔹 What Is Algebra?

Algebra means solving math problems using letters or symbols instead of just numbers.

**Example**  
Solve: `2x + 5 = 25`  
→ Subtract 5 from both sides: `2x = 20`  
→ Divide both sides by 2: `x = 10`

---

## 🔹 What Is Linear Algebra?

Linear algebra focuses on solving multiple linear equations at once.

A **linear equation** does **not** include:
- Powers (like `x²`)
- Roots (like `√x`)
- Curves

✅ `2x + 5 = 25` → Linear  
❌ `2x² + 5 = 25` → Nonlinear

**Three possible outcomes:**
- One solution → lines intersect once  
- No solution → lines are parallel  
- Infinite solutions → lines are the same

---

## 🔹 Real-World Example: YouTube Bot Race

**Scenario**  
Two bots scrape videos at different speeds:

- Bot A: 150 videos/hour → 2.5 videos/min, starts 5 minutes early  
- Bot B: 180 videos/hour → 3 videos/min

**Equations**
- Bot A: `V = 2.5T`  
- Bot B: `V = 3(T - 5)`

**Solve**
`2.5T = 3(T - 5)`  
→ `2.5T = 3T - 15`  
→ `-0.5T = -15`  
→ `T = 30 minutes`  
→ `V = 2.5 × 30 = 75 videos`

---

## 🔹 Linear Systems in Machine Learning

In machine learning, we use linear algebra to make predictions based on multiple inputs.

**Model Form**  
`Score = a×X₁ + b×X₂ + c×X₃ + ... + bias`

**Key Terms**
- `Score`: The predicted value  
- `X₁, X₂, ...`: Input features (e.g. video length, keyword count)  
- `a, b, c`: Weights (importance of each input)  
- `bias`: A constant added to the prediction

Each data row = one equation  
All rows = a system of linear equations to solve

## 🔹 Paper and Pencil Practice

Let’s revisit a real-world linear algebra problem with graphing.

**Example:**  
A cyclist rides at 20 km/h starting at time 0.  
A scooter starts 15 minutes later (0.25 hours) at 30 km/h.

We want to find:
- When the scooter catches up
- How far both have traveled by then

**Graph Setup:**
- X-axis: time (in hours)  
- Y-axis: distance (in km)  
- Cyclist line: starts at 0, slope = 20  
- Scooter line: starts at 0.25, slope = 30

**Interpretation:**  
The point where the two lines intersect gives the solution.

---

## 🔹 Solving the Same Problem in Python

We can also solve and plot this scenario using Python.

---

## 🔹 Key Output

From the math or graph:
- **Time** = 0.75 hours (45 minutes)
- **Distance** = 15 km

**Conclusion:**  
The scooter catches the cyclist 15 km down the road after 45 minutes.

---

## 🔹 Why This Practice Matters

- Solving by **graphing** shows how straight lines behave  
- Solving by **code** gives speed, precision, and visualization  
- Both deepen your understanding of linear systems


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a Time Range
time = np.linspace(0, 1.5, 1000)  # Simulate from 0 to 1.5 hours

# 2. Define Distance Equations
distance_cyclist = 20 * time
distance_scooter = 30 * (time - 0.25)
distance_scooter[time < 0.25] = 0  # Scooter hasn't started yet

# 3. Plot the Chart
plt.figure(figsize=(8, 5))
plt.plot(time, distance_cyclist, label="Cyclist", color="blue")
plt.plot(time, distance_scooter, label="Scooter", color="red")

# Add intersection point lines
plt.axvline(x=0.75, color='purple', linestyle='--')  # Vertical line at 0.75 hr
plt.axhline(y=15, color='purple', linestyle='--')    # Horizontal line at 15 km

plt.title("Scooter Catches Cyclist")
plt.xlabel("Time (hours)")
plt.ylabel("Distance (km)")
plt.xlim(0, 1.5)
plt.ylim(0, 40)
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sympy import symbols, Eq, solve

# Define time variable
t = symbols('t')

# Cyclist distance: 20 km/h * t
# Scooter distance: 30 km/h * (t - 0.25)
distance_cyclist = 20 * t
distance_scooter = 30 * (t - 0.25)

# Set up the equation: when distances are equal
equation = Eq(distance_cyclist, distance_scooter)

# Solve the equation
solution = solve(equation, t)[0]

# Calculate distance at that time
distance = distance_cyclist.subs(t, solution)

print(f"Scooter catches up at: {float(solution):.2f} hours")
print(f"Distance traveled: {float(distance):.2f} km")
```
