# 🧠 Lesson: Logic, Propositions, and Critical Thinking in Mathematics

---

## 🔹 1. Logic in Mathematics

### 🧠 Theory:
Logic is the study of reasoning—how we move from known facts to valid conclusions.  
In mathematics, logic underpins **proofs**, theorems, and formulas like the **quadratic formula** and **Pythagorean Theorem**.  
It allows us to express reasoning using **precise language** and **symbols**.

✅ Example:  
> If Judy likes all things round, and donuts are round, then Judy will love donuts.

This is an **if–then** logical statement where both premises are true, leading to a true conclusion.

---

## 🔹 2. Propositions

### 🧠 Definition:
A **proposition** is a statement that can be **labeled as either true or false**.

✅ Examples:
- "All squares are yellow."
- "Judy likes all things pink."
- "x = 2" (in algebra)
- "Line segment AB bisects line segment CD" (in geometry)

💡 In math, propositions can be:
- Words only  
- Words + symbols  
- Symbols only

---

## 🔹 3. True or False in Logic

### 🧠 Key Point:
When a problem **states** that something is true or false, **accept it** for the sake of the problem—even if it contradicts real-world knowledge.

✅ Example:  
If a problem says `2 + 2 = 5` is true, **treat it as true** for that scenario.

---

## 🔹 4. Critical Thinking in Math

### 🧠 Definition:
Critical thinking is using **what is known to be true** to form **new logical connections**.

✅ Example:  
Given:
- `x = 5` (true)
- `y = 1` (true)

We can deduce:
- If `z = x + y`, then `z = 6`
- If `z = x * y`, then `z = 5`

We **connect known truths** to derive **new propositions**.

---

## 🔹 5. Using If–Then Statements

Mathematicians often express connections in **conditional form**:

```
If [condition], then [result].
```

✅ Example:
> If a number is divisible by 6, then it is divisible by 3.

This can also be expressed in symbolic logic:
- P: "Number is divisible by 6"
- Q: "Number is divisible by 3"
- P → Q

---

## 🔹 6. Summary Table

| Concept         | Definition | Example |
|-----------------|------------|---------|
| Logic           | Study of reasoning with true/false statements | "If P then Q" |
| Proposition     | Statement that is true or false | "x = 2" |
| True/False      | Accept given truth values in problems | "2+2=5" (true for problem) |
| Critical Thinking | Drawing new conclusions from known truths | From x=5, y=1 deduce z=6 |
| If–Then         | Conditional statement form | "If divisible by 6 → divisible by 3" |

---

````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output, HTML

# (question, answer, PEMDAS-style mnemonic)


flashcards = [

("What is logic in mathematics?",
"Logic is the study of reasoning—how we move from known facts to valid conclusions. It underpins proofs, theorems, and formulas, allowing reasoning to be expressed with precise language and symbols."),

("Give an example of a logical if–then statement in everyday terms.",
"If Judy likes all things round, and donuts are round, then Judy will love donuts."),

("What is a proposition?",
"A proposition is a statement that can be labeled as either true or false."),

("Give three examples of propositions.",
"'All squares are yellow.', 'x = 2', 'Line segment AB bisects line segment CD'."),

("What forms can propositions take in mathematics?",
"Words only, words plus symbols, or symbols only."),

("How should you treat given truth values in a problem?",
"Accept them as stated for the sake of the problem, even if they contradict real-world facts."),

("Give an example of accepting a false real-world statement as true in logic.",
"If a problem says '2 + 2 = 5' is true, treat it as true for that scenario."),

("What is critical thinking in mathematics?",
"Using what is known to be true to form new logical connections and conclusions."),

("Provide an example of critical thinking in math.",
"Given x = 5 and y = 1, deduce z = x + y = 6 or z = x * y = 5."),

("What is an if–then statement?",
"A conditional statement in the form 'If [condition], then [result]'."),

("Give a mathematical example of an if–then statement.",
"If a number is divisible by 6, then it is divisible by 3."),

("How can the example 'If divisible by 6 → divisible by 3' be expressed in symbolic logic?",
"P: 'Number is divisible by 6', Q: 'Number is divisible by 3', P → Q."),

("Summarize the definition of 'Logic' from the lesson's table.",
"Study of reasoning with true/false statements."),

("Summarize the definition of 'Proposition' from the lesson's table.",
"Statement that is true or false."),

("Summarize the 'True/False' concept from the lesson's table.",
"Accept given truth values in problems regardless of real-world accuracy."),

("Summarize 'Critical Thinking' from the lesson's table.",
"Drawing new conclusions from known truths."),

("Summarize 'If–Then' from the lesson's table.",
"A conditional statement form connecting a condition to a result.")
]



index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'  # Hidden initially

def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]
        if len(card) == 3:
            question, answer, mnemonic = card
        else:
            question, answer = card
            mnemonic = ""

        html = f"""
        <div style='background:#f9f9f9;padding:16px;border-radius:10px;
                    border:1px solid #ddd;line-height:1.6;font-family:monospace;
                    white-space:pre-wrap;font-size:14px'>
        <b>Q:</b> {question}<br><br>
        {f"<b>A:</b><pre>{answer}</pre><br><b>🔁 Mnemonic:</b> {mnemonic}" if showing_answer else ""}
        </div>
        """
        display(HTML(html))

    # Update button states
    prev_button.disabled = index == 0
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"


def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Attach handlers
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Initialize display
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
````
