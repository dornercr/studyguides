# 📘 Section 1.1: What is Matplotlib?

---

## 🧠 What is it?

**Matplotlib** is a Python library for creating **2D visualizations** — like:

- Line graphs  
- Bar charts  
- Scatter plots  
- Histograms  
- Pie charts  
- And more...

Think of it like a **digital canvas** — you tell it what to draw.

---

## 🤝 Matplotlib Works With:

- **NumPy** → numerical data  
- **Pandas** → labeled datasets  
- **Seaborn** → built on Matplotlib, best for statistical plots

---

## 💡 Two Ways to Use Matplotlib

| Style             | Description                          | Example                    |
|-------------------|--------------------------------------|----------------------------|
| **pyplot**        | Easy, stateful, MATLAB-style         | `plt.plot(x, y)`           |
| **Object-Oriented** | Explicit figure/axes control        | `fig, ax = plt.subplots()` |

We'll master both, but start with **pyplot**, the most common and beginner-friendly.

---

## ✅ Basic Import (Always Use This)

```python
import matplotlib.pyplot as plt
```

This gives access to the **plt** interface — your main command center.

---

## 🔧 Behind the Scenes

Every time you call `plt.plot()`, you’re drawing onto an invisible **Figure**, which contains one or more **Axes**.

With pyplot:
- Matplotlib **manages the figure and axes for you**
- You just issue drawing commands (e.g. `plt.plot`, `plt.title`, `plt.grid`, etc.)

---

## 🔍 Quick Command Reference

| Task             | Pyplot Command             |
|------------------|----------------------------|
| Plot a line      | `plt.plot(x, y)`           |
| Show the chart   | `plt.show()`               |
| Set title        | `plt.title("My Chart")`    |
| Label axes       | `plt.xlabel("X")` / `plt.ylabel("Y")` |
| Show grid        | `plt.grid(True)`           |

````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

# (question, answer, PEMDAS-style mnemonic)
flashcards = [

("What is Matplotlib used for?",
"🧠 Theory:\n"
"Matplotlib is a Python library for 2D plotting and visualization.\n\n"
"🗣️ Plain English:\n"
"It’s how you draw graphs, charts, and plots in Python.\"\n\n"
"💡 Mnemonic:\n"
"Matplotlib = Plot Pretty Pictures\""),

("What does import matplotlib.pyplot as plt do?",
"🧠 Theory:\n"
"Imports the pyplot module from Matplotlib under the alias 'plt'.\n\n"
"🗣️ Plain English:\n"
"Gives you the tools to start plotting right away.\""),

("What is pyplot?",
"🧠 Theory:\n"
"A stateful, easy-to-use interface in Matplotlib that mimics MATLAB.\n\n"
"🗣️ Plain English:\n"
"The quick way to plot things using commands like plt.plot().\""),

("How is the object-oriented interface different from pyplot?",
"🧠 Theory:\n"
"It gives you more control by working with explicit Figure and Axes objects.\n\n"
"🗣️ Plain English:\n"
"It’s like building plots manually instead of auto-mode.\"\n\n"
"💡 Mnemonic:\n"
"OO = Own Objects\""),

("What are some common plot types in Matplotlib?",
"🧠 Theory:\n"
"Line plot, bar chart, scatter plot, histogram, pie chart.\n\n"
"🗣️ Plain English:\n"
"You can draw almost anything you'd see in a data dashboard.\""),

("Which libraries work closely with Matplotlib?",
"🧠 Theory:\n"
"• NumPy — for numerical data\n• Pandas — for tabular data\n• Seaborn — statistical plots built on top of Matplotlib\n\n"
"🗣️ Plain English:\n"
"They all speak the same plotting language.\""),

("What does plt.plot(x, y) do?",
"🧠 Theory:\n"
"Creates a line graph of y vs x.\n\n"
"✅ Code:\n"
"    plt.plot([1, 2, 3], [4, 5, 6])\n"
"    plt.show()\n\n"
"🗣️ Plain English:\n"
"It draws a line through the points you give it.\""),

("What does plt.show() do?",
"🧠 Theory:\n"
"Displays the figure window with your plot.\n\n"
"🗣️ Plain English:\n"
"Tells Python: 'I’m done, now draw it!'\""),

("What does plt.title() do?",
"🧠 Theory:\n"
"Sets the title of your plot.\n\n"
"✅ Code:\n"
"    plt.title(\"My Chart\")"),

("What’s the purpose of plt.xlabel() and plt.ylabel()?",
"🧠 Theory:\n"
"Labels the x- and y-axes on the chart.\n\n"
"✅ Code:\n"
"    plt.xlabel(\"Time\")\n"
"    plt.ylabel(\"Value\")"),

("What does import matplotlib.pyplot as plt do?",
"🧠 Theory:\n"
"Imports the pyplot interface from Matplotlib and gives it the alias `plt`, so you can use commands like `plt.plot()` and `plt.show()`.\n\n"
"✅ Syntax:\n"
"    import matplotlib.pyplot as plt"),

("How do you define x and y data points for a plot?",
"🧠 Theory:\n"
"x and y are Python lists (or arrays) that hold the coordinates of your points.\n\n"
"✅ Code:\n"
"    x = [0, 1, 2, 3, 4, 5]          # X-values (independent)\n"
"    y = [0, 1, 4, 9, 16, 25]       # Y-values (squares of x)"),

("What does plt.plot(x, y) do?",
"🧠 Theory:\n"
"Creates a line plot connecting the points from x and y.\n\n"
"✅ Code:\n"
"    plt.plot(x, y)"),

("How do you add a title to the plot?",
"🧠 Theory:\n"
"Sets the main title of your graph using a string.\n\n"
"✅ Code:\n"
"    plt.title(\"Line Chart Example\")"),

("How do you label the x- and y-axes?",
"🧠 Theory:\n"
"Provides descriptive text for each axis.\n\n"
"✅ Code:\n"
"    plt.xlabel(\"X Axis\")\n"
"    plt.ylabel(\"Y Axis (squared)\")"),

("How do you turn on gridlines in a Matplotlib plot?",
"🧠 Theory:\n"
"Adds a background grid to help read values more easily.\n\n"
"✅ Code:\n"
"    plt.grid(True)"),

("What does plt.show() do?",
"🧠 Theory:\n"
"Opens a window and displays the full plot you’ve built with previous commands.\n\n"
"✅ Code:\n"
"    plt.show()")
]


index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'  # Hidden initially

def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]

        if len(card) == 2:
            question, answer = card
            mnemonic = ""
        else:
            question, answer, mnemonic = card

        display(Markdown(f"**Q:** {question}"))
        if showing_answer:
            display(Markdown(f"\n---\n\n**A:** {answer}"))
            if mnemonic.strip():
                display(Markdown(mnemonic))

    # Update button states
    prev_button.disabled = index == 0
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"

def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Attach handlers
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Initialize display
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
````

````python
# Import the pyplot module from matplotlib and alias it as 'plt'
import matplotlib.pyplot as plt

# Define X-axis data (independent variable)
x = [0, 1, 2, 3, 4, 5]

# Define Y-axis data (dependent variable)
y = [0, 1, 4, 9, 16, 25]  # squares of x

# Create a line plot with x and y data
plt.plot(x, y)

# Set the title of the chart
plt.title("Line Chart Example")

# Label the x-axis
plt.xlabel("X Axis")

# Label the y-axis
plt.ylabel("Y Axis (squared)")

# Turn on the grid for better readability
plt.grid(True)

# Display the plot in a window
plt.show()
````
