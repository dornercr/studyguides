# 🧠 Section 1: Python Fundamentals — Syntax, Source Structure, and Variables

---

## 🔹 1. Python Source Files

### ✅ Definition  
Python source code is stored in `.py` files and interpreted by the Python runtime. Each file is a module.

### ✅ Default Encoding  
Python 3 source files are `UTF-8` by default. This allows Unicode characters (like emojis or foreign text).

### ✅ Custom Encoding (rare)  
To use a different encoding (e.g., ISO-8859-1), declare it on the first or second line:

```python
# -*- coding: iso-8859-1 -*-
```

If a non-UTF-8 character is used without this declaration, you'll get a `SyntaxError`.

---

## 🔹 2. Indentation and Block Structure

### ✅ Definition  
Python uses **whitespace indentation** to define code blocks instead of braces `{}`. This enforces readability.

### ✅ Rules
- You must be consistent: use either **4 spaces** or **tabs**, not both.
- Indentation is required after:
  - `if`, `for`, `while`, `def`, `class`, etc.

### ✅ Example

```python
if True:
    print("Indented block")  # 4 spaces

print("Outside block")
```

Bad example (`❌ SyntaxError`):

```python
if True:
print("This is bad")  # Not indented
```

---

## 🔹 3. Keywords and Identifiers

### ✅ Keywords  
Reserved words in Python that cannot be used as variable names.

```python
True, False, None, def, return, if, elif, else,
while, for, in, is, not, and, or
```

Using a keyword as a variable name raises a `SyntaxError`.

### ✅ Identifiers

- Must start with a letter (a–z, A–Z) or underscore `_`
- Followed by letters, digits, or underscores
- Cannot use symbols like `$`, `@`, `!`
- Case-sensitive: `Name` ≠ `name`

---

## 🔹 4. Variable Assignment

### ✅ Syntax

```python
x = 5
name = "Alice"
```

Python assigns variables using **reference binding** — the name points to an object in memory.

### ✅ Multiple Assignment

```python
a, b = 1, 2
x = y = z = 0
```

### ✅ Swapping

```python
a, b = b, a
```

---

## 🔹 5. `=` vs `==`

| Operator | Meaning       | Example     |
|----------|---------------|-------------|
| `=`      | Assignment     | `x = 10`    |
| `==`     | Equality test  | `x == 10`   |

```python
x = 5
print(x == 5)  # True
```

---

## 🔹 6. Comments

- **Single-line:** use `#`
- **Multiline:** use triple-quotes (not technically comments, but ignored by interpreter unless used as docstrings)

```python
# This is a comment

"""
This is a block of documentation,
but still executes unless ignored by interpreter.
"""
```

## 🔹 2.1 What Is a Data Type?

A **data type** defines:

- What kind of value an object holds  
- What operations can be performed on it

In Python, everything is an object, and each object has a type:

```python
type(42)  # <class 'int'>
```

| `complex` | Complex numbers (real + imag)  | `2 + 3j`     | ❌      |
| `bool`    | Subclass of int                | `True`       | ❌      |

# 🧠 Section 3: Control Flow and Truthiness (with Theory)

---

## 🔹 3.1 `if` / `elif` / `else`

### 🧠 Theory:

Python uses conditional statements to control which block of code runs based on the result of a Boolean expression.

```python
x = 5

if x > 10:
    print("Big")
elif x == 5:
    print("Exact match")
else:
    print("Small")
```

- Python checks conditions **top-down**
- The **first** condition that evaluates to `True` runs
- Remaining branches are skipped

---

## 🔹 3.2 Boolean Operators

### 🧠 Theory:

Boolean operators `and`, `or`, and `not` allow combining or negating expressions. Python uses **short-circuit logic**, meaning evaluation stops as soon as the result is known.

| Operator | Meaning                 | Example             | Result  |
|----------|-------------------------|---------------------|---------|
| `and`    | Both sides must be true | `True and False`    | `False` |
| `or`     | One side must be true   | `False or 5`        | `5`     |
| `not`    | Invert truth value      | `not []`            | `True`  |

---

## 🔹 3.3 Truthiness by Type

### 🧠 Theory:

Any object in Python can be used in a condition.  
Python internally uses:

```python
bool(obj)  # calls obj.__bool__() or obj.__len__()
```

### Falsy values include:

- `False`, `None`
- `0`, `0.0`, `0j`
- `""`, `[]`, `()`, `{}`, `set()`, `range(0)`
- `b""`, `bytearray()`, `memoryview(b"")`

| Type  | Falsy Example | Truthy Example    |
|-------|---------------|-------------------|
| `int` | `0`           | `-1`              |
| `str` | `""`          | `"hello"`         |
| `list`| `[]`          | `[0]`             |
| `dict`| `{}`          | `{"a": 1}`        |
| `bytes`| `b""`        | `b"a"`            |

---

## 🔹 3.4 Ternary Expressions

### 🧠 Theory:

Python allows a compact **one-line if/else** expression for assigning or returning values.

```python
msg = "even" if x % 2 == 0 else "odd"
```

This is equivalent to:

```python
if x % 2 == 0:
    msg = "even"
else:
    msg = "odd"
```

---

## 🔹 3.5 Nested Conditions

### 🧠 Theory:

You can place `if` blocks inside other `if` blocks to handle more complex logic.

```python
if user:
    if age > 18:
        print("Adult user")
```

- Nesting allows conditional branches within other branches
- Deep nesting should generally be avoided for clarity

---

## ✅ Summary

- Python uses `bool(x)` to evaluate conditions
- Most types are **truthy** unless **empty or explicitly falsy**
- `if`, `elif`, `else` statements control conditional execution
- Logical operators (`and`, `or`, `not`) are **expression-based** and **short-circuit**

# 🧠 Section 4: Loops and Iteration

This section covers:

- `for` and `while` loops  
- Loop control (`break`, `continue`, `else`)  
- Iterating over different data types  
- Basic loop internals  

---

## 🔹 4.1 `for` Loops

### 🧠 Theory:

A `for` loop iterates over any **iterable** (e.g., list, string, range) one item at a time.

```python
for x in [1, 2, 3]:
    print(x)
```

- Python internally calls `iter()` on the object, then repeatedly calls `next()` until `StopIteration`.

### ✅ Common Iterables:

```python
for c in "abc":        # characters
for i in range(3):     # 0, 1, 2
for key in {"a": 1}:   # keys in dict
for x in [10, 20, 30]: # list items
```

---

## 🔹 4.2 `while` Loops

### 🧠 Theory:

A `while` loop runs as long as its condition evaluates to `True`.

```python
x = 0
while x < 3:
    print(x)
    x += 1
```

- Use `while` when the number of iterations isn't predetermined.

---

## 🔹 4.3 Loop Control: `break` and `continue`

### 🧠 Theory:

- `break` exits the loop **immediately**
- `continue` skips to the **next iteration**

```python
for i in range(5):
    if i == 3:
        break
    print(i)
```

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

---

## 🔹 4.4 `else` on Loops

### 🧠 Theory:

Python supports an `else` block **after `for` or `while`**, which only runs **if the loop did not hit `break`**.

```python
for i in range(3):
    if i == 5:
        break
else:
    print("Loop completed")
```

- ✅ Useful for **loop-else search patterns** (e.g., checking if an item was found).

---

## 🔹 4.5 `enumerate()` and `zip()`

```python
for i, val in enumerate(["a", "b"]):
    print(i, val)

for a, b in zip([1, 2], ["x", "y"]):
    print(a, b)
```

- `enumerate()` gives `(index, item)`  
- `zip()` joins multiple iterables into parallel tuples

---

## 🔹 4.6 Infinite Loops (with `while True`)

```python
while True:
    cmd = input()
    if cmd == "exit":
        break
```

- Use with caution — always include a `break` condition to prevent hangs

---

| `continue`       | Skip rest of current iteration    |
