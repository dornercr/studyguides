# 🧠 Section 1: Python Fundamentals — Syntax, Source Structure, and Variables

---

## 🔹 1. Python Source Files

### ✅ Definition  
Python source code is stored in `.py` files and interpreted by the Python runtime. Each file is a module.

### ✅ Default Encoding  
Python 3 source files are `UTF-8` by default. This allows Unicode characters (like emojis or foreign text).

### ✅ Custom Encoding (rare)  
To use a different encoding (e.g., ISO-8859-1), declare it on the first or second line:

```python
# -*- coding: iso-8859-1 -*-
```

If a non-UTF-8 character is used without this declaration, you'll get a `SyntaxError`.

---

## 🔹 2. Indentation and Block Structure

### ✅ Definition  
Python uses **whitespace indentation** to define code blocks instead of braces `{}`. This enforces readability.

### ✅ Rules
- You must be consistent: use either **4 spaces** or **tabs**, not both.
- Indentation is required after:
  - `if`, `for`, `while`, `def`, `class`, etc.

### ✅ Example

```python
if True:
    print("Indented block")  # 4 spaces

print("Outside block")
```

Bad example (`❌ SyntaxError`):

```python
if True:
print("This is bad")  # Not indented
```

---

## 🔹 3. Keywords and Identifiers

### ✅ Keywords  
Reserved words in Python that cannot be used as variable names.

```python
True, False, None, def, return, if, elif, else,
while, for, in, is, not, and, or
```

Using a keyword as a variable name raises a `SyntaxError`.

### ✅ Identifiers

- Must start with a letter (a–z, A–Z) or underscore `_`
- Followed by letters, digits, or underscores
- Cannot use symbols like `$`, `@`, `!`
- Case-sensitive: `Name` ≠ `name`

---

## 🔹 4. Variable Assignment

### ✅ Syntax

```python
x = 5
name = "Alice"
```

Python assigns variables using **reference binding** — the name points to an object in memory.

### ✅ Multiple Assignment

```python
a, b = 1, 2
x = y = z = 0
```

### ✅ Swapping

```python
a, b = b, a
```

---

## 🔹 5. `=` vs `==`

| Operator | Meaning       | Example     |
|----------|---------------|-------------|
| `=`      | Assignment     | `x = 10`    |
| `==`     | Equality test  | `x == 10`   |

```python
x = 5
print(x == 5)  # True
```

---

## 🔹 6. Comments

- **Single-line:** use `#`
- **Multiline:** use triple-quotes (not technically comments, but ignored by interpreter unless used as docstrings)

```python
# This is a comment

"""
This is a block of documentation,
but still executes unless ignored by interpreter.
"""
```

# 🧠 Section 2: Python Data Types

---

## 🔹 2.1 What Is a Data Type?

A **data type** defines:

- What kind of value an object holds  
- What operations can be performed on it

In Python, everything is an object, and each object has a type:

```python
type(42)  # <class 'int'>
```

---

## 🔹 2.2 Classification of Python Data Types

### ✅ A. Scalar / Atomic Types

| Type      | Description                    | Example      | Mutable |
|-----------|--------------------------------|--------------|---------|
| `int`     | Arbitrary-precision integer    | `42`         | ❌      |
| `float`   | 64-bit IEEE floating point     | `3.14`       | ❌      |
| `complex` | Complex numbers (real + imag)  | `2 + 3j`     | ❌      |
| `bool`    | Subclass of int                | `True`       | ❌      |
| `NoneType`| Represents null / void         | `None`       | ❌      |

---

### ✅ B. Sequence Types

| Type     | Description                | Example        | Mutable |
|----------|----------------------------|----------------|---------|
| `str`    | Unicode text               | `"hello"`      | ❌      |
| `list`   | Ordered, indexed collection| `[1, 2, 3]`    | ✅      |
| `tuple`  | Immutable sequence         | `(1, 2, 3)`    | ❌      |
| `range`  | Lazy iterable of integers  | `range(5)`     | ❌      |

---

### ✅ C. Mapping Type

| Type   | Description             | Example                | Mutable |
|--------|-------------------------|------------------------|---------|
| `dict` | Key-value hash table    | `{"a": 1, "b": 2}`     | ✅      |

---

### ✅ D. Set Types

| Type        | Description               | Example             | Mutable |
|-------------|---------------------------|---------------------|---------|
| `set`       | Unordered unique elements | `{1, 2, 3}`         | ✅      |
| `frozenset` | Immutable version of set  | `frozenset([1, 2])` | ❌      |

---

### ✅ E. Binary Types

| Type         | Description                | Example                  | Mutable |
|--------------|----------------------------|--------------------------|---------|
| `bytes`      | Immutable binary sequence  | `b"abc"`                 | ❌      |
| `bytearray`  | Mutable binary sequence    | `bytearray(b"abc")`      | ✅      |
| `memoryview` | View into binary object    | `memoryview(b"abc")`     | ✅*     |

> \* `memoryview` is only partially mutable depending on its source.

---

## 🔹 2.3 Type Relationships

- `bool` is a subclass of `int`
  
```python
isinstance(True, int)  # True
```

- `tuple` and `list` are both iterable, indexed, and ordered  
- `dict`, `set`, `list`, etc. use **hashing** and **equality** rules internally

---

## 🔹 2.4 Mutability Overview

| Type Class                        | Mutable |
|----------------------------------|---------|
| `int`, `float`, `complex`, `bool`, `NoneType` | ❌      |
| `str`, `tuple`, `range`, `frozenset`, `bytes` | ❌      |
| `list`, `dict`, `set`, `bytearray`, `memoryview` | ✅  |

---

## 🔹 2.5 Type Conversion

```python
int("10")          # 10
str(3.14)          # "3.14"
list("abc")        # ['a', 'b', 'c']
set([1, 2, 2, 3])  # {1, 2, 3}
```

Edge case:

```python
int("10.0")  # ValueError
```

---

## 🔹 2.6 Edge Cases & Notes

- `int` can grow arbitrarily large (no overflow):

```python
10**100  # works fine
```

- `float('nan')`, `float('inf')` supported:

```python
float('inf') > 1e308  # True
```

- `set()` removes duplicates:

```python
set([1, 1, 2])  # {1, 2}
```

- `dict` keys must be hashable:

```python
d = { (1, 2): "ok" }    # ✅ tuple is hashable
d = { [1, 2]: "fail" }  # ❌ TypeError — list is unhashable
```
