## 📐 Geometric Interpretation

### 🔹 Tangent Line to a Function
Given a function $ y = f(x) $, and a point $ P = (x_0, f(x_0)) $, the tangent line at $ P $ has the form:

$ y - f(x_0) = m(x - x_0) $

Where:

- $ m $ is the slope of the tangent = $ f'(x_0) $

---

### 🔹 Secant Line and the Limit Process

Slope of a secant line between points $ P = (x_0, f(x_0)) $ and $ Q = (x_0 + \Delta x, f(x_0 + \Delta x)) $:

$ \frac{\Delta f}{\Delta x} = \frac{f(x_0 + \Delta x) - f(x_0)}{\Delta x} $

**Derivative Definition:**

$ f'(x_0) = \lim_{\Delta x \to 0} \frac{f(x_0 + \Delta x) - f(x_0)}{\Delta x} $

---

## 🧮 Example 1: Derivative of $ f(x) = \frac{1}{x} $

$ f'(x_0) = \lim_{\Delta x \to 0} \frac{\frac{1}{x_0 + \Delta x} - \frac{1}{x_0}}{\Delta x} $

Simplify:

$ = \frac{1}{\Delta x} \cdot \left( \frac{x_0 - (x_0 + \Delta x)}{x_0(x_0 + \Delta x)} \right)
= \frac{-\Delta x}{\Delta x \cdot x_0(x_0 + \Delta x)} $

Cancel $ \Delta x $ and take the limit:

$ f'(x_0) = \frac{-1}{x_0^2} $

---

## 🟨 Word Problem: Area Under Tangent Line to $ y = \frac{1}{x} $

Given point $ (x_0, y_0 = \frac{1}{x_0}) $, find area of triangle formed by:

- x-intercept of the tangent
- y-intercept of the tangent
- origin

### 🔹 Tangent Line Equation

$ y - \frac{1}{x_0} = \left( -\frac{1}{x_0^2} \right)(x - x_0) $

#### 🔸 x-intercept:
Set $ y = 0 $ and solve for $ x $:

$ x = 2x_0 $

#### 🔸 y-intercept:
By symmetry or solve with $ x = 0 $:

$ y = 2y_0 = \frac{2}{x_0} $

#### 🔸 Triangle Area:
$ A = \frac{1}{2} \cdot \text{base} \cdot \text{height} = \frac{1}{2} \cdot 2x_0 \cdot 2y_0 = 2x_0 y_0 = 2 $

📝 **Area is always 2, regardless of $ x_0 $!**

---

## ✍️ Notation Summary

- $ f'(x) $: Newton’s notation
- $ \frac{dy}{dx} $, $ \frac{df}{dx} $: Leibniz notation
- $ \frac{\Delta f}{\Delta x} $: finite difference (secant slope)
- $ \lim_{\Delta x \to 0} \frac{\Delta f}{\Delta x} $: derivative

---

## 🧮 Example 2: Derivative of $ x^n $

Let $ f(x) = x^n $

$ \frac{\Delta f}{\Delta x} = \frac{(x + \Delta x)^n - x^n}{\Delta x} $

Use **binomial expansion**:

$ (x + \Delta x)^n = x^n + nx^{n-1}\Delta x + \mathcal{O}((\Delta x)^2) $

Then:

$ \frac{\Delta f}{\Delta x} = nx^{n-1} + \mathcal{O}(\Delta x)
\Rightarrow f'(x) = nx^{n-1} $

✅ **This rule works for all polynomials**.

```python
#set up
from sympy import symbols, simplify, expand, limit, Function, Eq, solve, diff

# Define symbolic variables used in various computations
x, x0, dx, h, a, n = symbols('x x0 dx h a n')  # x = variable, x0 = point of evaluation, dx = small change in x
```

```python
#derivative of f(x) = 1/x using limit definition
def derivative_reciprocal(x_val):
    expr = 1 / x  # Define the function f(x) = 1/x
    dq = (expr.subs(x, x0 + dx) - expr.subs(x, x0)) / dx  # Compute the difference quotient at x0
    derivative = simplify(limit(dq, dx, 0))  # Take the limit as dx → 0 for the derivative f'(x0)
    return derivative.subs(x0, x_val)  # Evaluate the result at the specific x0 value
```

```python
#example
print(derivative_reciprocal(3))  # Output: -1/9
```

```python
#area under the tangent to f(x) = 1/x at point x0
def tangent_area(x0_val):
    y0 = 1 / x0_val             # Evaluate y0 = f(x0) = 1/x0
    slope = -1 / x0_val**2      # Derivative f'(x0) = -1/x0²

    # The x-intercept of the tangent line is 2 * x0 (can be derived algebraically)
    x_intercept = 2 * x0_val

    # The y-intercept is 2 * y0 by symmetry (again derived from solving tangent line for x = 0)
    y_intercept = 2 * y0

    # Use triangle area formula: (1/2) * base * height
    area = simplify(0.5 * x_intercept * y_intercept)
    return area  # Always returns 2 for this function regardless of x0

#example
print(tangent_area(5))  # Output: 2
```

```python
#derivative of x^n using limit (general power rule)
def derivative_power(x_val, power_val):
    # Apply binomial expansion approximation to get limit definition
    f_expr = (x + dx)**power_val - x**power_val  # (x + dx)^n - x^n
    dq = f_expr / dx                             # Difference quotient
    derivative = limit(dq, dx, 0)                # Take the limit as dx → 0 to get derivative
    return simplify(derivative.subs(x, x_val))   # Evaluate the simplified expression at x = x_val

#example
print(derivative_power(3, 4))  # Output: 108 (since d/dx x⁴ = 4x³ = 4*27)
```

```python
#derivative of any polynomial using symbolic diff
def polynomial_derivative(expr, x_val=None):
    deriv = diff(expr, x)  # Symbolically differentiate the input expression with respect to x
    return deriv.subs(x, x_val) if x_val is not None else deriv  # Optionally evaluate at a point

#example
print(polynomial_derivative(x**3 + 5*x**10, 2))  # Output: 808 (3*4 + 50*512)
```

```python
#limit definition for arbitrary functions(symbolic)
def derivative_by_limit(expr, point):
    # Manually apply the definition of derivative as a limit
    dq = (expr.subs(x, point + dx) - expr.subs(x, point)) / dx  # Difference quotient at arbitrary point
    return simplify(limit(dq, dx, 0))  # Take limit to get derivative symbolically

#example
print(derivative_by_limit(1/x, x0))  # Output: -1/x0**2
```

```python
#sample use cases
print(derivative_reciprocal(4))                    # -1/16
print(tangent_area(2))                             # 2
print(derivative_power(2, 5))                      # 80
print(polynomial_derivative(x**2 + 2*x, 1))        # 4
print(derivative_by_limit(x**2 + 3*x - 4, a))       # 2*a + 3
```
