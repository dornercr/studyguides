# 🔹 Section 1.1: NumPy’s Module Layout – Study Guide

NumPy is made up of many smaller parts called **submodules**.  
Most users rely on just a few of these regularly, while others are meant for **specialized tasks** or **legacy codebases**.

---

## 🟢 Commonly Used Namespaces

These are the main submodules you'll use in everyday NumPy workflows:

- **`numpy`**  
  The main toolbox. Core functions for arrays, math, broadcasting, etc.

- **`numpy.exceptions`**  
  Handles errors and warnings when something goes wrong.

- **`numpy.fft`**  
  Tools for Fast Fourier Transforms (FFT), converting wave signals into frequencies.

- **`numpy.linalg`**  
  Linear algebra tools: matrix decompositions, system solvers, vector norms, etc.

- **`numpy.polynomial`**  
  Work with polynomial equations, like:  
  `ax² + bx + c`

- **`numpy.random`**  
  Generate random numbers, samples, distributions (used in simulations and ML).

- **`numpy.strings`**  
  Utilities for manipulating character sequences and string data.

- **`numpy.testing`**  
  Test helpers for writing and validating NumPy-based code.

- **`numpy.typing`**  
  Type hint definitions for NumPy arrays and data structures.

---

## 🟡 Special Use Namespaces

These are helpful for more advanced or specialized tasks:

- **`numpy.ctypeslib`**  
  Interface NumPy arrays with external C libraries via `ctypes`.

- **`numpy.dtypes`**  
  Manages NumPy’s internal data types (e.g., `int32`, `float64`).

- **`numpy.emath`**  
  Extended math for edge cases (e.g., safe square roots of negative numbers).

- **`numpy.lib`**  
  A miscellaneous collection of utilities, often used internally.

- **`numpy.rec`**  
  For working with record arrays (like rows in a spreadsheet). Mostly outdated.

- **`numpy.version`**  
  Shows the installed version of NumPy.

---

## 🔴 Outdated or Rarely Needed Namespaces

These are deprecated or considered low-priority for modern projects:

- **`numpy.char`**  
  Older string manipulation tools, now superseded by better string libraries.

- **`numpy.distutils`** *(Deprecated)*  
  Old build system for compiling NumPy. No longer in active use.

- **`numpy.f2py`**  
  Bridge between Python and Fortran. Typically run from the terminal.

- **`numpy.ma`**  
  For handling missing data with masked arrays. Often replaced by `pandas`.

- **`numpy.matlib`** *(Deprecated)*  
  Used to create matrix objects. Now replaced by regular arrays and `numpy.linalg`.

# 🔹 Section 1.2: NumPy – Exceptions & Warnings (Study Guide)

When using NumPy, sometimes things don’t go as planned.  
That’s where **warnings** and **exceptions** come in:

- **Warnings** notify you that something unusual happened but let your code continue.
- **Exceptions** stop your code entirely and must be fixed before continuing.

---

## 🆕 New in NumPy 1.25+

A new module called `numpy.exceptions` was introduced in version **1.25**.  
Most common errors and warnings still work as part of the regular `numpy` namespace.

---

## ⚠️ Warnings (Do NOT stop your code)

Warnings alert you to potential issues but won’t crash your program.

---

### 🔸 ComplexWarning  
Occurs when you cast a complex number to a float.  
Only the real part is kept, and the imaginary part is discarded.

    import numpy as np  
    np.array([3 + 2j], dtype=float)  # Warning: imaginary part discarded

---

### 🔸 VisibleDeprecationWarning  
Raised when you create an array from nested lists of different lengths.

    np.array([[1, 2], [3]])  # Warning: creating an array from ragged nested sequences

---

### 🔸 RankWarning  
Occurs when using `np.polyfit()` and the matrix doesn't have full rank, making the polynomial fit unreliable.

    np.polyfit([1, 2, 3], [4, 5, 6], 4)  # Warning: poorly conditioned polynomial fit

---

## ❌ Exceptions (DO stop your code)

Exceptions indicate definite errors. Your program will crash unless you handle or fix them.

---

### 🔸 AxisError  
Occurs when referencing an axis that doesn't exist in the array.

    arr = np.array([[1, 2], [3, 4]])  
    np.sum(arr, axis=2)  # Error: axis 2 is out of bounds for array of dimension 2

---

### 🔸 DTypePromotionError  
Raised when NumPy can't safely promote two types to a compatible one.

    np.result_type(np.int64, 'string')  # Error: no common dtype between int64 and string

---

### 🔸 TooHardError  
Happens when a NumPy operation is too complex or exceeds internal computation limits.  
You won't usually hit this unless you're doing deep math or optimization tasks.

    # No simple reproduction; this occurs in advanced internal logic (e.g., linear solvers)

---

## ✅ Summary Table

| Category   | Name                    | Trigger Example                                             |
|------------|-------------------------|-------------------------------------------------------------|
| Warning    | `ComplexWarning`        | Casting complex to float drops imaginary part              |
| Warning    | `VisibleDeprecationWarning` | Irregular nested list when creating an array            |
| Warning    | `RankWarning`           | Low-rank matrix used in polynomial fitting                 |
| Exception  | `AxisError`             | Referencing a non-existent axis in a NumPy operation       |
| Exception  | `DTypePromotionError`   | Combining incompatible types (e.g., int + string)          |
| Exception  | `TooHardError`          | NumPy gave up due to task complexity (very rare)           |

---

**Remember:**  
- Warnings = fix later, code still runs.  
- Exceptions = fix now, code stops.

# Section 1.3: Discrete Fourier Transform (DFT) – Study Guide

---

## 🔹 Purpose

DFT breaks a signal (list of values over time) into its frequency components.  
You can then:

- Analyze what frequencies are present  
- Filter or modify the signal  
- Reconstruct the signal with the inverse DFT

---

## ⚙️ Key Idea

- Input signal lives in **time domain**  
- Output spectrum lives in **frequency domain**

🎵 Think of music: DFT tells you what notes (frequencies) are in a sound.

---

## 🔹 FFT = Fast Fourier Transform

Efficient algorithm to compute DFT.  
- NumPy uses: `numpy.fft`  
- SciPy uses: `scipy.fft` (preferred for modern use)

---

## 🔹 Core Functions

### 📏 Standard FFTs (for complex or real input)

| Function   | What it does                   |
|------------|--------------------------------|
| fft(a)     | 1D DFT → frequencies           |
| ifft(a)    | 1D inverse DFT → time signal   |
| fft2(a)    | 2D DFT (e.g., image)           |
| ifft2(a)   | 2D inverse                     |
| fftn(a)    | N-dimensional DFT              |
| ifftn(a)   | N-dimensional inverse          |

---

### 📏 Real FFTs (optimized for real input)

| Function   | What it does                     |
|------------|----------------------------------|
| rfft(a)    | 1D real DFT (positive freqs only)|
| irfft(a)   | Inverse of rfft                  |
| rfft2(a)   | 2D real DFT                      |
| irfft2(a)  | Inverse                          |
| rfftn(a)   | N-D real DFT                     |
| irfftn(a)  | Inverse                          |

---

### 📏 Hermitian FFTs (used when output is purely real)

| Function   | What it does                         |
|------------|--------------------------------------|
| hfft(a)    | Hermitian spectrum → real signal     |
| ihfft(a)   | Real signal → Hermitian spectrum     |

---

## 🛠 Helper Functions

| Function      | What it does                           |
|---------------|----------------------------------------|
| fftfreq(n)    | Returns frequency bins for fft         |
| rfftfreq(n)   | Frequency bins for rfft                |
| fftshift(x)   | Moves 0 freq to center of array        |
| ifftshift(x)  | Undoes fftshift                        |

---

## 🧠 Concepts to Understand

### 🔄 Spectrum Output (for real input)

- `fft(a)` gives complex values  
- A[0] = total sum (DC component)  
- Positive freqs → A[1 : n//2]  
- Negative freqs → A[n//2 + 1 :]  
- If `n` is even: A[n//2] is **Nyquist freq** (real only)  
- If `n` is odd: frequencies are unevenly split

---

### 📊 Spectrum Types

- **Amplitude spectrum** → `np.abs(A)`  
- **Power spectrum** → `np.abs(A) ** 2`  
- **Phase spectrum** → `np.angle(A)`

---

### 🔁 Normalization (`norm=`)

| Mode       | Meaning                        |
|------------|--------------------------------|
| "backward" | Default, scales inverse only   |
| "forward"  | Scales forward only            |
| "ortho"    | Scales both equally            |

---

### ⚠️ Type Promotion

- `float32` → `float64`  
- `complex64` → `complex128`  
- Use `scipy.fftpack` to avoid auto-promotion (legacy)

---

### 🔁 Real vs Hermitian

- Real signal ⇒ Hermitian spectrum (symmetric)
- Hermitian spectrum ⇒ Real signal

Efficiency notes:
- `rfft`: skips negative freqs (real input → n → n/2 + 1 output)
- `hfft`: assumes symmetric input, skips negatives

---

### 📐 Higher Dimensions

Use `fft2`, `fftn` for 2D+ data (e.g., images).  
DFT is faster than direct time-domain convolution — great for large filters.

---

## 🧠 Analogy Summary

- **Time domain** = how a signal changes over time  
- **Frequency domain** = what "waves" or periodic patterns it's made of  
- **DFT** converts between the two
