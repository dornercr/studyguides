# 🧭 LESSON 1: Your First Program — “Hello, World!”

---

## 🧠 The 20% That Teaches 80%

A **program** is a list of instructions for the computer.  
The simplest program just **prints to the screen**.

Every language has different **syntax**, but the same goal: show output.

---

## ✅ THE CODE

### 🔹 Python (no setup needed)

```python
print("Hello, world!")
```

---

### 🔹 C

```c
#include <stdio.h>

int main() {
    printf("Hello, world!\n");
    return 0;
}
```

---

### 🔹 C++

```cpp
#include <iostream>
using namespace std;

int main() {
    cout << "Hello, world!" << endl;
    return 0;
}
```

---

### 🔹 Rust

```rust
fn main() {
    println!("Hello, world!");
}
```

---

## 🔍 LINE-BY-LINE BREAKDOWN

| Part         | C                       | C++                     | Rust                  | Python            |
|--------------|-------------------------|--------------------------|------------------------|--------------------|
| Header/Import| `#include <stdio.h>`    | `#include <iostream>`    | None (built-in)        | None               |
| Entry Point  | `int main()`            | `int main()`             | `fn main()`            | Runs from top      |
| Print        | `printf()`              | `cout <<`                | `println!()` (macro)   | `print()`          |
| End          | `return 0;`             | `return 0;`              | Auto-end               | Nothing required   |
| Semicolons   | ✅ Yes                  | ✅ Yes                   | ✅ Yes                 | ❌ No              |

---

## 💡 KEY CONCEPTS

| Term                 | Meaning                                                      |
|----------------------|--------------------------------------------------------------|
| **Function**         | A named block of code that can be called (like `main()`)     |
| **Standard Output**  | Where print goes — the screen or terminal                    |
| **Header**           | Tells C/C++ what library or function definitions to use      |
| **Macro (Rust)**     | Special code-generation tool (used for `println!`)           |
| **Interpreter vs Compiler** | Python is interpreted; C, C++, Rust are compiled     |

# 🧱 LESSON 2: Variables & Data Types

---

## 🔑 Pareto Core (20% of effort → 80% of understanding)

- **Variable** = A named space in memory to store data  
- **Data type** = The kind of data (e.g., number, text, true/false)  
- **C, C++, Rust** = Statically typed (types required)  
- **Python** = Dynamically typed (figures it out)

---

## ✅ Example: Declare the Same Variables in 4 Languages

### Data:

- Name: `"Alice"`  
- Age: `30`  
- Height: `5.6`  
- Is student: `true`

---

### 🔹 Python (Dynamically typed)

```python
name = "Alice"
age = 30
height = 5.6
is_student = True
```

---

### 🔹 C (Statically typed, minimal)

```c
#include <stdio.h>
#include <stdbool.h>  // For bool type

int main() {
    char name[] = "Alice";     // Array of chars
    int age = 30;              // Integer
    float height = 5.6f;       // Float (use 'f' suffix)
    bool is_student = true;    // Boolean
    return 0;
}
```

---

### 🔹 C++ (Modern, with std::string)

```cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    string name = "Alice";     // String class
    int age = 30;
    float height = 5.6f;
    bool is_student = true;
    return 0;
}
```

---

### 🔹 Rust (Immutable by default)

```rust
fn main() {
    let name = "Alice";              // &str
    let age: i32 = 30;               // 32-bit integer
    let height: f32 = 5.6;           // 32-bit float
    let is_student: bool = true;     // Boolean
}
```

---

## 🧠 Type System Comparison

| Feature      | Python     | C                  | C++              | Rust           |
|--------------|------------|--------------------|------------------|----------------|
| Typing       | Dynamic    | Static             | Static           | Static         |
| String       | `"text"`   | `char[]`           | `string` class   | `&str`         |
| Integer      | `30`       | `int`              | `int`            | `i32`          |
| Float        | `5.6`      | `float`            | `float`          | `f32`          |
| Boolean      | `True`     | `bool` (via `stdbool.h`) | `bool`     | `bool`         |
| Mutable?     | Yes        | Yes                | Yes              | No by default  |

---

## 🚧 C String Warning

C does **not** have real strings — just **arrays of characters** (`char[]` or `char*`).  
String handling is done via functions like `strlen()`, `strcpy()`, etc.

---

## 💡 Rust Mutability Example

```rust
let x = 5;       // Immutable
let mut y = 10;  // Mutable
y = 15;
```

# 🧭 LESSON 4: Conditions (if / else / comparison logic)

---

## 🔑 20% That Drives 80% of Decisions

- **Conditions** let your program make decisions  
- **Comparisons** return true or false  
- `if / else` blocks control which code runs  

---

## 🧠 Logic Operators

| Operator | Meaning                 | Example     |
|----------|--------------------------|-------------|
| `==`     | Equal to                | `a == b`    |
| `!=`     | Not equal to            | `a != b`    |
| `<`      | Less than               | `a < b`     |
| `>`      | Greater than            | `a > b`     |
| `<=`     | Less than or equal to   | `a <= b`    |
| `>=`     | Greater than or equal to| `a >= b`    |

---

## ✅ Example: Compare Ages in All 4 Languages

---

### 🔹 Python

```python
age = 20

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

---

### 🔹 C

```c
#include <stdio.h>

int main() {
    int age = 20;

    if (age >= 18) {
        printf("You are an adult.\n");
    } else {
        printf("You are a minor.\n");
    }

    return 0;
}
```

---

### 🔹 C++

```cpp
#include <iostream>
using namespace std;

int main() {
    int age = 20;

    if (age >= 18) {
        cout << "You are an adult." << endl;
    } else {
        cout << "You are a minor." << endl;
    }

    return 0;
}
```

---

### 🔹 Rust

```rust
fn main() {
    let age = 20;

    if age >= 18 {
        println!("You are an adult.");
    } else {
        println!("You are a minor.");
    }
}
```

---

## 🎯 Rules to Remember

| Language | Requires `()`? | Braces `{}` required? | Else-if keyword    |
|----------|----------------|------------------------|---------------------|
| Python   | ❌              | ❌ (uses indentation)   | `elif`              |
| C/C++    | ✅              | ✅ Always              | `else if`           |
| Rust     | ❌              | ✅ Always              | `else if`           |

---

## 🔄 BONUS: Else-if Ladder (Multiple Conditions)

### 🔹 Python

```python
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
else:
    print("C or lower")
```

---

### 🔹 C / C++ / Rust

```c
if (score >= 90) {
    printf("A\n");
} else if (score >= 80) {
    printf("B\n");
} else {
    printf("C or lower\n");
}
```

# 🔁 LESSON 5: Loops (for, while, repeat-until)

---

## 🔑 20% That Powers 80% of Automation

**Loops repeat code** until a condition is false. Use them when:

- ✅ You want to run code **N times** → use `for`  
- ✅ You don’t know how many times → use `while`

| Loop Type | When to Use                |
|-----------|----------------------------|
| `for`     | Counted repetition (known times) |
| `while`   | Repeat until condition fails     |

---

## ✅ Print 1 to 5 in All Four Languages

---

### 🔹 Python — `for` and `while`

```python

    print(i)

    print(i)
    i += 1
```

---

### 🔹 C — `for` and `while`

```c
#include <stdio.h>

int main() {
    // for loop
    for (int i = 1; i <= 5; i++) {
        printf("%d\n", i);
    }

    // while loop
    int j = 1;
    while (j <= 5) {
        printf("%d\n", j);
        j++;
    }

    return 0;
}
```

---

### 🔹 C++ — identical to C

```cpp
#include <iostream>
using namespace std;

int main() {
    for (int i = 1; i <= 5; i++) {
        cout << i << endl;
    }

    int j = 1;
    while (j <= 5) {
        cout << j << endl;
        j++;
    }

    return 0;
}
```

---

### 🔹 Rust — `for`, `while`, `loop`

```rust
fn main() {
    // for loop (idiomatic)
    for i in 1..=5 {
        println!("{}", i);
    }

    // while loop
    let mut j = 1;
    while j <= 5 {
        println!("{}", j);
        j += 1;
    }

    // loop (infinite unless broken)
    let mut k = 1;
    loop {
        if k > 5 { break; }
        println!("{}", k);
        k += 1;
    }
}
```

---

## 🔍 Loop Structures Compared

| Language | `for` Loop Syntax         | `while` Loop      | Special                     |
|----------|----------------------------|--------------------|-----------------------------|
| Python   | `for x in range()`         | `while condition:` | No `do-while`               |
| C        | `for(init; cond; upd)`     | `while (cond)`     | Has `do { } while (cond);`  |
| C++      | Same as C                  | Same as C          | Adds `break`, `continue`    |
| Rust     | `for x in start..end`      | `while cond {}`    | `loop {}` = infinite loop   |

## ✅ Store 5 Scores: `95, 88, 76, 100, 67`

---

### 🔹 Python (list)

```python

```

---

### 🔹 C (fixed-size array)

```c
#include <stdio.h>

int main() {
    int scores[5] = {95, 88, 76, 100, 67};
```

### 🔹 C++ (vector)

```cpp

#include <vector>
using namespace std;

int main() {
    vector<int> scores = {95, 88, 76, 100, 67};

    cout << scores[0] << endl;     // Access
    scores.push_back(80);          // Add
    cout << scores.size() << endl; // Get size
    return 0;
}
```

---

### 🔹 Rust (array + vector)

```rust
fn main() {
    let scores = [95, 88, 76, 100, 67];        // fixed-size array
    println!("{}", scores[0]);                // Access

    let mut scores_vec = vec![95, 88, 76, 100, 67];  // dynamic
    scores_vec.push(80);                      // Add
    println!("{}", scores_vec.len());         // Length
}
```

| C++      | ✅ `int arr[5]`   | ✅ `vector`      | Use `std::vector` for ease   |
| Rust     | ✅ `[T; N]`       | ✅ `Vec<T>`      | Use `Vec` in real projects   |
