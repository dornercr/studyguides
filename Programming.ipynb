# 🧭 LESSON 1: Your First Program — “Hello, World!”

---

## 🧠 The 20% That Teaches 80%

A **program** is a list of instructions for the computer.  
The simplest program just **prints to the screen**.

Every language has different **syntax**, but the same goal: show output.

---

## ✅ THE CODE

### 🔹 Python (no setup needed)

```python
print("Hello, world!")
```

---

### 🔹 C

```c
#include <stdio.h>

int main() {
    printf("Hello, world!\n");
    return 0;
}
```

---

### 🔹 C++

```cpp
#include <iostream>
using namespace std;

int main() {
    cout << "Hello, world!" << endl;
    return 0;
}
```

---

### 🔹 Rust

```rust
fn main() {
    println!("Hello, world!");
}
```

---

## 🔍 LINE-BY-LINE BREAKDOWN

| Part         | C                       | C++                     | Rust                  | Python            |
|--------------|-------------------------|--------------------------|------------------------|--------------------|
| Header/Import| `#include <stdio.h>`    | `#include <iostream>`    | None (built-in)        | None               |
| Entry Point  | `int main()`            | `int main()`             | `fn main()`            | Runs from top      |
| Print        | `printf()`              | `cout <<`                | `println!()` (macro)   | `print()`          |
| End          | `return 0;`             | `return 0;`              | Auto-end               | Nothing required   |
| Semicolons   | ✅ Yes                  | ✅ Yes                   | ✅ Yes                 | ❌ No              |

---

## 💡 KEY CONCEPTS

| Term                 | Meaning                                                      |
|----------------------|--------------------------------------------------------------|
| **Function**         | A named block of code that can be called (like `main()`)     |
| **Standard Output**  | Where print goes — the screen or terminal                    |
| **Header**           | Tells C/C++ what library or function definitions to use      |
| **Macro (Rust)**     | Special code-generation tool (used for `println!`)           |
| **Interpreter vs Compiler** | Python is interpreted; C, C++, Rust are compiled     |

# 🧱 LESSON 2: Variables & Data Types  
### (with C, C++, Python, and Rust)

---

## 🔑 Pareto Core (20% of effort → 80% of understanding)

- **Variable** = A named space in memory to store data  
- **Data type** = The kind of data (e.g., number, text, true/false)  
- **C, C++, Rust** = Statically typed (types required)  
- **Python** = Dynamically typed (figures it out)

---

## ✅ Example: Declare the Same Variables in 4 Languages

### Data:

- Name: `"Alice"`  
- Age: `30`  
- Height: `5.6`  
- Is student: `true`

---

### 🔹 Python (Dynamically typed)

```python
name = "Alice"
age = 30
height = 5.6
is_student = True
```

---

### 🔹 C (Statically typed, minimal)

```c
#include <stdio.h>
#include <stdbool.h>  // For bool type

int main() {
    char name[] = "Alice";     // Array of chars
    int age = 30;              // Integer
    float height = 5.6f;       // Float (use 'f' suffix)
    bool is_student = true;    // Boolean
    return 0;
}
```

---

### 🔹 C++ (Modern, with std::string)

```cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    string name = "Alice";     // String class
    int age = 30;
    float height = 5.6f;
    bool is_student = true;
    return 0;
}
```

---

### 🔹 Rust (Immutable by default)

```rust
fn main() {
    let name = "Alice";              // &str
    let age: i32 = 30;               // 32-bit integer
    let height: f32 = 5.6;           // 32-bit float
    let is_student: bool = true;     // Boolean
}
```

---

## 🧠 Type System Comparison

| Feature      | Python     | C                  | C++              | Rust           |
|--------------|------------|--------------------|------------------|----------------|
| Typing       | Dynamic    | Static             | Static           | Static         |
| String       | `"text"`   | `char[]`           | `string` class   | `&str`         |
| Integer      | `30`       | `int`              | `int`            | `i32`          |
| Float        | `5.6`      | `float`            | `float`          | `f32`          |
| Boolean      | `True`     | `bool` (via `stdbool.h`) | `bool`     | `bool`         |
| Mutable?     | Yes        | Yes                | Yes              | No by default  |

---

## 🚧 C String Warning

C does **not** have real strings — just **arrays of characters** (`char[]` or `char*`).  
String handling is done via functions like `strlen()`, `strcpy()`, etc.

---

## 💡 Rust Mutability Example

```rust
let x = 5;       // Immutable
let mut y = 10;  // Mutable
y = 15;
```

# 🧭 LESSON 4: Conditions (if / else / comparison logic)

---

## 🔑 20% That Drives 80% of Decisions

- **Conditions** let your program make decisions  
- **Comparisons** return true or false  
- `if / else` blocks control which code runs  

---

## 🧠 Logic Operators

| Operator | Meaning                 | Example     |
|----------|--------------------------|-------------|
| `==`     | Equal to                | `a == b`    |
| `!=`     | Not equal to            | `a != b`    |
| `<`      | Less than               | `a < b`     |
| `>`      | Greater than            | `a > b`     |
| `<=`     | Less than or equal to   | `a <= b`    |
| `>=`     | Greater than or equal to| `a >= b`    |

---

## ✅ Example: Compare Ages in All 4 Languages

---

### 🔹 Python

```python
age = 20

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

---

### 🔹 C

```c
#include <stdio.h>

int main() {
    int age = 20;

    if (age >= 18) {
        printf("You are an adult.\n");
    } else {
        printf("You are a minor.\n");
    }

    return 0;
}
```

---

### 🔹 C++

```cpp
#include <iostream>
using namespace std;

int main() {
    int age = 20;

    if (age >= 18) {
        cout << "You are an adult." << endl;
    } else {
        cout << "You are a minor." << endl;
    }

    return 0;
}
```

---

### 🔹 Rust

```rust
fn main() {
    let age = 20;

    if age >= 18 {
        println!("You are an adult.");
    } else {
        println!("You are a minor.");
    }
}
```

---

## 🎯 Rules to Remember

| Language | Requires `()`? | Braces `{}` required? | Else-if keyword    |
|----------|----------------|------------------------|---------------------|
| Python   | ❌              | ❌ (uses indentation)   | `elif`              |
| C/C++    | ✅              | ✅ Always              | `else if`           |
| Rust     | ❌              | ✅ Always              | `else if`           |

---

## 🔄 BONUS: Else-if Ladder (Multiple Conditions)

### 🔹 Python

```python
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
else:
    print("C or lower")
```

---

### 🔹 C / C++ / Rust

```c
if (score >= 90) {
    printf("A\n");
} else if (score >= 80) {
    printf("B\n");
} else {
    printf("C or lower\n");
}
```

````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output, HTML

# (question, answer, PEMDAS-style mnemonic)

flashcards = [

# GENERAL LOGIC

("What does an `if` statement do?",
"It runs a block of code if the condition is true."),

("What is the difference between `==` and `=`?",
"`==` compares values, `=` assigns a value."),

("What does `else` do in an if-statement?",
"It runs a block of code if the `if` condition is false."),

("What is `else if` used for?",
"It adds a second condition to test if the first was false."),


# OPERATORS

("What does `==` mean?",
"Equal to"),

("What does `!=` mean?",
"Not equal to"),

("What does `<` mean?",
"Less than"),

("What does `>=` mean?",
"Greater than or equal to"),

("What is a common use for `if age >= 18`?",
"To check if someone is an adult."),


# PYTHON-SPECIFIC

("How do you write an if-statement in Python?",
'if condition:\n    # code'),

("Does Python use parentheses in if conditions?",
"No — they are optional."),

("What does Python use instead of curly braces?",
"Indentation"),

("How do you write else-if in Python?",
"`elif`"),

("What happens if Python code isn’t indented?",
"A syntax error occurs."),


# C / C++ SPECIFIC

("Does C require parentheses for if conditions?",
"Yes — always."),

("Does C++ use curly braces for if blocks?",
"Yes — always."),

("How do you write else-if in C or C++?",
"`else if`"),

("What does `if (score >= 90)` do in C?",
"Checks if score is at least 90 and runs the block if true."),

("What happens if you forget `{}` in C for a multi-line block?",
"Only the first line runs; the rest is not inside the condition."),


# RUST-SPECIFIC

("Does Rust require curly braces for if/else?",
"Yes — always."),

("Can Rust if-conditions use parentheses?",
"No — they are not needed and usually omitted."),

("How do you write else-if in Rust?",
"`else if`"),


# MULTIPLE CHOICE STYLE

("Which language uses indentation instead of `{}`?",
"Python"),

("Which two languages allow `else if`?",
"C/C++ and Rust"),

("Which language uses `elif` instead of `else if`?",
"Python"),

("Which language doesn't need parentheses in `if`?",
"Python and Rust"),

("How do you compare two values in all languages?",
"Use `==`"),

("Which symbol is used to test inequality?",
"`!=`"),

("What is the result of `if (10 > 5)` in C?",
"The condition is true, so the block runs."),


# CHALLENGE-BASED

("How do you check if score ≥ 90, then print 'Excellent'?",
"Use `if score >= 90:` in Python or `if (score >= 90)` in C/C++/Rust."),

("What structure handles 3+ outcomes?",
"An if/elif/else or if/else if/else ladder."),

("What does an else-if ladder help you avoid?",
"Nested if-else blocks and unreadable code.")
]


index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'  # Hidden initially

def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]
        if len(card) == 3:
            question, answer, mnemonic = card
        else:
            question, answer = card
            mnemonic = ""

        html = f"""
        <div style='background:#f9f9f9;padding:16px;border-radius:10px;
                    border:1px solid #ddd;line-height:1.6;font-family:monospace;
                    white-space:pre-wrap;font-size:14px'>
        <b>Q:</b> {question}<br><br>
        {f"<b>A:</b><pre>{answer}</pre><br><b>🔁 Mnemonic:</b> {mnemonic}" if showing_answer else ""}
        </div>
        """
        display(HTML(html))

    # Update button states
    prev_button.disabled = index == 0
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"


def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Attach handlers
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Initialize display
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
````
