````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output, HTML
from html import escape
import random

# ---------------------- Example decks (replace or extend) ----------------------
programming_basics = [

# ---------- 1. VARIABLES ----------
("Variables — Concept",
"A variable (named storage) holds a value in memory. \
Python is dynamically typed (type decided at runtime). \
C, Rust, C++ are statically typed (type fixed at compile time).",
"V M T → Variable, Memory, Type.\nKindergarten: 'Very Many Toys.'"),

("Variables — Examples",
"Python: x=10\nC: int x=10;\nRust: let x:i32=10;\nC++: int x=10;",
"X I → Example, Initialize.\nKindergarten: 'Xylophones Inspire.'"),

# ---------- 2. TYPES ----------
("Types — Concept",
"A type (classification of data) defines allowed values and operations. \
Common: int, float, string/char, bool.",
"T I F B → Types: Int, Float, Bool.\nKindergarten: 'Tigers In Forests Bite.'"),

("Types — Examples",
"Python: x=3.14\nC: float x=3.14;\nRust: let y:f64=3.14;\nC++: double x=3.14;",
"N D → Number, Decimal.\nKindergarten: 'Nice Duck.'"),

# ---------- 3. OPERATORS ----------
("Operators — Arithmetic",
"Arithmetic operators = + - * / % (modulus).\n\n\
Python: 5%2\nC: 5%2\nRust: 5%2\nC++: 5%2",
"A A M → Add, Arithmetic, Mod.\nKindergarten: 'All Ants March.'"),

("Operators — Comparison",
"Comparison operators = == != < > <= >=.\nReturn true/false.",
"C C T → Compare, Check, True.\nKindergarten: 'Cats Chase Tails.'"),

("Operators — Logic",
"Logical operators = and/&&, or/||, not/!.\n\
Python: if a and b: ...\nC/C++/Rust: if(a && b) { }",
"L A O → Logic, And, Or.\nKindergarten: 'Lions And Owls.'"),

# ---------- 4. CONTROL FLOW ----------
("Conditionals — If/Else",
"A conditional runs code if a condition is true.\n\n\
Python: if x>0: print(x)\nC: if(x>0){printf(\"%d\",x);}\
\nRust: if x>0 { println!(\"{}\",x); }\nC++: if(x>0){std::cout<<x;}",
"I E → If, Else.\nKindergarten: 'Ice Everywhere.'"),

("Loops — For",
"A for loop repeats for a known range.\n\n\
Python: for i in range(3): print(i)\nC: for(int i=0;i<3;i++){}\nRust: for i in 0..3{}\nC++: for(int i=0;i<3;i++){}",
"F R → For, Repeat.\nKindergarten: 'Fish Run.'"),

("Loops — While",
"A while loop repeats until a condition is false.\n\n\
Python: while x>0: x-=1\nC: while(x>0){x--;}\nRust: while x>0 {x-=1;}\nC++: while(x>0){x--;}",
"W C → While, Condition.\nKindergarten: 'Whales Cry.'"),

# ---------- 5. FUNCTIONS ----------
("Functions — Concept",
"A function (reusable code block) takes input (parameters) and returns output. \
Encourages modularity (breaking programs into parts).",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

("Functions — Examples",
"Python: def add(a,b): return a+b\nC: int add(int a,int b){return a+b;}\
\nRust: fn add(a:i32,b:i32)->i32{a+b}\nC++: int add(int a,int b){return a+b;}",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),

# ---------- 6. DATA STRUCTURES ----------
("Arrays — Concept",
"Array = fixed-size, contiguous memory block. \
Efficient but size cannot change.",
"A F → Array, Fixed.\nKindergarten: 'Apples Fall.'"),

("Arrays — Examples",
"C: int nums[3]={1,2,3};\nRust: let arr=[1,2,3];\nC++: int nums[3]={1,2,3};",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

("Lists/Vectors — Dynamic",
"List/Vector = resizable array.\n\n\
Python: nums=[1,2,3]\nRust: let v=vec![1,2,3];\nC++: std::vector<int> v={1,2,3};",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

("Maps/Dicts — Key-Value",
"Dict/Map = store key→value pairs.\n\n\
Python: ages={\"bob\":25}\nRust: HashMap::new()\nC++: std::map<string,int> ages;",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Sets — Unique",
"Set = collection of unique items.\n\n\
Python: {1,2,3}\nRust: use std::collections::HashSet;\nC++: std::set<int> s;",
"S U → Set, Unique.\nKindergarten: 'Snakes Unite.'"),

# ---------- 7. INPUT/OUTPUT ----------
("Input — User",
"Get input from user.\n\n\
Python: input()\nC: scanf()\nRust: stdin().read_line()\nC++: std::cin",
"I U → Input, User.\nKindergarten: 'Ice Underwear.'"),

("Output — Print",
"Show output to console.\n\n\
Python: print()\nC: printf()\nRust: println!()\nC++: std::cout",
"O P → Output, Print.\nKindergarten: 'Owls Play.'"),

# ---------- 8. ERROR HANDLING ----------
("Error Handling — Concept",
"Error handling = controlled way to manage unexpected events (e.g., divide by zero).\n\
Python: exceptions, C: return codes, Rust: Option/Result, C++: exceptions.",
"E C H → Error, Control, Handle.\nKindergarten: 'Elephants Catch Hats.'"),

("Error Handling — Examples",
"Python: try/except\nC: if(den==0){}\nRust: Option/Result\nC++: try/catch",
"T R → Try, Result.\nKindergarten: 'Turtles Run.'"),

# ---------- 9. COMPILATION ----------
("Compilation vs Interpretation",
"Compiler (C, C++, Rust): translates code into machine code before execution (fast, safe). \
Interpreter (Python): executes line by line at runtime (flexible, easy debugging).",
"C I → Compile, Interpret.\nKindergarten: 'Cats Ice-skate.'"),
]

data_structures = [

# ---------- 1. VARIABLES ----------
("Variables — Concept",
"""A variable (named storage) holds a value in memory.
Typing:
- Python is dynamically typed (type decided at runtime).
- C, Rust, C++ are statically typed (type fixed at compile time).""",
"V M T → Variable, Memory, Type.\nKindergarten: 'Very Many Toys.'"),

("Variables — Examples",
"""Python:
    x = 10
    name = "Alice"

C:
    int x = 10;
    char name[] = "Alice";

Rust:
    let x: i32 = 10;
    let mut name: &str = "Alice";

C++:
    int x = 10;
    std::string name = "Alice";""",
"X I → Example, Initialize.\nKindergarten: 'Xylophones Inspire.'"),

# ---------- 2. TYPES ----------
("Types — Concept",
"""A type (classification of data) defines allowed values and operations.
Common: int, float/double, string/char, bool.""",
"T I F B → Types: Int, Float, Bool.\nKindergarten: 'Tigers In Forests Bite.'"),

("Types — Examples",
"""Python:
    n = 42            # int
    x = 3.14          # float
    ok = True         # bool
    s = "hi"          # str

C:
    int n = 42;
    float x = 3.14f;
    _Bool ok = 1;     // or bool with <stdbool.h>
    char s[] = "hi";

Rust:
    let n: i32 = 42;
    let x: f64 = 3.14;
    let ok: bool = true;
    let s: &str = "hi";

C++:
    int n = 42;
    double x = 3.14;
    bool ok = true;
    std::string s = "hi";""",
"N D → Number, Decimal.\nKindergarten: 'Nice Duck.'"),

# ---------- 3. OPERATORS ----------
("Operators — Arithmetic",
"""Arithmetic: +  -  *  /  % (modulus)

Python:
    print(7 % 3)

C:
    int r = 7 % 3;

Rust:
    let r = 7 % 3;

C++:
    int r = 7 % 3;""",
"A A M → Add, Arithmetic, Mod.\nKindergarten: 'All Ants March.'"),

("Operators — Comparison",
"""Comparison: ==  !=  <  >  <=  >=  (returns boolean)

Python:
    print(5 > 2)

C:
    int b = (5 > 2);

Rust:
    let b = 5 > 2;

C++:
    bool b = (5 > 2);""",
"C C T → Compare, Check, True.\nKindergarten: 'Cats Chase Tails.'"),

("Operators — Logic",
"""Logic: and/&&   or/||   not/!

Python:
    if ok and ready:
        print("go")

C:
    if (ok && ready) { printf("go"); }

Rust:
    if ok && ready { println!("go"); }

C++:
    if (ok && ready) { std::cout << "go"; }""",
"L A O → Logic, And, Or.\nKindergarten: 'Lions And Owls.'"),

# ---------- 4. CONTROL FLOW ----------
("Conditionals — If/Else",
"""A conditional runs code when a condition is true.

Python:
    if x > 0:
        print(x)
    else:
        print("neg or zero")

C:
    if (x > 0) {
        printf("%d", x);
    } else {
        printf("neg or zero");
    }

Rust:
    if x > 0 {
        println!("{}", x);
    } else {
        println!("neg or zero");
    }

C++:
    if (x > 0) {
        std::cout << x;
    } else {
        std::cout << "neg or zero";
    }""",
"I E → If, Else.\nKindergarten: 'Ice Everywhere.'"),

("Loops — For",
"""A for loop repeats for a known range.

Python:
    for i in range(3):
        print(i)

C:
    for (int i = 0; i < 3; i++) {
        printf("%d\\n", i);
    }

Rust:
    for i in 0..3 {
        println!("{}", i);
    }

C++:
    for (int i = 0; i < 3; i++) {
        std::cout << i << "\\n";
    }""",
"F R → For, Repeat.\nKindergarten: 'Fish Run.'"),

("Loops — While",
"""A while loop repeats until the condition becomes false.

Python:
    while x > 0:
        x -= 1

C:
    while (x > 0) {
        x--;
    }

Rust:
    while x > 0 {
        x -= 1;
    }

C++:
    while (x > 0) {
        x--;
    }""",
"W C → While, Condition.\nKindergarten: 'Whales Cry.'"),

# ---------- 5. FUNCTIONS ----------
("Functions — Concept",
"""A function (reusable code block) takes inputs (parameters) and may return output.
Functions promote modularity (split programs into parts).""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

("Functions — Examples",
"""Python:
    def add(a, b):
        return a + b

C:
    int add(int a, int b) {
        return a + b;
    }

Rust:
    fn add(a: i32, b: i32) -> i32 {
        a + b
    }

C++:
    int add(int a, int b) {
        return a + b;
    }""",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),



# ---------- 6. DATA STRUCTURES (quick tour, detailed deck below) ----------
("Arrays — Concept",
"""Array = fixed size, contiguous memory block.
Pros: O(1) access. Cons: fixed size, costly insert/delete.""",
"A F → Array, Fixed.\nKindergarten: 'Apples Fall.'"),

("Arrays — Examples",
"""Python:
    nums = [1, 2, 3]      # behaves like dynamic array

C:
    int nums[3] = {1, 2, 3};

Rust:
    let arr = [1, 2, 3];

C++:
    int nums[3] = {1, 2, 3};""",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

("Lists/Vectors — Dynamic",
"""Dynamic array that can grow and shrink.

Python:
    nums = [1, 2, 3]
    nums.append(4)

C (manual via heap):
    int *arr = malloc(3 * sizeof(int));
    arr[0]=1; arr[1]=2; arr[2]=3;
    arr = realloc(arr, 5 * sizeof(int));
    free(arr);

Rust:
    let mut v = vec![1, 2, 3];
    v.push(4);

C++:
    std::vector<int> v = {1, 2, 3};
    v.push_back(4);""",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

("Maps/Dicts — Key Value",
"""Key to value mapping. Hash maps average O(1) lookup.

Python:
    ages = {"bob": 25}
    print(ages["bob"])

C (struct array demo):
    struct Pair { char key[20]; int value; };
    struct Pair dict[2] = { {"bob",25}, {"alice",30} };

Rust:
    use std::collections::HashMap;
    let mut m = HashMap::new();
    m.insert("bob", 25);

C++:
    std::map<std::string, int> m;     // ordered tree map
    m["bob"] = 25;""",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Sets — Unique",
"""Set holds unique items only.

Python:
    s = {1, 2, 3}
    s.add(4)

C (manual check):
    int arr[3] = {1,2,3};
    int val = 2, exists = 0;
    for (int i=0;i<3;i++) if (arr[i]==val) { exists = 1; break; }

Rust:
    use std::collections::HashSet;
    let mut s = HashSet::new();
    s.insert(1);

C++:
    std::set<int> s;
    s.insert(1);""",
"S U → Set, Unique.\nKindergarten: 'Snakes Unite.'"),

# ---------- 7. I/O ----------
("Input — User",
"""Python:
    name = input("Enter name: ")

C:
    char name[64];
    scanf("%63s", name);

Rust:
    use std::io::{self, Read};
    let mut s = String::new();
    io::stdin().read_line(&mut s).unwrap();

C++:
    std::string name;
    std::cin >> name;""",
"I U → Input, User.\nKindergarten: 'Ice Umbrella.'"),

("Output — Print",
"""Python:
    print("hello")

C:
    printf("hello\\n");

Rust:
    println!("hello");

C++:
    std::cout << "hello\\n";""",
"O P → Output, Print.\nKindergarten: 'Owls Play.'"),

# ---------- 8. ERROR HANDLING ----------
("Error Handling — Concept",
"""Handle unexpected events in a controlled way.
Python and C++ use exceptions. C uses return codes. Rust uses Option/Result.""",
"E C H → Error, Control, Handle.\nKindergarten: 'Elephants Catch Hats.'"),

("Error Handling — Examples",
"""Python:
    try:
        x = 1 / 0
    except ZeroDivisionError:
        print("err")

C:
    if (den == 0) {
        printf("divide by zero");
    }

Rust:
    fn divide(a: i32, b: i32) -> Option<i32> {
        if b == 0 { None } else { Some(a / b) }
    }

C++:
    try {
        int x = 1/0;   // example only
    } catch (...) {
        std::cout << "err";
    }""",
"T R → Try, Result.\nKindergarten: 'Turtles Run.'"),

# ---------- 9. RUNTIME MODEL ----------
("Compilation vs Interpretation",
"""Compiler (C, C++, Rust) turns code into machine code before running.
Interpreter (Python) executes line by line at runtime.""",
"C I → Compile, Interpret.\nKindergarten: 'Cats Ice-skate.'"),
]

# ========== DATA STRUCTURES (expanded, multi-line) ==========
data_structures = [

# ---------- ARRAYS ----------
("Array — Concept",
"""Array = fixed-size, contiguous memory collection.
Pros: O(1) random access.
Cons: fixed size, O(n) insert/delete.""",
"A F C → Array, Fixed, Contiguous.\nKindergarten: 'All Frogs Clap.'"),

("Array — Examples",
"""Python:
    nums = [1, 2, 3]
    print(nums[0])

C:
    int nums[3] = {1,2,3};
    printf("%d\\n", nums[0]);

Rust:
    let arr = [1, 2, 3];
    println!("{}", arr[0]);

C++:
    int nums[3] = {1, 2, 3};
    std::cout << nums[0] << "\\n";""",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

# ---------- LISTS / VECTORS ----------
("List/Vector — Concept",
"""Dynamic array. Resizable. Fast access and append.
Most used general-purpose structure.""",
"L D R → List, Dynamic, Resizable.\nKindergarten: 'Little Dogs Run.'"),

("List/Vector — Examples",
"""Python:
    nums = [1, 2, 3]
    nums.append(4)

C (manual dynamic array):
    int *arr = malloc(3 * sizeof(int));
    arr[0]=1; arr[1]=2; arr[2]=3;
    arr = realloc(arr, 5 * sizeof(int));
    free(arr);

Rust:
    let mut v = vec![1, 2, 3];
    v.push(4);

C++:
    std::vector<int> v = {1, 2, 3};
    v.push_back(4);""",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

# ---------- STACK ----------
("Stack — Concept",
"""Stack uses LIFO: last in, first out.
Use cases: recursion, undo, parsing.""",
"S L → Stack, LIFO.\nKindergarten: 'Snakes Leap.'"),

("Stack — Examples",
"""Python:
    stack = []
    stack.append(1)
    x = stack.pop()

C:
    int stack[100];
    int top = -1;
    stack[++top] = 1;       // push
    int x = stack[top--];   // pop

Rust:
    let mut s: Vec<i32> = Vec::new();
    s.push(1);
    let x = s.pop();

C++:
    std::stack<int> s;
    s.push(1);
    int x = s.top();
    s.pop();""",
"P P → Push, Pop.\nKindergarten: 'Puppies Play.'"),

# ---------- QUEUE ----------
("Queue — Concept",
"""Queue uses FIFO: first in, first out.
Use cases: scheduling, buffering, BFS.""",
"Q F → Queue, FIFO.\nKindergarten: 'Queen Flies.'"),

("Queue — Examples",
"""Python:
    from collections import deque
    q = deque()
    q.append(1)      # enqueue
    x = q.popleft()  # dequeue

C (circular buffer sketch):
    int q[100], front = 0, rear = 0, cap = 100;
    q[rear] = 1;            rear = (rear + 1) % cap;   // enqueue
    int x = q[front];       front = (front + 1) % cap; // dequeue

Rust:
    use std::collections::VecDeque;
    let mut q: VecDeque<i32> = VecDeque::new();
    q.push_back(1);
    let x = q.pop_front();

C++:
    std::queue<int> q;
    q.push(1);
    int x = q.front();
    q.pop();""",
"E D → Enqueue, Dequeue.\nKindergarten: 'Elephants Dance.'"),

# ---------- SET ----------
("Set — Concept",
"""Set stores unique items only.
Great for membership tests and deduplication.""",
"S U → Set, Unique.\nKindergarten: 'Silly Unicorn.'"),

("Set — Examples",
"""Python:
    s = {1, 2, 3}
    s.add(4)

C (simulate with search):
    int arr[3] = {1,2,3};
    int val = 2, exists = 0;
    for (int i=0;i<3;i++) if (arr[i]==val) { exists = 1; break; }

Rust:
    use std::collections::HashSet;
    let mut s: HashSet<i32> = HashSet::new();
    s.insert(1);

C++:
    std::set<int> s;
    s.insert(1);""",
"I U → Insert, Unique.\nKindergarten: 'Ice Unicorn.'"),

# ---------- MAP / DICT ----------
("Map/Dict — Concept",
"""Key to value store.
Hash map: average O(1) lookup. Tree map: O(log n) + ordering.""",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Map/Dict — Examples",
"""Python:
    ages = {"bob": 25}
    print(ages["bob"])

C (struct array demo):
    struct Pair { char key[20]; int value; };
    struct Pair dict[2] = { {"bob",25}, {"alice",30} };
    printf("%s is %d\\n", dict[0].key, dict[0].value);

Rust:
    use std::collections::HashMap;
    let mut m: HashMap<&str, i32> = HashMap::new();
    m.insert("bob", 25);
    println!("{}", m["bob"]);

C++:
    std::map<std::string, int> m;   // tree map (ordered)
    m["bob"] = 25;
    std::cout << m["bob"] << "\\n";""",
"K V → Key, Value.\nKindergarten: 'Kangaroos Visit.'"),
]

variables = [

# ---------- CONCEPT ----------
("Variables — Concept",
"""Definition: Named storage for a value (lives in memory).
Key ideas: declaration, initialization, type, scope, lifetime.""",
"V D T S L → Variable, Decl, Type, Scope, Lifetime.\nKindergarten: 'Very Daring Tigers Smile Loudly.'"),

# ---------- DECLARATION vs INITIALIZATION ----------
("Variables — Declare vs Initialize",
"""Declare = introduce name and type. Initialize = give first value.

Python:
    x = 0                 # declare+init (dynamic)

C:
    int x; x = 0;         # separate
    int y = 1;            # combined

Rust:
    let x: i32; x = 0;
    let y: i32 = 1;

C++:
    int x; x = 0;
    int y = 1;""",
"D I → Declare, Initialize.\nKindergarten: 'Dogs Ice-skate.'"),

# ---------- TYPE SYSTEM ----------
("Variables — Types",
"""Type = classification of data (int, float, bool, string).
Dynamic typing (Python) vs static typing (C, Rust, C++).

Python:
    a = 42          # int
    b = 3.14        # float
    ok = True       # bool

C:
    int a = 42;
    float b = 3.14f;
    _Bool ok = 1;   // or bool with <stdbool.h>

Rust:
    let a: i32 = 42;
    let b: f64 = 3.14;
    let ok: bool = true;

C++:
    int a = 42;
    double b = 3.14;
    bool ok = true;""",
"T D S → Type, Dynamic, Static.\nKindergarten: 'Turtles Dance Softly.'"),

# ---------- TYPE INFERENCE ----------
("Variables — Type Inference",
"""Inference = compiler deduces type from value.

Python:
    x = 10                 # runtime typing

Rust:
    let x = 10;            # infers i32
    let y = 3.14;          # infers f64

C++:
    auto x = 10;           # int
    auto y = 3.14;         # double""",
"I V → Inference, Value.\nKindergarten: 'Iguanas Vacation.'"),

# ---------- MUTABILITY ----------
("Variables — Mutability",
"""Mutable = can change after init. Immutable = cannot change.

Python:
    x = 1
    x = 2                   # rebinding allowed

C:
    int x = 1; x = 2;

Rust:
    let x = 1;              // immutable
    let mut y = 1; y = 2;   // mutable

C++:
    int x = 1; x = 2;""",
"M I → Mutable, Immutable.\nKindergarten: 'Mice Imagine.'"),

# ---------- CONSTANTS ----------
("Variables — Constants",
"""Constant = value that never changes.

Python:
    PI = 3.14159            # by convention uppercase

C:
    const int N = 10;

Rust:
    const N: i32 = 10;
    static MAX: i32 = 100;  // global static

C++:
    const int N = 10;""",
"C F S → Const, Fixed, Static.\nKindergarten: 'Cats Find Sun.'"),

# ---------- REFERENCES and POINTERS (preview) ----------
("Variables — Refs and Pointers (Preview)",
"""Reference = alias to another variable. Pointer = holds an address.

C:
    int v = 10;
    int *p = &v;     // pointer to v

Rust:
    let v = 10;
    let r = &v;      // reference (borrow)

C++:
    int v = 10;
    int &r = v;      // reference""",
"R P → Reference, Pointer.\nKindergarten: 'Rabbits Play.'"),

# ---------- SCOPE ----------
("Variables — Scope",
"""Scope = where a name is visible.

Python:
    x = 1
    def f():
        y = 2           # local

C:
    int x = 1;
    void f(){ int y = 2; }

Rust:
    let x = 1;
    { let y = 2; }      // y ends here

C++:
    int x = 1;
    void f(){ int y = 2; }""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- LIFETIME ----------
("Variables — Lifetime",
"""Lifetime = how long a value exists.

Python:
    def f():
        a = 1  # destroyed after f returns

C:
    void f(){ int a = 1; }  // stack lifetime

Rust:
    // compiler checks lifetimes via borrowing rules

C++:
    { int a = 1; }  // a destroyed at block end""",
"L B → Lifetime, Block.\nKindergarten: 'Lambs Bounce.'"),

# ---------- SHADOWING ----------
("Variables — Shadowing",
"""Shadowing = new declaration hides an older one in inner scope.

Python:
    x = 1
    def f():
        x = 2  # hides outer x inside f

Rust:
    let x = 1;
    let x = x + 1;  // new x shadows old

C++:
    int x = 1;
    { int x = 2; }  // inner x hides outer""",
"S H → Shadow, Hide.\nKindergarten: 'Shy Hedgehog.'"),

# ---------- INITIALIZATION PITFALLS ----------
("Variables — Init Pitfalls",
"""Uninitialized use can cause bugs.

C:
    int x; printf("%d", x);      // undefined

C++:
    int x; std::cout << x;       // undefined

Rust:
    let x: i32; // must assign before use -> compile error

Python:
    # NameError if you read a variable not yet defined""",
"U I → Uninitialized, Invalid.\nKindergarten: 'Umbrellas Itch.'"),

# ---------- NAMING ----------
("Variables — Naming",
"""Use clear, descriptive names. Prefer lower_snake_case in Python/Rust, camelCase or snake_case in C/C++ projects.

Python:
    max_count = 10

Rust:
    let max_count = 10;

C/C++:
    int max_count = 10;""",
"N C → Name, Clarity.\nKindergarten: 'New Car.'"),
]


functions = [

# ---------- CONCEPT ----------
("Functions — Concept",
"""A function is a reusable block of code.
- Takes input (parameters).
- May return output.
- Promotes modularity (splitting programs into parts).""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

# ---------- SIMPLE FUNCTION ----------
("Functions — Simple Example",
"""Python:
    def hello():
        print("Hello!")

C:
    void hello() {
        printf("Hello!\\n");
    }

Rust:
    fn hello() {
        println!("Hello!");
    }

C++:
    void hello() {
        std::cout << "Hello!\\n";
    }""",
"H → Hello.\nKindergarten: 'Happy Hippo.'"),

# ---------- PARAMETERS ----------
("Functions — Parameters & Return",
"""Python:
    def add(a, b):
        return a + b

C:
    int add(int a, int b) {
        return a + b;
    }

Rust:
    fn add(a: i32, b: i32) -> i32 {
        a + b
    }

C++:
    int add(int a, int b) {
        return a + b;
    }""",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),

# ---------- VOID / NO RETURN ----------
("Functions — No Return (Void)",
"""Python:
    def log(msg):
        print(msg)   # no explicit return

C:
    void log(const char* msg) {
        printf("%s", msg);
    }

Rust:
    fn log(msg: &str) {
        println!("{}", msg);
    }

C++:
    void log(const std::string& msg) {
        std::cout << msg;
    }""",
"V N → Void, No-return.\nKindergarten: 'Very Nice.'"),

# ---------- SCOPE ----------
("Functions — Scope",
"""Local variables exist only inside a function.
Global variables exist everywhere.

Python:
    x = 10
    def f():
        y = 20  # local

C:
    int x = 10;
    void f() {
        int y = 20;  // local
    }

Rust:
    let x: i32 = 10;
    fn f() {
        let y: i32 = 20;
    }

C++:
    int x = 10;
    void f() {
        int y = 20;
    }""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- CONCEPT ----------
("Functions — Concept",
"""Definition: Reusable block of code.
- Takes inputs (parameters).
- May return outputs.
- Improves modularity.""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

# ---------- CALL SEMANTICS ----------
("Functions — Call Semantics",
"""Value = copy of argument.
Reference = direct access to original.

Python:
    def inc(lst): lst.append(1)

C:
    void inc(int *x){ (*x)++; }

Rust:
    fn inc(x: &mut i32){ *x += 1; }

C++:
    void inc(int &x){ x++; }""",
"V R → Value, Ref.\nKindergarten: 'Very Round.'"),

# ---------- PURE vs SIDE EFFECT ----------
("Functions — Purity vs Side Effect",
"""Pure = same input → same output, no external change.
Side effect = modifies outside state (I/O, globals).""",
"P S → Pure, Side.\nKindergarten: 'Pink Snake.'"),

# ---------- SCOPE ----------
("Functions — Scope",
"""Local = variable inside a function.
Global = variable accessible everywhere.""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- RECURSION ----------
("Functions — Recursion",
"""Definition: Function calls itself.
Requires base case to stop recursion.

Python:
    def fact(n): return 1 if n<=1 else n*fact(n-1)""",
"R B → Recursion, Base.\nKindergarten: 'Red Balloon.'"),

# ---------- HIGHER ORDER ----------
("Functions — Higher Order",
"""Definition: Functions that take or return other functions.
Closure = captures surrounding variables.

Python:
    def make_adder(k): return lambda x:x+k""",
"H C → Higher, Closure.\nKindergarten: 'Happy Cat.'"),

# ---------- OVERLOAD / GENERIC ----------
("Functions — Overload & Generics",
"""Overload = same name, different parameters.
Generic = type-flexible functions.

C++:
    int add(int,int);
    double add(double,double);
    template<typename T>T maxT(T a,T b);""",
"O G → Overload, Generic.\nKindergarten: 'Orange Giraffe.'"),

# ---------- VARIADIC ----------
("Functions — Variadic",
"""Definition: Accept variable number of arguments.

Python:
    def total(*xs): return sum(xs)

C:
    int total(int n,...); // stdarg.h

C++:
    template<typename...T> int total(T...xs){ return (0+...+xs);}""",
"V A → Variadic Args.\nKindergarten: 'Very Antsy.'"),

# ---------- ERROR ----------
("Functions — Error Handling",
"""C = return codes.
C++ = exceptions.
Rust = Result/Option.
Python = exceptions.""",
"E R → Error, Result.\nKindergarten: 'Elephant Runs.'"),

# ---------- INLINE / LINK ----------
("Functions — Inline & Linkage",
"""Inline = hint to expand function at call site.
Linkage = scope across files (internal vs external).""",
"I L → Inline, Link.\nKindergarten: 'Ice Lion.'"),

# ---------- ASYNC ----------
("Functions — Asynchronous",
"""Sync = blocks until finished.
Async = returns immediately, completes later.

Python: async/await
Rust: async/await
C++20: coroutines
C: threads/event loops""",
"A S → Async, Sync.\nKindergarten: 'Ant Sings.'"),
]



# ---------------------- Topic registry ----------------------
# List your deck variable names here
topic_vars = [
    "programming_basics",
    "variables",
    "functions",
    "data_structures",

    # add more like: "flashcards_data_structures", "flashcards_algorithms"
]

# Validate topics exist
_missing = [name for name in topic_vars if name not in globals()]
if _missing:
    raise NameError(f"Missing deck variables: {_missing}. Define them or remove from topic_vars.")

# Build combined "All" deck by concatenating every listed deck
flashcards_all = []
for _name in topic_vars:
    flashcards_all.extend(globals()[_name])

# Prepend "All" to the dropdown list
topic_vars = ["flashcards_all"] + topic_vars

# ---------------------- State ----------------------
current_deck_name = topic_vars[0]  # default to All
flashcards = globals()[current_deck_name]
index = 0
showing_answer = False
output = widgets.Output()

# ---------------------- Widgets ----------------------
topic_dropdown = widgets.Dropdown(
    options=topic_vars,
    value=current_deck_name,
    description="Topic:"
)
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
shuffle_button = widgets.Button(description="Shuffle 🔀")
jump_input = widgets.BoundedIntText(value=1, min=1, description='Jump to:')
jump_button = widgets.Button(description="Go ⏩")
restart_button.layout.display = 'none'
counter_label = widgets.Label()

# ---------------------- Render ----------------------
def render_card(q_text: str, a_text: str, m_text: str) -> None:
    q_html = escape(q_text)
    m_html = escape(m_text) if m_text else ""

    # Show question in styled box
    display(HTML(f"""
    <div style="background:#fafafa;padding:12px;border-radius:10px;
                border:1px solid #e5e5e5;margin-bottom:8px;
                font-family:sans-serif;font-size:14px;line-height:1.4">
      <b>Q:</b> {q_html}
    </div>
    """))

    # Show answer as real code block
    if showing_answer:
        display(Markdown(f"**A:**\n\n```text\n{a_text}\n```"))  # code block here

        if m_text:
            display(Markdown(f"**Mnemonic:** {m_html}"))


def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]
        if len(card) == 3:
            question, answer, mnemonic = card
        else:
            question, answer = card
            mnemonic = ""
        render_card(question, answer, mnemonic)
    counter_label.value = f"Topic: {current_deck_name} - Card {index+1} of {len(flashcards)}"
    prev_button.disabled = (index == 0)
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"
    jump_input.max = len(flashcards)

# ---------------------- Handlers ----------------------
def on_topic_change(change):
    global current_deck_name, flashcards, index, showing_answer
    if change["name"] == "value" and change["new"] != change["old"]:
        current_deck_name = change["new"]
        flashcards = globals()[current_deck_name]
        index = 0
        showing_answer = False
        restart_button.layout.display = 'none'
        show_button.layout.display = ''
        next_button.layout.display = ''
        prev_button.layout.display = ''
        update_card()

def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You have reached the end of this topic.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

def shuffle_deck(_):
    global flashcards, index, showing_answer
    random.shuffle(flashcards)
    index = 0
    showing_answer = False
    update_card()

def jump_to_card(_):
    global index, showing_answer
    val = jump_input.value - 1
    if 0 <= val < len(flashcards):
        index = val
        showing_answer = False
        update_card()

# ---------------------- Wire up ----------------------
topic_dropdown.observe(on_topic_change, names="value")
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)
shuffle_button.on_click(shuffle_deck)
jump_button.on_click(jump_to_card)

# ---------------------- Launch ----------------------
update_card()
display(widgets.VBox([
    widgets.HBox([topic_dropdown]),
    widgets.HBox([prev_button, show_button, next_button, restart_button, shuffle_button]),
    widgets.HBox([counter_label]),
    widgets.HBox([jump_input, jump_button]),
    output
]))
````

````python

# Colab Flashcard Viewer (no flashcards bundled)
# - Mobile-friendly: wrapped code/answers (no horizontal overflow)
# - Plug in your own decks by defining them separately and listing them in `topic_vars`

import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output, HTML
from html import escape
import random

# ---------------------- Global CSS: wrap long output on mobile ----------------------
display(HTML("""
<style>
.output_subarea pre,
.output_subarea code,
pre, code {
  white-space: pre-wrap !important;
  word-break: break-word !important;
  overflow-wrap: anywhere !important;
}
.output_wrapper, .output, .jp-OutputArea-output {
  max-width: 100% !important;
}
</style>
"""))

# ---------------------- Example decks (replace or extend) ----------------------
programming_basics = [

# ---------- 1. VARIABLES ----------
("Variables — Concept",
"A variable (named storage) holds a value in memory. \
Python is dynamically typed (type decided at runtime). \
C, Rust, C++ are statically typed (type fixed at compile time).",
"V M T → Variable, Memory, Type.\nKindergarten: 'Very Many Toys.'"),

("Variables — Examples",
"Python: x=10\nC: int x=10;\nRust: let x:i32=10;\nC++: int x=10;",
"X I → Example, Initialize.\nKindergarten: 'Xylophones Inspire.'"),

# ---------- 2. TYPES ----------
("Types — Concept",
"A type (classification of data) defines allowed values and operations. \
Common: int, float, string/char, bool.",
"T I F B → Types: Int, Float, Bool.\nKindergarten: 'Tigers In Forests Bite.'"),

("Types — Examples",
"Python: x=3.14\nC: float x=3.14;\nRust: let y:f64=3.14;\nC++: double x=3.14;",
"N D → Number, Decimal.\nKindergarten: 'Nice Duck.'"),

# ---------- 3. OPERATORS ----------
("Operators — Arithmetic",
"Arithmetic operators = + - * / % (modulus).\n\n\
Python: 5%2\nC: 5%2\nRust: 5%2\nC++: 5%2",
"A A M → Add, Arithmetic, Mod.\nKindergarten: 'All Ants March.'"),

("Operators — Comparison",
"Comparison operators = == != < > <= >=.\nReturn true/false.",
"C C T → Compare, Check, True.\nKindergarten: 'Cats Chase Tails.'"),

("Operators — Logic",
"Logical operators = and/&&, or/||, not/!.\n\
Python: if a and b: ...\nC/C++/Rust: if(a && b) { }",
"L A O → Logic, And, Or.\nKindergarten: 'Lions And Owls.'"),

# ---------- 4. CONTROL FLOW ----------
("Conditionals — If/Else",
"A conditional runs code if a condition is true.\n\n\
Python: if x>0: print(x)\nC: if(x>0){printf(\"%d\",x);}\
\nRust: if x>0 { println!(\"{}\",x); }\nC++: if(x>0){std::cout<<x;}",
"I E → If, Else.\nKindergarten: 'Ice Everywhere.'"),

("Loops — For",
"A for loop repeats for a known range.\n\n\
Python: for i in range(3): print(i)\nC: for(int i=0;i<3;i++){}\nRust: for i in 0..3{}\nC++: for(int i=0;i<3;i++){}",
"F R → For, Repeat.\nKindergarten: 'Fish Run.'"),

("Loops — While",
"A while loop repeats until a condition is false.\n\n\
Python: while x>0: x-=1\nC: while(x>0){x--;}\nRust: while x>0 {x-=1;}\nC++: while(x>0){x--;}",
"W C → While, Condition.\nKindergarten: 'Whales Cry.'"),

# ---------- 5. FUNCTIONS ----------
("Functions — Concept",
"A function (reusable code block) takes input (parameters) and returns output. \
Encourages modularity (breaking programs into parts).",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

("Functions — Examples",
"Python: def add(a,b): return a+b\nC: int add(int a,int b){return a+b;}\
\nRust: fn add(a:i32,b:i32)->i32{a+b}\nC++: int add(int a,int b){return a+b;}",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),

# ---------- 6. DATA STRUCTURES ----------
("Arrays — Concept",
"Array = fixed-size, contiguous memory block. \
Efficient but size cannot change.",
"A F → Array, Fixed.\nKindergarten: 'Apples Fall.'"),

("Arrays — Examples",
"C: int nums[3]={1,2,3};\nRust: let arr=[1,2,3];\nC++: int nums[3]={1,2,3};",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

("Lists/Vectors — Dynamic",
"List/Vector = resizable array.\n\n\
Python: nums=[1,2,3]\nRust: let v=vec![1,2,3];\nC++: std::vector<int> v={1,2,3};",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

("Maps/Dicts — Key-Value",
"Dict/Map = store key→value pairs.\n\n\
Python: ages={\"bob\":25}\nRust: HashMap::new()\nC++: std::map<string,int> ages;",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Sets — Unique",
"Set = collection of unique items.\n\n\
Python: {1,2,3}\nRust: use std::collections::HashSet;\nC++: std::set<int> s;",
"S U → Set, Unique.\nKindergarten: 'Snakes Unite.'"),

# ---------- 7. INPUT/OUTPUT ----------
("Input — User",
"Get input from user.\n\n\
Python: input()\nC: scanf()\nRust: stdin().read_line()\nC++: std::cin",
"I U → Input, User.\nKindergarten: 'Ice Underwear.'"),

("Output — Print",
"Show output to console.\n\n\
Python: print()\nC: printf()\nRust: println!()\nC++: std::cout",
"O P → Output, Print.\nKindergarten: 'Owls Play.'"),

# ---------- 8. ERROR HANDLING ----------
("Error Handling — Concept",
"Error handling = controlled way to manage unexpected events (e.g., divide by zero).\n\
Python: exceptions, C: return codes, Rust: Option/Result, C++: exceptions.",
"E C H → Error, Control, Handle.\nKindergarten: 'Elephants Catch Hats.'"),

("Error Handling — Examples",
"Python: try/except\nC: if(den==0){}\nRust: Option/Result\nC++: try/catch",
"T R → Try, Result.\nKindergarten: 'Turtles Run.'"),

# ---------- 9. COMPILATION ----------
("Compilation vs Interpretation",
"Compiler (C, C++, Rust): translates code into machine code before execution (fast, safe). \
Interpreter (Python): executes line by line at runtime (flexible, easy debugging).",
"C I → Compile, Interpret.\nKindergarten: 'Cats Ice-skate.'"),
]

data_structures = [

# ---------- 1. VARIABLES ----------
("Variables — Concept",
"""A variable (named storage) holds a value in memory.
Typing:
- Python is dynamically typed (type decided at runtime).
- C, Rust, C++ are statically typed (type fixed at compile time).""",
"V M T → Variable, Memory, Type.\nKindergarten: 'Very Many Toys.'"),

("Variables — Examples",
"""Python:
    x = 10
    name = "Alice"

C:
    int x = 10;
    char name[] = "Alice";

Rust:
    let x: i32 = 10;
    let mut name: &str = "Alice";

C++:
    int x = 10;
    std::string name = "Alice";""",
"X I → Example, Initialize.\nKindergarten: 'Xylophones Inspire.'"),

# ---------- 2. TYPES ----------
("Types — Concept",
"""A type (classification of data) defines allowed values and operations.
Common: int, float/double, string/char, bool.""",
"T I F B → Types: Int, Float, Bool.\nKindergarten: 'Tigers In Forests Bite.'"),

("Types — Examples",
"""Python:
    n = 42            # int
    x = 3.14          # float
    ok = True         # bool
    s = "hi"          # str

C:
    int n = 42;
    float x = 3.14f;
    _Bool ok = 1;     // or bool with <stdbool.h>
    char s[] = "hi";

Rust:
    let n: i32 = 42;
    let x: f64 = 3.14;
    let ok: bool = true;
    let s: &str = "hi";

C++:
    int n = 42;
    double x = 3.14;
    bool ok = true;
    std::string s = "hi";""",
"N D → Number, Decimal.\nKindergarten: 'Nice Duck.'"),

# ---------- 3. OPERATORS ----------
("Operators — Arithmetic",
"""Arithmetic: +  -  *  /  % (modulus)

Python:
    print(7 % 3)

C:
    int r = 7 % 3;

Rust:
    let r = 7 % 3;

C++:
    int r = 7 % 3;""",
"A A M → Add, Arithmetic, Mod.\nKindergarten: 'All Ants March.'"),

("Operators — Comparison",
"""Comparison: ==  !=  <  >  <=  >=  (returns boolean)

Python:
    print(5 > 2)

C:
    int b = (5 > 2);

Rust:
    let b = 5 > 2;

C++:
    bool b = (5 > 2);""",
"C C T → Compare, Check, True.\nKindergarten: 'Cats Chase Tails.'"),

("Operators — Logic",
"""Logic: and/&&   or/||   not/!

Python:
    if ok and ready:
        print("go")

C:
    if (ok && ready) { printf("go"); }

Rust:
    if ok && ready { println!("go"); }

C++:
    if (ok && ready) { std::cout << "go"; }""",
"L A O → Logic, And, Or.\nKindergarten: 'Lions And Owls.'"),

# ---------- 4. CONTROL FLOW ----------
("Conditionals — If/Else",
"""A conditional runs code when a condition is true.

Python:
    if x > 0:
        print(x)
    else:
        print("neg or zero")

C:
    if (x > 0) {
        printf("%d", x);
    } else {
        printf("neg or zero");
    }

Rust:
    if x > 0 {
        println!("{}", x);
    } else {
        println!("neg or zero");
    }

C++:
    if (x > 0) {
        std::cout << x;
    } else {
        std::cout << "neg or zero";
    }""",
"I E → If, Else.\nKindergarten: 'Ice Everywhere.'"),

("Loops — For",
"""A for loop repeats for a known range.

Python:
    for i in range(3):
        print(i)

C:
    for (int i = 0; i < 3; i++) {
        printf("%d\\n", i);
    }

Rust:
    for i in 0..3 {
        println!("{}", i);
    }

C++:
    for (int i = 0; i < 3; i++) {
        std::cout << i << "\\n";
    }""",
"F R → For, Repeat.\nKindergarten: 'Fish Run.'"),

("Loops — While",
"""A while loop repeats until the condition becomes false.

Python:
    while x > 0:
        x -= 1

C:
    while (x > 0) {
        x--;
    }

Rust:
    while x > 0 {
        x -= 1;
    }

C++:
    while (x > 0) {
        x--;
    }""",
"W C → While, Condition.\nKindergarten: 'Whales Cry.'"),

# ---------- 5. FUNCTIONS ----------
("Functions — Concept",
"""A function (reusable code block) takes inputs (parameters) and may return output.
Functions promote modularity (split programs into parts).""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

("Functions — Examples",
"""Python:
    def add(a, b):
        return a + b

C:
    int add(int a, int b) {
        return a + b;
    }

Rust:
    fn add(a: i32, b: i32) -> i32 {
        a + b
    }

C++:
    int add(int a, int b) {
        return a + b;
    }""",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),



# ---------- 6. DATA STRUCTURES (quick tour, detailed deck below) ----------
("Arrays — Concept",
"""Array = fixed size, contiguous memory block.
Pros: O(1) access. Cons: fixed size, costly insert/delete.""",
"A F → Array, Fixed.\nKindergarten: 'Apples Fall.'"),

("Arrays — Examples",
"""Python:
    nums = [1, 2, 3]      # behaves like dynamic array

C:
    int nums[3] = {1, 2, 3};

Rust:
    let arr = [1, 2, 3];

C++:
    int nums[3] = {1, 2, 3};""",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

("Lists/Vectors — Dynamic",
"""Dynamic array that can grow and shrink.

Python:
    nums = [1, 2, 3]
    nums.append(4)

C (manual via heap):
    int *arr = malloc(3 * sizeof(int));
    arr[0]=1; arr[1]=2; arr[2]=3;
    arr = realloc(arr, 5 * sizeof(int));
    free(arr);

Rust:
    let mut v = vec![1, 2, 3];
    v.push(4);

C++:
    std::vector<int> v = {1, 2, 3};
    v.push_back(4);""",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

("Maps/Dicts — Key Value",
"""Key to value mapping. Hash maps average O(1) lookup.

Python:
    ages = {"bob": 25}
    print(ages["bob"])

C (struct array demo):
    struct Pair { char key[20]; int value; };
    struct Pair dict[2] = { {"bob",25}, {"alice",30} };

Rust:
    use std::collections::HashMap;
    let mut m = HashMap::new();
    m.insert("bob", 25);

C++:
    std::map<std::string, int> m;     // ordered tree map
    m["bob"] = 25;""",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Sets — Unique",
"""Set holds unique items only.

Python:
    s = {1, 2, 3}
    s.add(4)

C (manual check):
    int arr[3] = {1,2,3};
    int val = 2, exists = 0;
    for (int i=0;i<3;i++) if (arr[i]==val) { exists = 1; break; }

Rust:
    use std::collections::HashSet;
    let mut s = HashSet::new();
    s.insert(1);

C++:
    std::set<int> s;
    s.insert(1);""",
"S U → Set, Unique.\nKindergarten: 'Snakes Unite.'"),

# ---------- 7. I/O ----------
("Input — User",
"""Python:
    name = input("Enter name: ")

C:
    char name[64];
    scanf("%63s", name);

Rust:
    use std::io::{self, Read};
    let mut s = String::new();
    io::stdin().read_line(&mut s).unwrap();

C++:
    std::string name;
    std::cin >> name;""",
"I U → Input, User.\nKindergarten: 'Ice Umbrella.'"),

("Output — Print",
"""Python:
    print("hello")

C:
    printf("hello\\n");

Rust:
    println!("hello");

C++:
    std::cout << "hello\\n";""",
"O P → Output, Print.\nKindergarten: 'Owls Play.'"),

# ---------- 8. ERROR HANDLING ----------
("Error Handling — Concept",
"""Handle unexpected events in a controlled way.
Python and C++ use exceptions. C uses return codes. Rust uses Option/Result.""",
"E C H → Error, Control, Handle.\nKindergarten: 'Elephants Catch Hats.'"),

("Error Handling — Examples",
"""Python:
    try:
        x = 1 / 0
    except ZeroDivisionError:
        print("err")

C:
    if (den == 0) {
        printf("divide by zero");
    }

Rust:
    fn divide(a: i32, b: i32) -> Option<i32> {
        if b == 0 { None } else { Some(a / b) }
    }

C++:
    try {
        int x = 1/0;   // example only
    } catch (...) {
        std::cout << "err";
    }""",
"T R → Try, Result.\nKindergarten: 'Turtles Run.'"),

# ---------- 9. RUNTIME MODEL ----------
("Compilation vs Interpretation",
"""Compiler (C, C++, Rust) turns code into machine code before running.
Interpreter (Python) executes line by line at runtime.""",
"C I → Compile, Interpret.\nKindergarten: 'Cats Ice-skate.'"),
]

# ========== DATA STRUCTURES (expanded, multi-line) ==========
data_structures = [

# ---------- ARRAYS ----------
("Array — Concept",
"""Array = fixed-size, contiguous memory collection.
Pros: O(1) random access.
Cons: fixed size, O(n) insert/delete.""",
"A F C → Array, Fixed, Contiguous.\nKindergarten: 'All Frogs Clap.'"),

("Array — Examples",
"""Python:
    nums = [1, 2, 3]
    print(nums[0])

C:
    int nums[3] = {1,2,3};
    printf("%d\\n", nums[0]);

Rust:
    let arr = [1, 2, 3];
    println!("{}", arr[0]);

C++:
    int nums[3] = {1, 2, 3};
    std::cout << nums[0] << "\\n";""",
"N F → Numbers, Fixed.\nKindergarten: 'Nice Frog.'"),

# ---------- LISTS / VECTORS ----------
("List/Vector — Concept",
"""Dynamic array. Resizable. Fast access and append.
Most used general-purpose structure.""",
"L D R → List, Dynamic, Resizable.\nKindergarten: 'Little Dogs Run.'"),

("List/Vector — Examples",
"""Python:
    nums = [1, 2, 3]
    nums.append(4)

C (manual dynamic array):
    int *arr = malloc(3 * sizeof(int));
    arr[0]=1; arr[1]=2; arr[2]=3;
    arr = realloc(arr, 5 * sizeof(int));
    free(arr);

Rust:
    let mut v = vec![1, 2, 3];
    v.push(4);

C++:
    std::vector<int> v = {1, 2, 3};
    v.push_back(4);""",
"L V → List, Vector.\nKindergarten: 'Lions Visit.'"),

# ---------- STACK ----------
("Stack — Concept",
"""Stack uses LIFO: last in, first out.
Use cases: recursion, undo, parsing.""",
"S L → Stack, LIFO.\nKindergarten: 'Snakes Leap.'"),

("Stack — Examples",
"""Python:
    stack = []
    stack.append(1)
    x = stack.pop()

C:
    int stack[100];
    int top = -1;
    stack[++top] = 1;       // push
    int x = stack[top--];   // pop

Rust:
    let mut s: Vec<i32> = Vec::new();
    s.push(1);
    let x = s.pop();

C++:
    std::stack<int> s;
    s.push(1);
    int x = s.top();
    s.pop();""",
"P P → Push, Pop.\nKindergarten: 'Puppies Play.'"),

# ---------- QUEUE ----------
("Queue — Concept",
"""Queue uses FIFO: first in, first out.
Use cases: scheduling, buffering, BFS.""",
"Q F → Queue, FIFO.\nKindergarten: 'Queen Flies.'"),

("Queue — Examples",
"""Python:
    from collections import deque
    q = deque()
    q.append(1)      # enqueue
    x = q.popleft()  # dequeue

C (circular buffer sketch):
    int q[100], front = 0, rear = 0, cap = 100;
    q[rear] = 1;            rear = (rear + 1) % cap;   // enqueue
    int x = q[front];       front = (front + 1) % cap; // dequeue

Rust:
    use std::collections::VecDeque;
    let mut q: VecDeque<i32> = VecDeque::new();
    q.push_back(1);
    let x = q.pop_front();

C++:
    std::queue<int> q;
    q.push(1);
    int x = q.front();
    q.pop();""",
"E D → Enqueue, Dequeue.\nKindergarten: 'Elephants Dance.'"),

# ---------- SET ----------
("Set — Concept",
"""Set stores unique items only.
Great for membership tests and deduplication.""",
"S U → Set, Unique.\nKindergarten: 'Silly Unicorn.'"),

("Set — Examples",
"""Python:
    s = {1, 2, 3}
    s.add(4)

C (simulate with search):
    int arr[3] = {1,2,3};
    int val = 2, exists = 0;
    for (int i=0;i<3;i++) if (arr[i]==val) { exists = 1; break; }

Rust:
    use std::collections::HashSet;
    let mut s: HashSet<i32> = HashSet::new();
    s.insert(1);

C++:
    std::set<int> s;
    s.insert(1);""",
"I U → Insert, Unique.\nKindergarten: 'Ice Unicorn.'"),

# ---------- MAP / DICT ----------
("Map/Dict — Concept",
"""Key to value store.
Hash map: average O(1) lookup. Tree map: O(log n) + ordering.""",
"M K V → Map, Key, Value.\nKindergarten: 'Monkeys Keep Vines.'"),

("Map/Dict — Examples",
"""Python:
    ages = {"bob": 25}
    print(ages["bob"])

C (struct array demo):
    struct Pair { char key[20]; int value; };
    struct Pair dict[2] = { {"bob",25}, {"alice",30} };
    printf("%s is %d\\n", dict[0].key, dict[0].value);

Rust:
    use std::collections::HashMap;
    let mut m: HashMap<&str, i32> = HashMap::new();
    m.insert("bob", 25);
    println!("{}", m["bob"]);

C++:
    std::map<std::string, int> m;   // tree map (ordered)
    m["bob"] = 25;
    std::cout << m["bob"] << "\\n";""",
"K V → Key, Value.\nKindergarten: 'Kangaroos Visit.'"),
]

variables = [

# ---------- CONCEPT ----------
("Variables — Concept",
"""Definition: Named storage for a value (lives in memory).
Key ideas: declaration, initialization, type, scope, lifetime.""",
"V D T S L → Variable, Decl, Type, Scope, Lifetime.\nKindergarten: 'Very Daring Tigers Smile Loudly.'"),

# ---------- DECLARATION vs INITIALIZATION ----------
("Variables — Declare vs Initialize",
"""Declare = introduce name and type. Initialize = give first value.

Python:
    x = 0                 # declare+init (dynamic)

C:
    int x; x = 0;         # separate
    int y = 1;            # combined

Rust:
    let x: i32; x = 0;
    let y: i32 = 1;

C++:
    int x; x = 0;
    int y = 1;""",
"D I → Declare, Initialize.\nKindergarten: 'Dogs Ice-skate.'"),

# ---------- TYPE SYSTEM ----------
("Variables — Types",
"""Type = classification of data (int, float, bool, string).
Dynamic typing (Python) vs static typing (C, Rust, C++).

Python:
    a = 42          # int
    b = 3.14        # float
    ok = True       # bool

C:
    int a = 42;
    float b = 3.14f;
    _Bool ok = 1;   // or bool with <stdbool.h>

Rust:
    let a: i32 = 42;
    let b: f64 = 3.14;
    let ok: bool = true;

C++:
    int a = 42;
    double b = 3.14;
    bool ok = true;""",
"T D S → Type, Dynamic, Static.\nKindergarten: 'Turtles Dance Softly.'"),

# ---------- TYPE INFERENCE ----------
("Variables — Type Inference",
"""Inference = compiler deduces type from value.

Python:
    x = 10                 # runtime typing

Rust:
    let x = 10;            # infers i32
    let y = 3.14;          # infers f64

C++:
    auto x = 10;           # int
    auto y = 3.14;         # double""",
"I V → Inference, Value.\nKindergarten: 'Iguanas Vacation.'"),

# ---------- MUTABILITY ----------
("Variables — Mutability",
"""Mutable = can change after init. Immutable = cannot change.

Python:
    x = 1
    x = 2                   # rebinding allowed

C:
    int x = 1; x = 2;

Rust:
    let x = 1;              // immutable
    let mut y = 1; y = 2;   // mutable

C++:
    int x = 1; x = 2;""",
"M I → Mutable, Immutable.\nKindergarten: 'Mice Imagine.'"),

# ---------- CONSTANTS ----------
("Variables — Constants",
"""Constant = value that never changes.

Python:
    PI = 3.14159            # by convention uppercase

C:
    const int N = 10;

Rust:
    const N: i32 = 10;
    static MAX: i32 = 100;  // global static

C++:
    const int N = 10;""",
"C F S → Const, Fixed, Static.\nKindergarten: 'Cats Find Sun.'"),

# ---------- REFERENCES and POINTERS (preview) ----------
("Variables — Refs and Pointers (Preview)",
"""Reference = alias to another variable. Pointer = holds an address.

C:
    int v = 10;
    int *p = &v;     // pointer to v

Rust:
    let v = 10;
    let r = &v;      // reference (borrow)

C++:
    int v = 10;
    int &r = v;      // reference""",
"R P → Reference, Pointer.\nKindergarten: 'Rabbits Play.'"),

# ---------- SCOPE ----------
("Variables — Scope",
"""Scope = where a name is visible.

Python:
    x = 1
    def f():
        y = 2           # local

C:
    int x = 1;
    void f(){ int y = 2; }

Rust:
    let x = 1;
    { let y = 2; }      // y ends here

C++:
    int x = 1;
    void f(){ int y = 2; }""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- LIFETIME ----------
("Variables — Lifetime",
"""Lifetime = how long a value exists.

Python:
    def f():
        a = 1  # destroyed after f returns

C:
    void f(){ int a = 1; }  // stack lifetime

Rust:
    // compiler checks lifetimes via borrowing rules

C++:
    { int a = 1; }  // a destroyed at block end""",
"L B → Lifetime, Block.\nKindergarten: 'Lambs Bounce.'"),

# ---------- SHADOWING ----------
("Variables — Shadowing",
"""Shadowing = new declaration hides an older one in inner scope.

Python:
    x = 1
    def f():
        x = 2  # hides outer x inside f

Rust:
    let x = 1;
    let x = x + 1;  // new x shadows old

C++:
    int x = 1;
    { int x = 2; }  // inner x hides outer""",
"S H → Shadow, Hide.\nKindergarten: 'Shy Hedgehog.'"),

# ---------- INITIALIZATION PITFALLS ----------
("Variables — Init Pitfalls",
"""Uninitialized use can cause bugs.

C:
    int x; printf("%d", x);      // undefined

C++:
    int x; std::cout << x;       // undefined

Rust:
    let x: i32; // must assign before use -> compile error

Python:
    # NameError if you read a variable not yet defined""",
"U I → Uninitialized, Invalid.\nKindergarten: 'Umbrellas Itch.'"),

# ---------- NAMING ----------
("Variables — Naming",
"""Use clear, descriptive names. Prefer lower_snake_case in Python/Rust, camelCase or snake_case in C/C++ projects.

Python:
    max_count = 10

Rust:
    let max_count = 10;

C/C++:
    int max_count = 10;""",
"N C → Name, Clarity.\nKindergarten: 'New Car.'"),
]


functions = [

# ---------- CONCEPT ----------
("Functions — Concept",
"""A function is a reusable block of code.
- Takes input (parameters).
- May return output.
- Promotes modularity (splitting programs into parts).""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

# ---------- SIMPLE FUNCTION ----------
("Functions — Simple Example",
"""Python:
    def hello():
        print("Hello!")

C:
    void hello() {
        printf("Hello!\\n");
    }

Rust:
    fn hello() {
        println!("Hello!");
    }

C++:
    void hello() {
        std::cout << "Hello!\\n";
    }""",
"H → Hello.\nKindergarten: 'Happy Hippo.'"),

# ---------- PARAMETERS ----------
("Functions — Parameters & Return",
"""Python:
    def add(a, b):
        return a + b

C:
    int add(int a, int b) {
        return a + b;
    }

Rust:
    fn add(a: i32, b: i32) -> i32 {
        a + b
    }

C++:
    int add(int a, int b) {
        return a + b;
    }""",
"A A → Add, Arguments.\nKindergarten: 'Apples Add.'"),

# ---------- VOID / NO RETURN ----------
("Functions — No Return (Void)",
"""Python:
    def log(msg):
        print(msg)   # no explicit return

C:
    void log(const char* msg) {
        printf("%s", msg);
    }

Rust:
    fn log(msg: &str) {
        println!("{}", msg);
    }

C++:
    void log(const std::string& msg) {
        std::cout << msg;
    }""",
"V N → Void, No-return.\nKindergarten: 'Very Nice.'"),

# ---------- SCOPE ----------
("Functions — Scope",
"""Local variables exist only inside a function.
Global variables exist everywhere.

Python:
    x = 10
    def f():
        y = 20  # local

C:
    int x = 10;
    void f() {
        int y = 20;  // local
    }

Rust:
    let x: i32 = 10;
    fn f() {
        let y: i32 = 20;
    }

C++:
    int x = 10;
    void f() {
        int y = 20;
    }""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- CONCEPT ----------
("Functions — Concept",
"""Definition: Reusable block of code.
- Takes inputs (parameters).
- May return outputs.
- Improves modularity.""",
"F P R → Function, Parameter, Return.\nKindergarten: 'Funny Pink Rabbit.'"),

# ---------- CALL SEMANTICS ----------
("Functions — Call Semantics",
"""Value = copy of argument.
Reference = direct access to original.

Python:
    def inc(lst): lst.append(1)

C:
    void inc(int *x){ (*x)++; }

Rust:
    fn inc(x: &mut i32){ *x += 1; }

C++:
    void inc(int &x){ x++; }""",
"V R → Value, Ref.\nKindergarten: 'Very Round.'"),

# ---------- PURE vs SIDE EFFECT ----------
("Functions — Purity vs Side Effect",
"""Pure = same input → same output, no external change.
Side effect = modifies outside state (I/O, globals).""",
"P S → Pure, Side.\nKindergarten: 'Pink Snake.'"),

# ---------- SCOPE ----------
("Functions — Scope",
"""Local = variable inside a function.
Global = variable accessible everywhere.""",
"S L G → Scope, Local, Global.\nKindergarten: 'Silly Little Goat.'"),

# ---------- RECURSION ----------
("Functions — Recursion",
"""Definition: Function calls itself.
Requires base case to stop recursion.

Python:
    def fact(n): return 1 if n<=1 else n*fact(n-1)""",
"R B → Recursion, Base.\nKindergarten: 'Red Balloon.'"),

# ---------- HIGHER ORDER ----------
("Functions — Higher Order",
"""Definition: Functions that take or return other functions.
Closure = captures surrounding variables.

Python:
    def make_adder(k): return lambda x:x+k""",
"H C → Higher, Closure.\nKindergarten: 'Happy Cat.'"),

# ---------- OVERLOAD / GENERIC ----------
("Functions — Overload & Generics",
"""Overload = same name, different parameters.
Generic = type-flexible functions.

C++:
    int add(int,int);
    double add(double,double);
    template<typename T>T maxT(T a,T b);""",
"O G → Overload, Generic.\nKindergarten: 'Orange Giraffe.'"),

# ---------- VARIADIC ----------
("Functions — Variadic",
"""Definition: Accept variable number of arguments.

Python:
    def total(*xs): return sum(xs)

C:
    int total(int n,...); // stdarg.h

C++:
    template<typename...T> int total(T...xs){ return (0+...+xs);}""",
"V A → Variadic Args.\nKindergarten: 'Very Antsy.'"),

# ---------- ERROR ----------
("Functions — Error Handling",
"""C = return codes.
C++ = exceptions.
Rust = Result/Option.
Python = exceptions.""",
"E R → Error, Result.\nKindergarten: 'Elephant Runs.'"),

# ---------- INLINE / LINK ----------
("Functions — Inline & Linkage",
"""Inline = hint to expand function at call site.
Linkage = scope across files (internal vs external).""",
"I L → Inline, Link.\nKindergarten: 'Ice Lion.'"),

# ---------- ASYNC ----------
("Functions — Asynchronous",
"""Sync = blocks until finished.
Async = returns immediately, completes later.

Python: async/await
Rust: async/await
C++20: coroutines
C: threads/event loops""",
"A S → Async, Sync.\nKindergarten: 'Ant Sings.'"),
]



# ---------------------- Topic registry ----------------------
# List your deck variable names here
topic_vars = [
    "programming_basics",
    "variables",
    "functions",
    "data_structures",

    # add more like: "flashcards_data_structures", "flashcards_algorithms"
]
flashcards_all = []
for name in topic_vars:
    deck = globals()[name]
    if not isinstance(deck, list):
        raise TypeError(f"Deck '{name}' must be a list of tuples.")
    flashcards_all.extend(deck)

# If user provided decks, prepend All. If not, fall back to a safe empty state.
if topic_vars:
    topic_vars = ["flashcards_all"] + topic_vars
    globals()["flashcards_all"] = flashcards_all
else:
    # Create a harmless empty deck and a single "how-to" message
    topic_vars = ["flashcards_all"]
    globals()["flashcards_all"] = []

# ---------------------- State ----------------------
current_deck_name = topic_vars[0]
flashcards = globals()[current_deck_name]
index = 0
showing_answer = False
output = widgets.Output()
output.layout = widgets.Layout(width='100%')

# ---------------------- Widgets ----------------------
topic_dropdown = widgets.Dropdown(
    options=topic_vars,
    value=current_deck_name,
    description="Topic:"
)
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
shuffle_button = widgets.Button(description="Shuffle 🔀")
jump_input = widgets.BoundedIntText(value=1, min=1, description='Jump to:')
jump_button = widgets.Button(description="Go ⏩")
restart_button.layout.display = 'none'
counter_label = widgets.Label()

# ---------------------- Render helpers ----------------------
def render_card(q_text: str, a_text: str = "", m_text: str = "") -> None:
    q_html = escape(q_text) if q_text else ""
    m_html = escape(m_text) if m_text else ""
    display(HTML(f"""
    <div style="background:#fafafa;padding:12px;border-radius:10px;
                border:1px solid #e5e5e5;margin-bottom:8px;
                font-family:sans-serif;font-size:14px;line-height:1.4">
      <b>Q:</b> {q_html}
    </div>
    """))
    if showing_answer and a_text is not None:
        a_html = escape(a_text)
        display(HTML(f"""
          <div style="font-family:sans-serif;font-size:14px;margin:6px 0">
            <b>A:</b>
            <pre style="
              background:#0b1020; color:#e6edf3;
              padding:10px; border-radius:10px;
              border:1px solid #1f2638; margin-top:6px;
              white-space: pre-wrap; word-break: break-word; overflow-wrap:anywhere;
            ">{a_html}</pre>
          </div>
        """))
        if m_text:
            display(HTML(f"""
              <div style="font-family:sans-serif;font-size:14px;margin:6px 0">
                <b>Mnemonic:</b> {m_html}
              </div>
            """))

def update_card():
    with output:
        clear_output(wait=True)
        if not flashcards:
            # No decks loaded
            info = (
                "No cards loaded.\n\n"
                "How to use:\n"
                "1) Define one or more deck variables, each as a list of tuples: (question, answer[, mnemonic]).\n"
                "2) Set topic_vars = [\"deck_name1\", \"deck_name2\", ...].\n"
                "3) Re-run this cell."
            )
            render_card("Welcome", info, "")
            # Disable navigation
            prev_button.disabled = True
            next_button.disabled = True
            show_button.disabled = True
            shuffle_button.disabled = True
            jump_input.disabled = True
            jump_button.disabled = True
            counter_label.value = "Topic: none - 0 cards"
            return

        # Deck present
        prev_button.disabled = (index == 0)
        next_button.disabled = (index >= len(flashcards) - 1)
        show_button.disabled = False
        shuffle_button.disabled = False
        jump_input.disabled = False
        jump_button.disabled = False

        card = flashcards[index]
        if len(card) == 3:
            question, answer, mnemonic = card
        elif len(card) == 2:
            question, answer = card
            mnemonic = ""
        else:
            question, answer, mnemonic = "Invalid card format", "", ""

        render_card(question, answer, mnemonic)
        counter_label.value = f"Topic: {current_deck_name} - Card {index+1} of {len(flashcards)}"
        show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"
        jump_input.max = max(1, len(flashcards))

# ---------------------- Handlers ----------------------
def on_topic_change(change):
    global current_deck_name, flashcards, index, showing_answer
    if change["name"] == "value" and change["new"] != change["old"]:
        current_deck_name = change["new"]
        flashcards = globals()[current_deck_name]
        index = 0
        showing_answer = False
        restart_button.layout.display = 'none'
        show_button.layout.display = ''
        next_button.layout.display = ''
        prev_button.layout.display = ''
        update_card()

def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You have reached the end of this topic.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

def shuffle_deck(_):
    global flashcards, index, showing_answer
    random.shuffle(flashcards)
    index = 0
    showing_answer = False
    update_card()

def jump_to_card(_):
    global index, showing_answer
    if not flashcards:
        return
    val = jump_input.value - 1
    if 0 <= val < len(flashcards):
        index = val
        showing_answer = False
        update_card()

# ---------------------- Wire up ----------------------
topic_dropdown.observe(on_topic_change, names="value")
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)
shuffle_button.on_click(shuffle_deck)
jump_button.on_click(jump_to_card)

# ---------------------- Launch ----------------------
update_card()
display(widgets.VBox([
    widgets.HBox([topic_dropdown]),
    widgets.HBox([prev_button, show_button, next_button, restart_button, shuffle_button]),
    widgets.HBox([counter_label]),
    widgets.HBox([jump_input, jump_button]),
    output
]))
````
