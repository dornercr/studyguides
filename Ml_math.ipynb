## 🔹 1. What Is Machine Learning?

### 🧠 Theory:

**Machine Learning (ML)** is the science of making computers **learn patterns from data** instead of being explicitly programmed.

It answers questions like:

- “Is this email spam?”  
- “What number is in this image?”  
- “How much will this house sell for?”

ML combines **math** and **code** to build models that improve through **experience (data)**.

---

## 🔹 2. The 6 Core Areas of Math Used in ML

We’ll learn and code each of these:

| # | Math Area         | Used For                                              |
|---|--------------------|--------------------------------------------------------|
| 1 | **Linear Algebra** | Data representation (vectors, matrices), transformations |
| 2 | **Calculus**       | Optimization (e.g., gradient descent)                  |
| 3 | **Statistics**     | Understanding data, noise, and distributions           |
| 4 | **Probability**    | Modeling uncertainty (Bayes, stochasticity)            |
| 5 | **Discrete Math**  | Logic, trees, graphs, combinatorics                    |
| 6 | **Optimization**   | Tuning parameters, loss minimization                   |

### ✅ Python Starter Code

```python
import numpy as np              # Linear algebra
import pandas as pd             # Data handling
import matplotlib.pyplot as plt # Visualization
from sklearn.linear_model import LinearRegression  # Simple ML model

# Simple dataset: y = 2x + 1
X = np.array([[1], [2], [3], [4], [5]])  # Features (2D)
y = np.array([3, 5, 7, 9, 11])           # Labels

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict
print("Prediction for x=6:", model.predict([[6]]))
```

## 🔹 4. First Concept: Linear Algebra — The Language of ML

### 🧠 Why It Matters

- Images → matrices of pixels  
- Datasets → rows = samples, columns = features  
- We use **matrix operations** to make predictions, update weights, etc.

---

### 💻 Code Example: Vector and Matrix Operations

```python
import numpy as np

# A vector (1D)
v = np.array([1, 2, 3])

# A matrix (2D)
M = np.array([[1, 2],
              [3, 4],
              [5, 6]])

# Matrix-vector multiplication
result = M @ v[:2]  # Dot product with first 2 elements
print("Matrix-Vector Product:", result)
```

````python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

# (question, answer, PEMDAS-style mnemonic)

flashcards = [

# 1.1 What is ML

("What is machine learning?",
"🧠 Theory:\n"
"Machine Learning is the science of learning patterns from data instead of hardcoding rules.\n\n"
"🗣️ Plain English:\n"
"Instead of writing instructions, you let the computer learn from examples.\"\n\n"
"💡 Mnemonic:\n"
"ML = Make it Learn (from data).\""),

("How is ML different from traditional programming?",
"🧠 Theory:\n"
"Traditional programming requires explicit rules. ML lets the computer learn those rules from data.\n\n"
"🗣️ Plain English:\n"
"Programming: If spam → delete. ML: Show it lots of spam and let it figure that out.\"\n\n"
"💡 Mnemonic:\n"
"Code = rules → output; ML = data + output → rules\""),

# 1.2 6 Core Math Areas

("What are the 6 types of math used in ML?",
"🧠 Theory:\n"
"1. Linear Algebra\n"
"2. Calculus\n"
"3. Statistics\n"
"4. Probability\n"
"5. Discrete Math\n"
"6. Optimization\n\n"
"💡 Mnemonic:\n"
"Learn Calculus So Predictions Don’t Overfit\""),

("What is Linear Algebra used for in ML?",
"🧠 Theory:\n"
"For representing data (vectors, matrices), transforming it, and feeding it into models.\n\n"
"🗣️ Plain English:\n"
"It’s how we store and manipulate big tables of data.\"\n\n"
"💡 Mnemonic:\n"
"LA = Layout + Arithmetic\""),

("What is Calculus used for in ML?",
"🧠 Theory:\n"
"For optimization — to minimize loss and update model parameters using gradients.\n\n"
"🗣️ Plain English:\n"
"Helps us climb the hill or roll down the cost curve.\"\n\n"
"💡 Mnemonic:\n"
"Calc = Climb Accuracy, Lower Cost\""),

("What is Statistics used for in ML?",
"🧠 Theory:\n"
"To understand data distributions, central tendency, and variance.\n\n"
"🗣️ Plain English:\n"
"It helps you know what’s normal and what’s weird.\"\n\n"
"💡 Mnemonic:\n"
"Stats See Spread\""),

("What is Probability used for?",
"🧠 Theory:\n"
"For modeling uncertainty and making probabilistic predictions.\n\n"
"🗣️ Plain English:\n"
"It tells the model how confident it should be.\"\n\n"
"💡 Mnemonic:\n"
"Probability Predicts Possibilities\""),

("What is Discrete Math used for?",
"🧠 Theory:\n"
"To understand logic, graphs, trees, sets, and combinatorics.\n\n"
"🗣️ Plain English:\n"
"It’s the math behind decision trees and logical flow.\"\n\n"
"💡 Mnemonic:\n"
"Discrete = Decisions, Graphs, Rules\""),

("What is Optimization in ML?",
"🧠 Theory:\n"
"The process of adjusting model parameters to minimize error.\n\n"
"🗣️ Plain English:\n"
"It’s how we tweak the knobs to make better predictions.\"\n\n"
"💡 Mnemonic:\n"
"Optimize = Oops → Better\""),

# 1.3 Python Setup

("How do you set up Python for ML?",
"✅ Code:\n"
"    pip install numpy pandas matplotlib scikit-learn\n\n"
"🗣️ Plain English:\n"
"You’re installing the core math and ML tools.\""),

("What does this starter code do?",
"✅ Code:\n"
"    import numpy as np\n"
"    from sklearn.linear_model import LinearRegression\n\n"
"    X = np.array([[1], [2], [3]])\n"
"    y = np.array([3, 5, 7])\n"
"    model = LinearRegression()\n"
"    model.fit(X, y)\n"
"    print(model.predict([[4]]))\n\n"
"🖨️ Sample Output:\n"
"    [9.]\n\n"
"🧠 Theory:\n"
"It trains a model to learn y = 2x + 1 and predicts y for x = 4.\"\n\n"
"💡 Mnemonic:\n"
"Fit learns. Predict guesses.\""),

# 1.4 Linear Algebra Intro

("What is a vector?",
"🧠 Theory:\n"
"A 1D array of numbers — represents direction and magnitude.\n\n"
"✅ Code:\n"
"    v = np.array([1, 2, 3])\n\n"
"🗣️ Plain English:\n"
"Just a list of numbers like an arrow.\"\n\n"
"💡 Mnemonic:\n"
"Vector = Very Direct\""),

("What is a matrix?",
"🧠 Theory:\n"
"A 2D array of numbers — like a spreadsheet with rows and columns.\n\n"
"✅ Code:\n"
"    M = np.array([[1, 2], [3, 4]])\n\n"
"🗣️ Plain English:\n"
"A table of numbers.\""),

("How do you multiply matrix and vector?",
"🧠 Theory:\n"
"Use the dot product to combine matrix rows and vector values.\n\n"
"✅ Code:\n"
"    M = np.array([[1, 2], [3, 4]])\n"
"    v = np.array([1, 2])\n"
"    print(M @ v)\n\n"
"🖨️ Output:\n"
"    [ 5 11 ]\n\n"
"🗣️ Plain English:\n"
"You're combining a table with a list of weights.\"\n\n"
"💡 Mnemonic:\n"
"@ = apply vector to matrix\""),

("What does the `@` operator do?",
"🧠 Theory:\n"
"Performs matrix multiplication (dot product).\n\n"
"🗣️ Plain English:\n"
"It applies inputs to weights — like feeding data through the model.\"\n\n"
"💡 Mnemonic:\n"
"@ = Action Through Weights\""),
]
index = 0
showing_answer = False
output = widgets.Output()

# Buttons
prev_button = widgets.Button(description="◀️ Previous")
show_button = widgets.Button(description="Show Answer 👁️")
next_button = widgets.Button(description="Next ▶️")
restart_button = widgets.Button(description="🔁 Restart")
restart_button.layout.display = 'none'

def update_card():
    with output:
        clear_output(wait=True)
        card = flashcards[index]
        if len(card) == 2:
            question, answer = card
            mnemonic = ""
        else:
            question, answer, mnemonic = card

        display(Markdown(f"**Q:** {question}"))
        if showing_answer:
            display(Markdown(f"\n---\n\n**A:**\n{answer}"))
            if mnemonic.strip():
                display(Markdown(mnemonic))

    prev_button.disabled = index == 0
    next_button.disabled = False
    show_button.description = "Hide Answer 🙈" if showing_answer else "Show Answer 👁️"

def show_toggle(_):
    global showing_answer
    showing_answer = not showing_answer
    update_card()

def next_card(_):
    global index, showing_answer
    if index < len(flashcards) - 1:
        index += 1
        showing_answer = False
        update_card()
    else:
        with output:
            clear_output()
            display(Markdown("🎉 **You've reached the end of the flashcards.**"))
        show_button.layout.display = 'none'
        next_button.layout.display = 'none'
        prev_button.layout.display = 'none'
        restart_button.layout.display = ''

def prev_card(_):
    global index, showing_answer
    if index > 0:
        index -= 1
        showing_answer = False
        update_card()

def restart(_):
    global index, showing_answer
    index = 0
    showing_answer = False
    show_button.layout.display = ''
    next_button.layout.display = ''
    prev_button.layout.display = ''
    restart_button.layout.display = 'none'
    update_card()

# Hook up events
show_button.on_click(show_toggle)
next_button.on_click(next_card)
prev_button.on_click(prev_card)
restart_button.on_click(restart)

# Display the first flashcard
update_card()
display(widgets.HBox([prev_button, show_button, next_button, restart_button]), output)
````
