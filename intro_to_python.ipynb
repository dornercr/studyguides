The program runs but gives the **wrong answer**.

# 2.6 Computers and Programs (General) – Study Guide

This guide explains how computers work "under the hood" using 9th-grade language and technical terms.

---

## 💡 Switches and Bits

- Early computers used **switches** to control electricity.
- A switch can be **on (1)** or **off (0)** — these are called **bits** (binary digits).
- Engineers built circuits of switches to do math like addition and multiplication.

```text
Bit = 0 or 1 (like a light switch)
```

---

## 🖥️ Early Computers

- 1930s–40s computers used **10,000+ switches** and filled entire rooms.
- They could do **thousands of calculations per second**.
- Used for military tasks like ballistic trajectory tables.

---

## 🧠 Processors and Memory

- A **processor** runs instructions (math steps).
- A **memory** stores those instructions and numbers.
- Originally built from switches — now billions fit on a tiny chip.

### Analogy:
Processor = Chef  
Instructions = Recipe  
Memory = Counter with ingredients

---

## 📦 What Memory Does

- Memory stores:
  - **Instructions** (what to do)
  - **Data** (the numbers used or created)

- Each memory location has an **address** (like a mailbox).

---

## 🔁 How Computers Work

Example:
```text
20 (wage) → ×40 → ×52 → Output = 41600
```

- Processor follows steps one at a time.
- Instructions:
  - Multiply wage by hours/week
  - Multiply result by weeks/year
  - Show final result

---

## 🔣 Sample Instructions

| Command | What It Does |
|--------|----------------|
| Add X, #num, Y | Y = X + num |
| Sub X, #num, Y | Y = X - num |
| Mul X, #num, Y | Y = X × num |
| Div X, #num, Y | Y = X ÷ num |
| Jmp Z | Jump to instruction at location Z |

Example:
```text
Mul 97, #9, 98  # Multiplies value in memory 97 by 9, stores result in 98
```

---

## 🧮 Stored as 0s and 1s

Instructions and data are saved in memory as **binary**:

```text
011 1100001 001001 1100010
```

- This means:
  - Instruction type (Mul)
  - Inputs/outputs as memory locations

---

## ⚙️ How It Executes

- Starts at memory location 0
- Executes each instruction one at a time
- Uses memory to store results
- Loops with Jmp 0 to start over

# 2.7 Computer Tour – Study Guide

This guide explains computer parts using 9th-grade-friendly language and technical accuracy.

---

## 💻 What is a Computer?

The meaning of "computer" has changed:
- **Old meaning**: A person doing math by hand
- **1940s–1950s**: Big room-sized machines
- **1970s–1980s**: Personal Computers (PCs)
- **2000s–today**: Includes phones, tablets, smart devices, and even cars or appliances

---

## 🧱 Hardware vs. Software

- **Hardware** = Physical parts of the computer
- **Software** = Programs that run on the hardware

---

## 🧩 Main Computer Components

### 1. Input/Output (I/O) Devices
- Used to talk **to** and **from** the computer
- Examples: screen, keyboard, mouse, mic, printer, touchscreen, USB
- Also called **peripherals**

### 2. Storage (Disk)
- Long-term memory (even when power is off)
- Hard drive: uses magnets to store 0s and 1s
- Flash drive: stores data using electric circuits
- **Non-volatile** (doesn’t erase when powered off)
- Sizes: **Terabytes (TB)**

### 3. Memory (RAM)
- Short-term, fast memory
- Loses everything when powered off = **volatile**
- Faster than storage, but smaller
- Sizes: **Gigabytes (GB)**
- Used to hold data and instructions temporarily

### 4. Processor (CPU)
- Runs programs and performs calculations
- Starts with BIOS → loads OS → runs user programs
- Has small built-in memory = **cache**
- Sometimes called **microprocessor**

### 5. Clock
- Controls how fast the processor runs
- Speed is measured in **MHz** or **GHz**
- 1 GHz = 1 billion instructions per second

---

## 🧠 Multitasking

Computers can run multiple programs at once (like web + music + games).  
The **Operating System (OS)** switches between programs super fast (thousands of times/second).

---

## 📊 Quiz Practice

- A disk stores terabytes? → ✅ True  
- RAM is volatile and fast? → ✅ True  
- Cache is the fastest memory and is on the processor? → ✅ True  
- Clock controls processor speed? → ✅ True  
- OS manages programs and peripherals? → ✅ True  

---

## 🧬 Moore's Law

Moore’s Law = **IC capacity doubles about every 18 months**

- Smaller **transistors** → faster, cheaper computers
- 1971: Intel 4004 had 2,300 transistors  
- Today: Billions of transistors per chip (multiple cores)

---

## 💡 Side Tip

Adding **more RAM** makes your PC faster because:
- RAM is closer to processor than storage
- Like keeping ingredients close to the stove

---

## 🧠 Key Terms

| Term | Meaning |
|------|---------|
| Operating System | Manages programs, talks to devices |
| Disk | Non-volatile, slow storage |
| RAM | Volatile, fast memory |
| Cache | Very fast, tiny memory on CPU |
| Clock | Controls instruction speed (GHz) |
| Moore’s Law | ICs double in power ~every 18 months |

- Bad formatting makes your program harder to understand

# 2.10 Python Example: Salary Calculation – Study Guide

## 🧮 Example Program: Salary Calculation

This program assumes:
- 40 work hours per week
- 50 work weeks per year
- A given hourly wage

### ✅ Python Code

```python
hourly_wage = 20

print('Annual salary is:')
print(hourly_wage * 40 * 50)  # 20 * 40 * 50 = 40000
print()

print('Monthly salary is:')
print((hourly_wage * 40 * 50) / 12)  # 40000 / 12 = 3333.333...
print()
```

## 🛠️ Common Mistake (FIXME)

If you see something like:

You’ll get the same annual salary again, NOT monthly.
