### What is code?

**1. Core Definition**
Code is a set of precise, human-readable instructions written in a programming language (a formal language designed to communicate instructions to a machine) to perform a specific task or solve a problem.

**2. Why It Matters**
Understanding that code is an abstraction layer between human logic and machine hardware is fundamental. Exams test this by asking about the relationship between source code and machine execution.

**3. Core Concepts**

* **Source Code:** The text written by the programmer.
* **Syntax:** The set of rules defining the combination of symbols that are considered correctly structured programs.
* **Semantics:** The meaning of the instructions (what the code actually does).
* **Abstraction:** Hiding complex background details so the programmer can focus on logic.

**4. Formulas / Rules**

* **Rule of Precision:** Computers do exactly what is written, not what is intended.
* **Syntax Rule:** If syntax is incorrect, the program will fail to run (SyntaxError).

**5. Step-by-Step Process**

1. Problem identification.
2. Algorithm design (logic).
3. Coding (writing syntax).
4. Translation (converting code to machine-readable format).
5. Execution.

**6. Common Exam Traps**

* **Trap:** Thinking "Code" and "Software" are exactly the same. **Fact:** Code is the component; software is the finished product including assets and documentation.
* **Trap:** Assuming computers "understand" English-like commands. **Fact:** They only execute binary; code must be translated.

**7. Worked Example**

* **Scenario:** A student writes `print("Hello"` (missing a parenthesis).
* **Result:** The computer cannot translate this because it violates the syntax rules of Python. It is "broken code."

**8. Memory Hooks**

* **S.A.L.T.:** **S**yntax, **A**bstraction, **L**ogic, **T**ranslation.

**9. Active Recall Questions**

1. What is the difference between syntax and semantics?
2. Why is code considered an "abstraction"?
3. True or False: Computers can interpret vague instructions if the logic is mostly correct.
**Answers: 1. Syntax is the grammar/rules; Semantics is the meaning/logic. 2. It hides the complexity of binary/hardware operations. 3. False.**

### Compiled vs interpreted languages

**1. Core Definition**

* **Compiled Languages:** Languages where the entire source code is converted into machine code (binary) by a compiler (a translation program) before execution.
* **Interpreted Languages:** Languages where an interpreter reads and executes the code line-by-line at runtime.

**2. Why It Matters**
High-yield topic. Exams frequently compare performance (speed) versus portability (the ability to run on different operating systems).

**3. Core Concepts**

* **Compilation:** Happens once; produces an executable file (e.g., .exe).
* **Interpretation:** Happens every time the program runs.
* **Portability:** Interpreted languages are usually more portable because the interpreter handles the OS-specific details.
* **Speed:** Compiled languages are generally faster because translation is finished before the program starts.

**4. Formulas / Rules**

* **Compiled:** Source Code -> Compiler -> Machine Code -> Execution.
* **Interpreted:** Source Code -> Interpreter -> Execution (Line by Line).

**5. Step-by-Step Process**

1. Write Source Code.
2. If Compiled: Run Compiler -> Get Binary -> Run Binary.
3. If Interpreted: Run Interpreter on Source Code directly.

**6. Common Exam Traps**

* **Trap:** Saying Python is "only" interpreted. **Fact:** Modern Python uses a hybrid approach (compiles to bytecode first).
* **Trap:** Compiled code is always better. **Fact:** Interpreted code is often easier to debug and develop quickly.

**7. Worked Example**

* **C++ (Compiled):** You write code, wait 2 minutes for it to compile, then it runs in 1 second.
* **Python (Interpreted):** You write code, it starts running instantly, but takes 5 seconds to finish.

**8. Memory Hooks**

* **Compiled = Cake:** You bake it entirely (compile) before you eat it (run).
* **Interpreted = Buffet:** You pick up and eat one item at a time (line-by-line).

**9. Active Recall Questions**

1. Which type of language produces a standalone executable file?
2. Why are interpreted languages often called "portable"?
3. Which method identifies syntax errors across the whole file before any code runs?
**Answers: 1. Compiled. 2. Because the same source code can run on any system that has the appropriate interpreter installed. 3. Compilation.**

### How Python executes code (CPython, bytecode, VM)

**1. Core Definition**
Python uses a hybrid model where source code is compiled into **Bytecode** (a low-level, platform-independent representation) and then executed by the **Python Virtual Machine (PVM)** (the runtime engine).

**2. Why It Matters**
Crucial for understanding Python's internal architecture. Exams test the specific order of execution and the roles of the PVM and CPython.

**3. Core Concepts**

* **CPython:** The default and most widely used implementation of Python (written in C).
* **Bytecode:** The intermediate stage between source code and machine code (stored in `.pyc` files).
* **PVM:** The "engine" that iterates through the bytecode and executes the instructions.

**4. Formulas / Rules**

* **The Python Pipeline:** Source (.py) -> Compiler -> Bytecode (.pyc) -> PVM -> Machine Output.

**5. Step-by-Step Process**

1. Programmer writes `.py` file.
2. The CPython compiler checks syntax.
3. CPython translates source to Bytecode.
4. Bytecode is shipped to the PVM.
5. PVM translates Bytecode to machine-specific instructions.

**6. Common Exam Traps**

* **Trap:** Thinking Bytecode is Machine Code. **Fact:** Machine code is for the CPU; Bytecode is for the PVM.
* **Trap:** Thinking the PVM is a physical piece of hardware. **Fact:** It is software.

**7. Worked Example**

* When you run `hello.py`, Python creates a `__pycache__` folder. Inside is a `.pyc` file. This file contains the **Bytecode** that allows Python to skip the compilation step if the source hasn't changed.

**8. Memory Hooks**

* **S-B-V:** **S**ource -> **B**ytecode -> **V**irtual Machine.

**9. Active Recall Questions**

1. What is the standard implementation of Python called?
2. Where is Python bytecode stored on the disk?
3. What is the role of the PVM?
**Answers: 1. CPython. 2. In the `__pycache__` folder as `.pyc` files. 3. To execute the bytecode instructions on the local machine.**

### REPL vs script execution

**1. Core Definition**

* **REPL (Read-Eval-Print Loop):** An interactive shell where you type one line of code, and it executes immediately.
* **Script Execution:** Writing code in a text file (`.py`) and running the entire file at once.

**2. Why It Matters**
Exams test when to use which. REPL is for testing; Scripts are for production/reusability.

**3. Core Concepts**

* **Interactive Mode:** Triggered by typing `python` in the terminal.
* **Prompt:** Usually indicated by `>>>`.
* **Volatility:** REPL code is lost once the session ends.
* **Persistence:** Script code is saved on the hard drive.

**4. Formulas / Rules**

* **REPL:** Type -> Enter -> Result -> Repeat.
* **Script:** Write -> Save -> Run `python filename.py` -> Result.

**5. Step-by-Step Process**

* **To use REPL:** Open Terminal -> Type `python` -> Type `print("Hi")`.
* **To use Script:** Open Editor -> Write `print("Hi")` -> Save as `test.py` -> Type `python test.py` in terminal.

**6. Common Exam Traps**

* **Trap:** Thinking you can save a REPL session as a file. **Fact:** You must manually copy code into a script to save it.
* **Trap:** Confusing the `>>>` prompt with the system command prompt (`$` or `C:\>`).

**7. Worked Example**

* **REPL Use Case:** Testing if `5 + 10` equals `15` quickly.
* **Script Use Case:** Creating a program that calculates payroll for 500 employees every month.

**8. Memory Hooks**

* **REPL = Conversation:** Quick back-and-forth.
* **Script = Letter:** You write the whole thing, mail it, and they read it all at once.

**9. Active Recall Questions**

1. What does the acronym REPL stand for?
2. Which mode is better for long-term project development?
3. What symbol indicates you are in the Python interactive shell?
**Answers: 1. Read-Eval-Print Loop. 2. Script execution. 3. `>>>`.**

### Running Python from terminal

**1. Core Definition**
Using the Command Line Interface (CLI) to invoke the Python interpreter to run a file or enter interactive mode.

Fundamental skill for automation and server-side work. Exams test command structure and arguments.

**7. Worked Example**

* File: `app.py`
* Terminal Command: `python3 app.py`
* Result: The PVM executes the bytecode of `app.py` and prints output to the terminal.

### File extensions and shebang

**1. Core Definition**

* **File Extension:** The `.py` suffix that identifies a file as a Python source file.
* **Shebang:** A character sequence (`#!`) at the very beginning of a script that tells a Unix-like OS (Linux/macOS) which interpreter to use.

**2. Why It Matters**
Important for cross-platform development and script automation.

**3. Core Concepts**

* **.py:** Standard source file.
* **.pyc:** Compiled bytecode file.
* **Shebang Location:** Must be the very first line (Line 1).
* **Portability:** Use `#!/usr/bin/env python3` for better portability.

**4. Formulas / Rules**

* **Standard Shebang:** `#!/usr/bin/python3` (Hardcoded path).
* **Recommended Shebang:** `#!/usr/bin/env python3` (Searches PATH).

**5. Step-by-Step Process**

1. Create a text file.
2. Add `#!/usr/bin/env python3` to the first line.
3. Write Python code below.
4. Save with `.py` extension.
5. (On Linux/macOS) Make executable: `chmod +x file.py`.

**6. Common Exam Traps**

* **Trap:** Using a Shebang on Windows. **Fact:** Windows ignores Shebangs (it uses the file extension association).
* **Trap:** Putting a space before the `#!`. **Fact:** It must be the first two characters of the file.

**7. Worked Example**

* A script starts with `#! /usr/bin/python3`. This tells a Linux server: "Don't run this as a shell script; find the Python 3 interpreter and let it handle this file."

**8. Memory Hooks**

* **Shebang = Sharp-Bang:** `#` (Hash/Sharp) and `!` (Bang).

**9. Active Recall Questions**

1. What is the file extension for Python bytecode?
2. On which operating systems is the Shebang relevant?
3. Write the most portable version of a Python 3 shebang.
**Answers: 1. `.pyc`. 2. Unix-like systems (Linux, macOS). 3. `#!/usr/bin/env python3`.**

### What is a variable?

**1. Core Definition**
A variable is a **named identifier** (a label) that acts as a reference to a value stored in the computer's memory. In Python, variables are not "containers" that hold values, but rather **tags** or **labels** that point to objects.

**2. Why It Matters**
Variables allow programmers to store, retrieve, and manipulate data by name instead of memory addresses. Exams test the understanding of variables as references rather than buckets.

**3. Core Concepts**

* **Assignment:** The process of using the `=` operator to link a name to a value.
* **Object-Oriented:** In Python, everything (numbers, strings, functions) is an object.
* **Reference:** A variable "points" to the memory location of an object.

**4. Formulas / Rules**

* **Assignment Rule:** `variable_name = value`. The right side is evaluated first, then assigned to the left.
* **Reassignment:** You can point an existing variable name to a completely new object at any time.

**5. Step-by-Step Process**

1. Python evaluates the expression on the right of the `=`.
2. Python creates an object in memory for that value.
3. Python creates (or updates) a name and points it to that object.

**6. Common Exam Traps**

* **Trap:** Thinking `=` means mathematical equality. **Fact:** `=` is the assignment operator; `==` is for equality.
* **Trap:** Thinking a variable has a fixed "type." **Fact:** In Python, the *object* has a type, the *variable* is just a pointer.

**7. Worked Example**

```python
x = 5
y = x
x = 10
print(y) # Output is 5, because y points to the object '5', not to the name 'x'.

```

**8. Memory Hooks**

* **Variable = Luggage Tag:** The tag (variable) can be moved from one suitcase (object) to another.

**9. Active Recall Questions**

1. Does a Python variable hold a value or point to a value?
2. In the statement `a = 10 + 5`, which side is evaluated first?
3. What happens to the old value if a variable is reassigned?
**Answers: 1. Points to a value (reference). 2. The right side (`10 + 5`). 3. It is eventually removed from memory by the Garbage Collector if no other variables point to it.**

### Naming rules

**1. Core Definition**
The set of syntax requirements and stylistic conventions (PEP 8) that dictate how variable names must be formed to be valid and readable.

**2. Why It Matters**
Exams frequently include "Which of these is a valid variable name?" questions. Mixing up rules and conventions is a high-yield trap.

**3. Core Concepts**

* **Case Sensitivity:** `myVar` and `myvar` are different variables.
* **Characters:** Must start with a letter (a-z, A-Z) or an underscore (`_`).
* **Numbers:** Can contain numbers but **cannot** start with one.
* **Reserved Words:** Cannot use keywords like `if`, `else`, `while`, `True`, etc.
* **Snake Case:** Python convention (PEP 8) uses `all_lowercase_with_underscores`.

**4. Formulas / Rules**

* **Legal:** `_score`, `total_2`, `UserAddress`.
* **Illegal:** `2_total` (starts with number), `user-name` (hyphens not allowed), `class` (reserved word).

**5. Common Exam Traps**

* **Trap:** `True = 10` (Illegal: `True` is a keyword).
* **Trap:** `my-var = 5` (Illegal: Python sees the `-` as subtraction).
* **Trap:** `Var` vs `var` (Valid, but risky as they are separate identifiers).

**6. Memory Hooks**

* **L.U.N.:** **L**etters, **U**nderscores, NO **N**umbers (at the start).

**7. Active Recall Questions**

1. Is `3_count` a valid variable name?
2. Is `total_count` and `Total_Count` the same variable in Python?
3. Can you name a variable `import`?
**Answers: 1. No (cannot start with a number). 2. No (Python is case-sensitive). 3. No (`import` is a reserved keyword).**

### Dynamic typing

**1. Core Definition**
A feature of Python where the type of a variable is determined at **runtime** (during execution) based on the object it currently points to, rather than being explicitly declared by the programmer.

**2. Why It Matters**
Contrasts with Static Typing (like C++ or Java). Exams test the flexibility and risks of changing a variable's type mid-script.

**3. Core Concepts**

* **No Declaration:** You don't write `int x;`. You just write `x = 5`.
* **Type Flexibility:** A variable can point to an `int`, then later point to a `str`.
* **Runtime Checking:** Errors related to types (e.g., adding a string to an integer) happen when the code runs, not before.

**4. Formulas / Rules**

* **The Assignment Rule:** The type of the variable is the type of the object it currently references.

**5. Common Exam Traps**

* **Trap:** Thinking Python is "untyped." **Fact:** Python is **Strongly Typed** (it cares about types) but **Dynamically Typed** (it checks them at runtime).
* **Trap:** Assuming `x = 5` means `x` will always be an integer.

**6. Worked Example**

```python
data = 10      # data is currently an int
data = "Ten"   # data is now a str (this is perfectly legal in Python)

```

**7. Memory Hooks**

* **Dynamic = Dancer:** A dancer can change costumes (types) while the show (program) is running.

**8. Active Recall Questions**

1. In dynamic typing, when is the type of a variable determined?
2. Does the programmer need to declare the type of a variable in Python?
3. If `x = 5`, can you later set `x = "Hello"`?
**Answers: 1. At runtime. 2. No. 3. Yes.**

### Memory model basics

**1. Core Definition**
Python manages memory through a private **heap** space where all objects and data structures reside. It uses **References** to link variable names to these objects and **Garbage Collection** to clean up unused memory.

**2. Why It Matters**
Understanding memory helps explain why certain objects change (mutability) and how the "identity" of a variable works.

**3. Core Concepts**

* **The Heap:** The pool of memory where Python stores objects.
* **Reference Counting:** Python keeps track of how many variables point to an object.
* **Garbage Collection (GC):** When an object's reference count hits zero, Python automatically deletes the object to free memory.
* **ID:** Every object has a unique integer ID (essentially its memory address).

**4. Formulas / Rules**

* **Reference Increase:** `x = [1, 2]`; `y = x` (Reference count for the list object increases to 2).
* **Reference Decrease:** `x = 5` (Reference count for the list object decreases).

**5. Common Exam Traps**

* **Trap:** Thinking variables "contain" the data. **Fact:** Variables are just pointers.
* **Trap:** Manual memory management. **Fact:** Python handles memory automatically; programmers rarely use `del` to free memory.

**6. Memory Hooks**

* **The Balloon Hook:** The **object** is the balloon; the **variable** is the string. If all strings are cut, the balloon floats away (Garbage Collection).

**7. Active Recall Questions**

1. Where are objects stored in Python's memory?
2. What happens when an object's reference count reaches zero?
3. What function returns the unique memory address of an object?
**Answers: 1. The private heap. 2. It is deleted by the Garbage Collector. 3. `id()`.**

### Core Data Types (int, float, bool, str, NoneType)

**1. Core Definition**
These are the basic building blocks (primitive-like objects) used to represent different kinds of information in Python.

**2. Why It Matters**
The foundation of all data manipulation. Exams test limits, precision, and literal syntax.

**3. Core Concepts**

* **int:** Whole numbers (positive, negative, or zero). Python ints have "arbitrary precision" (can be as large as your RAM).
* **float:** Numbers with a decimal point (floating-point representation).
* **bool:** Logical values `True` or `False`. (Note: Subtype of `int`; `True == 1`, `False == 0`).
* **str:** A sequence of Unicode characters enclosed in quotes (`'`, `"`, `'''`).
* **NoneType:** A special type with only one value: `None`. Represents the absence of a value or a null pointer.

**4. Formulas / Rules**

### Immutability vs mutability

**1. Core Definition**

* **Immutable:** An object whose state/value cannot be changed after it is created (e.g., `int`, `float`, `str`, `bool`, `tuple`).
* **Mutable:** An object that can be modified "in-place" without creating a new object (e.g., `list`, `dict`, `set`).

**2. Why It Matters**
High-yield topic. Affects how data is shared between variables and how functions modify arguments.

**3. Core Concepts**

* **In-place change:** If you change a list, the memory ID stays the same.
* **New object:** If you "change" a string, Python actually creates a brand new string and points the variable to it.

**4. Formulas / Rules**

* **Immutable types:** `int`, `float`, `bool`, `str`, `tuple`, `frozenset`.
* **Mutable types:** `list`, `dict`, `set`, `bytearray`.

* **Trap:** Modifying a list that multiple variables point to.

**6. Worked Example**

```python
a = [1, 2]
b = a
a.append(3)
print(b) # Output: [1, 2, 3] (Because lists are mutable)

x = 10
y = x
x = x + 1
print(y) # Output: 10 (Because ints are immutable)

```

### Identity vs equality (is vs ==)

**1. Core Definition**

* **Equality (`==`):** Checks if two objects have the same **value/content**.
* **Identity (`is`):** Checks if two variables point to the **exact same object** in memory (same ID).

**2. Why It Matters**
A classic exam favorite. Students often confuse "looking the same" with "being the same."

**3. Core Concepts**

* `a == b` asks: "Do they look alike?"
* `a is b` asks: "Are they the same physical object?"
* **Interning:** For small integers ( -5 to 256) and some strings, Python reuses the same memory object, which can make `is` return `True` unexpectedly.

**4. Formulas / Rules**

* `is` check: `id(a) == id(b)`
* `==` check: `a.__eq__(b)`

**5. Common Exam Traps**

* **Trap:** Using `is` to compare numbers or strings. **Rule:** Always use `==` for values; use `is` only for `None` or singleton checks.
* **Trap:** `list1 = [1, 2]; list2 = [1, 2]`. `list1 == list2` is `True`, but `list1 is list2` is `False`.

**6. Worked Example**

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b) # True (Same values)
print(a is b) # False (Different objects in memory)
print(a is c) # True (c points to the same object as a)

```

**7. Memory Hooks**

* **`==` = Identical Twins:** They look the same, but they are two different people.
* **`is` = Two names for the same person:** (e.g., "Clark Kent" is "Superman").

**8. Active Recall Questions**

3. In the same example, what is the result of `x is y`?
**Answers: 1. `is`. 2. `True`. 3. `False`.**

### Arithmetic operators

**1. Core Definition**
Symbols used to perform common mathematical operations on numeric values (integers and floats).

**2. Why It Matters**
Fundamental for data manipulation. Exams test the distinction between classic division (`/`), floor division (`//`), and the modulo operator (`%`).

**3. Core Concepts**

* **Addition (`+`):** Sum of two values.
* **Subtraction (`-`):** Difference between two values.
* **Multiplication (`*`):** Product of two values.
* **Division (`/`):** Always returns a **float** (e.g., `10 / 2` is `5.0`).
* **Floor Division (`//`):** Returns the largest integer less than or equal to the result (truncates the decimal).
* **Modulo (`%`):** Returns the **remainder** after division.
* **Exponentiation (`**`):** Raises the first number to the power of the second.

**4. Formulas / Rules**

* `a % b` results in a value between `0` and `b-1`.
* `a // b` is equivalent to `floor(a / b)`.

**5. Common Exam Traps**

* **Trap:** Forgetting that `/` always produces a float, even with whole numbers.

**6. Worked Example**

```python
x = 17 % 3   # 17 / 3 = 5 with a remainder of 2. x = 2.
y = 17 // 3  # 17 / 3 = 5.66... y = 5.
z = 2 ** 3   # 2 * 2 * 2. z = 8.

```

**7. Memory Hooks**

* **Modulo = Remainder:** Think of it as "What's left over?"
* **Floor = Bottom:** Floor division always drops to the "floor" (the lower integer).

### Short-circuit evaluation

**1. Core Definition**
A behavior in Python where the second operand of a logical expression is only evaluated if the first operand is not sufficient to determine the final result.

**2. Why It Matters**
Used for safety (checking if an object is not None before accessing an attribute) and performance.

**3. Core Concepts**

* **AND Short-circuit:** If the first value is `False`, the whole expression is `False`. Python **stops** and doesn't look at the second value.
* **OR Short-circuit:** If the first value is `True`, the whole expression is `True`. Python **stops**.

**4. Formulas / Rules**

* `False and [Anything]` -> returns `False` immediately.
* `True or [Anything]` -> returns `True` immediately.

**5. Common Exam Traps**

* **Trap:** Code that would normally crash (like `1/0`) won't crash if it's on the right side of a short-circuited expression.

**6. Worked Example**

```python
x = 0
if x != 0 and (10 / x) > 1:
    print("This won't crash!")
# Because x != 0 is False, Python stops. It never tries to divide by zero.

```

### if, elif, else

**1. Core Definition**
Conditional statements used to execute different blocks of code based on whether a specific condition (a Boolean expression) evaluates to `True` or `False`.

**2. Why It Matters**
This is the primary way programs make decisions. Exams test "exclusive" logic—knowing that once one block in an `if/elif/else` chain runs, the others are skipped.

**3. Core Concepts**

* **if:** The starting condition.
* **elif (else if):** Checked only if the previous `if` or `elif` was `False`. You can have multiple `elif` blocks.
* **else:** The "catch-all" block. It runs only if all previous conditions in the chain were `False`.
* **Indentation:** Python uses 4 spaces (standard) to define which code belongs to which block.

**4. Formulas / Rules**

* **The Chain Rule:** Only **one** block in a single `if/elif/else` structure can execute.
* **The Optional Rule:** `elif` and `else` are optional. An `if` can stand alone.

**5. Step-by-Step Process**

1. Evaluate `if` condition.
2. If `True`: Execute block and exit the chain.
3. If `False`: Move to the next `elif` (if any).
4. Repeat until a `True` condition is found or the `else` block is reached.

**6. Common Exam Traps**

* **Trap:** Multiple `if` statements vs. `if/elif`. **Fact:** Multiple `if` statements are independent; more than one can run. In an `if/elif` chain, only the first `True` one runs.
* **Trap:** Forgetting the colon (`:`) at the end of the line.

**7. Worked Example**

```python
score = 85
if score >= 90:
    print("A")
elif score >= 80:
    print("B")  # This runs, then the structure exits.
elif score >= 70:
    print("C")
# Output: B (even though 85 is also >= 70)

```

3. How many `elif` blocks are allowed in one sequence?
**Answers: 1. No. 2. The block executes, and the rest of the chain is skipped. 3. Unlimited.**

### for loops

**1. Core Definition**
A "definite" iteration statement used to iterate over a **sequence** (like a list, string, or range) or any other iterable object.

**2. Why It Matters**
Used when the number of iterations is known or defined by a collection. Exams test loop variables and sequence unpacking.

**3. Core Concepts**

* **Iterable:** The object being looped over.
* **Loop Variable:** The temporary name assigned to each item in the sequence during each step.
* **Body:** The indented code that runs for every item.

**4. Formulas / Rules**

**5. Common Exam Traps**

### while loops

**1. Core Definition**
An "indefinite" iteration statement that repeats a block of code as long as a specified condition remains `True`.

**2. Why It Matters**
Used when you don't know how many times you need to loop (e.g., waiting for user input). Exams test your ability to spot potential infinite loops.

**3. Core Concepts**

* **Condition:** Checked *before* every iteration.
* **Infinite Loop:** A loop where the condition never becomes `False`.
* **Iteration:** One single pass through the loop body.

**4. Formulas / Rules**

* `while condition:`
* If the condition is `False` at the start, the loop never runs.

**5. Common Exam Traps**

### break, continue, pass

### Nested loops

### String methods

**1. Core Definition**
Built-in functions that are part of the string class (`str`) used to perform common operations on string objects. Since strings are **immutable**, these methods always return a **new** string rather than modifying the original.

### String slicing

**1. Core Definition**
The process of extracting a subset (substring) of a string using a specific range of indices.

**2. Why It Matters**
Fundamental for string manipulation. Exams test your ability to handle positive/negative indices and the "exclusive" stop index.

**3. Core Concepts**

* **Syntax:** `string[start:stop:step]`
* **Zero-based Indexing:** The first character is `0`.
* **Negative Indexing:** The last character is `-1`.
* **Slicing Rule:** Includes `start`, but **excludes** `stop`.

**4. Formulas / Rules**

* **Default Start:** `0`
* **Default Stop:** Length of the string.
* **Default Step:** `1`
* **Reversing a string:** `s[::-1]`

**5. Step-by-Step Process**

1. Identify the `start` index (inclusive).
2. Identify the `stop` index (exclusive).
3. Determine the `step` (direction and jump size).

**6. Common Exam Traps**

* **Trap:** Out-of-bounds error. **Fact:** Slicing is forgiving. `s[0:100]` on a 5-char string just returns the whole string; it does **not** crash.
* **Trap:** `s[5:2]` returns an empty string because the step is positive but the range moves backward.

**7. Worked Example**

```python
word = "Python"
print(word[0:2])  # "Py" (indices 0 and 1)
print(word[1:-1]) # "ytho" (index 1 to second-to-last)
print(word[::2])  # "Pto" (every second char)

```

### String formatting (f-strings, .format())

**1. Core Definition**
Techniques used to inject variables or expressions into a string template for dynamic display.

**2. Why It Matters**
f-strings are the modern standard (Python 3.6+) and are high-yield for their speed and readability. Exams may still test the older `.format()` method.

**3. Core Concepts**

* **f-strings:** Prefixed with `f` or `F`. Variables/expressions are placed inside curly braces `{}`.
* **.format():** Uses `{}` as placeholders, filled by arguments passed to the method.
* **Padding/Alignment:** `{:>10}` (Right align), `{:.<10}` (Left align with dots).

**4. Formulas / Rules**

* **f-string:** `f"Text {variable} text"`
* **.format():** `"Text {} text".format(variable)`

**5. Common Exam Traps**

### Immutability

**1. Core Definition**
A tuple is an ordered, **immutable** collection of items. Once a tuple is created, its elements and their order cannot be changed, added, or removed.

**2. Why It Matters**
Immutability provides "data integrity." If you have a collection of constants (like coordinates or configuration settings), using a tuple ensures they aren't accidentally modified by other parts of the program.

**3. Core Concepts**

* **Parentheses `()`:** The standard way to define a tuple.
* **The Trailing Comma Rule:** To create a tuple with only **one** item, you *must* include a trailing comma (e.g., `(5,)`). Without it, Python treats `(5)` as an integer in parentheses.
* **Hashability:** Because they are immutable, tuples containing only immutable elements can be used as keys in a dictionary (unlike lists).

**4. Formulas / Rules**

* **Legal:** `t = (1, 2, 3)`; `x = t[0]` (Accessing is fine).

**5. Common Exam Traps**

* **Trap:** The "Mutable Element" loophole. If a tuple contains a mutable object (like a list), you **can** modify the contents of that list, but you cannot replace the list object itself within the tuple.
* **Trap:** `t = (1)`. **Fact:** `type(t)` is `int`. `t = (1,)` is a `tuple`.

**6. Worked Example**

```python
point = (10, 20)
# point[0] = 15  # ERROR: 'tuple' object does not support item assignment

# Loophole Example
nested = ([1, 2], 3)
nested[0].append(3) # LEGAL: Modifying the list inside the tuple
print(nested)       # ([1, 2, 3], 3)

```

### Packing/unpacking

**1. Core Definition**

* **Packing:** The process of grouping several values into a single tuple.
* **Unpacking:** The process of assigning the elements of a tuple back into individual variables.

**2. Why It Matters**
High-yield for returning multiple values from a function and for "Pythonic" variable swapping.

**3. Core Concepts**

* **Implicit Packing:** You don't actually need parentheses to create a tuple; commas are the true trigger (e.g., `t = 1, 2, 3`).
* **Unpacking Symmetry:** The number of variables on the left must match the number of items in the tuple on the right.
* **Extended Unpacking (`*`):** Using an asterisk to capture "the rest" of the items into a list.

**4. Formulas / Rules**

* **Variable Swap:** `a, b = b, a` (Python packs the right side into a temporary tuple, then unpacks it into the left side).

**5. Step-by-Step Process (Unpacking)**

1. Evaluate the tuple on the right.
2. Assign the first item to the first variable, second to second, etc.
3. If using `*`, collect all remaining items into a list.

**6. Worked Example**

```python
# Unpacking
data = ("Charles", 30, "Developer")
name, age, job = data

# Extended Unpacking
nums = (1, 2, 3, 4, 5)
first, *middle, last = nums
# first = 1, middle = [2, 3, 4], last = 5

```

**7. Memory Hooks**

* **Packing = Suitcase:** Putting all your clothes into one bag.
* **Unpacking = Arrival:** Taking each item out of the bag to put them in specific drawers.

### Set operations

**1. Core Definition**
A set is an unordered collection of **unique**, hashable elements. Sets are primarily used for mathematical set theory operations like union, intersection, and difference.

**2. Why It Matters**
High-yield for comparing two data sources (e.g., "Find all users who are in Group A but not in Group B"). Exams test both the method names (e.g., `.union()`) and the shorthand operators (e.g., `|`).

**3. Core Concepts**

* **Curly Braces `{}`:** Used to define a set with initial values.
* **Unordered:** Items have no index. You cannot access `my_set[0]`.
* **Unique:** Duplicate values are automatically removed.

**4. Formulas / Rules**

* **Union (`|`):** All unique elements from both sets. `A | B`
* **Intersection (`&`):** Only elements present in **both** sets. `A & B`
* **Difference (`-`):** Elements in the first set but **not** the second. `A - B`
* **Symmetric Difference (`^`):** Elements in either set, but **not both**. `A ^ B`

**5. Common Exam Traps**

* **Trap:** Using `{}` to create an empty set. **Fact:** `{}` creates an empty **dictionary**. You must use `set()` for an empty set.
* **Trap:** Expecting a specific order. Sets do not preserve the order in which items were added.

**6. Worked Example**

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a & b) # {3} (Intersection)
print(a | b) # {1, 2, 3, 4, 5} (Union)
print(a - b) # {1, 2} (Difference)

```

**Answers: 1. `&`. 2. `set()`. 3. Symmetric Difference (`^`).**

### Hashing

**1. Core Definition**
Hashing is the process of converting an object into a fixed-size integer (a "hash value") using a hash function. This value acts as a unique fingerprint for the object's data.

**2. Why It Matters**
This is the technical "engine" that makes sets and dictionaries extremely fast. Sets use the hash value to instantly locate an item without scanning the entire collection.

**3. Core Concepts**

* **Hashable:** An object is hashable if its hash value never changes during its lifetime.
* **Immutability:** Only immutable objects (strings, ints, tuples of immutables) are hashable.
* **Non-hashable:** Lists, sets, and dictionaries are **not** hashable because they are mutable.

**4. Formulas / Rules**

* **Equality Rule:** If `a == b`, then `hash(a)` **must** equal `hash(b)`.

**5. Common Exam Traps**

* **Trap:** Trying to put a list into a set. **Result:** `TypeError: unhashable type: 'list'`.
* **Trap:** Thinking two different objects can't have the same hash. **Fact:** Rare "hash collisions" can occur, but Python handles them internally.

### Removing duplicates

**1. Core Definition**
The process of filtering a collection to ensure it contains only unique items by taking advantage of the set's "unique elements only" property.

**2. Why It Matters**
This is the single most common real-world use for sets. It is significantly faster than using a `for` loop to check for duplicates manually.

**3. Step-by-Step Process**

1. Take a list with duplicates: `data = [1, 2, 2, 3]`
2. Convert to a set: `unique_set = set(data)`
3. (Optional) Convert back to a list: `clean_list = list(unique_set)`

**4. Common Exam Traps**

* **Trap:** Order loss. **Fact:** Converting a list to a set and back to a list **will** lose the original order of the items.

**5. Worked Example**

```python
names = ["Alice", "Bob", "Alice", "Charlie"]
unique_names = list(set(names))
# Output: ['Bob', 'Alice', 'Charlie'] (Order may vary)

```

### Membership testing

**1. Core Definition**
Checking whether a specific value exists within a collection using the `in` or `not in` keywords.

**2. Why It Matters**
Sets are optimized for this operation. Checking `x in my_set` takes **Constant Time ($O(1)$)**, whereas checking `x in my_list` takes **Linear Time ($O(n)$)**.

**3. Core Concepts**

* **Complexity:** In a set of 1,000,000 items, finding one item is nearly instantaneous.
* **Efficiency:** Because of hashing, Python knows exactly "where" to look for the item's fingerprint.

**4. Formulas / Rules**

* **Syntax:** `value in collection` -> Returns `True` or `False`.

**5. Common Exam Traps**

* **Trap:** Thinking sets are better for *everything*. **Fact:** Sets are faster for searching, but lists are better if you need to maintain order or allow duplicates.

**6. Active Recall Questions**

1. Is `x in my_set` faster than `x in my_list` for large datasets?
2. What is the Big-O time complexity of a membership test in a set?
**Answers: 1. Yes, significantly. 2. $O(1)$ (Constant Time).**

### Frozen sets

**1. Core Definition**
A `frozenset` is an **immutable** version of a set. Once created, elements cannot be added, removed, or changed.

**2. Why It Matters**
Since `frozenset` is immutable, it is **hashable**. This allows you to use a set as a key in a dictionary or as an element inside another set.

**3. Core Concepts**

* **Creation:** `fs = frozenset([1, 2, 3])`
* **Static:** No `.add()`, `.remove()`, or `.pop()` methods.
* **Operations:** Supports all non-modifying operations like `union` and `intersection`.

**4. Worked Example**

```python
# A set of sets is impossible with normal sets:
# sets = {{1, 2}, {3, 4}} # TypeError

# Use frozensets instead:
sets = {frozenset([1, 2]), frozenset([3, 4])} # Legal!

```

### Key-value pairs

**1. Core Definition**
A dictionary (`dict`) is a mutable, unordered (by default in logic, though ordered by insertion since Python 3.7) collection that stores data in **key-value pairs**. Each **key** maps to a specific **value**.

**2. Why It Matters**
Dictionaries are the most important data structure for organizing data by "label" rather than "position." Exams test your understanding that keys must be unique and hashable.

**3. Core Concepts**

* **Key:** Must be an immutable/hashable type (string, int, float, or tuple of immutables).
* **Value:** Can be any Python object (including lists or other dictionaries).
* **Syntax:** `{key: value, key2: value2}`.

**4. Formulas / Rules**

* **Uniqueness Rule:** If you assign a value to an existing key, the old value is overwritten.
* **Access Rule:** `dict[key]` returns the value. If the key is missing, it raises a `KeyError`.

**5. Worked Example**

```python
user = {"name": "Charles", "age": 30}
user["age"] = 31 # Updates existing key
user["city"] = "Voorhees" # Adds new key-value pair

```

**6. Active Recall Questions**

1. Can a list be used as a dictionary key?
2. What happens if you try to access a key that doesn't exist?

**Answers: 1. No (lists are mutable/unhashable). 2. A `KeyError` is raised. 3. `my_dict = {}` or `my_dict = dict()`.**

### Hash tables (how dict works internally)

**1. Core Definition**
Python dictionaries are implemented as **Hash Tables**. A hash function converts the key into an integer (hash value), which is then used as an index in an underlying array (the "bucket") to store the value.

**2. Why It Matters**
Understanding the "under the hood" mechanics explains why dictionary lookups are so fast. Exams often ask why keys must be hashable.

**3. Core Concepts**

* **Hash Function:** Maps a key to a specific index.
* **Collision:** When two different keys produce the same hash/index (Python handles this using "open addressing").
* **O(1) Access:** Because the index is calculated directly from the key, finding a value doesn't require searching through the whole dictionary.

**4. Common Exam Traps**

* **Trap:** Thinking dictionaries search linearly like a list. **Fact:** They jump directly to the memory location via the hash.

**5. Active Recall Questions**

1. What internal data structure powers a Python dictionary?
2. Why is a dictionary lookup faster than a list search?
3. What is a "hash collision"?
**Answers: 1. A Hash Table. 2. Because dictionaries use hashing for direct access ($O(1)$) instead of scanning ($O(n)$). 3. When two different keys result in the same hash index.**

### Dictionary methods

**1. Core Definition**
Built-in functions used to safely access, modify, or view the contents of a dictionary.

**2. Why It Matters**
Crucial for writing robust code. Methods like `.get()` prevent program crashes when a key might be missing.

**3. Core Concepts**

* **`.get(key, default)`:** Returns the value for a key; returns `None` (or a custom default) if the key is missing. **Safe access.**
* **`.keys()`:** Returns a view of all keys.
* **`.values()`:** Returns a view of all values.
* **`.items()`:** Returns a view of all key-value pairs (as tuples).
* **`.update(other_dict)`:** Merges another dictionary into the current one.
* **`.pop(key)`:** Removes the key and returns its value.

**4. Worked Example**

```python
counts = {"apples": 5}
print(counts.get("oranges", 0)) # Output: 0 (No KeyError!)

```

### Iteration patterns

**1. Core Definition**
The specific ways to use loops to traverse the data stored in a dictionary.

**2. Why It Matters**
Frequently tested in code-writing or trace-the-output questions. Understanding unpacking in loops is essential.

**3. Step-by-Step Process**

* **Iterate over Keys (Default):** `for k in my_dict:`
* **Iterate over Values:** `for v in my_dict.values():`
* **Iterate over Both:** `for k, v in my_dict.items():` (Most common for processing).

**4. Worked Example**

```python
prices = {"Bread": 2, "Milk": 3}
for item, price in prices.items():
    print(f"{item} costs ${price}")

```

**5. Active Recall Questions**

### Dictionary comprehensions

**1. Core Definition**
A concise syntax to create a dictionary from an iterable in a single line.

**2. Why It Matters**
Identical logic to list comprehensions but creates key-value pairs. High-yield for data transformation (e.g., swapping keys and values).

**3. Formulas / Rules**

* **Syntax:** `{key_expression: value_expression for item in iterable}`

**4. Worked Example**

```python
# Create a dictionary of squares
squares = {x: x**2 for x in range(5)}
# Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

# Swap keys and values
original = {"a": 1, "b": 2}
swapped = {v: k for k, v in original.items()}

```

### Ordered dict behavior

**1. Core Definition**
Starting in Python 3.7, the standard `dict` maintains the **insertion order** of keys by default.

**2. Why It Matters**
Older versions of Python (pre-3.6) had "unordered" dictionaries where the loop order was random. This is an "edge case" for legacy code but a "feature" in modern Python.

**3. Core Concepts**

* **Insertion Order:** The order in which you add keys is the order they appear when looping.
* **`collections.OrderedDict`:** A specialized class used in older versions or for specific behaviors (like equality checking where order matters).

**4. Active Recall Questions**

### Performance complexity

**1. Core Definition**
The analysis of time and memory efficiency for dictionary operations.

**2. Why It Matters**
Crucial for big-data optimization. Dictionaries sacrifice memory (to keep the hash table sparse) to gain extreme speed.
