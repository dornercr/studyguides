# 🧠 SDLC Quick Study Guide (6 Phases)

## 🔁 SDLC = Software Development Life Cycle
Steps to **build**, **test**, and **maintain** software.

---

## 1️⃣ Planning
- Set **goals**, **timeline**, **budget**
- Organize **team and tasks**
- Identify **risks**

🧠 Why: Avoid confusion and stay on track

---

## 2️⃣ Requirements
- Ask what the **software must do**
- Includes **features**, **performance**, **security**

🧠 Why: Build what users really need

---

## 3️⃣ Design
- Make a plan: **UI layout**, **database schema**, **algorithms**
- Create **high-level** and **detailed** designs

🧠 Why: Clear plans help developers build faster and better

---

## 4️⃣ Development
- Write the actual **code**
- Build **front-end**, **back-end**, and **database logic**

🧠 Why: Code turns plans into a working app

---

## 5️⃣ Testing
- Find **bugs**
- Check if everything works correctly
- Use **unit**, **integration**, and **user testing**

🧠 Why: Catch problems before release

---

## 6️⃣ Deployment & Maintenance
- Release the software to users
- Fix bugs and push updates over time

🧠 Why: Keep software working and useful long-term

---

## 🔑 Key Benefits of SDLC
✅ Keeps projects **organized**  
✅ Improves **teamwork and communication**  
✅ Builds **higher-quality** software  
✅ **Reduces risk**  
✅ Helps improve **future projects**

# 🧠 Defining Requirements – Study Guide

## 📌 What Is the Requirements Phase?
The requirements phase is where we determine **what the software must do** and **how well it must perform** to support many users and tasks.

---

## 🎯 Goal of This Phase
✅ To gather and define the **needs and expectations** of users, developers, and stakeholders  
🛠️ Helps the team build the **right software** from the beginning

---

## 🔑 Steps to Identify Requirements
- **Analyze workload and users**  
  → How many users? How much data? What kind of activity?

- **Set performance metrics (KPIs)**  
  → Speed, uptime, system load, etc.

- **Plan for future growth**  
  → Ensure scalability

- **Talk with stakeholders**  
  → Collect input from developers, designers, users, etc.

- **Prioritize requirements**  
  → Focus first on what matters most

---

## 🔍 Two Types of Requirements

### 📘 Functional Requirements
Describe **what the system does**  
Examples:
- Logging in  
- Processing payments  
- Generating reports  

🎯 Define **tasks and features**

---

### 📗 Nonfunctional Requirements
Describe **how the system performs**  
Examples:
- Handles 1,000 users at once  
- Loads pages in under 2 seconds  
- Keeps data safe and secure  

🎯 Define **quality attributes** like performance, security, and reliability

---

## ✅ Knowledge Check Answers

✅ To gather and define the needs and expectations of users and stakeholders

# 🧠 Unit 1.3 – The Design Phase

## 📌 What Happens in the Design Phase?
This is when developers **plan how the software will be built**.

It includes:
- Big-picture decisions
- Small details that ensure security, usability, and scalability

---

## 🔑 Key Focus Areas

🔒 **Security**  
- Design must protect data and stop unauthorized access

🎨 **User Interface (UI)**  
- Must be clear, simple, and easy to use

🧱 **Architecture**  
- Plan how different parts of the system connect  
- Ensure **scalability** (growth) and **flexibility** (change)

---

## 🧩 UML Diagrams
**UML = Unified Modeling Language**

- Visualize system **structure and interactions**
- Used to communicate **how parts work together**

---

## 🔧 Two Levels of Design

### 1️⃣ High-Level Design (HLD)
- Big-picture view of the system  
- Describes main **modules** and **their relationships**

### 2️⃣ Low-Level Design (LLD)
- Zoomed-in view of **component internals**  
- Specifies:
  - Detailed logic
  - Functions
  - Class-level design
  - Interactions between parts

---
